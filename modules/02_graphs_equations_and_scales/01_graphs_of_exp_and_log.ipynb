# Graphs of $b^x$ and $\log_b x$

This notebook draws the two functions side by side and reads everything off the picture: why $\log_b x$ is the mirror image of $b^x$ in the line $y = x$, why the logarithm only accepts positive inputs, where the asymptotes come from, what the base does to the shape, how shifts and stretches move the curves, and just how **slowly** a logarithm grows.

This is notebook 01 of module 02 (**Graphs, equations and scales**) in the **Learning Logarithms** series.

## 1. Introduction

A googol is $10^{100}$ — a $1$ followed by a hundred zeros, vastly more than the roughly $10^{80}$ atoms in the observable universe. Try to draw the graph of $y = 10^x$ on ordinary paper with $1$ cm per unit on both axes. At $x = 2$ the curve is already $1$ metre high; at $x = 9$ it is $10{,}000$ km high, close to the diameter of the Earth; at $x = 100$ it is far beyond any distance in astronomy.

Now draw the **other** graph, $y = \log_{10} x$. Walk to the right for a googol centimetres — about $10^{82}$ light-years — and the curve has climbed to height

$$
\log_{10} 10^{100} = 100 \text{ cm} = 1 \text{ metre}.
$$

The two curves are made of exactly the same points, with $x$ and $y$ swapped. One explodes, the other barely moves. Understanding one graph *is* understanding the other, and that mirror relationship explains nearly every property of logarithms that you need on a graph: the domain, the asymptote, the effect of the base, and the speed of growth.

## 2. Theory

### 2.1 One set of points, two readings

By definition

$$
b^y = x \quad\Longleftrightarrow\quad y = \log_b x \qquad (b > 0,\; b \neq 1).
$$

So the point $(p, q)$ lies on $y = b^x$ exactly when $(q, p)$ lies on $y = \log_b x$. Swapping the coordinates of every point is the reflection in the diagonal $y = x$. This is true for **every** inverse pair $f, f^{-1}$; for $b^x$ and $\log_b x$ it gives the table

| $y = 2^x$ | $(-2, \tfrac14)$ | $(-1, \tfrac12)$ | $(0, 1)$ | $(1, 2)$ | $(2, 4)$ | $(3, 8)$ |
|---|---|---|---|---|---|---|
| $y = \log_2 x$ | $(\tfrac14, -2)$ | $(\tfrac12, -1)$ | $(1, 0)$ | $(2, 1)$ | $(4, 2)$ | $(8, 3)$ |

### 2.2 Domain, range, intercept, asymptote

Every property of $\log_b x$ is a property of $b^x$ with the roles of $x$ and $y$ exchanged:

| | $y = b^x$ | $y = \log_b x$ |
|---|---|---|
| domain | all real $x$ | $x > 0$ |
| range | $y > 0$ | all real $y$ |
| fixed point | passes through $(0, 1)$ | passes through $(1, 0)$ |
| second landmark | $(1, b)$ | $(b, 1)$ |
| asymptote | horizontal, $y = 0$ | vertical, $x = 0$ |

**Why $x > 0$?** Because $b^y$ is positive for every real $y$, there is no exponent that produces $0$ or a negative number. The question "$2^{\,?} = -4$" has no real answer, so $\log_2(-4)$ is undefined (over the reals; module 07 revisits it with complex numbers).

**Why an asymptote?** As $y \to -\infty$, $b^y \to 0$ (for $b > 1$) without ever reaching it: $2^{-10} \approx 0.001$, $2^{-20} \approx 10^{-6}$. Reflected, $\log_b x \to -\infty$ as $x \to 0^+$: the log curve plunges down along the $y$-axis.

### 2.3 The effect of the base

**Case $b > 1$.** $b^x$ is increasing, so $\log_b x$ is increasing. The larger $b$, the steeper $b^x$ and the *flatter* $\log_b x$. Change of base makes this precise:

$$
\log_b x = \frac{\ln x}{\ln b}.
$$

Every logarithm graph is the **same curve** $y = \ln x$, stretched vertically by the factor $1/\ln b$. That is why they all pass through $(1, 0)$ and differ only in steepness.

**Case $0 < b < 1$.** Now $\ln b < 0$, so $\log_b x = -\ln x / |\ln b|$ is $\log_{1/b} x$ upside down:

$$
\log_{1/b} x = -\log_b x, \qquad (1/b)^x = b^{-x}.
$$

So $y = (1/2)^x$ is $y = 2^x$ reflected in the $y$-axis (exponential **decay**), and $y = \log_{1/2} x$ is $y = \log_2 x$ reflected in the $x$-axis (a decreasing logarithm). Domain, range and asymptotes are unchanged.

**Why $b \neq 1$?** $1^x = 1$ for every $x$: the graph is a horizontal line, not one-to-one, so it has no inverse. In the formula, $\ln 1 = 0$ would sit in the denominator.

### 2.4 Transformations

The standard moves apply as for any function. For

$$
y = a \log_b(x - h) + k \qquad\text{and}\qquad y = a\, b^{\,x - h} + k,
$$

- $h$ shifts right, $k$ shifts up, $a$ stretches vertically (and reflects if $a < 0$).
- The **log** curve needs $x - h > 0$: domain $x > h$, vertical asymptote $x = h$.
- The **exponential** curve has range $y > k$ (for $a > 0$), horizontal asymptote $y = k$.

Two moves are special to these functions, because the laws of logarithms turn them into other moves:

- A horizontal **stretch** of a log graph is a vertical **shift**: $\log_b(cx) = \log_b x + \log_b c$.
- A horizontal **shift** of an exponential graph is a vertical **stretch**: $b^{x - h} = b^{-h} \cdot b^x$.

### 2.5 How slowly the logarithm grows

To raise $\log_{10} x$ by $1$ you must multiply $x$ by $10$. Equal steps upward cost equal **ratios** to the right — the core picture of this series, now as a graph. Concretely:

| $x$ | $10$ | $1000$ | $10^6$ | $10^{100}$ (googol) |
|---|---|---|---|---|
| $\log_{10} x$ | $1$ | $3$ | $6$ | $100$ |
| $\ln x$ | $2.30$ | $6.91$ | $13.8$ | $230.3$ |

The logarithm is **unbounded** ($\log_b x$ eventually passes any number $M$, at $x = b^M$) yet it grows more slowly than any positive power of $x$. Its slope makes this visible: $\frac{d}{dx}\ln x = \frac1x$, so at $x = 10^6$ the curve rises one millionth per unit step. Notebook 04 of this module turns this into the growth hierarchy $\ln n \ll n^\varepsilon$.

## 3. Worked Examples

### Example 1 — key points of $y = \log_3 x$

Read the exponential table $3^{-1} = \tfrac13$, $3^0 = 1$, $3^1 = 3$, $3^2 = 9$ backwards: the log curve passes through $(\tfrac13, -1)$, $(1, 0)$, $(3, 1)$, $(9, 2)$. Domain $x > 0$, range all reals, asymptote $x = 0$.

### Example 2 — domain of $f(x) = \log_2(x^2 - 4x + 3)$

The argument must be positive: $x^2 - 4x + 3 = (x - 1)(x - 3) > 0$. A product of two factors is positive when both have the same sign: $x < 1$ or $x > 3$. Domain: $(-\infty, 1) \cup (3, \infty)$, with vertical asymptotes at $x = 1$ and $x = 3$.

### Example 3 — the transformed curve $y = 2\log_{10}(x + 3) - 1$

- $h = -3$: shift left by $3$. Domain $x > -3$, asymptote $x = -3$.
- Landmarks: $x + 3 = 1 \Rightarrow x = -2$, $y = -1$; $x + 3 = 10 \Rightarrow x = 7$, $y = 2 - 1 = 1$.
- $x$-intercept: $\log_{10}(x + 3) = \tfrac12 \Rightarrow x + 3 = \sqrt{10} \Rightarrow x = \sqrt{10} - 3 \approx 0.162$.

### Example 4 — the inverse of $f(x) = 3 \cdot 2^x - 5$

Solve $y = 3 \cdot 2^x - 5$ for $x$: $2^x = \frac{y + 5}{3}$, so $x = \log_2\frac{y + 5}{3}$. Hence

$$
f^{-1}(x) = \log_2\!\left(\frac{x + 5}{3}\right), \qquad \text{domain } x > -5.
$$

The horizontal asymptote $y = -5$ of $f$ reflects to the vertical asymptote $x = -5$ of $f^{-1}$, and the range $y > -5$ of $f$ becomes the domain of $f^{-1}$. Check: $f(1) = 1$, and $f^{-1}(1) = \log_2 2 = 1$ ✓.

### Example 5 — a horizontal stretch that is a shift

$y = \log_2(8x) = \log_2 8 + \log_2 x = 3 + \log_2 x$. Squeezing the graph of $\log_2 x$ horizontally by a factor $8$ lifts it by exactly $3$.

### Example 6 — how far right for $\ln x = 100$?

$x = e^{100} = 10^{100 / \ln 10} = 10^{43.43} \approx 2.7 \times 10^{43}$. With $1$ cm per unit, that is about $3 \times 10^{25}$ light-years — and the curve is only $1$ metre high.

## 4. Python Examples

### 4.1 Swapping coordinates

Points on $y = 2^x$, swapped, land on $y = \log_2 x$ — checked with a tolerance, not with `==`.

```python
import math

for x in [-2, -1, 0, 0.5, 1, 2, 3]:
    y = 2**x
    # (x, y) on y = 2^x  ->  (y, x) should be on y = log2(x)
    swapped_ok = math.isclose(math.log2(y), x, abs_tol=1e-12)
    print(f"({x:>4}, {y:8.4f}) on 2^x   ->  ({y:8.4f}, {x:>4}) on log2 x: {swapped_ok}")
```

### 4.2 Outside the domain

`math.log` refuses non-positive inputs; NumPy returns `nan` (and `-inf` at zero) with a warning, which we silence here so the output stays readable.

```python
import math

import numpy as np

for value in [8, 0, -4]:
    try:
        print(f"math.log2({value}) = {math.log2(value)}")
    except ValueError as error:
        print(f"math.log2({value}) -> ValueError: {error}")

with np.errstate(divide="ignore", invalid="ignore"):
    print("np.log2([8, 0, -4]) =", np.log2(np.array([8.0, 0.0, -4.0])))
```

### 4.3 Every log is a stretched $\ln$

Check $\log_b x = \ln x / \ln b$ and see the stretch factor $1/\ln b$ for several bases. For $b < 1$ the factor is negative: the curve is flipped.

```python
import math

import numpy as np

x_values = np.array([0.25, 0.5, 2.0, 10.0, 1000.0])
for base in [2, math.e, 10, 0.5]:
    stretch = 1 / math.log(base)
    via_change_of_base = np.log(x_values) * stretch
    direct = np.array([math.log(x, base) for x in x_values])
    print(f"b = {base:6.4f}: stretch 1/ln b = {stretch:+.4f}, matches math.log(x, b): {np.allclose(via_change_of_base, direct)}")
```

### 4.4 Domain of a composite argument with SymPy

SymPy can solve the inequality "argument $> 0$" for Example 2 and verify the inverse of Example 4.

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)
argument = x**2 - 4 * x + 3
print("domain of log2(x^2 - 4x + 3):", sp.solve_univariate_inequality(argument > 0, x, relational=False))

f = 3 * 2**x - 5
inverse_candidates = sp.solve(sp.Eq(y, f), x)
print("solve y = 3*2^x - 5 for x:", inverse_candidates)
f_inverse = sp.log((x + 5) / 3, 2)
print("f(f_inverse(x)) simplifies to:", sp.simplify(f.subs(x, f_inverse)))
```

### 4.5 How slowly: the logarithm of huge numbers

Python integers are exact, so `math.log10` of a googol works even though the number has 101 digits.

```python
import math

googol = 10**100
print("digits of a googol:", len(str(googol)))
print("log10(googol) =", math.log10(googol))
print("ln(googol)    =", round(math.log(googol), 2))
print("log2(googol)  =", round(math.log2(googol), 2))
print()
print("x with ln x = 100: e^100 =", f"{math.exp(100):.3e}")
print("x with log10 x = 1000 has", 1000 + 1, "digits")
```

## 5. Visualizations

### 5.1 The mirror in $y = x$

Both curves, the mirror line, and a few pairs of matching points joined by dashed segments perpendicular to $y = x$.

```python
import matplotlib.pyplot as plt
import numpy as np

x_exp = np.linspace(-4, 3.2, 400)
x_log = np.linspace(0.01, 9, 400)

fig, ax = plt.subplots(figsize=(7, 7))
ax.plot(x_exp, 2**x_exp, color="tab:blue", label=r"$y = 2^x$")
ax.plot(x_log, np.log2(x_log), color="tab:orange", label=r"$y = \log_2 x$")
ax.plot([-4, 9], [-4, 9], "k--", linewidth=1, label=r"$y = x$")

for p in [-2, 0, 1, 2, 3]:
    q = 2.0**p
    ax.plot([p, q], [q, p], ":", color="gray")
    ax.plot(p, q, "o", color="tab:blue")
    ax.plot(q, p, "o", color="tab:orange")

ax.axhline(0, color="black", linewidth=0.6)
ax.axvline(0, color="black", linewidth=0.6)
ax.set_xlim(-4, 9)
ax.set_ylim(-4, 9)
ax.set_aspect("equal")
ax.grid(True, alpha=0.3)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend(loc="upper left")
ax.set_title("Each point on $2^x$ has a mirror point on $\\log_2 x$")
plt.show()
```

The horizontal asymptote $y = 0$ of $2^x$ (far left) reflects into the vertical asymptote $x = 0$ of $\log_2 x$ (bottom). The intercept $(0, 1)$ reflects into $(1, 0)$.

### 5.2 The effect of the base

Left: exponentials with bases above and below $1$. Right: their logarithms. Bases $b$ and $1/b$ are drawn in the same colour — solid for $b > 1$, dashed for $1/b$.

```python
import matplotlib.pyplot as plt
import numpy as np

bases = [1.5, 2, np.e, 10]
colours = ["tab:green", "tab:blue", "tab:purple", "tab:red"]
labels = ["1.5", "2", "e", "10"]
x_exp = np.linspace(-3, 3, 400)
x_log = np.linspace(0.02, 6, 400)

fig, (ax_exp, ax_log) = plt.subplots(1, 2, figsize=(12, 5))
for base, colour, label in zip(bases, colours, labels):
    ax_exp.plot(x_exp, base**x_exp, color=colour, label=f"$b = {label}$")
    ax_exp.plot(x_exp, (1 / base) ** x_exp, color=colour, linestyle="--")
    ax_log.plot(x_log, np.log(x_log) / np.log(base), color=colour, label=f"$b = {label}$")
    ax_log.plot(x_log, np.log(x_log) / np.log(1 / base), color=colour, linestyle="--")

ax_exp.set_ylim(0, 6)
ax_exp.set_title("$y = b^x$ (solid) and $y = (1/b)^x$ (dashed)")
ax_log.set_ylim(-4, 4)
ax_log.set_title("$y = \\log_b x$ (solid) and $y = \\log_{1/b} x$ (dashed)")
for ax in (ax_exp, ax_log):
    ax.axhline(0, color="black", linewidth=0.6)
    ax.axvline(0, color="black", linewidth=0.6)
    ax.grid(True, alpha=0.3)
    ax.set_xlabel("x")
    ax.legend(loc="lower right")
ax_exp.plot(0, 1, "ko")
ax_log.plot(1, 0, "ko")
plt.tight_layout()
plt.show()
```

All exponentials pass through $(0, 1)$, all logarithms through $(1, 0)$. A larger base makes $b^x$ steeper and $\log_b x$ flatter. The dashed curves are mirror images of the solid ones: in the $y$-axis on the left, in the $x$-axis on the right.

### 5.3 Transformations and the moving asymptote

The curve $y = 2\log_{10}(x + 3) - 1$ of Example 3, built up step by step from $y = \log_{10} x$. The asymptote moves with the horizontal shift and with nothing else.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(9, 5.5))
x_base = np.linspace(0.005, 12, 600)
x_shifted = np.linspace(-3 + 0.005, 12, 600)

ax.plot(x_base, np.log10(x_base), color="gray", label=r"$\log_{10} x$")
ax.plot(x_shifted, np.log10(x_shifted + 3), color="tab:blue", linestyle="--", label=r"$\log_{10}(x+3)$")
ax.plot(x_shifted, 2 * np.log10(x_shifted + 3), color="tab:green", linestyle="-.", label=r"$2\log_{10}(x+3)$")
ax.plot(x_shifted, 2 * np.log10(x_shifted + 3) - 1, color="tab:red", linewidth=2, label=r"$2\log_{10}(x+3) - 1$")
ax.axvline(-3, color="tab:red", linestyle=":", label="asymptote $x = -3$")
ax.axvline(0, color="black", linewidth=0.6)
ax.axhline(0, color="black", linewidth=0.6)

for x_point, y_point in [(-2, -1), (7, 1), (np.sqrt(10) - 3, 0)]:
    ax.plot(x_point, y_point, "o", color="tab:red")

ax.set_xlim(-4, 12)
ax.set_ylim(-4, 3)
ax.grid(True, alpha=0.3)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend(loc="lower right")
ax.set_title("Shift left 3, stretch by 2, shift down 1")
plt.show()
```

### 5.4 How slowly the logarithm grows

Left: $\ln x$ against $\sqrt{x}$ and $x$ on ordinary axes up to $x = 1000$ — the logarithm looks almost flat. Right: $\log_{10} x$ against $x$ all the way to a googol, with a **logarithmic** $x$-axis. On a log axis, $\log_{10} x$ is a straight line (it *is* the axis position), and it reaches only $100$.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, (ax_linear, ax_logx) = plt.subplots(1, 2, figsize=(12, 4.8))

x_values = np.linspace(1, 1000, 1000)
ax_linear.plot(x_values, np.log(x_values), label=r"$\ln x$")
ax_linear.plot(x_values, np.sqrt(x_values), label=r"$\sqrt{x}$")
ax_linear.plot(x_values, x_values / 10, label=r"$x / 10$")
ax_linear.set_xlabel("x (linear scale)")
ax_linear.set_ylabel("y (linear scale)")
ax_linear.set_title("Up to x = 1000")
ax_linear.grid(True, alpha=0.3)
ax_linear.legend()

exponents = np.linspace(0, 100, 500)
huge_x = 10.0**exponents
ax_logx.plot(huge_x, np.log10(huge_x), color="tab:red", label=r"$\log_{10} x$")
ax_logx.plot(huge_x, np.log(huge_x), color="tab:purple", label=r"$\ln x$")
ax_logx.set_xscale("log")
ax_logx.set_xlabel("x (log scale)")
ax_logx.set_ylabel("y (linear scale)")
ax_logx.set_title("Up to a googol, $10^{100}$")
ax_logx.grid(True, which="both", alpha=0.3)
ax_logx.set_xlim(1, 1e100)
ax_logx.set_xticks([10.0**k for k in range(0, 101, 20)])
ax_logx.annotate("googol: log10 = 100", xy=(1e100, 100), xytext=(1e50, 20),
                 arrowprops=dict(arrowstyle="->"))
ax_logx.legend(loc="upper left")
plt.tight_layout()
plt.show()
```

## 7. Solutions

1. $5^{-1} = \tfrac15$, $5^0 = 1$, $5^1 = 5$, $5^2 = 25$ give $(\tfrac15, -1)$, $(1, 0)$, $(5, 1)$, $(25, 2)$. Domain $x > 0$, range all reals, vertical asymptote $x = 0$.
2. $6 - 2x > 0 \iff x < 3$: domain $(-\infty, 3)$. $x^2 - 9 > 0 \iff |x| > 3$: domain $(-\infty, -3) \cup (3, \infty)$.
3. $\log_2 x = \ln x / \ln 2$ and $\log_{10} x = \ln x / \ln 10$, so $\log_2 x = \frac{\ln 10}{\ln 2}\log_{10} x \approx 3.32\,\log_{10} x$. The base-2 curve is steeper everywhere, by the factor $\log_2 10 \approx 3.32$ — at $x = 10$ and at every other $x$.
4. Estimate: $2^{50} = 2^{10 \cdot 5} \approx (10^3)^5 = 10^{15}$ mm $= 10^{12}$ m, about $7$ times the Earth–Sun distance. Exactly: $2^{50} = 1{,}125{,}899{,}906{,}842{,}624$ mm $\approx 1.13 \times 10^{12}$ m, about $7.5$ astronomical units — between the orbits of Jupiter and Saturn.
5. $y = \log_3(x - 2) + 4 \Rightarrow x - 2 = 3^{y - 4} \Rightarrow f^{-1}(x) = 3^{x - 4} + 2$. $f$: domain $x > 2$, range all reals, vertical asymptote $x = 2$. $f^{-1}$: domain all reals, range $y > 2$, horizontal asymptote $y = 2$.
6. $\log_2(x^3) = 3\log_2 x$ (vertical stretch by $3$, domain $x > 0$ for both). $\log_2(4x) = 2 + \log_2 x$ (shift up by $2$). But $\log_2(x^2)$ has domain $x \neq 0$, while $2\log_2 x$ has domain $x > 0$. They agree for $x > 0$; for $x < 0$ only the first exists, and it equals $2\log_2|x|$. The law $\log_b(x^k) = k\log_b x$ needs $x > 0$.
7. $(\sqrt2)^2 = 2$ and $(\sqrt2)^4 = 4$ ✓. Touching: from $b^x = x$ and $b^x \ln b = 1$ we get $x \ln b = 1$, so $\ln x = x\ln b = 1$, i.e. $x = e$ and $\ln b = 1/e$: $b = e^{1/e} \approx 1.4447$. For $b > e^{1/e}$ the exponential stays above the diagonal and the two graphs never meet; for $1 < b < e^{1/e}$ they meet twice.
8. $(1/16)^{1/4} = 16^{-1/4} = \tfrac12$ and $(1/16)^{1/2} = \tfrac14$, so $(\tfrac14, \tfrac12)$ and $(\tfrac12, \tfrac14)$ are on $y = (1/16)^x$; swapping coordinates puts them on $y = \log_{1/16} x$ too. The diagonal point solves $16^{-x} = x$; bisection gives $x \approx 0.36425$. (Three intersections occur exactly when $0 < b < e^{-e} \approx 0.066$.)
9. For $x_1, x_2 > 0$, AM–GM says $\frac{x_1 + x_2}{2} \ge \sqrt{x_1 x_2}$. Since $\log_b$ is increasing for $b > 1$, $\log_b\frac{x_1 + x_2}{2} \ge \log_b\sqrt{x_1x_2} = \frac{\log_b x_1 + \log_b x_2}{2}$. The curve lies above its chords at midpoints, so it is concave. (Reflecting in $y = x$ swaps "concave increasing" and "convex increasing": that is why $b^x$ is convex.)

## 8. Conclusion

- $y = \log_b x$ is $y = b^x$ reflected in $y = x$: $\;b^y = x \iff y = \log_b x$. Every property of one graph is a property of the other with $x$ and $y$ swapped.
- $b^x$: domain all reals, range $y > 0$, through $(0, 1)$, horizontal asymptote $y = 0$. $\log_b x$: domain $x > 0$, range all reals, through $(1, 0)$, vertical asymptote $x = 0$. The domain restriction exists because no real power of a positive base is $\le 0$.
- Every logarithm graph is $\ln x$ stretched by $1/\ln b$. For $b > 1$ it increases; for $0 < b < 1$ it is flipped, since $\log_{1/b} x = -\log_b x$ and $(1/b)^x = b^{-x}$.
- In $y = a\log_b(x - h) + k$ only $h$ moves the asymptote. A horizontal stretch of a log curve is a vertical shift; a horizontal shift of an exponential is a vertical stretch.
- The logarithm is unbounded but astonishingly slow: $\log_{10}$ of a googol is $100$, and $\ln x = 100$ needs $x \approx 2.7 \times 10^{43}$.
