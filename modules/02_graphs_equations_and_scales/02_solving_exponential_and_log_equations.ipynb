# Solving Exponential and Logarithmic Equations

This notebook is about the question logarithms were made to answer: **when?** When will the bacteria reach a million, when will the loan double, when will two growth curves cross? Taking a logarithm brings the unknown down from the exponent. We then meet the two classic traps — candidate solutions that fall outside the domain, and solutions that silently disappear — and finally equations such as $x e^x = 5$ that no amount of algebra can solve, which we solve numerically instead.

This is notebook 02 of module 02 (**Graphs, equations and scales**) in the **Learning Logarithms** series.

## 1. Introduction

A culture starts with $1000$ bacteria and doubles every $20$ minutes. After $t$ minutes there are

$$
N(t) = 1000 \cdot 2^{t/20}
$$

bacteria. **When does it reach one million?** We need $2^{t/20} = 1000$. Trial and error works ($2^{10} = 1024$, so a bit under $200$ minutes), but it is guesswork, and it gets much worse with a messier question: a loan of `$10,000` at $5\%$ annual interest grows as $10000 \cdot 1.05^n$ — after how many years has it doubled? You need the $n$ with $1.05^n = 2$, and no table of whole powers will hand it to you.

The unknown is stuck **in the exponent**. Addition, multiplication, powers and roots cannot get it out. The logarithm can — that is precisely what it is for:

$$
2^{t/20} = 1000 \;\Longleftrightarrow\; \frac{t}{20} = \log_2 1000 \;\Longrightarrow\; t = 20 \log_2 1000 \approx 199.3 \text{ minutes},
$$

$$
1.05^n = 2 \;\Longleftrightarrow\; n = \log_{1.05} 2 = \frac{\ln 2}{\ln 1.05} \approx 14.2 \text{ years}.
$$

The same tool solves equations with the unknown inside a logarithm, but there it comes with a warning: the logarithm only accepts positive inputs, so algebra can produce "solutions" that are not solutions at all.

## 2. Theory

### 2.1 The basic move: take a logarithm of both sides

The logarithm is one-to-one on $(0, \infty)$, so for positive $A$ and $B$

$$
A = B \;\Longleftrightarrow\; \ln A = \ln B.
$$

Applied to $b^{x} = c$ (with $c > 0$):

$$
\underbrace{b^{x} = c}_{\text{exponent form}} \quad\Longleftrightarrow\quad \underbrace{x = \log_b c = \frac{\ln c}{\ln b}}_{\text{log form}}.
$$

If $c \le 0$ there is **no** real solution: $b^x$ is always positive (its graph lies above the asymptote $y = 0$). Any base works for the logarithm you take; $\ln$ is the usual choice because it is what Python and calculators provide.

**Isolate first.** For $a \cdot b^{kx + m} + d = c$, move everything else away from the power before taking logs: $b^{kx + m} = \frac{c - d}{a}$. Taking $\ln$ of a sum is a mistake — $\ln(A + B) \neq \ln A + \ln B$.

**Different bases.** For $2^x = 3^{x - 1}$, take $\ln$ of both sides and use the power law: $x \ln 2 = (x - 1)\ln 3$. That is a **linear** equation in $x$.

### 2.2 Hidden quadratics: substitution

In $3^{2x} - 4 \cdot 3^x + 3 = 0$, notice $3^{2x} = (3^x)^2$. With $u = 3^x$ (and therefore $u > 0$):

$$
u^2 - 4u + 3 = 0 \;\Longrightarrow\; (u - 1)(u - 3) = 0 \;\Longrightarrow\; u = 1 \text{ or } u = 3.
$$

Back-substitute: $3^x = 1 \Rightarrow x = 0$; $3^x = 3 \Rightarrow x = 1$. Any root with $u \le 0$ must be **discarded**, because $3^x$ can never be zero or negative. The same trick handles $e^x + e^{-x} = c$ (multiply by $u = e^x$).

### 2.3 Logarithmic equations and the domain

To solve an equation with the unknown inside a logarithm:

1. **State the domain** — every argument must be positive.
2. **Combine** logs with the laws into a single $\log_b(\text{something}) = k$, or $\log_b A = \log_b B$.
3. **Exponentiate**: $\log_b A = k \iff A = b^k$; $\;\log_b A = \log_b B \iff A = B$.
4. **Check** each candidate against the domain from step 1.

Step 4 is not optional. The law $\log_b x + \log_b y = \log_b(xy)$ holds only when $x > 0$ **and** $y > 0$, but the right-hand side only needs $xy > 0$, which is also true when both are negative. Combining therefore **enlarges** the domain and can let in **extraneous roots**.

*Example:* $\log_2 x + \log_2(x - 2) = 3$. Combining gives $\log_2(x(x - 2)) = 3$, so $x^2 - 2x - 8 = 0$, i.e. $x = 4$ or $x = -2$. But $\log_2(-2)$ is undefined: $x = -2$ satisfies the *combined* equation, not the original. Only $x = 4$ survives.

### 2.4 The opposite trap: lost roots

Going the other way **shrinks** the domain. $\ln(x^2) = \ln 9$ has domain $x \neq 0$ and two solutions, $x = \pm 3$. Rewriting it as $2\ln x = \ln 9$ restricts to $x > 0$ and finds only $x = 3$. The correct power law for even powers is

$$
\ln(x^2) = 2 \ln|x|.
$$

Rule of thumb: **combining logs can add roots; splitting logs can lose roots.** Check against the original equation's domain either way.

### 2.5 Equations with no elementary solution

Some equations mix the unknown in and out of the exponent:

$$
x e^x = 5, \qquad x^x = 100, \qquad 2^x = x^2, \qquad \ln n = n^{0.1}.
$$

Taking logs does not separate $x$: from $x e^x = 5$ you get $\ln x + x = \ln 5$, still tangled. It can be proved that these equations have no solution in terms of the usual functions (roots, $\exp$, $\ln$, trig). The solution of $x e^x = c$ is so common that it has its own name, the **Lambert W function**: $x = W(c)$. Python's standard library does not have it, but solving numerically is easy:

**Bisection.** If a continuous $f$ changes sign on $[a, b]$, it has a root there. Halve the interval, keep the half with the sign change, repeat. Each step gains one binary digit — about $3.3$ steps per decimal digit, since $\log_2 10 \approx 3.32$. Slow but guaranteed.

**Newton's method.** Replace $f$ by its tangent line at $x_k$ and jump to where the tangent crosses zero:

$$
x_{k+1} = x_k - \frac{f(x_k)}{f'(x_k)}.
$$

Near a simple root the number of correct digits roughly **doubles** each step. It needs a derivative and a decent starting point. For $f(x) = x e^x - 5$, $f'(x) = (x + 1)e^x$.

A graph (notebook 02/01) is the best first step: it tells you **how many** solutions there are and roughly where, which gives both methods their starting intervals.

## 3. Worked Examples

### Example 1 — the bacteria

$1000 \cdot 2^{t/20} = 10^6 \Rightarrow 2^{t/20} = 1000 \Rightarrow t = 20\log_2 1000 = 20 \cdot \frac{\ln 1000}{\ln 2} \approx 20 \times 9.966 = 199.3$ minutes. Sanity check: $2^{10} = 1024 \approx 1000$, so about $10$ doublings, $200$ minutes ✓.

### Example 2 — the loan

$10000 \cdot 1.05^n = 20000 \Rightarrow 1.05^n = 2 \Rightarrow n = \frac{\ln 2}{\ln 1.05} = \frac{0.6931}{0.04879} \approx 14.21$. With yearly compounding the balance first exceeds `$20,000` after $15$ years. (The "rule of 72" says $72/5 = 14.4$ — notebook 04/03 explains why.)

### Example 3 — different bases: $2^x = 3^{x - 1}$

$x\ln 2 = (x - 1)\ln 3 \Rightarrow x(\ln 3 - \ln 2) = \ln 3 \Rightarrow x = \dfrac{\ln 3}{\ln 3 - \ln 2} = \dfrac{\ln 3}{\ln 1.5} \approx \dfrac{1.0986}{0.4055} \approx 2.7095$.

### Example 4 — hidden quadratic: $3^{2x} - 4\cdot 3^x + 3 = 0$

As in section 2.2: $u = 3^x$, $u \in \{1, 3\}$, so $x \in \{0, 1\}$. Check $x = 1$: $9 - 12 + 3 = 0$ ✓. Check $x = 0$: $1 - 4 + 3 = 0$ ✓.

### Example 5 — hidden quadratic with a rejected root: $4^x - 2^{x+1} - 8 = 0$

$4^x = (2^x)^2$ and $2^{x+1} = 2 \cdot 2^x$. With $u = 2^x > 0$: $u^2 - 2u - 8 = 0 \Rightarrow (u - 4)(u + 2) = 0$. The root $u = -2$ is impossible ($2^x > 0$), so $u = 4$ and $x = 2$.

### Example 6 — extraneous root: $\log_2 x + \log_2(x - 2) = 3$

Domain: $x > 0$ and $x > 2$, i.e. $x > 2$. Combine: $x(x - 2) = 2^3 = 8 \Rightarrow x^2 - 2x - 8 = 0 \Rightarrow x \in \{4, -2\}$. Reject $-2$ (outside the domain). Check $x = 4$: $\log_2 4 + \log_2 2 = 2 + 1 = 3$ ✓.

### Example 7 — logs on both sides: $2\ln x = \ln(x + 6)$

Domain $x > 0$. Then $\ln(x^2) = \ln(x + 6) \Rightarrow x^2 = x + 6 \Rightarrow (x - 3)(x + 2) = 0$. Reject $-2$; the solution is $x = 3$. Check: $2\ln 3 = \ln 9 = \ln(3 + 6)$ ✓.

### Example 8 — no algebraic route: $x e^x = 5$

$g(x) = x e^x$ is negative for $x < 0$ and increasing for $x > 0$, so there is exactly one positive solution. $g(1) = e \approx 2.72 < 5$ and $g(2) = 2e^2 \approx 14.8 > 5$: the root is in $[1, 2]$. Newton from $x_0 = 1$:

$$
x_1 = 1 - \frac{e - 5}{2e} \approx 1.4197, \qquad x_2 \approx 1.3326, \qquad x_3 \approx 1.32675, \qquad x_4 \approx 1.3267247,
$$

so $x = W(5) \approx 1.32672$. Check: $1.32672 \cdot e^{1.32672} \approx 5.0000$ ✓.

## 4. Python Examples

### 4.1 "When?" questions in one line each

Each answer is $\log_b(\text{target ratio})$, computed with change of base.

```python
import math

# Bacteria: 1000 * 2^(t/20) = 1e6
t_minutes = 20 * math.log(1e6 / 1000) / math.log(2)
print(f"bacteria reach one million after {t_minutes:.1f} minutes")

# Loan: 10000 * 1.05^n = 20000
n_years = math.log(2) / math.log(1.05)
print(f"loan doubles after {n_years:.2f} years -> first exceeds double after {math.ceil(n_years)} whole years")
print(f"check: 1.05^14 = {1.05**14:.4f}, 1.05^15 = {1.05**15:.4f}")

# Different bases: 2^x = 3^(x - 1)
x_crossing = math.log(3) / (math.log(3) - math.log(2))
print(f"2^x = 3^(x-1) at x = {x_crossing:.4f}; both sides = {2**x_crossing:.4f}, {3**(x_crossing - 1):.4f}")
```

### 4.2 Hidden quadratics by substitution

Solve the quadratic in $u$, throw away non-positive $u$, then take the logarithm.

```python
import math

import numpy as np


def solve_hidden_quadratic(a, b, c, base):
    # Solves a*(base^x)^2 + b*base^x + c = 0 for real x
    u_roots = np.roots([a, b, c])
    solutions = []
    for u in u_roots:
        if abs(u.imag) > 1e-12:
            print(f"  u = {u}: complex, discarded")
        elif u.real <= 0:
            print(f"  u = {u.real:g}: base^x must be positive, discarded")
        else:
            x = math.log(u.real) / math.log(base)
            solutions.append(x)
            print(f"  u = {u.real:g}: x = log_{base}({u.real:g}) = {x:.6f}")
    return sorted(solutions)


print("3^(2x) - 4*3^x + 3 = 0")
print("  solutions:", solve_hidden_quadratic(1, -4, 3, 3))
print("4^x - 2*2^x - 8 = 0")
print("  solutions:", solve_hidden_quadratic(1, -2, -8, 2))
```

### 4.3 Extraneous roots: solve, then check the domain

SymPy's `solve` works on the combined equation; we check each candidate in the **original** one. Note that SymPy itself is careful: solving the original form directly already discards $-2$.

```python
import sympy as sp

x = sp.symbols("x")
original_lhs = sp.log(x, 2) + sp.log(x - 2, 2)
combined_equation = sp.Eq(x * (x - 2), 2**3)

candidates = sp.solve(combined_equation, x)
print("candidates from the combined equation:", candidates)
for candidate in candidates:
    in_domain = candidate > 0 and candidate - 2 > 0
    if in_domain:
        value = sp.nsimplify(original_lhs.subs(x, candidate))
        print(f"  x = {candidate}: in domain, original LHS = {value}")
    else:
        print(f"  x = {candidate}: log of a negative number -> extraneous")

print("sympy on the original equation:", sp.solve(sp.Eq(original_lhs, 3), x))
```

### 4.4 Lost roots: $\ln(x^2)$ versus $2\ln x$

```python
import math

for candidate in [3, -3]:
    left_correct = math.log(candidate**2)
    try:
        left_split = f"{2 * math.log(candidate):.4f}"
    except ValueError:
        left_split = "undefined"
    print(f"x = {candidate:>2}: ln(x^2) = {left_correct:.4f}, 2 ln x = {left_split}, "
          f"2 ln|x| = {2 * math.log(abs(candidate)):.4f}, ln 9 = {math.log(9):.4f}")
```

### 4.5 Bisection and Newton for $x e^x = 5$

Both methods side by side, printing the error at each step. Bisection halves the error; Newton squares it.

```python
import math


def f(x):
    return x * math.exp(x) - 5


def f_prime(x):
    return (x + 1) * math.exp(x)


reference = 1.3267246652422002  # W(5), see the SymPy cell below

print("Bisection on [1, 2]")
low, high = 1.0, 2.0
for step in range(1, 11):
    middle = (low + high) / 2
    if f(low) * f(middle) <= 0:
        high = middle
    else:
        low = middle
    print(f"  step {step:2d}: midpoint {middle:.10f}, error {abs(middle - reference):.1e}")

print("Newton from x0 = 1")
x_current = 1.0
for step in range(1, 6):
    x_current = x_current - f(x_current) / f_prime(x_current)
    print(f"  step {step}: x = {x_current:.15f}, error {abs(x_current - reference):.1e}")
```

SymPy knows the Lambert W function and can confirm the root (and recognise that the solution is $W(5)$):

```python
import sympy as sp

x = sp.symbols("x")
exact_solutions = sp.solve(sp.Eq(x * sp.exp(x), 5), x)
print("sympy solution:", exact_solutions)
print("numerical value:", sp.N(exact_solutions[0], 15))
print("x^x = 100      :", sp.nsolve(x**x - 100, x, 3.5))
```

## 5. Visualizations

### 5.1 "When?" is a horizontal line meeting a curve

Solving $N(t) = 10^6$ means intersecting the growth curve with a horizontal line. On a logarithmic $y$-axis the exponential becomes a straight line — which is exactly what taking logs of both sides does algebraically.

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(0, 240, 400)
bacteria = 1000 * 2 ** (t / 20)
t_target = 20 * np.log2(1000)

fig, (ax_linear, ax_log) = plt.subplots(1, 2, figsize=(12, 4.6))
for ax in (ax_linear, ax_log):
    ax.plot(t, bacteria, color="tab:green", label=r"$N(t) = 1000 \cdot 2^{t/20}$")
    ax.axhline(1e6, color="tab:red", linestyle="--", label="target $10^6$")
    ax.axvline(t_target, color="gray", linestyle=":")
    ax.plot(t_target, 1e6, "o", color="tab:red")
    ax.set_xlabel("t (minutes, linear scale)")
    ax.legend(loc="upper left")

ax_linear.set_ylabel("bacteria (linear scale)")
ax_linear.set_title("Linear axis: the answer hides in a steep curve")
ax_linear.grid(True, alpha=0.3)
ax_log.set_yscale("log")
ax_log.set_ylabel("bacteria (log scale)")
ax_log.set_title(f"Log axis: a straight line, crossing at t = {t_target:.1f}")
ax_log.grid(True, which="both", alpha=0.3)
plt.tight_layout()
plt.show()
```

### 5.2 Seeing the extraneous root

Left: the original function $\log_2 x + \log_2(x - 2)$ exists only for $x > 2$. Right: the combined function $\log_2(x(x - 2))$ exists for $x < 0$ too, and reaches $3$ at both $x = 4$ and $x = -2$. The extra branch is where the false root lives.

```python
import matplotlib.pyplot as plt
import numpy as np

x_right = np.linspace(2.001, 7, 400)
x_left = np.linspace(-5, -0.001, 400)

fig, (ax_original, ax_combined) = plt.subplots(1, 2, figsize=(12, 4.6), sharey=True)
ax_original.plot(x_right, np.log2(x_right) + np.log2(x_right - 2), color="tab:blue")
ax_original.plot(4, 3, "o", color="tab:green", markersize=8, label="x = 4 (solution)")
ax_original.set_title(r"original: $\log_2 x + \log_2(x-2)$, domain $x > 2$")

ax_combined.plot(x_right, np.log2(x_right * (x_right - 2)), color="tab:orange")
ax_combined.plot(x_left, np.log2(x_left * (x_left - 2)), color="tab:orange")
ax_combined.plot(4, 3, "o", color="tab:green", markersize=8, label="x = 4 (solution)")
ax_combined.plot(-2, 3, "X", color="tab:red", markersize=10, label="x = -2 (extraneous)")
ax_combined.set_title(r"combined: $\log_2(x(x-2))$, domain $x < 0$ or $x > 2$")

for ax in (ax_original, ax_combined):
    ax.axhline(3, color="gray", linestyle="--", linewidth=1)
    ax.axvspan(0, 2, color="lightgray", alpha=0.5)
    ax.set_xlim(-5, 7)
    ax.set_ylim(-4, 6)
    ax.set_xlabel("x")
    ax.grid(True, alpha=0.3)
    ax.legend(loc="lower right")
ax_original.set_ylabel("y")
plt.tight_layout()
plt.show()
```

### 5.3 Newton's method on $x e^x = 5$

Each tangent line (dashed) is followed down to the $x$-axis; that crossing is the next guess. Two tangents already land almost on the root. The right panel shows the error on a log scale: bisection zig-zags down along a straight trend (its bracket shrinks by exactly $\tfrac12$ per step, about $0.3$ decades), Newton falls off a cliff. Newton's last points are exact zeros, drawn at $10^{-17}$ so that the log axis can show them.

```python
import matplotlib.pyplot as plt
import numpy as np

def f(x):
    return x * np.exp(x) - 5

def f_prime(x):
    return (x + 1) * np.exp(x)

reference = 1.3267246652422002
x_grid = np.linspace(0.6, 2.1, 300)

fig, (ax_tangent, ax_error) = plt.subplots(1, 2, figsize=(12, 4.8))
ax_tangent.plot(x_grid, f(x_grid), color="tab:blue", label=r"$f(x) = x e^x - 5$")
ax_tangent.axhline(0, color="black", linewidth=0.7)

x_current = 2.0
for step in range(3):
    x_next = x_current - f(x_current) / f_prime(x_current)
    ax_tangent.plot([x_current, x_next], [f(x_current), 0], "--", color="tab:orange")
    ax_tangent.plot([x_current, x_current], [0, f(x_current)], ":", color="gray")
    ax_tangent.plot(x_current, f(x_current), "o", color="tab:orange")
    ax_tangent.text(x_current, -1.4, f"$x_{step}$", ha="center")
    x_current = x_next
ax_tangent.plot(reference, 0, "*", color="tab:red", markersize=12, label="root W(5)")
ax_tangent.set_xlabel("x")
ax_tangent.set_ylabel("f(x)")
ax_tangent.set_ylim(-4, 12)
ax_tangent.grid(True, alpha=0.3)
ax_tangent.legend(loc="upper left")
ax_tangent.set_title("Newton's tangent steps from $x_0 = 2$")

# error histories
bisection_errors = []
low, high = 1.0, 2.0
for _ in range(40):
    middle = (low + high) / 2
    bisection_errors.append(abs(middle - reference))
    if f(low) * f(middle) <= 0:
        high = middle
    else:
        low = middle

newton_errors = []
x_current = 2.0
for _ in range(8):
    newton_errors.append(abs(x_current - reference))
    x_current = x_current - f(x_current) / f_prime(x_current)
newton_errors = [max(error, 1e-17) for error in newton_errors]   # keep exact zeros visible on the log axis

ax_error.plot(range(len(bisection_errors)), bisection_errors, "o-", markersize=3, label="bisection")
ax_error.plot(range(len(newton_errors)), newton_errors, "s-", label="Newton")
ax_error.set_yscale("log")
ax_error.set_xlabel("step (linear scale)")
ax_error.set_ylabel("absolute error (log scale)")
ax_error.grid(True, which="both", alpha=0.3)
ax_error.legend()
ax_error.set_title("Error per step")
plt.tight_layout()
plt.show()
```

## 8. Conclusion

- To get an unknown out of an exponent, isolate the power and take a logarithm of both sides: $\;b^x = c \iff x = \log_b c = \ln c / \ln b$, for $c > 0$; for $c \le 0$ there is no real solution.
- Equations with different bases become linear after taking $\ln$; expressions like $b^{2x}$ and $b^x$ together are quadratics in $u = b^x$, where only roots with $u > 0$ count.
- For logarithmic equations, **write down the domain first** and check every candidate against it. Combining logs ($\log_b x + \log_b y \to \log_b xy$) can introduce extraneous roots; splitting logs ($\ln x^2 \to 2\ln x$) can lose genuine ones.
- Equations such as $x e^x = 5$ (solution $W(5) \approx 1.32672$), $x^x = 100$ and $2^x = x^2$ mix $x$ inside and outside the exponent and have no elementary solution. A graph to locate the roots, then bisection (one binary digit per step) or Newton's method (digits roughly double per step), solves them to machine precision.
- The picture: "when?" is the intersection of a horizontal line with an exponential curve — and on a log axis that curve is a straight line.
