# Growth Rates: $\ln n \ll n^\varepsilon \ll n \ll n\ln n \ll n^2 \ll e^n$

This notebook ranks the functions that measure how expensive a computation is, how fast a population grows, or how slowly a series converges. Logarithms sit at the bottom of the ladder — slower than any power of $n$, however small — and exponentials at the top — faster than any power, however large. We prove the ranking with limits (by L'Hôpital's rule and with SymPy), find out **when** the asymptotically faster function actually takes over (sometimes absurdly late), and draw the race on log-scaled axes.

This is notebook 04 of module 02 (**Graphs, equations and scales**) in the **Learning Logarithms** series.

## 1. Introduction

You have to process $n = 10^6$ records on a machine that does $10^9$ simple operations per second. Five algorithms are on offer, with these operation counts:

| algorithm | operations | time for $n = 10^6$ |
|---|---|---|
| binary search | $\log_2 n \approx 20$ | $20$ nanoseconds |
| one pass over the data | $n = 10^6$ | $1$ millisecond |
| merge sort | $n\log_2 n \approx 2 \times 10^7$ | $0.02$ seconds |
| comparing all pairs | $n^2 = 10^{12}$ | $1000$ seconds, about $17$ minutes |
| trying all subsets | $2^n = 2^{1{,}000{,}000}$, a $301{,}030$-digit number | about $10^{301{,}021}$ seconds |

The gap between the rows is not a constant factor; it is whole **orders of magnitude**, and it widens as $n$ grows. Buying a computer $1000$ times faster turns $17$ minutes into one second — but does nothing for the last row. Knowing which row you are in matters far more than how fast your machine is.

The comparison raises precise questions. Is $n\ln n$ really "between" $n$ and $n^2$? Does $\ln n$ lose to $n^{0.01}$? (Yes — but only after $n$ passes about $10^{281}$.) Does $n^{100}$ lose to $1.01^n$? (Yes — beyond $n \approx 117{,}000$.) The tool for all of them is the limit of a ratio, and the trick for most of them is to **take a logarithm**.

## 2. Theory

### 2.1 Comparing growth: the ratio test

For positive functions $f$ and $g$ we write

$$
f \ll g \quad (\text{"} f \text{ grows more slowly than } g\text{"}, \text{ also written } f = o(g)) \quad\Longleftrightarrow\quad \lim_{n\to\infty} \frac{f(n)}{g(n)} = 0.
$$

If the ratio tends to a positive constant, $f$ and $g$ grow **at the same rate** ($f = \Theta(g)$): $3n^2 + n$ and $n^2$ do. Constant factors never change the rank.

The ladder this notebook establishes, for any fixed $\varepsilon > 0$, $k > 1$ and $b > 1$:

$$
1 \;\ll\; \ln n \;\ll\; (\ln n)^k \;\ll\; n^\varepsilon \;\ll\; n \;\ll\; n\ln n \;\ll\; n^{k} \;\ll\; b^n \;\ll\; n! \;\ll\; n^n.
$$

### 2.2 The logarithm loses to every power

**Claim:** $\displaystyle\lim_{n\to\infty}\frac{\ln n}{n^\varepsilon} = 0$ for every $\varepsilon > 0$.

Both numerator and denominator tend to $\infty$, so L'Hôpital's rule applies (treat $n$ as a real variable):

$$
\lim_{n\to\infty}\frac{\ln n}{n^\varepsilon} = \lim_{n\to\infty}\frac{1/n}{\varepsilon n^{\varepsilon - 1}} = \lim_{n\to\infty}\frac{1}{\varepsilon\, n^{\varepsilon}} = 0.
$$

Powers of the log lose too: $(\ln n)^k / n^\varepsilon = \left(\ln n / n^{\varepsilon/k}\right)^k \to 0^k = 0$.

Consequences: $n \ll n\ln n$ (the ratio is $1/\ln n \to 0$) and $n\ln n \ll n^{1+\varepsilon}$ (the ratio is $\ln n / n^\varepsilon \to 0$). So $n\ln n$ sits strictly between $n$ and every $n^{1 + \varepsilon}$ — "almost linear".

### 2.3 Every power loses to an exponential

**Claim:** $\displaystyle\lim_{n\to\infty}\frac{n^k}{b^n} = 0$ for every $k$ and every $b > 1$.

**By logarithms.** $\ln\dfrac{n^k}{b^n} = k\ln n - n\ln b = n\left(k\frac{\ln n}{n} - \ln b\right)$. Since $\ln n / n \to 0$ (section 2.2 with $\varepsilon = 1$), the bracket tends to $-\ln b < 0$, so the logarithm of the ratio tends to $-\infty$ and the ratio to $0$.

**By L'Hôpital** (for whole $k$): differentiating $k$ times turns $n^k$ into $k!$ and $b^n$ into $(\ln b)^k b^n$, so the ratio becomes $k!/((\ln b)^k b^n) \to 0$.

The two results are the **same fact** seen through the mirror $y = x$ of notebook 02/01: substitute $n = e^m$ in $\ln n / n^\varepsilon$ and it becomes $m / e^{\varepsilon m}$.

### 2.4 Comparing by taking logarithms — and its pitfall

Since $\ln$ is increasing, $\dfrac{f}{g} \to 0 \iff \ln f - \ln g \to -\infty$. That gives a quick test:

| $f$ | $\ln f$ |
|---|---|
| $(\ln n)^k$ | $k\ln\ln n$ |
| $n^k$ | $k\ln n$ |
| $n^{\ln n}$ | $(\ln n)^2$ |
| $b^n$ | $n\ln b$ |
| $n!$ | $\approx n\ln n - n$ (Stirling, notebook 04/04) |
| $n^n$ | $n\ln n$ |

**Pitfall:** it is the **difference** of the logs that must go to $-\infty$, not their ratio. $\ln(n^2) = 2\ln n$ and $\ln(n^3) = 3\ln n$ have ratio $2/3$ — "comparable" logs — yet $n^2 \ll n^3$ because the difference $-\ln n \to -\infty$. Taking logs **compresses** growth: every polynomial becomes a multiple of $\ln n$.

### 2.5 Bases: irrelevant for logs, decisive for exponentials

- $\log_2 n = \dfrac{\ln n}{\ln 2} \approx 1.443\ln n$ and $\log_{10} n \approx 0.434\ln n$. All logarithms differ by a **constant factor**, so they have the same growth rate. Computer scientists write $O(\log n)$ without a base for this reason.
- $2^n / 3^n = (2/3)^n \to 0$: exponentials with different bases have **different** growth rates. Equivalently $3^n = 2^{n\log_2 3}$: changing the base changes the exponent by a factor, and a factor in the exponent is not a constant factor.

### 2.6 Crossover points: "eventually" can be very late

$\ln n \ll n^{0.1}$, yet $\ln n > n^{0.1}$ for an enormous range of $n$. Where do they cross? Put $n = e^t$: the equation $\ln n = n^{0.1}$ becomes $t = e^{0.1 t}$, which (like the equations of notebook 02/02) has no elementary solution. Bisection gives two roots: $t \approx 1.118$ and $t \approx 35.77$, i.e.

$$
n \approx 3.06 \quad\text{and}\quad n \approx 3.43 \times 10^{15}.
$$

For all $n$ between them — every input size you will ever meet in practice — the "slower" function $\ln n$ is the **larger** one.

A clean way to see it: the ratio $\ln n / n^\varepsilon$ has its maximum where its derivative vanishes, at $n = e^{1/\varepsilon}$, with value $\dfrac{1}{\varepsilon e}$. For $\varepsilon = 0.1$ that is $3.68$ at $n = e^{10} \approx 22{,}026$: the log is ahead by a factor $3.68$ there, and it takes until $3.4 \times 10^{15}$ to fall behind. For $\varepsilon = \tfrac12$ the maximum is $2/e < 1$: $\ln n < \sqrt n$ for **every** $n$.

## 3. Worked Examples

### Example 1 — $\displaystyle\lim_{n\to\infty}\frac{(\ln n)^2}{\sqrt n}$

L'Hôpital once: $\dfrac{2\ln n \cdot \frac1n}{\frac{1}{2}n^{-1/2}} = \dfrac{4\ln n}{\sqrt n}$. Again: $\dfrac{4/n}{\frac12 n^{-1/2}} = \dfrac{8}{\sqrt n} \to 0$. So $(\ln n)^2 \ll \sqrt n$.

### Example 2 — $\displaystyle\lim_{n\to\infty}\frac{n^{3}}{2^n}$

By logs: $3\ln n - n\ln 2 \to -\infty$, so the limit is $0$. At $n = 10$: $1000/1024 \approx 0.98$; at $n = 30$: $27000/1.07\times10^{9} \approx 2.5 \times 10^{-5}$. (The two curves cross at $n \approx 9.94$, as in notebook 02/03.)

### Example 3 — where does $n^{100}$ fall behind $1.01^n$?

Compare logs: $100\ln n$ versus $n\ln 1.01 \approx 0.00995\,n$. Solving $100\ln n = n\ln 1.01$ numerically gives $n \approx 117{,}308$. Beyond that point $1.01^n$ leads forever; at the crossing both are about $10^{507}$.

### Example 4 — sorting a billion items

$n = 10^9$, $10^9$ operations per second. Merge sort: $n\log_2 n \approx 10^9 \times 29.9 \approx 3 \times 10^{10}$ operations, about $30$ seconds. All-pairs: $n^2 = 10^{18}$ operations, $10^9$ seconds, about $32$ years.

### Example 5 — constants matter below the crossover

Algorithm A costs $100\, n\log_2 n$, algorithm B costs $n^2$. B is faster while $n^2 < 100\, n\log_2 n$, i.e. $n < 100\log_2 n$, which holds up to $n \approx 996$. Above about a thousand items the better growth rate wins and keeps winning by more and more. Real sorting libraries exploit exactly this: they switch to a simple quadratic method for short runs.

## 4. Python Examples

### 4.1 The operation-count table

The table from the introduction for several values of $n$. The $2^n$ column is shown as its number of decimal digits, $\lfloor n\log_{10} 2\rfloor + 1$, because the number itself is far too large to print.

```python
import math

operations_per_second = 1e9


def human_time(seconds):
    if seconds < 1e-6:
        return f"{seconds * 1e9:.0f} ns"
    if seconds < 1e-3:
        return f"{seconds * 1e6:.1f} us"
    if seconds < 1:
        return f"{seconds * 1e3:.1f} ms"
    if seconds < 3600:
        return f"{seconds:.1f} s"
    if seconds < 3.156e7:
        return f"{seconds / 3600:.1f} h"
    return f"{seconds / 3.156e7:.3g} years"


print(f"{'n':>8} | {'log2 n':>8} | {'n':>8} | {'n log2 n':>9} | {'n^2':>11} | digits of 2^n")
for n in [10, 1_000, 1_000_000, 1_000_000_000]:
    costs = [math.log2(n), n, n * math.log2(n), n**2]
    times = [human_time(cost / operations_per_second) for cost in costs]
    digits_of_two_to_n = math.floor(n * math.log10(2)) + 1
    print(f"{n:>8.0e} | {times[0]:>8} | {times[1]:>8} | {times[2]:>9} | {times[3]:>11} | {digits_of_two_to_n:,}")
```

### 4.2 Limits with SymPy

`sympy.limit` confirms each rung of the ladder. Each ratio is (slower function) / (faster function), so every limit should be $0$.

```python
import sympy as sp

n = sp.symbols("n", positive=True)
epsilon = sp.Rational(1, 10)
comparisons = [
    (sp.log(n), n**epsilon),
    (sp.log(n) ** 3, sp.sqrt(n)),
    (n, n * sp.log(n)),
    (n * sp.log(n), n ** sp.Rational(11, 10)),
    (n**100, sp.Rational(101, 100) ** n),
    (2**n, sp.factorial(n)),
]
for slower, faster in comparisons:
    print(f"lim ({str(slower)}) / ({str(faster)}) = {sp.limit(slower / faster, n, sp.oo)}")

print()
print("same growth rate: lim log2(n)/ln(n) =", sp.limit(sp.log(n, 2) / sp.log(n), n, sp.oo))
print("different rates : lim 2^n / 3^n     =", sp.limit(2**n / 3**n, n, sp.oo))
```

### 4.3 Crossover points by bisection

For huge values we compare **logarithms** of the two sides so nothing overflows. A small bisection routine (as in notebook 02/02) finds where the difference changes sign.

```python
import math


def bisect(function, low, high, steps=200):
    for _ in range(steps):
        middle = (low + high) / 2
        if function(low) * function(middle) <= 0:
            high = middle
        else:
            low = middle
    return (low + high) / 2


# ln n = n^0.1, written with t = ln n:  t - e^(0.1 t) = 0
for low, high in [(1, 2), (20, 60)]:
    t = bisect(lambda t: t - math.exp(0.1 * t), low, high)
    print(f"ln n = n^0.1 at t = {t:.4f}, n = e^t = {math.exp(t):.4g}")

# ln n = n^0.01: the large crossover, reported as a power of ten
t = bisect(lambda t: t - math.exp(0.01 * t), 100, 2000)
print(f"ln n = n^0.01 at n = 10^{t / math.log(10):.1f}")

# n^100 = 1.01^n, compared through logs: 100 ln n - n ln 1.01
n_cross = bisect(lambda n: 100 * math.log(n) - n * math.log(1.01), 1e3, 1e7)
print(f"n^100 = 1.01^n at n = {n_cross:,.0f}; both sides are 10^{100 * math.log10(n_cross):.1f}")

# 100 n log2 n = n^2
print(f"100 n log2 n = n^2 at n = {bisect(lambda n: n - 100 * math.log2(n), 100, 1e4):.1f}")
```

### 4.4 The maximum of $\ln n / n^\varepsilon$

The formula $\max = 1/(\varepsilon e)$ at $n = e^{1/\varepsilon}$ tells us immediately whether $\ln n$ ever overtakes $n^\varepsilon$: exactly when $1/(\varepsilon e) > 1$, i.e. $\varepsilon < 1/e \approx 0.368$.

```python
import math

import numpy as np

for epsilon in [0.5, 1 / math.e, 0.3, 0.1, 0.01]:
    n_at_max = math.exp(1 / epsilon)
    formula_max = 1 / (epsilon * math.e)
    # check the formula against a brute-force search on a log-spaced grid of t = ln n
    t_grid = np.linspace(0.01, 3 / epsilon, 200_001)
    numeric_max = np.max(t_grid * np.exp(-epsilon * t_grid))
    overtakes = "yes" if formula_max > 1 + 1e-12 else "no"
    print(f"eps = {epsilon:.4f}: max of ln n / n^eps = {formula_max:8.4f} (grid: {numeric_max:8.4f}) "
          f"at n = {n_at_max:.3g}; ln n ever ahead: {overtakes}")
```

### 4.5 Counting comparisons: insertion sort versus merge sort

Growth rates are about real programs, so count the comparisons two sorting algorithms actually make on random lists. Insertion sort needs about $n^2/4$ on average; merge sort needs at most $n\log_2 n$. Doubling $n$ should roughly **quadruple** the first count and only slightly more than **double** the second.

```python
import math
import random


def insertion_sort_comparisons(values):
    items = list(values)
    comparisons = 0
    for i in range(1, len(items)):
        current = items[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            if items[j] <= current:
                break
            items[j + 1] = items[j]
            j -= 1
        items[j + 1] = current
    return comparisons


def merge_sort_comparisons(values):
    if len(values) <= 1:
        return list(values), 0
    middle = len(values) // 2
    left, left_count = merge_sort_comparisons(values[:middle])
    right, right_count = merge_sort_comparisons(values[middle:])
    merged, comparisons = [], left_count + right_count
    i = j = 0
    while i < len(left) and j < len(right):
        comparisons += 1
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1
    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged, comparisons


random.seed(4)
print(f"{'n':>6} | {'insertion':>10} | {'n^2/4':>10} | {'merge':>8} | {'n log2 n':>9}")
for n in [100, 200, 400, 800, 1600, 3200]:
    data = [random.random() for _ in range(n)]
    insertion_count = insertion_sort_comparisons(data)
    sorted_data, merge_count = merge_sort_comparisons(data)
    assert sorted_data == sorted(data)
    print(f"{n:>6} | {insertion_count:>10,} | {n * n // 4:>10,} | {merge_count:>8,} | {n * math.log2(n):>9,.0f}")
```

## 5. Visualizations

### 5.1 The race on a log-scaled $y$-axis

Six cost functions for $n$ from $1$ to $10^6$, with both axes logarithmic. On log-log axes every power $n^k$ is a straight line of slope $k$ (notebook 02/03); $\ln n$ and $n\ln n$ bend slightly; $2^n$ shoots off the top almost immediately. The dashed horizontal line marks $10^{12}$ operations, about $17$ minutes at $10^9$ operations per second.

```python
import matplotlib.pyplot as plt
import numpy as np

n = np.logspace(0, 6, 400)
curves = {
    r"$\ln n$": np.log(n),
    r"$\sqrt{n}$": np.sqrt(n),
    r"$n$": n,
    r"$n \ln n$": n * np.log(n),
    r"$n^2$": n**2,
    r"$2^n$": 2.0 ** np.minimum(n, 100),   # capped: beyond the top of the plot anyway
}

fig, ax = plt.subplots(figsize=(10, 6))
for label, values in curves.items():
    ax.plot(n, np.where(values > 0, values, np.nan), label=label, linewidth=2)
ax.axhline(1e12, color="gray", linestyle="--")
ax.text(1e2, 2.5e12, "$10^{12}$ operations: about 17 minutes", color="gray")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlim(1, 1e6)
ax.set_ylim(1e-1, 1e14)
ax.set_xlabel("n (log scale)")
ax.set_ylabel("operations (log scale)")
ax.grid(True, which="both", alpha=0.3)
ax.legend(loc="upper left", bbox_to_anchor=(1.01, 1.0))
ax.set_title("Growth rates on log-log axes")
plt.tight_layout()
plt.show()
```

### 5.2 $\ln n$ against $n^{0.1}$: the very late crossover

Left: the two functions on a logarithmic $n$-axis up to $10^{18}$. Right: their ratio, which rises to its maximum $1/(0.1\,e) \approx 3.68$ at $n = e^{10} \approx 22{,}026$ and only drops below $1$ at $n \approx 3.4 \times 10^{15}$.

```python
import matplotlib.pyplot as plt
import numpy as np

n = np.logspace(0, 18, 600)
log_n = np.log(n)
power = n**0.1
crossings = [3.0597, 3.4306e15]

fig, (ax_values, ax_ratio) = plt.subplots(1, 2, figsize=(12, 4.8))
ax_values.plot(n, log_n, label=r"$\ln n$", color="tab:blue")
ax_values.plot(n, power, label=r"$n^{0.1}$", color="tab:orange")
ax_values.set_ylabel("value (linear scale)")
ax_values.set_title(r"$\ln n$ leads from $n \approx 3$ to $n \approx 3.4 \times 10^{15}$")

ratio = np.divide(log_n, power)
ax_ratio.plot(n, ratio, color="tab:purple", label=r"$\ln n \,/\, n^{0.1}$")
ax_ratio.axhline(1, color="black", linewidth=0.8)
ax_ratio.plot(np.exp(10), 1 / (0.1 * np.e), "o", color="tab:purple")
ax_ratio.annotate(r"max $1/(0.1e) \approx 3.68$ at $n = e^{10}$", (np.exp(10), 1 / (0.1 * np.e)),
                  textcoords="offset points", xytext=(15, -5))
ax_ratio.set_ylabel("ratio (linear scale)")
ax_ratio.set_ylim(0, 4.3)
ax_ratio.set_title("Their ratio")

for ax in (ax_values, ax_ratio):
    for crossing in crossings:
        ax.axvline(crossing, color="gray", linestyle=":")
    ax.set_xscale("log")
    ax.set_xlabel("n (log scale)")
    ax.grid(True, which="both", alpha=0.3)
    ax.legend(loc="upper left")
plt.tight_layout()
plt.show()
```

### 5.3 $n^{100}$ against $1.01^n$, through their logarithms

Both numbers overflow a float long before the crossing, so we plot $\log_{10}$ of each: $100\log_{10} n$ (a slowly bending curve) and $n\log_{10} 1.01$ (a straight line through the origin). Taking logs turned the comparison into "logarithm versus line" — the line wins in the end.

```python
import matplotlib.pyplot as plt
import numpy as np

n = np.linspace(1, 200_000, 2000)
log10_polynomial = 100 * np.log10(n)
log10_exponential = n * np.log10(1.01)
n_cross = 117_308

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(n, log10_polynomial, label=r"$\log_{10}(n^{100}) = 100\log_{10} n$", color="tab:blue")
ax.plot(n, log10_exponential, label=r"$\log_{10}(1.01^n) = n\log_{10} 1.01$", color="tab:red")
ax.axvline(n_cross, color="gray", linestyle=":")
ax.text(n_cross * 1.03, 150, f"crossing at n = {n_cross:,}", color="gray")
ax.set_xlabel("n (linear scale)")
ax.set_ylabel("log10 of the value (linear scale)")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left")
ax.set_title(r"$n^{100}$ versus $1.01^n$")
plt.show()
```

4. Estimate: $n\log_2 n \approx 3 \times 10^{10}$ operations, $30$ s; $n^2 = 10^{18}$ operations, $10^9$ s $\approx 33$ years. Computed: $29.9$ s and $31.7$ years.
5. The crossover is at $n \approx 996$. For $n = 50$ use B ($2500$ versus about $28{,}000$ operations); for $n = 10^5$ use A (about $1.7 \times 10^{8}$ versus $10^{10}$).
6. The limit is $0$ by section 2.2 with $\varepsilon = 0.01$. With $n = e^t$: $t = e^{0.01 t}$, whose large root is $t \approx 647.3$, so $n \approx 10^{281.1}$. No computer will ever see an input that large, yet the limit statement is true.
7. $f'(n) = \dfrac{1 - \varepsilon\ln n}{n^{1+\varepsilon}}$, zero at $\ln n = 1/\varepsilon$; $f$ increases before and decreases after, so the maximum is $f(e^{1/\varepsilon}) = \dfrac{1/\varepsilon}{e} = \dfrac{1}{\varepsilon e}$. Hence $\ln n \le n^\varepsilon/(\varepsilon e)$. $\ln n < n^\varepsilon$ for all $n$ exactly when $1/(\varepsilon e) < 1$, i.e. $\varepsilon > \varepsilon^* = 1/e \approx 0.368$. (At $\varepsilon = 1/e$ the curves touch at $n = e^e \approx 15.15$.)
8. $\ln(n^{10}) = 10\ln n$, $\;\ln(n^{\ln n}) = (\ln n)^2$, $\;\ln(2^n) = n\ln 2$. Differences: $(\ln n)^2 - 10\ln n \to +\infty$ and $(\ln n)^2 - n\ln 2 \to -\infty$. So $n^{10} \ll n^{\ln n} \ll 2^n$: faster than every polynomial, slower than every exponential ("quasi-polynomial"). Another late crossover: $(\ln n)^2 > 10\ln n$ only once $\ln n > 10$, i.e. $n > e^{10} \approx 22{,}026$ — at $n = 1000$ the polynomial $n^{10}$ is still ahead.
9. $\tfrac{n}{2}\ln\tfrac n2 \le \ln n! \le n\ln n$, and $\dfrac{\frac n2\ln\frac n2}{n\ln n} = \dfrac12\left(1 - \dfrac{\ln 2}{\ln n}\right) \to \tfrac12$. So the ratio $\ln n!/(n\ln n)$ stays between (nearly) $\tfrac12$ and $1$: same growth rate. Numerically the ratio is $0.656$ at $n = 10$, $0.856$ at $10^3$, $0.928$ at $10^6$, $0.952$ at $10^9$ — it tends to $1$, but only like $1 - 1/\ln n$, because $\ln n! \approx n\ln n - n$ (notebook 04/04).

Numerical checks:

```python
import math

import sympy as sp

n = sp.symbols("n", positive=True)
order = [sp.log(n) ** 3, sp.sqrt(n), n * sp.log(n), n ** sp.Rational(3, 2), n**2 / sp.log(n), sp.Rational(11, 10) ** n]
print("Ex 1: consecutive limits:", [sp.limit(order[i] / order[i + 1], n, sp.oo) for i in range(len(order) - 1)])
print("Ex 2:", sp.limit(n**3 / 2**n, n, sp.oo), " log2/log10 =", round(math.log(10) / math.log(2), 4))
print("Ex 3: ln 5 =", round(math.log(5), 4), " lim 5^ln(n) / n^2 =", sp.limit(5 ** sp.log(n) / n**2, n, sp.oo))

seconds_nlogn = 1e9 * math.log2(1e9) / 1e9
seconds_square = 1e18 / 1e9
print(f"Ex 4: n log2 n -> {seconds_nlogn:.1f} s, n^2 -> {seconds_square / 3.156e7:.1f} years")


def bisect(function, low, high, steps=200):
    for _ in range(steps):
        middle = (low + high) / 2
        if function(low) * function(middle) <= 0:
            high = middle
        else:
            low = middle
    return (low + high) / 2


print(f"Ex 5: crossover n = {bisect(lambda m: m - 100 * math.log2(m), 100, 1e4):.1f};"
      f" n=50: A {100 * 50 * math.log2(50):.0f}, B {50**2};"
      f" n=1e5: A {100 * 1e5 * math.log2(1e5):.3g}, B {1e5**2:.0e}")
t6 = bisect(lambda t: t - math.exp(0.01 * t), 100, 2000)
print(f"Ex 6: t = {t6:.1f}, n = 10^{t6 / math.log(10):.1f}")
print(f"Ex 7: threshold 1/e = {1 / math.e:.4f}; at eps = 1/e: ln(e^e) = {math.e:.4f}, (e^e)^(1/e) = {math.exp(math.e) ** (1 / math.e):.4f}")
for value in [10, 1000, 22_026, 100_000]:
    print(f"Ex 8: n = {value:>6}: ln-values 10 ln n = {10 * math.log(value):7.1f}, (ln n)^2 = {math.log(value)**2:7.1f}, n ln 2 = {value * math.log(2):7.1f}")
for value in [10, 1e3, 1e6, 1e9]:
    print(f"Ex 9: n = {value:.0e}: ln n! / (n ln n) = {math.lgamma(value + 1) / (value * math.log(value)):.3f}")
```

## 8. Conclusion

- $f \ll g$ means $f(n)/g(n) \to 0$. The ladder: $\;\ln n \ll (\ln n)^k \ll n^\varepsilon \ll n \ll n\ln n \ll n^k \ll b^n \ll n! \ll n^n$.
- $\ln n \ll n^\varepsilon$ for every $\varepsilon > 0$ (L'Hôpital: the ratio becomes $1/(\varepsilon n^\varepsilon)$), and $n^k \ll b^n$ for every $k$ and $b > 1$ — the same fact seen through the mirror $n = e^m$.
- Comparing logarithms is a fast test: $f \ll g$ when $\ln f - \ln g \to -\infty$. The *difference* must diverge; a ratio of logs is not enough, because logs compress every polynomial to a multiple of $\ln n$.
- The base of a logarithm only changes a constant factor ($\log_2 n = \ln n / \ln 2$); the base of an exponential changes the growth rate ($2^n \ll 3^n$).
- "Eventually" can be absurdly late: $\ln n$ exceeds $n^{0.1}$ until $n \approx 3.4 \times 10^{15}$, and $n^{0.01}$ until about $10^{281}$. The maximum of $\ln n / n^\varepsilon$ is $1/(\varepsilon e)$, so $\ln n < n^\varepsilon$ everywhere only for $\varepsilon > 1/e$. Constants and crossovers decide practical choices; growth rates decide what happens as $n$ grows.
- On log-log axes powers are straight lines of slope $k$, logarithms flatten out and exponentials bend up and leave the plot.
