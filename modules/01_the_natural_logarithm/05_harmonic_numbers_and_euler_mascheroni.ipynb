# Harmonic Numbers, $\ln n$ and the Euler–Mascheroni Constant

This notebook studies the sum $H_n = 1 + \frac12 + \frac13 + \cdots + \frac1n$. It appears when you stack books over a table edge, when you collect coupons, and at the divergent end of Mercator's series. The sum has no closed formula — but the area under $1/x$ pins it down: $H_n \approx \ln n + \gamma$, where $\gamma = 0.5772\ldots$ is the Euler–Mascheroni constant.

This is notebook 05 of module 01 (**The natural logarithm and $e$**) in the **Learning Logarithms** series.

## 1. Introduction

**The overhanging books.** You have a pile of identical books, each of length $1$, and you stack them one per level at the edge of a table, sliding each one out as far as it will go without the stack toppling. How far can the top book stick out past the table edge?

Work from the top down. The top book can stick out $\frac12$ past the book below it (its centre of mass sits right over the edge). The top two books together have their centre of mass in the middle of the pair, which lets the second book stick out $\frac14$ past the third. The third can stick out $\frac16$, and in general the $k$-th book from the top can stick out $\frac{1}{2k}$. With $n$ books the total overhang is

$$
\frac12 + \frac14 + \frac16 + \cdots + \frac{1}{2n} = \frac12\left(1 + \frac12 + \frac13 + \cdots + \frac1n\right) = \frac{H_n}{2}.
$$

With $4$ books the top one is already entirely beyond the table ($H_4/2 = \frac{25}{24} > 1$). How many books do you need for an overhang of $2$ book lengths? Of $3$? Of $10$? Is there a limit?

**The coupon collector.** A cereal box contains one of $n$ different toys, each equally likely. How many boxes do you expect to buy to get the whole set? When you already have $k$ toys, a new box brings a new one with probability $\frac{n-k}{n}$, so you wait on average $\frac{n}{n-k}$ boxes. Adding up,

$$
\frac{n}{n} + \frac{n}{n-1} + \cdots + \frac{n}{1} = n\,H_n.
$$

Both questions need $H_n$ for large $n$, and adding a million fractions by hand is not an option. The tool that tames the sum is the logarithm: $H_n$ is a staircase approximation of the area under $1/x$, and that area is $\ln n$.

**History.** Around 1350 **Nicole Oresme** proved that the harmonic series diverges, by grouping its terms (Exercise 5). In the 1730s **Leonhard Euler** showed that $H_n - \ln n$ settles down to a constant and computed it as $0.577218$; **Lorenzo Mascheroni** published a longer computation in 1790. Today it is called the Euler–Mascheroni constant $\gamma$. Nobody knows whether $\gamma$ is rational.

## 2. Theory

### 2.1 The staircase and the hyperbola

Draw a bar of height $\frac1k$ over the interval $[k, k+1]$ for $k = 1, \dots, n$. The bars have total area $H_n$. Because $1/x$ is decreasing, on $[k, k+1]$ the curve lies **below** the top of the bar (height $\frac1k$ at the left end). So the staircase covers the area under $1/x$ from $1$ to $n + 1$:

$$
H_n \;\ge\; \int_1^{n+1} \frac{dx}{x} = \ln(n+1).
$$

Now shift the staircase one step to the left: put the bar of height $\frac1k$ over $[k-1, k]$ for $k = 2, \dots, n$. Each bar now lies **under** the curve (height $\frac1k$ is the right-end value). So

$$
H_n - 1 \;\le\; \int_1^{n} \frac{dx}{x} = \ln n.
$$

Together, the **integral sandwich**:

$$
\boxed{\;\ln(n+1) \;\le\; H_n \;\le\; 1 + \ln n.\;}
$$

The two bounds differ by less than $1$ for every $n$. So $H_n$ grows exactly like $\ln n$: slowly, but **without bound**.

### 2.2 The harmonic series diverges

Since $\ln(n + 1) \to \infty$, so does $H_n$. In exponent and log form, the sandwich says how long you wait for $H_n$ to reach a level $L$:

$$
H_n \ge L \quad\text{needs}\quad n \approx e^{L - \gamma}, \qquad\text{i.e.}\quad L \approx \ln n + \gamma.
$$

Each extra unit of $H_n$ costs a factor of $e \approx 2.718$ more terms: reaching $10$ takes about $12{,}000$ terms, reaching $20$ about $270$ million. On a log axis $H_n$ is a straight line; on a linear axis it looks as if it stops growing — but it never does.

This is the divergent endpoint of notebook 04: Mercator's series at $x = -1$ is $-H_\infty$, matching $\ln(1 + (-1)) = \ln 0 = -\infty$.

### 2.3 The Euler–Mascheroni constant

Let $\gamma_n = H_n - \ln n$. Geometrically, $\gamma_n$ is the staircase area (over $[1, n+1]$) minus the area under the curve over $[1, n]$; that is,

$$
\gamma_n = \underbrace{\sum_{k=1}^{n-1}\left(\frac1k - \int_k^{k+1}\frac{dx}{x}\right)}_{\text{curvy triangles between bars and curve}} + \frac1n.
$$

- Each curvy triangle has positive area, so the sum **increases** with $n$; the extra $\frac1n$ shrinks to $0$.
- More precisely, $\gamma_{n+1} - \gamma_n = \frac{1}{n+1} - \ln\frac{n+1}{n} < 0$ (since $\ln(1 + \frac1n) > \frac{1}{n+1}$, notebook 02), so $\gamma_n$ is **decreasing**.
- And $\gamma_n \ge \ln(n+1) - \ln n > 0$, so it is **bounded below**.

A decreasing sequence bounded below converges:

$$
\gamma = \lim_{n \to \infty}\left(H_n - \ln n\right) = 0.5772156649\ldots
$$

**Picture: $\frac12 < \gamma < 1$.** Slide every curvy triangle to the left until it sits over $[0, 1]$. Triangle $k$ lives between heights $\frac{1}{k+1}$ and $\frac1k$, so they stack without overlapping inside the unit square: $\gamma < 1$. And because $1/x$ is convex, each curvy triangle contains the straight triangle cut off by the chord, with area $\frac12\left(\frac1k - \frac{1}{k+1}\right)$; these straight triangles add up to exactly $\frac12$, so $\gamma > \frac12$. The curvy triangles fill $57.7\%$ of the unit square.

### 2.4 A sharper approximation

The error $H_n - \ln n - \gamma$ is about $\frac{1}{2n}$, and a finer expansion (the Euler–Maclaurin formula, beyond this notebook) gives

$$
H_n = \ln n + \gamma + \frac{1}{2n} - \frac{1}{12n^2} + \cdots
$$

The $\frac{1}{2n}$ has a simple picture. $\gamma_n$ contains the curvy triangles $k = 1, \dots, n-1$ plus the extra $\frac1n$, while $\gamma$ contains all of them. The missing triangles $k \ge n$ are nearly straight triangles of area $\frac12\left(\frac1k - \frac{1}{k+1}\right)$, which add up to about $\frac{1}{2n}$. So $\gamma_n - \gamma \approx \frac1n - \frac{1}{2n} = \frac{1}{2n}$.

## 3. Worked Examples

### Example 1 — small harmonic numbers

$H_1 = 1$, $H_2 = \frac32$, $H_3 = \frac{11}{6}$, $H_4 = \frac{25}{12}$, $H_5 = \frac{137}{60}$, $H_6 = \frac{49}{20}$. Four books give an overhang of $\frac{H_4}{2} = \frac{25}{24} \approx 1.04$ book lengths: the top book is completely past the edge.

### Example 2 — the sandwich at $n = 100$

$\ln 101 \approx 4.615 \le H_{100} \le 1 + \ln 100 \approx 5.605$. The estimate $\ln 100 + \gamma + \frac{1}{200} \approx 4.6052 + 0.5772 + 0.005 = 5.1874$; the exact value is $H_{100} = 5.187378\ldots$

### Example 3 — books for an overhang of $3$ lengths

Need $\frac{H_n}{2} \ge 3$, i.e. $H_n \ge 6$: $n \approx e^{6 - \gamma} = e^{5.4228} \approx 226.5$, so $n = 227$ books. (Direct summation confirms: $H_{226} < 6 \le H_{227}$.) For $5$ book lengths, $H_n \ge 10$ needs $n = 12{,}367$.

### Example 4 — rolling a die until all six faces appear

The coupon collector with $n = 6$: expected rolls $6\,H_6 = 6 \cdot \frac{49}{20} = 14.7$.

### Example 5 — collecting $50$ stickers

$50\,H_{50} \approx 50(\ln 50 + \gamma + \frac{1}{100}) = 50(3.9120 + 0.5772 + 0.01) \approx 224.96$ packets. The last sticker alone takes $50$ packets on average — the logarithm counts the slowly growing cost of the stragglers.

### Example 6 — $\gamma$ from $H_{10}$

$H_{10} = 2.928968$, $\ln 10 = 2.302585$, so $\gamma_{10} = 0.626383$ — still $0.049$ above $\gamma$. Subtracting the correction terms: $0.626383 - \frac{1}{20} + \frac{1}{1200} = 0.577216$, correct to six decimals from only ten terms.

## 4. Python Examples

### 4.1 Harmonic numbers, exactly and in floating point

```python
from fractions import Fraction


def harmonic_exact(n):
    return sum(Fraction(1, k) for k in range(1, n + 1))


for n in range(1, 11):
    value = harmonic_exact(n)
    print(f"H_{n:<2} = {str(value):>12} = {float(value):.6f}")
```

The denominators grow quickly and the numerators are never divisible by them: $H_n$ is not an integer for any $n > 1$ (a classical result). The fractions are exact but unwieldy — another reason to approximate with $\ln$.

### 4.2 The sandwich and the $\ln n + \gamma$ approximation

```python
import math

import numpy as np

euler_gamma = float(np.euler_gamma)
print(f"gamma = {euler_gamma:.12f}")
print(f"{'n':>9}{'ln(n+1)':>12}{'H_n':>14}{'1+ln n':>12}{'ln n+gamma+1/(2n)':>20}")
for n in [10, 100, 1000, 10**4, 10**6]:
    harmonic = np.sum(1 / np.arange(1, n + 1))
    approximation = math.log(n) + euler_gamma + 1 / (2 * n)
    print(f"{n:>9}{math.log(n + 1):>12.6f}{harmonic:>14.9f}{1 + math.log(n):>12.6f}{approximation:>20.9f}")
```

### 4.3 How many books? Exact search and logarithmic estimate

For each target overhang $d$ (in book lengths), find the smallest $n$ with $H_n \ge 2d$ by summing, and compare with the estimate $e^{2d - \gamma}$.

```python
import math

import numpy as np

euler_gamma = float(np.euler_gamma)
targets = [1, 2, 3, 4, 5]
harmonic = 0.0
n = 0
for overhang in targets:
    while harmonic < 2 * overhang:
        n += 1
        harmonic += 1 / n
    estimate = math.exp(2 * overhang - euler_gamma)
    print(f"overhang {overhang} book lengths: {n:>6} books   (estimate e^(2d - gamma) = {estimate:10.2f})")
```

The estimate is always within one book, and is cheap: an overhang of $10$ book lengths would need about $e^{20 - \gamma} \approx 2.7 \times 10^8$ books (Exercise 4).

### 4.4 Watching $H_n - \ln n$ converge to $\gamma$

$H_n - \ln n$ decreases towards $\gamma$ from above, and $H_n - \ln(n+1)$ increases towards it from below. The gap above $\gamma$ is close to $\frac{1}{2n}$.

```python
import math

import numpy as np

euler_gamma = float(np.euler_gamma)
for n in [1, 10, 100, 1000, 10**5]:
    harmonic = np.sum(1 / np.arange(1, n + 1))
    from_above = harmonic - math.log(n)
    from_below = harmonic - math.log(n + 1)
    print(f"n = {n:>6}: {from_below:.8f} < gamma < {from_above:.8f}   gap above = {from_above - euler_gamma:.2e}, 1/(2n) = {1 / (2 * n):.2e}")
```

### 4.5 Simulating the coupon collector

Simulate buying cereal boxes until all $n$ toys have appeared, many times, and compare the average with $n H_n$ and with $n(\ln n + \gamma)$.

```python
import math

import numpy as np

rng = np.random.default_rng(seed=1683)
euler_gamma = float(np.euler_gamma)


def boxes_until_complete(number_of_toys):
    seen = np.zeros(number_of_toys, dtype=bool)
    boxes = 0
    distinct = 0
    while distinct < number_of_toys:
        toy = rng.integers(number_of_toys)
        boxes += 1
        if not seen[toy]:
            seen[toy] = True
            distinct += 1
    return boxes


trials = 2000
for number_of_toys in [6, 20, 50]:
    results = [boxes_until_complete(number_of_toys) for _ in range(trials)]
    exact_expectation = number_of_toys * np.sum(1 / np.arange(1, number_of_toys + 1))
    log_estimate = number_of_toys * (math.log(number_of_toys) + euler_gamma)
    print(f"n = {number_of_toys:>2}: simulated mean {np.mean(results):7.2f}   n*H_n = {exact_expectation:7.2f}   n(ln n + gamma) = {log_estimate:7.2f}")
```

## 5. Visualizations

### 5.1 The book stack

Eight books stacked with maximal overhang. Each book sticks out $\frac{1}{2k}$ past the one below it ($k$ counted from the top), and the total overhang is $\frac{H_8}{2} \approx 1.36$ book lengths.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

number_of_books = 8
book_height = 0.18

# Right edge of the book k levels from the top (k = 1 is the top book), measured from the table edge at x = 0.
right_edges = []
for k in range(1, number_of_books + 1):
    overhang_below_this_book = sum(1 / (2 * j) for j in range(k + 1, number_of_books + 1))
    right_edges.append(overhang_below_this_book + 1 / (2 * k))

fig, ax = plt.subplots(figsize=(10, 4.5))
ax.add_patch(Rectangle((-2.2, -0.6), 2.2, 0.6, color="saddlebrown", alpha=0.6))
for k, right_edge in enumerate(right_edges, start=1):
    level = number_of_books - k
    ax.add_patch(Rectangle((right_edge - 1, level * book_height), 1, book_height, facecolor="tab:blue", edgecolor="white", alpha=0.8))
    ax.text(right_edge + 0.03, level * book_height + book_height / 2, f"+1/{2 * k}", va="center", fontsize=9)

top_overhang = right_edges[0]
ax.axvline(0, color="black", linestyle=":")
ax.annotate("", xy=(top_overhang, number_of_books * book_height + 0.12), xytext=(0, number_of_books * book_height + 0.12), arrowprops=dict(arrowstyle="<->"))
ax.text(top_overhang / 2, number_of_books * book_height + 0.2, f"overhang = H_8 / 2 = {top_overhang:.3f}", ha="center")
ax.text(-1.1, -0.35, "table", ha="center", color="white", fontsize=12)
ax.set_xlim(-2.3, 2.0)
ax.set_ylim(-0.7, number_of_books * book_height + 0.45)
ax.set_aspect("equal")
ax.axis("off")
ax.set_title("Harmonic book stacking: book k from the top sticks out 1/(2k)")
plt.show()
```

### 5.2 The integral sandwich

Left: bars of height $\frac1k$ on $[k, k+1]$ cover the area under $1/x$ from $1$ to $n + 1$, so $H_n \ge \ln(n+1)$. The orange curvy triangles between bars and curve add up (as $n \to \infty$) to $\gamma$. Right: the same bars shifted one step left fit under the curve, so $H_n - 1 \le \ln n$.

```python
import matplotlib.pyplot as plt
import numpy as np

n = 8
x = np.linspace(0.6, n + 1.3, 500)
fig, axes = plt.subplots(1, 2, figsize=(13, 4.8), sharey=True)

ax = axes[0]
for k in range(1, n + 1):
    ax.bar(k, 1 / k, width=1, align="edge", color="tab:blue", alpha=0.35, edgecolor="tab:blue")
    strip = np.linspace(k, k + 1, 50)
    ax.fill_between(strip, 1 / strip, 1 / k, color="tab:orange", alpha=0.8)
ax.plot(x, 1 / x, color="black", linewidth=2, label="y = 1/x")
ax.set_title(f"bars on [k, k+1]: H_{n} >= ln({n + 1}); orange gaps -> gamma")
ax.set_xlabel("x")
ax.set_ylabel("height")

ax = axes[1]
for k in range(2, n + 1):
    ax.bar(k - 1, 1 / k, width=1, align="edge", color="tab:green", alpha=0.45, edgecolor="tab:green")
ax.plot(x, 1 / x, color="black", linewidth=2, label="y = 1/x")
ax.set_title(f"bars on [k-1, k]: H_{n} - 1 <= ln {n}")
ax.set_xlabel("x")

for ax in axes:
    ax.set_xlim(0.6, n + 1.3)
    ax.set_ylim(0, 1.15)
    ax.grid(True, alpha=0.3)
    ax.legend(loc="upper right")
fig.tight_layout()
plt.show()
```

### 5.3 $H_n$ against $\ln n$ on a logarithmic axis

With a log-scaled $n$-axis, $\ln n$ is a straight line, and $H_n$ is the parallel line shifted up by $\gamma$ (with a visible bend only for the first few $n$). The shaded band is the sandwich $[\ln(n+1), 1 + \ln n]$.

```python
import matplotlib.pyplot as plt
import numpy as np

euler_gamma = float(np.euler_gamma)
n_values = np.arange(1, 10**5 + 1)
harmonic_numbers = np.cumsum(1 / n_values)

fig, ax = plt.subplots(figsize=(10, 5.5))
ax.fill_between(n_values, np.log(n_values + 1), 1 + np.log(n_values), color="tab:gray", alpha=0.2, label="sandwich [ln(n+1), 1 + ln n]")
ax.plot(n_values, harmonic_numbers, color="tab:blue", linewidth=2.5, label=r"$H_n$")
ax.plot(n_values, np.log(n_values) + euler_gamma, color="tab:red", linestyle="--", label=r"$\ln n + \gamma$")
ax.plot(n_values, np.log(n_values), color="black", linestyle=":", label=r"$\ln n$")
ax.set_xscale("log")
ax.set_xlabel("n (log scale)")
ax.set_ylabel("value")
ax.grid(True, which="both", alpha=0.3)
ax.legend(loc="upper left")
ax.set_title("H_n grows like ln n: a straight line on a log axis, offset by gamma")
plt.show()
```

### 5.4 Squeezing $\gamma$

$H_n - \ln n$ (from above) and $H_n - \ln(n+1)$ (from below) close in on $\gamma$. On log–log axes their distances from $\gamma$ are straight lines of slope $-1$: both errors behave like $\frac{1}{2n}$.

```python
import matplotlib.pyplot as plt
import numpy as np

euler_gamma = float(np.euler_gamma)
n_values = np.arange(1, 10**5 + 1)
harmonic_numbers = np.cumsum(1 / n_values)
from_above = harmonic_numbers - np.log(n_values)
from_below = harmonic_numbers - np.log(n_values + 1)

fig, (ax_left, ax_right) = plt.subplots(1, 2, figsize=(13, 4.8))
ax_left.plot(n_values, from_above, color="tab:blue", label=r"$H_n - \ln n$")
ax_left.plot(n_values, from_below, color="tab:green", label=r"$H_n - \ln(n+1)$")
ax_left.axhline(euler_gamma, color="tab:red", linestyle="--", label=r"$\gamma = 0.5772\ldots$")
ax_left.set_xscale("log")
ax_left.set_xlabel("n (log scale)")
ax_left.set_ylabel("value")
ax_left.grid(True, which="both", alpha=0.3)
ax_left.legend(loc="upper right")
ax_left.set_title("Two monotone sequences squeeze gamma")

ax_right.loglog(n_values, from_above - euler_gamma, color="tab:blue", label=r"$H_n - \ln n - \gamma$")
ax_right.loglog(n_values, euler_gamma - from_below, color="tab:green", label=r"$\gamma - (H_n - \ln(n+1))$")
ax_right.loglog(n_values, 1 / (2 * n_values), color="black", linestyle=":", label="1/(2n)")
ax_right.set_xlabel("n (log scale)")
ax_right.set_ylabel("distance from gamma (log scale)")
ax_right.grid(True, which="both", alpha=0.3)
ax_right.legend(loc="upper right")
ax_right.set_title("Both errors are about 1/(2n)")
fig.tight_layout()
plt.show()
```

## 7. Solutions

3. $10\,H_{10} = 29.29$ packets on average. The last sticker has probability $\frac{1}{10}$ per packet, so it alone takes $10$ packets on average — about a third of the total.
4. $n \approx e^{10 - \gamma} \approx e^{9.42} \approx 12{,}370$ and $n \approx e^{20 - \gamma} \approx e^{19.42} \approx 2.7 \times 10^8$. Exact: $H_n > 10$ first at $n = 12{,}367$; for $20$, solving $\ln n + \gamma + \frac{1}{2n} \ge 20$ gives $n = 272{,}400{,}600$.
5. The group $\frac{1}{2^{j}+1} + \cdots + \frac{1}{2^{j+1}}$ has $2^j$ terms, each at least $\frac{1}{2^{j+1}}$, so the group is at least $\frac12$. After $1 + \frac12$ and $k - 1$ more groups: $H_{2^k} \ge 1 + \frac{k}{2}$. The true growth is $\ln 2^k \approx 0.693k$ — Oresme's bound has the right shape (linear in $k = \log_2 n$) with a smaller slope.
6. $e^{8 - \gamma} \approx 1673.7$, so $n = 1674$; summation confirms $H_{1673} < 8 \le H_{1674}$.
7. $\gamma_{n+1} - \gamma_n = \frac{1}{n+1} - \left(\ln(n+1) - \ln n\right) = \frac{1}{n+1} - \ln\left(1 + \frac1n\right) < 0$, since the area under $1/x$ over $[n, n+1]$ exceeds the rectangle of height $\frac{1}{n+1}$. Positivity: $\gamma_n = H_n - \ln n \ge \ln(n+1) - \ln n > 0$ by the sandwich.
8. $H_{2n} - H_n \approx (\ln 2n + \gamma) - (\ln n + \gamma) = \ln 2$. And $1 - \frac12 + \frac13 - \cdots - \frac{1}{2n} = H_{2n} - 2\cdot\frac12 H_n = H_{2n} - H_n$ (subtracting twice the even terms). So the alternating harmonic series converges to $\ln 2$ because the $\gamma$'s cancel. For $n = 1000$: $H_{2000} - H_{1000} = 0.692897$, $\ln 2 = 0.693147$.
9. From $H_{10}$: $\gamma \approx H_{10} - \ln 10 - \frac{1}{20} + \frac{1}{1200} = 0.5772165$, an error of $8.3 \times 10^{-7}$ — six correct decimals from ten terms. From $H_{100}$ the error drops to about $8 \times 10^{-11}$: ten decimals. In both cases the error equals the next term of the expansion, $+\frac{1}{120n^4}$, which shrinks by $10^4$ when $n$ grows by $10$.

Numerical checks:

```python
import math
from fractions import Fraction

import numpy as np

euler_gamma = float(np.euler_gamma)


def harmonic(n):
    return float(np.sum(1 / np.arange(1, n + 1)))


h7 = sum(Fraction(1, k) for k in range(1, 8))
h8 = h7 + Fraction(1, 8)
print("Ex 1:", h7, float(h7), h8, float(h8), " overhang", h8 / 2, round(float(h8 / 2), 4))
print(f"Ex 2: {math.log(1001):.4f} <= {harmonic(1000):.6f} <= {1 + math.log(1000):.4f};  ln n + gamma = {math.log(1000) + euler_gamma:.4f}")
print(f"Ex 3: 10*H_10 = {10 * harmonic(10):.2f}")

running_sum = 0.0
n = 0
while running_sum <= 10:
    n += 1
    running_sum += 1 / n
print("Ex 4: H_n > 10 first at n =", n, "  estimate", round(math.exp(10 - euler_gamma)))
n_twenty = math.floor(math.exp(20 - euler_gamma))
while math.log(n_twenty) + euler_gamma + 1 / (2 * n_twenty) < 20:
    n_twenty += 1
print("      H_n > 20 first at n ~", n_twenty, "  (approximation at n-1:", round(math.log(n_twenty - 1) + euler_gamma + 1 / (2 * (n_twenty - 1)) - 20, 12), ")")

print("Ex 5: H_{2^k} vs 1 + k/2:", [(k, round(harmonic(2**k), 3), 1 + k / 2) for k in [1, 4, 10, 16]])
print("Ex 6:", round(math.exp(8 - euler_gamma), 1), harmonic(1673) < 8 <= harmonic(1674))
print("Ex 7: gamma_n decreasing for n = 1..10^4:", bool(np.all(np.diff(np.cumsum(1 / np.arange(1, 10**4 + 1)) - np.log(np.arange(1, 10**4 + 1))) < 0)))
print(f"Ex 8: H_2000 - H_1000 = {harmonic(2000) - harmonic(1000):.6f}, ln 2 = {math.log(2):.6f}")
for n in [10, 100]:
    gamma_estimate = harmonic(n) - math.log(n) - 1 / (2 * n) + 1 / (12 * n**2)
    print(f"Ex 9: from H_{n}: gamma ~ {gamma_estimate:.12f}, error {abs(gamma_estimate - euler_gamma):.1e}, next term 1/(120 n^4) = {1 / (120 * n**4):.1e}")
```

## 8. Conclusion

- The **harmonic numbers** $H_n = 1 + \frac12 + \cdots + \frac1n$ count the overhang of a book stack ($H_n / 2$) and the expected cost of collecting $n$ coupons ($n H_n$).
- Comparing the staircase of bars with the area under $1/x$ gives the **integral sandwich** $\ln(n+1) \le H_n \le 1 + \ln n$. So the harmonic series diverges, but only logarithmically: each extra unit of sum costs a factor $e$ more terms.
- $H_n - \ln n$ decreases to the **Euler–Mascheroni constant** $\gamma = 0.5772156649\ldots$, the total area of the curvy triangles between the staircase and the hyperbola. Whether $\gamma$ is rational is still unknown.
- In exponent and log form: $H_n \approx \ln n + \gamma \iff n \approx e^{H_n - \gamma}$. The refinement $H_n \approx \ln n + \gamma + \frac{1}{2n} - \frac{1}{12n^2}$ gives many digits from a few terms.
- Overhanging $3$ book lengths needs $227$ books, $5$ lengths needs $12{,}367$, $10$ lengths about $2.7 \times 10^8$ — unbounded, but logarithmically slow.
