# Derivatives of $\ln x$ and $e^x$

This notebook explains the word *natural* in "natural logarithm". Every exponential function $b^x$ grows at a rate proportional to its own size, $\frac{d}{dx} b^x = c_b \cdot b^x$ — but the constant $c_b$ looks like a random decimal ($0.693\ldots$ for $b = 2$, $2.302\ldots$ for $b = 10$). Those constants are $\ln 2$ and $\ln 10$. The natural logarithm *is* the growth-rate constant, and $e$ is the one base for which that constant is exactly $1$.

This is notebook 03 of module 01 (**The natural logarithm and $e$**) in the **Learning Logarithms** series.

## 1. Introduction

A bacterial culture doubles every hour: $N(t) = N_0 \cdot 2^t$. How fast is it growing *right now*, in bacteria per hour?

The average growth over the next hour is $N_0 2^{t+1} - N_0 2^t = N_0 2^t$ — but that overstates the instantaneous rate, because the culture keeps speeding up during the hour. Shrink the time step $h$:

$$
\frac{2^{t+h} - 2^{t}}{h} = 2^{t} \cdot \frac{2^{h} - 1}{h}.
$$

The factor $2^t$ comes out: **the growth rate is proportional to the current size**. The constant of proportionality is $c_2 = \lim_{h \to 0} \frac{2^h - 1}{h}$. Computing it numerically:

| $h$ | $0.1$ | $0.001$ | $10^{-6}$ |
|---|---|---|---|
| $\frac{2^h - 1}{h}$ | $0.7177$ | $0.69339$ | $0.6931474$ |

The limit is $0.693147\ldots$ For base $10$ the same experiment gives $2.302585\ldots$, for base $3$ it gives $1.098612\ldots$ Where do these numbers come from? Is there a base where the constant is exactly $1$ — a function that is its **own derivative**?

The answer ties the two previous notebooks together: the constants are $\ln 2$, $\ln 10$, $\ln 3$ — the areas under the hyperbola of notebook 02 — and the base with constant $1$ is $e$ from notebook 01. Without the natural logarithm, every exponential carries its own unexplained constant; with it, they all become $e^{kx}$ with a rate $k = \ln b$ you can read off.

## 2. Theory

### 2.1 Every exponential grows in proportion to itself

For any base $b > 0$:

$$
\frac{d}{dx} b^x = \lim_{h \to 0} \frac{b^{x+h} - b^x}{h} = b^x \cdot \underbrace{\lim_{h \to 0} \frac{b^h - 1}{h}}_{c_b} = c_b\, b^x.
$$

The constant $c_b$ is the slope of $b^x$ at $x = 0$. The question is what $c_b$ is.

### 2.2 $\frac{d}{dx}\ln x = \frac{1}{x}$, from the area

With $\ln x = \int_1^x \frac{dt}{t}$ (notebook 02), increasing $x$ by a small $h$ adds a thin strip of width $h$ whose height is between $\frac{1}{x+h}$ and $\frac{1}{x}$:

$$
\frac{h}{x + h} \le \ln(x + h) - \ln x \le \frac{h}{x}
\quad\Longrightarrow\quad
\frac{\ln(x+h) - \ln x}{h} \longrightarrow \frac{1}{x}.
$$

This is the fundamental theorem of calculus for this one integral: **the rate at which area accumulates equals the height of the curve.**

### 2.3 The same result from the limit definition

Without areas, using only the log laws and $e = \lim_{n\to\infty}(1 + 1/n)^n$ (which also holds for $n \to \infty$ through real values):

$$
\frac{\ln(x+h) - \ln x}{h} = \frac{1}{h}\ln\left(1 + \frac{h}{x}\right) = \frac{1}{x}\ln\left(1 + \frac{h}{x}\right)^{x/h} \longrightarrow \frac{1}{x}\ln e = \frac{1}{x}.
$$

Here $n = x/h \to \infty$. The slope of $\ln$ at $x = 1$ being exactly $1$ is the same fact as $\ln e = 1$.

### 2.4 $e^x$ is its own derivative — the inverse-function picture

$e^x$ is the inverse of $\ln x$. Reflecting a graph across $y = x$ swaps the roles of rise and run, so slopes become **reciprocals**: if $y = \ln x$ has slope $m$ at $(a, b)$, then $x = e^y$ has slope $1/m$ at $(b, a)$. Formally, with $f = \ln$ and $f^{-1} = \exp$:

$$
\frac{d}{dx} e^x = \frac{1}{f'(f^{-1}(x))} = \frac{1}{1 / e^x} = e^x.
$$

So $c_e = 1$: **at every point, $e^x$ is growing at a rate equal to its own value.** On the graph: the tangent to $e^x$ at $x = a$ has slope $e^a$, its height.

**Uniqueness.** If $f' = f$ and $f(0) = 1$, then $g(x) = f(x)\,e^{-x}$ has $g' = f'e^{-x} - fe^{-x} = 0$, so $g$ is constant, $g = g(0) = 1$, and $f = e^x$.

### 2.5 Any base: $\ln b$ is the growth-rate constant

Every positive base is a power of $e$: $b = e^{\ln b}$. So

$$
b^x = e^{x \ln b}, \qquad \frac{d}{dx} b^x = e^{x\ln b} \cdot \ln b = b^x \ln b.
$$

This identifies the mysterious constant: $c_b = \ln b$. The puzzle from the introduction is solved:

| base $b$ | $c_b = $ slope of $b^x$ at $0$ | $\ln b$ |
|---|---|---|
| $2$ | $0.693147\ldots$ | $\ln 2$ |
| $e$ | $1$ | $\ln e = 1$ |
| $3$ | $1.098612\ldots$ | $\ln 3$ |
| $10$ | $2.302585\ldots$ | $\ln 10$ |

**This is what makes the natural logarithm natural.** For growth $N(t) = N_0 b^t$, the number $\ln b$ is the **continuous growth rate**: $N'(t) = (\ln b)\, N(t)$. Doubling every hour is a continuous rate of $\ln 2 \approx 69.3\%$ per hour. In exponent and log form:

$$
b = e^{k} \quad\Longleftrightarrow\quad k = \ln b.
$$

Any other base's logarithm carries the conversion factor. Since $\log_b x = \frac{\ln x}{\ln b}$,

$$
\frac{d}{dx}\log_b x = \frac{1}{x \ln b}.
$$

Only for $b = e$ do both formulas lose their constant: $\frac{d}{dx}e^x = e^x$ and $\frac{d}{dx}\ln x = \frac1x$.

### 2.6 Negative arguments

For $x < 0$, $\ln|x| = \ln(-x)$, and the chain rule gives $\frac{1}{-x}\cdot(-1) = \frac1x$. So $\frac{d}{dx}\ln|x| = \frac1x$ for all $x \neq 0$, which is why $\int \frac{dx}{x} = \ln|x| + C$.

## 3. Worked Examples

### Example 1 — how fast does $2^x$ grow at $x = 10$?

$\frac{d}{dx}2^x\big|_{x=10} = 2^{10}\ln 2 = 1024 \times 0.693147 \approx 709.78$. Not $1024$: the slope is always $69.3\%$ of the height.

### Example 2 — rate of change of $\log_{10} x$ at $x = 100$

$\frac{d}{dx}\log_{10} x = \frac{1}{x\ln 10}$, so at $x = 100$: $\frac{1}{100 \times 2.302585} \approx 0.004343$. Moving from $100$ to $101$ raises $\log_{10}$ by about $0.0043$.

### Example 3 — $\frac{d}{dx}\ln(3x)$

$\ln(3x) = \ln 3 + \ln x$, so the derivative is $\frac1x$ — the constant factor $3$ becomes an additive constant and disappears. (Chain rule check: $\frac{1}{3x}\cdot 3 = \frac1x$.) On a graph: $\ln(3x)$ is $\ln x$ shifted up by $\ln 3$, so the slopes match.

### Example 4 — the tangent to $\ln x$ at $x = e$

Slope $\frac1e$, point $(e, 1)$: $y = 1 + \frac{1}{e}(x - e) = \frac{x}{e}$. It passes through the origin — $e$ is the only point where a tangent to $\ln$ goes through $(0, 0)$.

### Example 5 — mirror slopes

At $x = \ln 4 \approx 1.386$, $e^x$ has height $4$ and slope $4$. The mirror point on $y = \ln x$ is $(4, \ln 4)$, with slope $\frac14$. Reciprocal slopes at mirror points.

### Example 6 — writing $5^x$ as $e^{kx}$

$5^x = e^{x\ln 5} = e^{1.6094x}$: a continuous growth rate of $160.94\%$ per unit, and $\frac{d}{dx}5^x = 1.6094 \cdot 5^x$.

## 4. Python Examples

### 4.1 Measuring $c_b$ and comparing with $\ln b$

The difference quotient $\frac{b^h - 1}{h}$ for shrinking $h$. The error shrinks like $h$; at $h = 10^{-6}$ we have about six correct digits.

```python
import math

print(f"{'base':>8}{'h=0.1':>12}{'h=0.001':>12}{'h=1e-6':>14}{'ln b':>14}")
for base in [2, math.e, 3, 10]:
    quotients = [(base**h - 1) / h for h in [0.1, 0.001, 1e-6]]
    print(f"{base:>8.4f}{quotients[0]:>12.6f}{quotients[1]:>12.6f}{quotients[2]:>14.8f}{math.log(base):>14.8f}")
```

### 4.2 Difference quotients of $\ln$ against $1/x$

A symmetric difference quotient $\frac{\ln(x+h) - \ln(x-h)}{2h}$ is accurate to about $h^2$, so even $h = 10^{-4}$ gives eight or more digits.

```python
import math

step = 1e-4
for x in [0.25, 1, math.e, 10, 1000]:
    slope = (math.log(x + step) - math.log(x - step)) / (2 * step)
    print(f"x = {x:>9.4f}: numerical slope {slope:.10f}   1/x = {1 / x:.10f}   close: {math.isclose(slope, 1 / x, rel_tol=1e-7)}")
```

### 4.3 SymPy confirms the rules

```python
import sympy as sp

x = sp.symbols("x", positive=True)
b = sp.symbols("b", positive=True)

print("d/dx ln x       =", sp.diff(sp.log(x), x))
print("d/dx e^x        =", sp.diff(sp.exp(x), x))
print("d/dx b^x        =", sp.diff(b**x, x))
print("d/dx log_b x    =", sp.simplify(sp.diff(sp.log(x, b), x)))
print("d/dx ln(3x)     =", sp.simplify(sp.diff(sp.log(3 * x), x)))
print("lim (b^h-1)/h   =", sp.limit((b**x - 1) / x, x, 0))

f = sp.Function("f")
print("solve f' = f, f(0) = 1:", sp.dsolve(sp.Eq(f(x).diff(x), f(x)), f(x), ics={f(0): 1}))
```

### 4.4 Euler's method for $y' = y$ rebuilds compound interest

Solve $y' = y$, $y(0) = 1$ by taking $n$ steps of size $\frac1n$: each step does $y \leftarrow y + \frac1n y = (1 + \frac1n)\,y$. After $n$ steps, $y(1) \approx (1 + \frac1n)^n$ — exactly Bernoulli's compounding. "Growing at a rate equal to your size" and "compounding continuously at 100%" are the same statement.

```python
import math


def euler_method_exponential(steps, rate=1.0, end_time=1.0):
    step_size = end_time / steps
    y = 1.0
    for _ in range(steps):
        y = y + step_size * rate * y
    return y


for steps in [1, 2, 4, 12, 365, 10_000]:
    euler_value = euler_method_exponential(steps)
    compound_value = (1 + 1 / steps) ** steps
    print(f"n = {steps:>6}: Euler {euler_value:.8f}   (1+1/n)^n = {compound_value:.8f}")
print(f"exact y(1) = e = {math.e:.8f}")
```

## 5. Visualizations

### 5.1 Tangents at $x = 0$: the slope is $\ln b$

The curves $2^x$, $e^x$ and $3^x$ all pass through $(0, 1)$. Their tangent lines there have slopes $\ln 2 \approx 0.69$, $1$ and $\ln 3 \approx 1.10$. The base $e$ sits between $2$ and $3$ exactly where the tangent slope hits $1$.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-2, 1.6, 400)
fig, ax = plt.subplots(figsize=(10, 6))
for base, colour, name in [(2, "tab:blue", "2"), (math.e, "black", "e"), (3, "tab:red", "3")]:
    slope = math.log(base)
    ax.plot(x, base**x, color=colour, linewidth=2, label=f"y = {name}^x")
    ax.plot(x, 1 + slope * x, color=colour, linestyle="--", linewidth=1, label=f"tangent at 0, slope ln {name} = {slope:.3f}")
ax.plot(0, 1, "o", color="black")
ax.set_xlim(-2, 1.6)
ax.set_ylim(-1, 5)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left")
ax.set_title("All exponentials pass through (0, 1); the slope there is ln b")
plt.show()
```

### 5.2 Mirror images, reciprocal slopes

$y = e^x$ and $y = \ln x$ reflected across $y = x$. At each pair of mirror points the tangent slopes are reciprocals: $e^a$ on the exponential, $e^{-a}$ on the logarithm.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(7.5, 7.5))
x_exp = np.linspace(-3, 1.9, 300)
x_log = np.linspace(0.05, 6.5, 300)
ax.plot(x_exp, np.exp(x_exp), color="tab:orange", linewidth=2, label=r"$y = e^x$")
ax.plot(x_log, np.log(x_log), color="tab:purple", linewidth=2, label=r"$y = \ln x$")
ax.plot([-3, 6.5], [-3, 6.5], color="gray", linestyle=":", label="y = x")

for a, colour in [(0.0, "tab:green"), (math.log(4), "tab:blue")]:
    height = math.exp(a)
    span = np.linspace(-0.9, 0.9, 2)
    ax.plot(a + span, height + height * span, color=colour, linewidth=1.2)
    ax.plot(height + span, a + span / height, color=colour, linewidth=1.2)
    ax.plot([a, height], [height, a], "o", color=colour)
    ax.plot([a, height], [height, a], color=colour, linestyle=":", alpha=0.6)
    mirror_label = "slope 1" if height == 1 else f"slope 1/{height:g}"
    ax.annotate(f"slope {height:g}", xy=(a, height), xytext=(a - 1.7, height + 0.2), color=colour)
    ax.annotate(mirror_label, xy=(height, a), xytext=(height + 0.2, a - 0.7), color=colour)

ax.set_xlim(-3, 6.5)
ax.set_ylim(-3, 6.5)
ax.set_aspect("equal")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower right")
ax.set_title("Reflection across y = x turns slope m into 1/m")
plt.show()
```

### 5.3 The growth constant $c_b$ as a function of $b$

Measured slopes $\frac{b^h - 1}{h}$ (dots, $h = 10^{-6}$) for many bases, against the curve $\ln b$. Left: linear $b$-axis. Right: logarithmic $b$-axis, where $\ln b$ is a straight line and the bases $b$ and $1/b$ sit symmetrically with opposite growth rates (growth versus decay).

```python
import math

import matplotlib.pyplot as plt
import numpy as np

bases = np.array([0.1, 0.2, 0.5, 1, 2, math.e, 3, 5, 10])
step = 1e-6
measured = (bases**step - 1) / step
b_curve = np.logspace(-1.1, 1.1, 300)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
for ax in axes:
    ax.plot(b_curve, np.log(b_curve), color="tab:purple", label=r"$\ln b$")
    ax.plot(bases, measured, "o", color="black", label=r"measured $(b^h - 1)/h$")
    ax.axhline(1, color="gray", linestyle="--")
    ax.axvline(math.e, color="gray", linestyle=":")
    ax.set_ylabel("growth constant c_b")
    ax.legend(loc="upper left")
axes[0].set_xlabel("base b (linear scale)")
axes[0].grid(True, alpha=0.3)
axes[0].set_title("c_b = ln b; c_b = 1 at b = e")
axes[1].set_xscale("log")
axes[1].set_xlabel("base b (log scale)")
axes[1].grid(True, which="both", alpha=0.3)
axes[1].set_title("On a log axis: a straight line, odd about b = 1")
fig.tight_layout()
plt.show()
```

### 5.4 The slope of $\ln$ is the height of $1/x$

Top: $\ln x$ with tangent lines at $x = \frac12, 1, 2, 4$. Bottom: the hyperbola $1/x$ with the same four points marked. The slope of each tangent above equals the height of the point below — the area function grows exactly as fast as the curve is tall.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0.2, 6, 400)
points = [0.5, 1, 2, 4]
colours = ["tab:red", "tab:orange", "tab:green", "tab:blue"]

fig, (ax_top, ax_bottom) = plt.subplots(2, 1, figsize=(9, 8), sharex=True)
ax_top.plot(x, np.log(x), color="tab:purple", linewidth=2, label=r"$y = \ln x$")
for point, colour in zip(points, colours):
    span = np.linspace(-0.6, 0.6, 2) * point
    ax_top.plot(point + span, np.log(point) + span / point, color=colour, linewidth=1.5)
    ax_top.plot(point, np.log(point), "o", color=colour, label=f"slope at {point} = {1 / point:g}")
ax_top.set_ylabel("ln x")
ax_top.grid(True, alpha=0.3)
ax_top.legend(loc="lower right")
ax_top.set_title("Tangent slopes of ln x ...")

ax_bottom.plot(x, 1 / x, color="black", linewidth=2, label="y = 1/x")
for point, colour in zip(points, colours):
    ax_bottom.vlines(point, 0, 1 / point, color=colour, linewidth=2)
    ax_bottom.plot(point, 1 / point, "o", color=colour)
ax_bottom.set_xlabel("x")
ax_bottom.set_ylabel("1/x")
ax_bottom.set_ylim(0, 2.2)
ax_bottom.grid(True, alpha=0.3)
ax_bottom.legend(loc="upper right")
ax_bottom.set_title("... equal the heights of 1/x")
fig.tight_layout()
plt.show()
```

## 7. Solutions

1. $\frac{d}{dx}5^x = 5^x \ln 5 \approx 1.6094 \cdot 5^x$; $\frac{d}{dx}\log_2 x = \frac{1}{x\ln 2}$; $\ln(x^3) = 3\ln x$, so the derivative is $\frac3x$.
2. Slope of $e^x$ at $\ln 7$ is $e^{\ln 7} = 7$; slope of $\ln$ at $7$ is $\frac17$. The points $(\ln 7, 7)$ and $(7, \ln 7)$ are mirror images across $y = x$, so the slopes are reciprocals.
3. $y = x - 1$ and $y = x + 1$. They are mirror images of each other across $y = x$ (swap $x$ and $y$ in $y = x - 1$ to get $x = y - 1$), both with slope $1$.
4. $2^{20} = (2^{10})^2 \approx 10^6$, so the slope is $\approx 0.7 \times 10^6 = 7 \times 10^5$; exact: $2^{20}\ln 2 \approx 726{,}817$. Change of $\log_{10}$: $\approx \frac{1}{1000 \ln 10} \approx \frac{1}{2300} \approx 4.3 \times 10^{-4}$; exact: $\log_{10} 1001 - 3 = 4.341 \times 10^{-4}$.
5. For $x < 0$, $\ln|x| = \ln(-x)$; the chain rule gives $\frac{1}{-x}\cdot(-1) = \frac1x$. At $x = -2$ the slope is $-\frac12$.
6. $\ln b = 2 \Rightarrow b = e^2 \approx 7.389$. $\ln b = -1 \Rightarrow b = e^{-1} \approx 0.3679$ (a decaying exponential).
7. Let $g(t) = N(t)e^{-kt}$. Then $g' = N'e^{-kt} - kNe^{-kt} = 0$, so $g(t) = g(0) = N_0$ and $N(t) = N_0 e^{kt}$. Tripling every 5 hours: $e^{5k} = 3$, so $k = \frac{\ln 3}{5} \approx 0.2197$ per hour.
8. Let $f(x) = b^x$ with $f'(x) = b^x \ln b$ and inverse $f^{-1}(x) = \log_b x$. Then $\frac{d}{dx}\log_b x = \frac{1}{f'(\log_b x)} = \frac{1}{b^{\log_b x}\ln b} = \frac{1}{x\ln b}$.
9. Each step multiplies $y$ by $1 + \frac{k}{n}$, so after $n$ steps $y = (1 + k/n)^n$. With $k = \ln 2$, $n = 12$: $(1 + 0.057762)^{12} \approx 1.9618$, about $1.9\%$ short of $2$.

## 8. Conclusion

- Every exponential grows in proportion to itself: $\frac{d}{dx}b^x = c_b\,b^x$, where $c_b$ is the slope at $0$.
- $\frac{d}{dx}\ln x = \frac1x$: the area under $1/t$ grows as fast as the curve is tall (or, from the limit, because $\ln e = 1$).
- Reflecting across $y = x$ inverts slopes, so $\frac{d}{dx}e^x = e^x$: $e$ is the unique base whose exponential is its own derivative, and $e^x$ is the unique solution of $f' = f$, $f(0) = 1$.
- Writing $b^x = e^{x\ln b}$ gives $\frac{d}{dx}b^x = b^x\ln b$ and $\frac{d}{dx}\log_b x = \frac{1}{x\ln b}$. The **natural logarithm of the base is the continuous growth rate**: $b = e^k \iff k = \ln b$. That is the "natural" in natural logarithm.
- Euler's method for $y' = y$ reproduces $(1 + 1/n)^n$: continuous growth and continuous compounding are the same idea.
