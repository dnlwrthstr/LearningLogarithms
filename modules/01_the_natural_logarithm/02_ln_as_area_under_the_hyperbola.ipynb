# $\ln x$ as the Area Under the Hyperbola $1/x$

This notebook finds a logarithm hiding in a picture. Take the hyperbola $y = 1/x$ and measure the area underneath it from $1$ to $x$. That area turns products into sums — exactly the property Napier built his tables for — so it *is* a logarithm. Its base is the number where the area reaches $1$, and that number is the $e$ of notebook 01.

This is notebook 02 of module 01 (**The natural logarithm and $e$**) in the **Learning Logarithms** series.

## 1. Introduction

By the 1640s mathematicians could find the area under $y = x^n$ for the powers $n \neq -1$: Fermat and others had shown, for instance, that for $n \ge 0$ the area from $0$ to $x$ is $\frac{x^{n+1}}{n+1}$. The one exception was the **hyperbola** $y = 1/x$ ($n = -1$), where the formula would divide by zero. What *is* the area under $1/x$?

Try to compute it. The area from $1$ to $2$, from $2$ to $4$, from $4$ to $8$… Look at the picture in section 5.1 before reading on: these three regions have very different shapes — one tall and narrow, one short and wide — but they look suspiciously similar in size.

They are equal. **Grégoire de Saint-Vincent** proved this in his *Opus geometricum* (1647): the areas under the hyperbola over intervals $[a, b]$ and $[ka, kb]$ are always the same. Two years later his student **Alphonse Antonio de Sarasa** pointed out what this means: if the intervals are geometrically spaced ($1, 2, 4, 8, \dots$), the areas grow arithmetically ($A, 2A, 3A, \dots$). A function that turns a geometric sequence into an arithmetic one is a **logarithm** — Stifel's two rows again (notebook 00/01).

So the missing area formula is not a power of $x$ at all; it is a logarithm. For decades afterwards the natural logarithm was called the **hyperbolic logarithm**.

This picture gives a definition of $\ln x$ that needs no exponent, no base, no limit of $(1 + 1/n)^n$ — just an area. Everything else about the natural logarithm follows from it.

## 2. Theory

### 2.1 The area function

For $x > 0$ define

$$
L(x) = \int_1^x \frac{dt}{t}.
$$

For $x > 1$ this is the area under $y = 1/t$ between $t = 1$ and $t = x$. For $0 < x < 1$ the integral runs backwards, so $L(x)$ is **minus** the area between $x$ and $1$. Directly from the definition:

- $L(1) = 0$ (no area).
- $L$ is strictly increasing: moving $x$ to the right adds positive area.
- $L$ is continuous (a small change in $x$ adds a thin sliver of area).

### 2.2 Stretching the region: the key lemma

**Lemma.** For $a, b > 0$:

$$
\int_a^{ab} \frac{dt}{t} = \int_1^{b} \frac{dt}{t}.
$$

**Picture proof.** Take the region under $1/t$ over $[1, b]$. Stretch it **horizontally** by the factor $a$ and squash it **vertically** by the factor $1/a$, i.e. map each point $(t, y) \mapsto (at, y/a)$.

- The base $[1, b]$ becomes $[a, ab]$.
- The top curve $y = 1/t$ becomes $y/a$ over $x = at$, which is height $\frac{1}{a t} = \frac1x$ — the **same hyperbola**.
- Areas get multiplied by $a \cdot \frac1a = 1$.

So the stretched region is exactly the region under $1/x$ over $[a, ab]$, and it has the same area. The hyperbola is the one curve that maps to itself under this stretch-and-squash, which is why it carries a logarithm.

**Algebra proof.** Substitute $t = a u$, $dt = a\,du$: $\int_a^{ab} \frac{dt}{t} = \int_1^{b} \frac{a\,du}{a u} = \int_1^b \frac{du}{u}$.

### 2.3 Products become sums

Split the area from $1$ to $ab$ at $a$, and use the lemma on the second piece:

$$
L(ab) = \int_1^{a} \frac{dt}{t} + \int_a^{ab} \frac{dt}{t} = L(a) + L(b).
$$

This is the defining law of a logarithm. Consequences follow exactly as for $\log_b$:

| Law | Proof from $L(ab) = L(a) + L(b)$ |
|---|---|
| $L(1/a) = -L(a)$ | $L(a) + L(1/a) = L(1) = 0$ |
| $L(a/b) = L(a) - L(b)$ | $L(a/b) = L(a) + L(1/b)$ |
| $L(a^n) = n L(a)$ | repeat the product law $n$ times |
| $L(a^{p/q}) = \tfrac{p}{q} L(a)$ | $q\,L(a^{p/q}) = L(a^p) = p\,L(a)$ |

Saint-Vincent's equal areas are the special case $L(2^{k+1}) - L(2^k) = L(2)$ for every $k$.

### 2.4 Which logarithm? Finding the base

$L$ is continuous, increasing, and satisfies $L(ab) = L(a) + L(b)$. Since $L(2) > 0$ and $L(2^n) = nL(2) \to \infty$, $L$ takes every value, in particular the value $1$ at exactly one point. Call that point $B$. Then for every rational $r$, $L(B^r) = r$, and by continuity

$$
L(x) = \log_B x \qquad\text{for all } x > 0.
$$

**What is $B$?** Compare the area from $1$ to $1 + \frac1n$ with two rectangles of width $\frac1n$: the curve is between heights $\frac{1}{1 + 1/n}$ and $1$, so

$$
\frac{1}{n+1} \;\le\; L\!\left(1 + \tfrac1n\right) \;\le\; \frac1n
\qquad\Longrightarrow\qquad
\frac{n}{n+1} \;\le\; L\!\left(\left(1 + \tfrac1n\right)^n\right) \;\le\; 1.
$$

The middle tends to $1$ as $n \to \infty$, and $(1 + 1/n)^n \to e$, so by continuity $L(e) = 1$. The base is Bernoulli's number:

$$
\boxed{\;\ln x = \int_1^x \frac{dt}{t} = \log_e x, \qquad \int_1^{e} \frac{dt}{t} = 1.\;}
$$

In exponent form: $e^{y} = x \iff y = \ln x \iff$ the area under $1/t$ from $1$ to $x$ equals $y$.

### 2.5 Two inequalities for free

On $[1, x]$ (for $x > 1$) the curve $1/t$ lies between $\frac1x$ and $1$, so the area lies between two rectangles:

$$
1 - \frac1x = \frac{x - 1}{x} \;\le\; \ln x \;\le\; x - 1.
$$

The same inequalities hold for $0 < x < 1$ (with the backwards integral), with equality only at $x = 1$. They say that the tangent line $y = x - 1$ at $x = 1$ lies above the graph of $\ln$ — the slope of $\ln$ at $1$ is exactly $1$, a fact notebook 03 uses.

## 3. Worked Examples

### Example 1 — the area from $3$ to $12$

By the lemma with $a = 3$, $b = 4$: $\int_3^{12} \frac{dt}{t} = \int_1^4 \frac{dt}{t} = \ln 4 = 2\ln 2 \approx 1.3863$. Equivalently $\ln 12 - \ln 3 = \ln(12/3)$.

### Example 2 — bounding $\ln 2$ with four rectangles

Split $[1, 2]$ into four strips of width $\frac14$ at $1, \frac54, \frac32, \frac74, 2$. The curve is decreasing, so the left endpoints give rectangles that are too tall and the right endpoints rectangles that are too short:

$$
\text{upper} = \tfrac14\left(1 + \tfrac45 + \tfrac23 + \tfrac47\right) = \tfrac{319}{420} \approx 0.7595,
\qquad
\text{lower} = \tfrac14\left(\tfrac45 + \tfrac23 + \tfrac47 + \tfrac12\right) = \tfrac{533}{840} \approx 0.6345.
$$

So $0.6345 < \ln 2 < 0.7595$. The midpoint rule (heights $1/t$ at $\frac98, \frac{11}8, \frac{13}8, \frac{15}8$) gives $0.6912$, already within $0.002$ of $\ln 2 = 0.693147\ldots$

### Example 3 — using the product law

Given $\ln 2 \approx 0.6931$ and $\ln 3 \approx 1.0986$:

- $\ln 6 = \ln 2 + \ln 3 \approx 1.7918$,
- $\ln 1.5 = \ln 3 - \ln 2 \approx 0.4055$,
- $\ln \tfrac{8}{9} = 3\ln 2 - 2\ln 3 \approx -0.1178$ (negative because $\tfrac89 < 1$).

### Example 4 — locating $e$ as an area

Area from $1$ to $2$ is $0.693$; from $1$ to $3$ is $1.099$. The area reaches $1$ somewhere between $2$ and $3$, and closer to $3$: linear interpolation gives $2 + \frac{1 - 0.693}{1.099 - 0.693} \approx 2.76$, near the true $e = 2.718$. (Interpolating overshoots because $\ln$ curves downward.)

### Example 5 — the inequalities at $x = 2$

$1 - \frac12 = 0.5 \le \ln 2 \le 2 - 1 = 1$. Coarse, but sharp near $x = 1$: at $x = 1.1$ they give $0.0909 \le \ln 1.1 \le 0.1$, and $\ln 1.1 = 0.0953$.

## 4. Python Examples

### 4.1 Riemann sums for $\ln 2$

Left, right, midpoint and trapezoid sums with $n$ strips. Since $1/t$ is decreasing, left $>$ $\ln 2$ $>$ right. Since $1/t$ is convex, the trapezoid overestimates and the midpoint underestimates.

```python
import math

import numpy as np


def riemann_sums(a, b, strips):
    edges = np.linspace(a, b, strips + 1)
    width = (b - a) / strips
    heights = 1 / edges
    left = width * heights[:-1].sum()
    right = width * heights[1:].sum()
    midpoint = width * (1 / (edges[:-1] + width / 2)).sum()
    trapezoid = (left + right) / 2
    return left, right, midpoint, trapezoid


print(f"{'n':>5}{'left':>12}{'right':>12}{'midpoint':>12}{'trapezoid':>12}")
for strips in [4, 10, 100, 1000]:
    left, right, midpoint, trapezoid = riemann_sums(1, 2, strips)
    print(f"{strips:>5}{left:>12.6f}{right:>12.6f}{midpoint:>12.6f}{trapezoid:>12.6f}")
print(f"{'ln 2':>5}{math.log(2):>12.6f}")
```

The gap between left and right sums is exactly $\frac{1}{n}\left(1 - \frac12\right) = \frac{1}{2n}$: the rectangles differ only in the first and last strip. The midpoint and trapezoid errors shrink like $1/n^2$ — a hundred times smaller for ten times more strips.

### 4.2 Saint-Vincent's equal areas

Areas under $1/t$ over $[1, 2], [2, 4], [4, 8], \dots$ and over $[1, 3], [3, 9], [9, 27]$, computed numerically with the trapezoid rule on a fine grid.

```python
import math

import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz


def area_under_hyperbola(a, b, points=200_001):
    t = np.linspace(a, b, points)
    return np.trapezoid(1 / t, t)


for ratio in [2, 3]:
    print(f"ratio {ratio}: ln {ratio} = {math.log(ratio):.6f}")
    for k in range(4):
        a = ratio**k
        b = ratio ** (k + 1)
        print(f"   area over [{a:>2}, {b:>3}] = {area_under_hyperbola(a, b):.6f}")
```

### 4.3 The stretching lemma, numerically and symbolically

For any $a$, the area over $[a, ab]$ equals the area over $[1, b]$. SymPy confirms the general case with symbols.

```python
import numpy as np
import sympy as sp

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

b_value = 2.5
for a_value in [0.2, 1, 7, 1000]:
    t = np.linspace(a_value, a_value * b_value, 200_001)
    area = np.trapezoid(1 / t, t)
    print(f"a = {a_value:>6}: area over [a, a*b] = {area:.8f}")

t, a, b = sp.symbols("t a b", positive=True)
stretched = sp.integrate(1 / t, (t, a, a * b))
original = sp.integrate(1 / t, (t, 1, b))
print("SymPy: integral over [a, ab] =", sp.simplify(sp.expand_log(stretched, force=True)), "  integral over [1, b] =", original)
```

### 4.4 Finding the base by bisection

Search for the point $B$ where the area first equals $1$, using only the area function — no exponentials. The answer is $e$.

```python
import math

import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz


def area_from_one(x, points=400_001):
    t = np.linspace(1, x, points)
    return np.trapezoid(1 / t, t)


low, high = 2.0, 3.0
for step in range(40):
    middle = (low + high) / 2
    if area_from_one(middle) < 1:
        low = middle
    else:
        high = middle

print(f"area = 1 at x = {low:.10f}")
print(f"e             = {math.e:.10f}")
print("close:", math.isclose(low, math.e, rel_tol=1e-9))
```

### 4.5 Checking $1 - 1/x \le \ln x \le x - 1$

```python
import numpy as np

x = np.linspace(0.05, 10, 2000)
lower = 1 - 1 / x
upper = x - 1
natural_log = np.log(x)

print("lower <= ln x everywhere:", bool(np.all(lower <= natural_log + 1e-15)))
print("ln x <= upper everywhere:", bool(np.all(natural_log <= upper + 1e-15)))
for value in [0.5, 1.1, 2, 10]:
    print(f"x = {value:>4}: {1 - 1 / value:8.4f} <= {np.log(value):8.4f} <= {value - 1:8.4f}")
```

## 5. Visualizations

### 5.1 Saint-Vincent's picture: equal areas over doubling intervals

The three shaded regions over $[1, 2]$, $[2, 4]$, $[4, 8]$ have different shapes but the same area, $\ln 2$. Doubling the width halves the height.

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(0.5, 9, 600)
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(t, 1 / t, color="black", linewidth=2, label="y = 1/t")

colours = ["tab:blue", "tab:orange", "tab:green"]
for k, colour in enumerate(colours):
    left, right = 2**k, 2 ** (k + 1)
    strip = np.linspace(left, right, 200)
    ax.fill_between(strip, 1 / strip, color=colour, alpha=0.45, label=f"[{left}, {right}]: area = ln 2 = 0.693")

ax.set_xlim(0.5, 9)
ax.set_ylim(0, 1.6)
ax.set_xlabel("t")
ax.set_ylabel("1/t")
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_title("Geometric intervals, equal areas: the area under 1/t is a logarithm")
plt.show()
```

### 5.2 Stretch and squash

Left: the region under $1/t$ over $[1, 2.5]$. Right: the same region stretched horizontally by $a = 3$ and squashed vertically by $1/3$. The dots show where the corners go; the stretched region fits exactly under the same hyperbola, over $[3, 7.5]$.

```python
import matplotlib.pyplot as plt
import numpy as np

a = 3
b = 2.5
t_curve = np.linspace(0.4, 9, 600)
t_region = np.linspace(1, b, 200)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5), sharey=True)
for ax in axes:
    ax.plot(t_curve, 1 / t_curve, color="black", linewidth=1.5)
    ax.set_xlim(0, 9)
    ax.set_ylim(0, 1.3)
    ax.set_xlabel("t")
    ax.grid(True, alpha=0.3)

axes[0].fill_between(t_region, 1 / t_region, color="tab:blue", alpha=0.5)
axes[0].plot([1, b], [1, 1 / b], "o", color="tab:blue")
axes[0].set_title(f"over [1, {b}]: area = ln {b} = {np.log(b):.4f}")
axes[0].set_ylabel("1/t")

stretched_t = a * t_region
stretched_y = (1 / t_region) / a
axes[1].fill_between(stretched_t, stretched_y, color="tab:orange", alpha=0.5)
axes[1].plot([a * 1, a * b], [1 / a, 1 / (a * b)], "o", color="tab:orange")
axes[1].set_title(f"(t, y) -> ({a}t, y/{a}): over [{a}, {a * b}], same area")

fig.suptitle("Stretching by a and squashing by 1/a keeps both the hyperbola and the area")
fig.tight_layout()
plt.show()
```

### 5.3 Rectangles squeeze the area

Upper (left-endpoint) and lower (right-endpoint) rectangles for $\ln 2$ with $n = 4$ and $n = 16$ strips. The orange slivers — the difference between them — stack into a single column of total height $1 - \frac12$ and width $\frac1n$.

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(1, 2, 400)
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5), sharey=True)
for ax, strips in zip(axes, [4, 16]):
    edges = np.linspace(1, 2, strips + 1)
    width = 1 / strips
    upper = width * (1 / edges[:-1]).sum()
    lower = width * (1 / edges[1:]).sum()
    ax.bar(edges[:-1], 1 / edges[:-1], width=width, align="edge", color="tab:orange", alpha=0.5, edgecolor="tab:orange", label=f"upper sum = {upper:.4f}")
    ax.bar(edges[:-1], 1 / edges[1:], width=width, align="edge", color="tab:blue", alpha=1.0, edgecolor="white", label=f"lower sum = {lower:.4f}")
    ax.plot(t, 1 / t, color="black", linewidth=2)
    ax.set_xlim(0.95, 2.05)
    ax.set_ylim(0, 1.25)
    ax.set_xlabel("t")
    ax.set_title(f"n = {strips}: {lower:.4f} < ln 2 < {upper:.4f}")
    ax.legend(loc="upper right")
    ax.grid(True, alpha=0.3)
axes[0].set_ylabel("1/t")
fig.tight_layout()
plt.show()
```

### 5.4 The area function, on a linear and a logarithmic axis

Left: $L(x) = \int_1^x dt/t$ with the dashed line at height $1$ meeting it at $x = e$. Right: the same function against a **logarithmic** $x$-axis is a straight line — equal ratios in $x$ give equal steps in area, which is Saint-Vincent's discovery in one picture.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0.1, 10, 800)
area = np.log(x)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
for ax in axes:
    ax.plot(x, area, color="tab:purple", linewidth=2, label=r"$L(x) = \int_1^x dt/t$")
    ax.axhline(1, color="gray", linestyle="--")
    ax.axvline(math.e, color="gray", linestyle=":")
    ax.plot(math.e, 1, "o", color="black")
    ax.annotate("(e, 1)", xy=(math.e, 1), xytext=(math.e + 0.4, 0.3), arrowprops=dict(arrowstyle="->"))
    ax.set_ylabel("area L(x)")
    ax.legend(loc="lower right")

axes[0].set_xlabel("x (linear scale)")
axes[0].grid(True, alpha=0.3)
axes[0].set_title("L(x) on a linear axis")
axes[1].set_xscale("log")
axes[1].set_xlabel("x (log scale)")
axes[1].grid(True, which="both", alpha=0.3)
axes[1].set_title("L(x) on a log axis: a straight line")
fig.tight_layout()
plt.show()
```

So $M < \ln 2 < T$. For $n = 10$: $M = 0.692835 < 0.693147 < 0.693771 = T$. Simpson: $\frac{2M + T}{3} = 0.6931474$, an error of about $2 \times 10^{-7}$ instead of $3 \times 10^{-4}$.

## 8. Conclusion

- The area under the hyperbola, $\ln x = \int_1^x \frac{dt}{t}$, is the one area the power rule could not handle — and it turns out to be a logarithm.
- The reason is a symmetry: stretching by $a$ horizontally and squashing by $1/a$ vertically maps the hyperbola to itself and preserves area. Hence $\int_a^{ab} \frac{dt}{t} = \int_1^b \frac{dt}{t}$, and so $\ln(ab) = \ln a + \ln b$ (Saint-Vincent 1647, de Sarasa 1649).
- A continuous increasing function with the product law is $\log_B$ for the base $B$ where it equals $1$. Squeezing the area over $[1, 1 + \frac1n]$ between rectangles shows $B = \lim (1 + 1/n)^n = e$: the area from $1$ to $e$ is exactly $1$.
- In exponent and log form: $e^y = x \iff y = \ln x = $ area from $1$ to $x$.
- Riemann sums bracket $\ln x$ from above and below, and the rectangle picture gives $1 - \frac1x \le \ln x \le x - 1$.
