# Logarithms of Negative and Complex Numbers: $\ln(-1) = i\pi$

Every rule in this series so far came with the warning "$x > 0$". This notebook removes it. Once the exponential is allowed complex exponents, Euler's formula $e^{i\pi} = -1$ says at once what $\ln(-1)$ must be — and it also says that the answer is not unique. We meet the **multi-valued** logarithm, the **principal value** that Python returns, the **branch cut** where that principal value jumps, and the price paid: the law $\ln(zw) = \ln z + \ln w$ survives only "up to $2\pi i$".

This is notebook 01 of module 07 (**Beyond the reals**) in the **Learning Logarithms** series.

## 1. Introduction

What is $\ln(-1)$? On a calculator the question is an error. Two of the best mathematicians of the early 18th century gave it two different answers.

In letters exchanged in 1712–1713, **Johann Bernoulli** argued that $\ln(-x) = \ln x$. His reasoning was calculus: the derivative of $\ln(-x)$ is $\frac{-1}{-x} = \frac{1}{x}$, the same as the derivative of $\ln x$, so the two "must be the same function". In particular $\ln(-1) = \ln 1 = 0$.

**Gottfried Leibniz** disagreed: he held that logarithms of negative numbers are not real at all but "imaginary". An objection in his spirit runs: if $\ln(-1) = 0$, then $\ln\sqrt{-1} = \tfrac12 \ln(-1) = 0$ as well, which would make $\sqrt{-1} = e^0 = 1$ — absurd.

Neither convinced the other. The question was settled only in the 1740s by **Leonhard Euler**, using his formula

$$
e^{i\theta} = \cos\theta + i \sin\theta, \qquad\text{in particular}\qquad e^{i\pi} = -1.
$$

Read in log form, $e^{i\pi} = -1$ says $\ln(-1) = i\pi$. So Leibniz was right that the answer is imaginary. But $e^{3i\pi} = -1$ too, and $e^{-i\pi} = -1$: there are **infinitely many** candidates, $\ln(-1) = i\pi(2k + 1)$ for every integer $k$. And Bernoulli's calculus was not wrong either — it only proves that $\ln(-x)$ and $\ln x$ differ by a **constant**, and that constant is $i\pi$, not $0$.

| Exponent form | Log form |
|---|---|
| $e^{0} = 1$ | $\ln 1 = 0$ (and $2\pi i k$) |
| $e^{i\pi} = -1$ | $\ln(-1) = i\pi$ (and $i\pi + 2\pi i k$) |
| $e^{i\pi/2} = i$ | $\ln i = i\pi/2$ (and $i\pi/2 + 2\pi i k$) |
| $e^{1 + i\pi} = -e$ | $\ln(-e) = 1 + i\pi$ (and $1 + i\pi + 2\pi i k$) |

This matters in practice: every numerical library must pick **one** value, and the choice creates a line in the plane where `cmath.log` jumps by $2\pi$. Phase calculations in signal processing, complex powers such as $i^i$, and inverse trigonometric functions all inherit that jump.

This notebook is the **logarithm-centred** view. For the full complex-analysis treatment — Riemann surfaces, other branches, signed zeros in detail — see the Complex Numbers series notebooks linked below.

## 2. Theory

### 2.1 A notation warning

In this notebook $\ln z$ (lower case) means the **set** of all logarithms of $z$, and $\operatorname{Log} z$ (capital L) means the single **principal value**. For a positive real $x$ the principal value is the ordinary real logarithm, $\operatorname{Log} x = \ln x$. Python's `cmath.log` and `np.log` (on complex input) return $\operatorname{Log}$.

### 2.2 Solving $e^w = z$

Write the unknown as $w = u + iv$ and the target in polar form $z = r e^{i\theta}$ with $r = |z| > 0$. Then

$$
e^{w} = e^{u} e^{iv} = r e^{i\theta}.
$$

Two complex numbers in polar form are equal when their moduli agree and their angles agree **up to whole turns**:

$$
e^{u} = r \quad\Longrightarrow\quad u = \ln r = \ln|z| \quad(\text{an ordinary real log}),
\qquad
v = \theta + 2\pi k, \quad k \in \mathbb{Z}.
$$

So

$$
\boxed{\;\ln z = \ln|z| + i\,(\arg z) = \ln|z| + i(\theta + 2\pi k), \quad k \in \mathbb{Z}.\;}
$$

The **real part** of a logarithm is the logarithm of the size; the **imaginary part** is the angle. The logarithm splits a complex number into "how big" (on a log scale) and "which direction". The size part is single-valued; the direction part is only defined up to $2\pi$, because $e^{w}$ has period $2\pi i$. Only $z = 0$ has no logarithm at all, exactly as for real numbers.

### 2.3 The principal branch

To get a function we must pick one value. The standard choice restricts the angle to the **principal argument** $\operatorname{Arg} z \in (-\pi, \pi]$:

$$
\operatorname{Log} z = \ln|z| + i \operatorname{Arg} z, \qquad -\pi < \operatorname{Im}(\operatorname{Log} z) \le \pi.
$$

Examples: $\operatorname{Log}(-1) = i\pi$, $\operatorname{Log}(i) = i\pi/2$, $\operatorname{Log}(-i) = -i\pi/2$, $\operatorname{Log}(1 + i) = \tfrac12 \ln 2 + i\pi/4$.

### 2.4 The branch cut

Walk once around the unit circle, $z = e^{it}$ for $t$ from $0$ to $2\pi$. A **continuous** choice of logarithm is simply $it$: it grows steadily from $0$ to $2\pi i$ and ends one full turn higher than it started. The principal value cannot do this, because its imaginary part must stay in $(-\pi, \pi]$. It follows $it$ until $t = \pi$, and the moment the path crosses the **negative real axis** it jumps from $+i\pi$ down to $-i\pi$.

That ray $(-\infty, 0]$ is the **branch cut** of $\operatorname{Log}$. Approaching $-1$ from above gives values near $+i\pi$; approaching from below gives values near $-i\pi$. No choice of single-valued logarithm can be continuous on a full circle around $0$ — the cut can be moved, but not removed.

On a computer the two sides of the cut are distinguished by the **sign of zero**: `cmath.log(complex(-1, 0.0))` is $+i\pi$ but `cmath.log(complex(-1, -0.0))` is $-i\pi$.

### 2.5 The product law, up to $2\pi i$

The real law $\ln(xy) = \ln x + \ln y$ came from $e^{a}e^{b} = e^{a+b}$, which still holds for complex exponents. So as **sets**,

$$
\ln(zw) = \ln z + \ln w \qquad(\text{every sum of a value of } \ln z \text{ and a value of } \ln w \text{ is a value of } \ln(zw)).
$$

For **principal values**, however, the imaginary parts add: $\operatorname{Arg} z + \operatorname{Arg} w$ can leave the strip $(-\pi, \pi]$, and then the law is off by $2\pi i$:

$$
\operatorname{Log}(zw) = \operatorname{Log} z + \operatorname{Log} w + 2\pi i\, n, \qquad n \in \{-1, 0, 1\}.
$$

The simplest counterexample is $z = w = -1$:

$$
\operatorname{Log}(-1) + \operatorname{Log}(-1) = i\pi + i\pi = 2\pi i, \qquad\text{but}\qquad \operatorname{Log}\big((-1)(-1)\big) = \operatorname{Log} 1 = 0.
$$

This is precisely the trap behind Leibniz's argument: halving $\ln(-1)$ to get $\ln\sqrt{-1}$ silently uses $\ln(z^{1/2}) = \tfrac12 \ln z$, a rule that, like the product law, holds only up to the multi-valuedness of the logarithm (Exercise 8).

### 2.6 Complex powers, and $i^i$

For complex $z \ne 0$ and any $w$, define $z^{w} = e^{w \ln z}$. Each value of $\ln z$ gives a value of $z^w$; the principal power uses $\operatorname{Log}$. For $i^i$:

$$
\ln i = i\Big(\frac{\pi}{2} + 2\pi k\Big)
\quad\Longrightarrow\quad
i^{i} = e^{i \cdot i(\pi/2 + 2\pi k)} = e^{-\pi/2 - 2\pi k}.
$$

Every value is **real** and positive. The principal value ($k = 0$) is $e^{-\pi/2} \approx 0.2079$. An imaginary number raised to an imaginary power lands on the positive real axis — because multiplying an angle by $i$ turns it into a (real) size.

## 3. Worked Examples

### Example 1 — all logarithms of $-e$

$|-e| = e$ and $\arg(-e) = \pi + 2\pi k$. So $\ln(-e) = \ln e + i(\pi + 2\pi k) = 1 + i\pi(2k + 1)$. Principal value: $\operatorname{Log}(-e) = 1 + i\pi \approx 1 + 3.1416i$.

Check (exponent form): $e^{1 + i\pi} = e \cdot e^{i\pi} = e \cdot (-1) = -e$ ✓

### Example 2 — the principal logarithm of $-1 - i$

$|-1 - i| = \sqrt2$, so the real part is $\ln\sqrt 2 = \tfrac12 \ln 2 \approx 0.3466$. The point is in the third quadrant, so $\operatorname{Arg}(-1 - i) = -3\pi/4$ (not $+5\pi/4$, which is outside $(-\pi, \pi]$).

$$
\operatorname{Log}(-1 - i) = \tfrac12 \ln 2 - \tfrac{3\pi}{4} i \approx 0.3466 - 2.3562i.
$$

### Example 3 — a product-law failure that does not involve $-1$

Take $z = w = -1 + i$, with $\operatorname{Arg} z = 3\pi/4$. Then $zw = (-1 + i)^2 = 1 - 2i - 1 = -2i$.

- $\operatorname{Log} z + \operatorname{Log} w = 2\big(\tfrac12 \ln 2 + \tfrac{3\pi}{4} i\big) = \ln 2 + \tfrac{3\pi}{2} i \approx 0.6931 + 4.7124i$.
- $\operatorname{Log}(-2i) = \ln 2 - \tfrac{\pi}{2} i \approx 0.6931 - 1.5708i$.

The real parts agree (sizes always multiply correctly). The imaginary parts differ by $\tfrac{3\pi}{2} - (-\tfrac{\pi}{2}) = 2\pi$: the sum of the angles, $3\pi/2$, left the strip $(-\pi, \pi]$.

### Example 4 — solve $e^{z} = -2$

$z = \ln(-2) = \ln 2 + i(\pi + 2\pi k) \approx 0.6931 + i(3.1416 + 6.2832k)$. There is no real solution (as the graph of $e^x > 0$ predicts), but infinitely many complex ones, stacked vertically $2\pi$ apart.

### Example 5 — Bernoulli's constant

Bernoulli's derivative argument shows $\frac{d}{dx}\big[\ln(-x) - \ln x\big] = 0$ for $x > 0$, so the difference is a constant $C$. Evaluate at $x = 1$: $C = \ln(-1) - \ln 1 = \ln(-1) = i\pi(2k + 1)$. On the principal branch, $\operatorname{Log}(-x) = \ln x + i\pi$ for every $x > 0$. Calculus was right; the constant of integration was the whole story.

## 4. Python Examples

### 4.1 `cmath.log` returns the principal value

`math.log(-1)` raises an error; `cmath.log(-1)` answers. We check the table from the Introduction and round-trip each value through `cmath.exp`.

```python
import cmath
import math

try:
    math.log(-1)
except ValueError as error:
    print("math.log(-1) ->", error)

for z in [1, -1, 1j, -1j, -math.e, 1 + 1j, -1 - 1j]:
    principal_log = cmath.log(z)
    back = cmath.exp(principal_log)
    print(f"Log({complex(z):.4f}) = {principal_log:.4f}   exp(...) = {back:.4f}   round trip ok: {abs(back - z) < 1e-12}")
```

### 4.2 All the logarithms

The multi-valued logarithm is the principal value plus $2\pi i k$. Every one of them exponentiates back to the same $z$.

```python
import cmath
import math


def all_logs(z, k_values):
    principal = cmath.log(z)
    return [principal + 2j * math.pi * k for k in k_values]


z = -1
for k, value in zip(range(-2, 3), all_logs(z, range(-2, 3))):
    print(f"k = {k:+d}:  log(-1) = {value.imag / math.pi:+.0f} * i*pi   exp(value) = {cmath.exp(value):.3f}")
```

### 4.3 The branch cut and the sign of zero

Points just above and just below $-1$ have principal logarithms whose imaginary parts differ by almost $2\pi$, although the points themselves are $2 \times 10^{-9}$ apart. Exactly on the axis, the sign of the floating-point zero picks the side.

```python
import cmath
import math

above = complex(-1, 1e-9)
below = complex(-1, -1e-9)
print("Log(-1 + 1e-9 i) =", cmath.log(above))
print("Log(-1 - 1e-9 i) =", cmath.log(below))
print("jump in imaginary part / pi =", (cmath.log(above) - cmath.log(below)).imag / math.pi)
print()
print("Log(complex(-1,  0.0)) =", cmath.log(complex(-1, 0.0)))
print("Log(complex(-1, -0.0)) =", cmath.log(complex(-1, -0.0)))
```

### 4.4 Hunting for product-law failures

We sample pairs of random complex numbers, compare $\operatorname{Log}(zw)$ with $\operatorname{Log} z + \operatorname{Log} w$, and record the difference divided by $2\pi i$. It is always an integer in $\{-1, 0, 1\}$, and it is nonzero exactly when $\operatorname{Arg} z + \operatorname{Arg} w$ leaves $(-\pi, \pi]$.

```python
import cmath
import math

import numpy as np

rng = np.random.default_rng(seed=7)
counts = {-1: 0, 0: 0, 1: 0}
prediction_correct = True

for _ in range(10000):
    z = complex(*rng.normal(size=2))
    w = complex(*rng.normal(size=2))
    difference = cmath.log(z * w) - (cmath.log(z) + cmath.log(w))
    n = round(difference.imag / (2 * math.pi))
    assert abs(difference - 2j * math.pi * n) < 1e-9
    counts[n] += 1

    angle_sum = cmath.phase(z) + cmath.phase(w)
    predicted_n = 0
    if angle_sum > math.pi:
        predicted_n = -1
    elif angle_sum <= -math.pi:
        predicted_n = 1
    prediction_correct = prediction_correct and (predicted_n == n)

print("multiple of 2*pi*i : count", counts)
print("predicted from Arg z + Arg w in every case:", prediction_correct)

z = w = -1 + 1j
print()
print("Log(z) + Log(w) =", cmath.log(z) + cmath.log(w))
print("Log(z * w)      =", cmath.log(z * w), "  with z * w =", z * w)
```

About a quarter of random pairs break the law, split evenly between $+2\pi i$ and $-2\pi i$.

### 4.5 $i^i$ is real

Python's `**` on complex numbers uses the principal logarithm. We compare it with $e^{-\pi/2}$ and list a few other values $e^{-\pi/2 - 2\pi k}$.

```python
import cmath
import math

principal_power = 1j**1j
print("1j ** 1j          =", principal_power)
print("exp(-pi/2)        =", math.exp(-math.pi / 2))
print("exp(1j * Log(1j)) =", cmath.exp(1j * cmath.log(1j)))
print("imaginary part is zero:", abs(principal_power.imag) < 1e-15)
print()
for k in [-1, 0, 1, 2]:
    log_i = 1j * (math.pi / 2 + 2 * math.pi * k)
    value = cmath.exp(1j * log_i)
    print(f"k = {k:+d}: i^i = {value.real:.6g}")
```

### 4.6 SymPy works exactly

SymPy's `log` is the principal branch too, and it knows the exact values.

```python
import sympy as sp

print("log(-1)   =", sp.log(-1))
print("log(I)    =", sp.log(sp.I))
print("log(-1-I) =", sp.simplify(sp.expand_complex(sp.log(-1 - sp.I))))
print("I**I      =", sp.I**sp.I, "=", sp.exp(sp.I * sp.log(sp.I)))
print("exp(I*pi) =", sp.exp(sp.I * sp.pi))
```

## 5. Visualizations

### 5.1 The logarithms of $-1$, $i$ and $1 + i$ in the $w$-plane

Each nonzero $z$ (left, an Argand diagram) has a whole vertical **column** of logarithms (right), spaced $2\pi$ apart. The real coordinate is $\ln|z|$; the heights are the possible angles. The shaded band $-\pi < \operatorname{Im} w \le \pi$ is where the principal values live. (The $w$-plane is drawn with unequal axis scales so that three full columns fit.)

```python
import cmath
import math

import matplotlib.pyplot as plt
import numpy as np

points = {"-1": -1 + 0j, "i": 1j, "1+i": 1 + 1j}
colours = {"-1": "tab:red", "i": "tab:blue", "1+i": "tab:green"}

fig, (ax_z, ax_w) = plt.subplots(1, 2, figsize=(12, 5.5), gridspec_kw={"width_ratios": [1, 1]})

for label, z in points.items():
    ax_z.annotate("", xy=(z.real, z.imag), xytext=(0, 0), arrowprops=dict(arrowstyle="->", color=colours[label], linewidth=2))
    ax_z.text(z.real + 0.08, z.imag + 0.08, f"z = {label}", color=colours[label], fontsize=11)
ax_z.axhline(0, color="black", linewidth=0.8)
ax_z.axvline(0, color="black", linewidth=0.8)
ax_z.set_xlim(-1.6, 1.6)
ax_z.set_ylim(-1.6, 1.6)
ax_z.set_aspect("equal")
ax_z.set_xlabel("Re")
ax_z.set_ylabel("Im")
ax_z.set_title("z-plane (Argand diagram)")
ax_z.grid(True, alpha=0.3)

ax_w.axhspan(-math.pi, math.pi, color="gold", alpha=0.25, label="principal strip (-pi, pi]")
for label, z in points.items():
    principal = cmath.log(z)
    for k in range(-1, 2):
        value = principal + 2j * math.pi * k
        marker = "o" if k == 0 else "s"
        ax_w.plot(value.real, value.imag, marker, color=colours[label], markersize=9 if k == 0 else 7)
    ax_w.text(principal.real + 0.06, principal.imag + 0.35, f"Log({label})", color=colours[label], fontsize=10)
ax_w.axhline(0, color="black", linewidth=0.8)
ax_w.axvline(0, color="black", linewidth=0.8)
ax_w.set_yticks([k * math.pi for k in range(-3, 4)])
ax_w.set_yticklabels([f"{k}pi" if k not in (0, 1, -1) else {0: "0", 1: "pi", -1: "-pi"}[k] for k in range(-3, 4)])
ax_w.set_xlim(-1.5, 1.5)
ax_w.set_ylim(-10.5, 10.5)
ax_w.set_xlabel("Re w = ln|z|")
ax_w.set_ylabel("Im w = angle")
ax_w.set_title("w-plane (stretched vertically): all logarithms\ncircles = principal value, squares = k = +1, -1")
ax_w.legend(loc="upper left", fontsize=9)
ax_w.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### 5.2 Walking around the circle: the jump at the cut

Left: the path $z = e^{it}$, $t \in [0, 2\pi]$, with the branch cut drawn in red. Right: the imaginary part of the principal logarithm (orange) against the continuous logarithm $it$ (blue dashed). They agree until the path crosses the negative real axis at $t = \pi$; then the principal value drops by $2\pi$.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(0, 2 * np.pi, 801)
path = np.exp(1j * t)

fig, (ax_path, ax_angle) = plt.subplots(1, 2, figsize=(12, 5))

ax_path.plot(path.real, path.imag, color="tab:blue")
ax_path.annotate("", xy=(path[60].real, path[60].imag), xytext=(path[40].real, path[40].imag), arrowprops=dict(arrowstyle="->", color="tab:blue", linewidth=2))
ax_path.plot([-1.8, 0], [0, 0], color="crimson", linewidth=4, label="branch cut (-inf, 0]")
ax_path.plot(1, 0, "ko")
ax_path.text(1.05, 0.08, "start t = 0", fontsize=10)
ax_path.axhline(0, color="black", linewidth=0.8)
ax_path.axvline(0, color="black", linewidth=0.8)
ax_path.set_xlim(-1.8, 1.8)
ax_path.set_ylim(-1.5, 1.5)
ax_path.set_aspect("equal")
ax_path.set_xlabel("Re")
ax_path.set_ylabel("Im")
ax_path.set_title("z = exp(it): once around the origin")
ax_path.legend(loc="lower left", fontsize=9)

principal_angle = np.log(path).imag   # what np.log / cmath.log return
ax_angle.plot(t, t, "--", color="tab:blue", label="continuous log: Im = t")
ax_angle.plot(t[t <= np.pi], principal_angle[t <= np.pi], color="tab:orange", linewidth=2.5, label="principal Log: Im in (-pi, pi]")
ax_angle.plot(t[t > np.pi], principal_angle[t > np.pi], color="tab:orange", linewidth=2.5)
ax_angle.axvline(np.pi, color="crimson", linestyle=":", label="crossing the cut")
ax_angle.set_xticks([0, np.pi / 2, np.pi, 3 * np.pi / 2, 2 * np.pi])
ax_angle.set_xticklabels(["0", "pi/2", "pi", "3pi/2", "2pi"])
ax_angle.set_yticks([-np.pi, 0, np.pi, 2 * np.pi])
ax_angle.set_yticklabels(["-pi", "0", "pi", "2pi"])
ax_angle.set_xlabel("t")
ax_angle.set_ylabel("imaginary part of the logarithm")
ax_angle.set_title("The principal value jumps by 2pi")
ax_angle.grid(True, alpha=0.3)
ax_angle.legend(fontsize=9)

plt.tight_layout()
plt.show()
```

### 5.3 The logarithm unrolls the plane

The core picture of this series is that a logarithm turns multiplication into addition. In the complex plane, multiplying by $r e^{i\theta}$ **scales and rotates**; after taking $\operatorname{Log}$ it becomes a **shift**: $\ln r$ to the right and $\theta$ up. Circles $|z| = $ const (equal ratios $1, 2, 4, 8$) become equally spaced vertical lines, and rays from the origin become horizontal lines. The negative real axis, the cut, is sent to the top and bottom edges $\operatorname{Im} w = \pm\pi$.

```python
import matplotlib.pyplot as plt
import numpy as np

radii = [0.5, 1, 2, 4, 8]
ray_angles = np.linspace(-np.pi, np.pi, 9)[1:]   # 8 rays, last one is the negative real axis
circle_t = np.linspace(-np.pi, np.pi, 400)
ray_r = np.geomspace(0.4, 10, 200)
cmap = plt.get_cmap("viridis")

fig, (ax_z, ax_w) = plt.subplots(1, 2, figsize=(12, 5.5))

for index, r in enumerate(radii):
    colour = cmap(index / (len(radii) - 1))
    circle = r * np.exp(1j * circle_t)
    ax_z.plot(circle.real, circle.imag, color=colour)
    log_circle = np.log(circle)
    ax_w.plot(log_circle.real, log_circle.imag, color=colour, label=f"|z| = {r}")

for angle in ray_angles:
    ray = ray_r * np.exp(1j * angle)
    is_cut = np.isclose(angle, np.pi)
    style = dict(color="crimson", linewidth=2.5, zorder=5) if is_cut else dict(color="gray", linewidth=1)
    ax_z.plot(ray.real, ray.imag, **style)
    log_ray = np.log(ray)
    ax_w.plot(log_ray.real, np.full_like(log_ray.real, angle), **style)
    if is_cut:
        ax_w.plot(log_ray.real, np.full_like(log_ray.real, -np.pi), color="crimson", linewidth=2.5, linestyle="--")

ax_z.axhline(0, color="black", linewidth=0.6)
ax_z.axvline(0, color="black", linewidth=0.6)
ax_z.set_xlim(-9, 9)
ax_z.set_ylim(-9, 9)
ax_z.set_aspect("equal")
ax_z.set_xlabel("Re")
ax_z.set_ylabel("Im")
ax_z.set_title("z-plane: circles |z| = 0.5,...,8 and rays (red = cut)")

ax_w.axhline(0, color="black", linewidth=0.6)
ax_w.axvline(0, color="black", linewidth=0.6)
ax_w.set_xlim(-1.2, 2.6)
ax_w.set_ylim(-3.6, 3.6)
ax_w.set_aspect("equal")
ax_w.set_yticks([-np.pi, -np.pi / 2, 0, np.pi / 2, np.pi])
ax_w.set_yticklabels(["-pi", "-pi/2", "0", "pi/2", "pi"])
ax_w.set_xlabel("Re w = ln|z|")
ax_w.set_ylabel("Im w = Arg z")
ax_w.set_title("w = Log z: circles -> vertical lines, rays -> horizontal")
ax_w.legend(loc="center left", bbox_to_anchor=(1.02, 0.5), fontsize=9)

plt.tight_layout()
plt.show()
```

The vertical lines for $|z| = 0.5, 1, 2, 4, 8$ are equally spaced ($\ln 2 \approx 0.693$ apart): doubling the size is one fixed step right, exactly as on the real log ruler of notebook 00/01. The angle direction is new — and because angles wrap around, the strip has to be cut somewhere. The red cut appears twice on the right, once at $+\pi$ (solid, where $\operatorname{Log}$ sends it) and once at $-\pi$ (dashed, the edge it is glued to).

## 7. Solutions

1. $\operatorname{Log}(-e^2) = 2 + i\pi \approx 2 + 3.1416i$. $\operatorname{Log}(2i) = \ln 2 + i\pi/2 \approx 0.6931 + 1.5708i$. $\operatorname{Log}(1 - i) = \tfrac12 \ln 2 - i\pi/4 \approx 0.3466 - 0.7854i$.
2. $\ln i = i(\pi/2 + 2\pi k)$: $-\tfrac{7\pi}{2} i, -\tfrac{3\pi}{2} i, \tfrac{\pi}{2} i, \tfrac{5\pi}{2} i, \tfrac{9\pi}{2} i$. The principal value is $k = 0$: $\operatorname{Log} i = i\pi/2$, the only one with imaginary part in $(-\pi, \pi]$.
3. $e^{i\pi/3} = \tfrac12 + \tfrac{\sqrt3}{2}i \iff \operatorname{Log}\big(\tfrac12 + \tfrac{\sqrt3}{2}i\big) = i\pi/3$ (the modulus is $1$, so the real part is $0$). $\operatorname{Log}(-10) = \ln 10 + i\pi \iff e^{\ln 10 + i\pi} = 10 \cdot (-1) = -10$.
4. Estimate: $e^{-1.57}$ is a little below $e^{-1.5} \approx 0.22$, so about $0.21$. Python: `1j**1j` $= 0.2079 + 0i$.
5. $z = \ln 2 + i\pi(2k + 1)$. $|z| < 10$ needs $|\pi(2k + 1)| < \sqrt{100 - (\ln 2)^2} \approx 9.976$, i.e. $|2k + 1| \le 3$: $k = -2, -1, 0, 1$, giving imaginary parts $-3\pi, -\pi, \pi, 3\pi$ (all with real part $\ln 2 \approx 0.6931$).
6. Worked in Example 3: $\operatorname{Log} z + \operatorname{Log} w = \ln 2 + \tfrac{3\pi}{2}i$, $\operatorname{Log}(-2i) = \ln 2 - \tfrac{\pi}{2}i$, so $n = -1$. The law holds whenever $\operatorname{Arg} z + \operatorname{Arg} w \in (-\pi, \pi]$, e.g. $z = w = 1 + i$: $\operatorname{Arg}$ sum $= \pi/2$, and $\operatorname{Log}(2i) = \ln 2 + i\pi/2 = 2\operatorname{Log}(1 + i)$.
7. $|1/z| = 1/|z|$ gives real part $-\ln|z|$. The argument of $1/z$ is $-\operatorname{Arg} z$, which lies in $(-\pi, \pi]$ as long as $\operatorname{Arg} z \ne \pi$ (because $-\pi$ is excluded from the interval). On the negative real axis $\operatorname{Arg} z = \pi$, so $-\operatorname{Arg} z = -\pi$ is not a principal argument. At $z = -1$: $\operatorname{Log}(1/(-1)) = \operatorname{Log}(-1) = i\pi$, but $-\operatorname{Log}(-1) = -i\pi$ — off by $2\pi i$.
8. $\tfrac12 \ln(-1) = i\pi(k + \tfrac12)$, and $e^{i\pi(k + 1/2)}$ alternates between $i$ (even $k$) and $-i$ (odd $k$): exactly the two square roots of $-1$. "$\sqrt{-1} = 1$" came from choosing the value $\ln(-1) = 0$, which is not a logarithm of $-1$ at all ($e^0 = 1 \ne -1$). Leibniz's conclusion was right; Bernoulli's starting value was the error.
9. $f$ is a sawtooth: it rises with slope $1$ from $0$ to $\pi$, drops to $-\pi$, rises to $\pi$, and so on (jumps at $t = \pi, 3\pi, 5\pi$). The unwrap function below adds $2\pi$ each time the value drops by more than $\pi$ between samples.

Numerical checks:

```python
import cmath
import math

import numpy as np

print("Ex 1:", cmath.log(-math.e**2), cmath.log(2j), cmath.log(1 - 1j))
print("Ex 2:", [round((cmath.log(1j) + 2j * math.pi * k).imag / math.pi, 3) for k in range(-2, 3)], "(times pi)")
print("Ex 3:", cmath.log(complex(0.5, math.sqrt(3) / 2)), cmath.exp(math.log(10) + 1j * math.pi))
print("Ex 4:", 1j**1j, " estimate 0.21")

solutions = [complex(math.log(2), math.pi * (2 * k + 1)) for k in range(-5, 5)]
inside = [s for s in solutions if abs(s) < 10]
print("Ex 5:", [round(s.imag / math.pi) for s in inside], "(imag / pi),  all exp(z) = -2:", all(abs(cmath.exp(s) + 2) < 1e-12 for s in inside))

z = -1 + 1j
n = round((cmath.log(z * z) - 2 * cmath.log(z)).imag / (2 * math.pi))
print("Ex 6: n =", n, "  law holds for 1+i:", abs(cmath.log((1 + 1j) ** 2) - 2 * cmath.log(1 + 1j)) < 1e-12)

print("Ex 7: Log(1/(-1)) =", cmath.log(1 / -1), "  -Log(-1) =", -cmath.log(-1))
print("      Log(1/z) == -Log(z) for z = 2-3j:", abs(cmath.log(1 / (2 - 3j)) + cmath.log(2 - 3j)) < 1e-12)

print("Ex 8:", [cmath.exp(1j * math.pi * (k + 0.5)) for k in range(4)])


def unwrap(angles):
    unwrapped = [angles[0]]
    offset = 0.0
    for previous, current in zip(angles[:-1], angles[1:]):
        if current - previous < -math.pi:
            offset += 2 * math.pi
        elif current - previous > math.pi:
            offset -= 2 * math.pi
        unwrapped.append(current + offset)
    return np.array(unwrapped)


t = np.linspace(0, 6 * np.pi, 2001)
sawtooth = np.angle(np.exp(1j * t))
print("Ex 9: unwrapped equals t:", np.allclose(unwrap(sawtooth), t), "  same as np.unwrap:", np.allclose(unwrap(sawtooth), np.unwrap(sawtooth)))
```

## 8. Conclusion

- Negative and complex numbers **do** have logarithms. Solving $e^{w} = z$ gives $\ln z = \ln|z| + i\arg z$: the real part is the (ordinary) logarithm of the size, the imaginary part is the angle.
- Because $e^{w}$ has period $2\pi i$, the logarithm is **multi-valued**: $\ln(-1) = i\pi(2k + 1)$. Euler's $e^{i\pi} = -1$ settled the Leibniz–Bernoulli dispute — Leibniz was right that $\ln(-1)$ is imaginary, and Bernoulli's derivative argument only fixes $\ln(-x)$ up to the constant $i\pi$.
- The **principal value** $\operatorname{Log} z = \ln|z| + i\operatorname{Arg} z$ with $\operatorname{Arg} z \in (-\pi, \pi]$ is what `cmath.log`, `np.log` and SymPy return. It jumps by $2\pi i$ across the **branch cut** $(-\infty, 0]$; on the cut, the sign of the floating-point zero chooses the side.
- The product law holds for the full sets of values, but only **up to $2\pi i$** for principal values: $\operatorname{Log}((-1)(-1)) = 0 \ne 2\pi i = \operatorname{Log}(-1) + \operatorname{Log}(-1)$.
- Complex powers are defined through the logarithm, $z^w = e^{w \ln z}$; so $i^i = e^{-\pi/2 - 2\pi k}$ is real, with principal value $\approx 0.2079$.
- The picture: $\operatorname{Log}$ unrolls the punctured plane into a strip — circles of equal ratio become equally spaced lines, rotations become vertical shifts.
