# Logarithmic Spirals: Self-Similarity in Shells, Galaxies and Hawks

A **logarithmic spiral** is the curve whose logarithm is a straight line. In polar coordinates $r = a e^{b\theta}$, so $\ln r = \ln a + b\theta$: every turn multiplies the distance from the centre by the same factor. That one property makes the curve **equiangular** (it cuts every ray from the centre at the same angle) and **self-similar** (scaling it is the same as rotating it), and it gives it a finite length even though it winds around its centre infinitely often.

This is notebook 03 of module 07 (**Beyond the reals**) in the **Learning Logarithms** series, and the last notebook of the series.

## 1. Introduction

A falcon high above a pigeon wants to dive at it. Its sharpest vision is not straight ahead but off to the side, so to see the prey clearly it would like to keep the prey at a **fixed angle** $\alpha$ away from its own direction of flight — and to fly fast it would like to keep its head straight. Suppose it does exactly that: at every moment the line of sight to the prey makes the angle $\alpha$ with the flight direction. **What path does it fly?** How long is that path, and how many times does it circle the prey?

This is hard to answer by drawing. Each moment's direction depends on where the bird is; the curve has to be "solved for". The answer is a logarithmic spiral centred on the prey, and once logarithms are involved the questions have one-line answers: the path is $r_0 / \cos\alpha$ long, and in principle it winds around infinitely often while still having that finite length.

The same curve turned up long before falcons were measured. Descartes studied it in 1638 as the curve that meets all rays from a point at a constant angle; Torricelli found its arc length. **Jacob Bernoulli** was so taken with its self-reproducing properties that he called it the *spira mirabilis*, "the marvellous spiral", and asked for it to be carved on his tombstone with the motto *Eadem mutata resurgo* — "changed, I rise again the same". (The stonemason in Basel carved an Archimedean spiral instead, with equally spaced turns — the wrong curve.)

| Exponent form | Log form |
|---|---|
| $r = a e^{b\theta}$ | $\ln r = \ln a + b\theta$ (a straight line in the $(\theta, \ln r)$ plane) |
| one turn: $r \mapsto e^{2\pi b}\, r$ | one turn: $\ln r \mapsto \ln r + 2\pi b$ |
| complex form: $z = a e^{(b + i)\theta}$ | $\operatorname{Log} z = \ln a + (b + i)\theta$ (a straight line in the $w$-plane, up to $2\pi i$) |

The last row ties this notebook to the rest of the module: a logarithmic spiral is the **exponential of a straight line** in the complex plane (notebook 01), and it is the path traced by $e^{At}\mathbf{x}_0$ for a rotation-plus-decay generator (notebook 02, section 5.2).

## 2. Theory

### 2.1 The curve

In polar coordinates $(r, \theta)$ a logarithmic spiral is

$$
r(\theta) = a\, e^{b\theta}, \qquad a > 0,\ b \neq 0,\ \theta \in \mathbb{R}.
$$

For $b > 0$ it grows counter-clockwise; for $b < 0$ it shrinks counter-clockwise. As $\theta \to -\infty$ (for $b > 0$) the radius tends to $0$ but never reaches it: the spiral winds around the centre (the **pole**) infinitely many times.

Taking logarithms,

$$
\ln r = \ln a + b\,\theta.
$$

Equal **steps** in angle give equal **ratios** in radius. Compare the Archimedean spiral $r = c\,\theta$, where equal steps in angle give equal **differences** — Bernoulli's tombstone shows the additive spiral where he wanted the multiplicative one.

The **growth factor per turn** is $W = e^{2\pi b}$, equivalently $b = \ln W / (2\pi)$.

### 2.2 The equiangular property

Let $\psi$ be the angle between the tangent and the outward radius. In polar coordinates, a short step $d\theta$ moves the point $r\,d\theta$ sideways and $dr$ outwards, so

$$
\tan\psi = \frac{r\, d\theta}{dr} = \frac{r}{dr/d\theta} = \frac{a e^{b\theta}}{b\, a e^{b\theta}} = \frac{1}{b}, \qquad\text{i.e.}\qquad \cot\psi = b.
$$

The $\theta$-dependence cancels: **the angle is the same everywhere**. The converse also holds — if $r/(dr/d\theta)$ is a constant $1/b$, then $dr/d\theta = br$, whose solutions are exactly $r = a e^{b\theta}$. So the logarithmic spiral is **the** equiangular curve (the circle, $\psi = 90^\circ$, $b = 0$, is the limiting case). Astronomers describe spiral arms by the **pitch angle** $90^\circ - \psi$, the angle between the curve and a circle, so $b = \tan(\text{pitch})$.

### 2.3 Self-similarity: scaling is rotation

Multiply the whole spiral by $\lambda > 0$:

$$
\lambda\, r(\theta) = \lambda a e^{b\theta} = a e^{b(\theta + \ln\lambda / b)} = r\Big(\theta + \frac{\ln\lambda}{b}\Big).
$$

The scaled spiral is the original spiral **rotated** by $-\ln\lambda / b$. Enlarging a logarithmic spiral never produces a new shape — it reproduces the same spiral, turned. In particular, scaling by the per-turn factor $W = e^{2\pi b}$ is a rotation by a full turn: the spiral maps exactly onto itself. This is Bernoulli's *eadem mutata resurgo*, and it is the logarithm at work: on the $(\theta, \ln r)$ line, a scaling is a vertical shift and a rotation is a horizontal shift, and for a straight line these are the same thing.

### 2.4 Finite length, infinitely many turns

The arc length element in polar coordinates is $ds = \sqrt{r^2 + (dr/d\theta)^2}\, d\theta = r\sqrt{1 + b^2}\, d\theta$. From the pole ($\theta = -\infty$) to the point at angle $\theta$, with $b > 0$:

$$
s(\theta) = \int_{-\infty}^{\theta} a e^{bt}\sqrt{1 + b^2}\, dt = \frac{\sqrt{1 + b^2}}{b}\, a e^{b\theta} = \frac{\sqrt{1 + b^2}}{b}\, r(\theta) = \frac{r}{\cos\psi}.
$$

(The last step uses $\cot\psi = b$, so $\cos\psi = b / \sqrt{1 + b^2}$.) Infinitely many turns, finite total length: the turns shrink geometrically, so their lengths form a convergent geometric series. Torricelli obtained this result in the 1640s; it was one of the first curved lengths found exactly.

### 2.5 The complex picture

Write the point as a complex number $z = r e^{i\theta}$. On the spiral,

$$
z(\theta) = a e^{b\theta} e^{i\theta} = a\, e^{(b + i)\theta}.
$$

So the spiral is $e^{w}$ for $w$ running along the straight line $w = \ln a + (b + i)\theta$. Reading this backwards with notebook 01: $\operatorname{Log}$ maps the spiral onto pieces of a straight line of slope $1/b$ in the $w$-plane, chopped into the strip $-\pi < \operatorname{Im} w \le \pi$. The branch cut breaks the line into parallel segments, one per turn — unwrapping the angle (exercise 9 of notebook 01) reassembles the full line. Multiplying by a complex constant $c = \lambda e^{i\phi}$ (scale and rotate) shifts the line, which gives a one-line proof of 2.3.

### 2.6 The golden spiral

The **golden spiral** grows by the golden ratio $\varphi = \frac{1 + \sqrt5}{2} \approx 1.618$ every **quarter turn**:

$$
e^{b\pi/2} = \varphi \quad\Longrightarrow\quad b = \frac{\ln\varphi}{\pi/2} \approx 0.3063, \qquad \psi = \operatorname{arccot} b \approx 72.97^\circ, \qquad W = \varphi^4 \approx 6.85 \text{ per turn}.
$$

The familiar picture of quarter circles drawn in Fibonacci squares ($1, 1, 2, 3, 5, 8, \dots$) is only an **approximation** of it: each arc is a circle (constant curvature), and the ratio of consecutive radii, $F_{n+1}/F_n$, approaches $\varphi$ without equalling it.

### 2.7 Spirals in nature (with care)

- **Nautilus shells** grow at their opening without remodelling the older shell, so the shell keeps its shape as it grows — the self-similarity of 2.3 — and is well approximated by a logarithmic spiral. But it is **not** the golden spiral, despite the popular claim. Measured shells expand by a factor of roughly $3$ per turn (values vary between shells and measurement methods), i.e. $b \approx \ln 3 / (2\pi) \approx 0.175$ and $\psi \approx 80^\circ$ — far from the golden spiral's $\varphi^4 \approx 6.85$ per turn.
- **Spiral galaxies** have arms that are often fitted reasonably well by logarithmic spirals, with pitch angles that differ from galaxy to galaxy. The fit is an empirical description, not an exact law.
- **Diving falcons.** Vance Tucker and colleagues (2000) proposed that a peregrine falcon, whose sharpest vision looks sideways at roughly $40^\circ$ from straight ahead, keeps its head straight for aerodynamic reasons and therefore holds the prey at a constant angle to its flight direction. A constant angle between the velocity and the line to a (stationary) target is exactly the equiangular property, so the path is a logarithmic spiral centred on the prey. The hypothesis is supported by some flight observations, but it is a hypothesis about a behaviour, not a law, and real prey moves.

For the pursuit: if the velocity makes angle $\alpha$ with the direction **to** the target, then per unit path length the distance shrinks at rate $\cos\alpha$ and the bearing turns at rate $\sin\alpha / r$, so

$$
\frac{dr}{d\theta} = -r\cot\alpha \quad\Longrightarrow\quad r = r_0\, e^{-\cot\alpha\,(\theta - \theta_0)},
$$

a logarithmic spiral with $b = -\cot\alpha$, and total path length (from 2.4) $r_0 / \cos\alpha$.

## 3. Worked Examples

### Example 1 — reading off a spiral

$r = 2e^{0.1\theta}$. Log form: $\ln r = \ln 2 + 0.1\theta$, a line with intercept $\ln 2 \approx 0.693$ and slope $0.1$. Equiangular angle: $\psi = \operatorname{arccot} 0.1 = \arctan 10 \approx 84.29^\circ$ (nearly a circle). Growth per turn: $W = e^{0.2\pi} \approx 1.874$.

### Example 2 — which rotation equals doubling?

For $b = 0.2$, doubling every radius has the same effect as rotating the spiral **clockwise** by $\ln 2 / 0.2 \approx 3.466$ rad $\approx 198.6^\circ$ (by 2.3, the rotation is $-\ln\lambda / b$). Check: $e^{0.2 \times 3.466} = e^{0.693} = 2$ ✓.

### Example 3 — the length of the golden spiral from its centre to $r = 1$

$b \approx 0.3063$: $s = \sqrt{1 + 0.3063^2} / 0.3063 \approx 1.0459 / 0.3063 \approx 3.414$. Infinitely many turns inside $r = 1$, total length about $3.4$ — about half the circumference of the unit circle ($2\pi \approx 6.28$).

### Example 4 — the falcon's path

Take $\alpha = 40^\circ$ and a start $r_0 = 100$ m from the prey. Then $\cot 40^\circ \approx 1.192$ and $\cos 40^\circ \approx 0.766$:

- path length $= 100 / 0.766 \approx 130.5$ m, about $30\%$ longer than the straight dive;
- angle swept around the prey while closing from $100$ m to $1$ m: $\Delta\theta = \ln(100) / \cot 40^\circ \approx 4.605 / 1.192 \approx 3.86$ rad $\approx 221^\circ$.

The infinitely many turns all happen in the last few centimetres; what an observer sees is a single sweeping curve.

### Example 5 — golden spiral versus nautilus

| | per quarter turn | per turn $W$ | $b$ | $\psi$ |
|---|---|---|---|---|
| golden spiral | $\varphi \approx 1.618$ | $\varphi^4 \approx 6.85$ | $0.3063$ | $\approx 73.0^\circ$ |
| typical nautilus (approx.) | $3^{1/4} \approx 1.316$ | $\approx 3$ | $\approx 0.175$ | $\approx 80.1^\circ$ |

## 4. Python Examples

### 4.1 Building the spiral and checking the constant angle

We sample $r = ae^{b\theta}$, convert to $(x, y)$, estimate the tangent direction numerically with `np.gradient`, and measure the angle between tangent and radius at every sample.

```python
import math

import numpy as np

a = 1.0
b = 0.2
theta = np.linspace(-4 * np.pi, 2 * np.pi, 20001)
r = a * np.exp(b * theta)
x = r * np.cos(theta)
y = r * np.sin(theta)

dx = np.gradient(x, theta)
dy = np.gradient(y, theta)
cos_psi = (x * dx + y * dy) / (np.hypot(x, y) * np.hypot(dx, dy))
psi_degrees = np.degrees(np.arccos(cos_psi[1:-1]))   # skip the one-sided end differences

print(f"predicted psi = arccot(b) = {math.degrees(math.atan(1 / b)):.4f} deg")
print(f"measured psi: min {psi_degrees.min():.4f}, max {psi_degrees.max():.4f} deg")
print(f"growth factor per turn W = e^(2 pi b) = {math.exp(2 * math.pi * b):.4f}")
```

### 4.2 The log form is a straight line

Fitting a line to $(\theta, \ln r)$ recovers $\ln a$ and $b$ exactly — the practical way to measure a spiral from a photograph (mark points, unwrap the angle, fit a line).

```python
import math

import numpy as np

a_true, b_true = 2.0, 0.1
theta = np.linspace(0, 6 * np.pi, 50)
r = a_true * np.exp(b_true * theta)

slope, intercept = np.polyfit(theta, np.log(r), 1)
print(f"fitted slope b     = {slope:.6f}   (true {b_true})")
print(f"fitted intercept   = {intercept:.6f}   (ln a = {math.log(a_true):.6f})")
print(f"psi = {math.degrees(math.atan(1 / slope)):.2f} deg, W per turn = {math.exp(2 * math.pi * slope):.4f}")
```

### 4.3 Scaling equals rotation

Scale the spiral by $\lambda = 2$, and separately rotate it by $-\ln 2 / b$. Both produce the same set of points. Using the complex form $z = ae^{(b+i)\theta}$, scaling is multiplication by $2$ and rotation is multiplication by $e^{-i\ln 2 / b}$.

```python
import math

import numpy as np

a, b = 1.0, 0.2
theta = np.linspace(-20, 20, 4001)
spiral = a * np.exp((b + 1j) * theta)

scale = 2.0
shift = math.log(scale) / b
scaled = scale * spiral                                                      # every point doubled
rotated = a * np.exp((b + 1j) * (theta + shift)) * np.exp(-1j * shift)       # spiral points further out, turned back by shift

print(f"doubling = rotating by -ln(2)/b = {-shift:.4f} rad = {math.degrees(-shift):.1f} deg")
print("scale * z(theta) == exp(-i shift) * z(theta + shift):", np.allclose(scaled, rotated))
print(f"scaling by W = e^(2 pi b) = {math.exp(2 * math.pi * b):.4f} is a rotation by a full turn: the spiral maps onto itself")
```

### 4.4 Finite arc length

We integrate $ds = r\sqrt{1 + b^2}\, d\theta$ from far inside (where the remaining length is negligible) out to $r = 1$, and compare with $r\sqrt{1 + b^2} / b$ for the golden spiral. The numpy shim keeps `np.trapezoid` working on numpy 1.x and 2.x.

```python
import math

import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

golden_ratio = (1 + math.sqrt(5)) / 2
b = math.log(golden_ratio) / (math.pi / 2)

theta = np.linspace(-200, 0, 2_000_001)   # r(-200) = e^(-61), effectively the centre
r = np.exp(b * theta)
numerical_length = np.trapezoid(r * math.sqrt(1 + b**2), theta)
formula_length = math.sqrt(1 + b**2) / b

print(f"golden spiral: b = {b:.4f}, psi = {math.degrees(math.atan(1 / b)):.2f} deg, W = {math.exp(2 * math.pi * b):.3f}")
print(f"turns between r = e^-61 and r = 1: {200 / (2 * math.pi):.1f}")
print(f"length (numerical) = {numerical_length:.6f}")
print(f"length (formula)   = {formula_length:.6f}")
```

### 4.5 Golden spiral, Fibonacci squares and the nautilus

The ratio of consecutive Fibonacci numbers approaches $\varphi$, so the Fibonacci quarter-circle spiral approaches the golden spiral's growth rate. The nautilus figure of about $3$ per turn sits nowhere near.

```python
import math

golden_ratio = (1 + math.sqrt(5)) / 2
fibonacci = [1, 1]
while len(fibonacci) < 12:
    fibonacci.append(fibonacci[-1] + fibonacci[-2])
ratios = [fibonacci[n + 1] / fibonacci[n] for n in range(len(fibonacci) - 1)]
print("F(n+1)/F(n):", [round(q, 4) for q in ratios])
print("phi        :", round(golden_ratio, 4))
print()

for name, growth_per_turn in [("golden spiral", golden_ratio**4), ("nautilus (approx.)", 3.0)]:
    b = math.log(growth_per_turn) / (2 * math.pi)
    psi = math.degrees(math.atan(1 / b))
    quarter = growth_per_turn**0.25
    print(f"{name:20s}: per quarter turn {quarter:.3f}, per turn {growth_per_turn:.3f}, b = {b:.4f}, psi = {psi:.2f} deg")
```

### 4.6 Simulating the falcon

Step the falcon forward at unit speed, always flying at angle $\alpha$ to the line of sight to the prey at the origin. Compare its distance with the formula $r_0 e^{-\cot\alpha\,\Delta\theta}$ and its path length with $r_0 / \cos\alpha$.

```python
import math

import numpy as np

alpha = math.radians(40)
r0 = 100.0
position = np.array([r0, 0.0])
step = 0.001
path_length = 0.0
unwrapped_angle = 0.0
previous_angle = 0.0
max_formula_error = 0.0

while np.linalg.norm(position) > 1.0:
    to_prey = -position / np.linalg.norm(position)
    # rotate the line of sight clockwise by alpha: the falcon then circles the prey counter-clockwise
    direction = np.array([
        math.cos(alpha) * to_prey[0] + math.sin(alpha) * to_prey[1],
        -math.sin(alpha) * to_prey[0] + math.cos(alpha) * to_prey[1],
    ])
    position = position + step * direction
    path_length += step

    angle = math.atan2(position[1], position[0])
    change = angle - previous_angle
    if change < -math.pi:
        change += 2 * math.pi
    elif change > math.pi:
        change -= 2 * math.pi
    unwrapped_angle += change
    previous_angle = angle

    formula_r = r0 * math.exp(-unwrapped_angle / math.tan(alpha))
    max_formula_error = max(max_formula_error, abs(np.linalg.norm(position) - formula_r))

print(f"path length to r = 1   : {path_length:.2f} m   formula (r0 - 1)/cos(alpha) = {(r0 - 1) / math.cos(alpha):.2f} m")
print(f"angle swept            : {math.degrees(unwrapped_angle):.1f} deg   formula ln(100)/cot(alpha) = {math.degrees(math.log(r0) * math.tan(alpha)):.1f} deg")
print(f"max |r - r0 e^(-theta cot alpha)| along the way: {max_formula_error:.4f} m")
print(f"full dive to r = 0 would be r0/cos(alpha) = {r0 / math.cos(alpha):.2f} m (vs {r0:.0f} m straight)")
```

## 5. Visualizations

### 5.1 The spiral and its logarithm

Left: $r = e^{0.2\theta}$ with several radii and the tangent at each crossing — every crossing makes the same angle $\psi \approx 78.7^\circ$. Right: the same curve as $\ln r$ against $\theta$ is a straight line of slope $b$; an Archimedean spiral $r = c\theta$ (Bernoulli's tombstone) is not. This is the core picture of the series: the logarithm turns a multiplicative pattern (each turn $\times W$) into an additive one (each turn $+2\pi b$).

```python
import math

import matplotlib.pyplot as plt
import numpy as np

b = 0.2
theta = np.linspace(-3 * np.pi, 2.2 * np.pi, 3000)
r = np.exp(b * theta)

fig, (ax_spiral, ax_log) = plt.subplots(1, 2, figsize=(12, 5.5))
ax_spiral.plot(r * np.cos(theta), r * np.sin(theta), color="tab:blue", linewidth=2)
for ray_angle in np.arange(0, 2 * np.pi, np.pi / 3):
    ax_spiral.plot([0, 4.5 * np.cos(ray_angle)], [0, 4.5 * np.sin(ray_angle)], color="gray", linewidth=0.8)
    for turn in [0, 2 * np.pi]:
        t0 = ray_angle + turn - 2 * np.pi
        r0 = math.exp(b * t0)
        point = np.array([r0 * math.cos(t0), r0 * math.sin(t0)])
        tangent = np.array([b * math.cos(t0) - math.sin(t0), b * math.sin(t0) + math.cos(t0)])
        tangent = tangent / np.linalg.norm(tangent) * 0.35 * r0
        ax_spiral.plot([point[0] - tangent[0], point[0] + tangent[0]], [point[1] - tangent[1], point[1] + tangent[1]], color="crimson", linewidth=1.5)
ax_spiral.plot(0, 0, "k+", markersize=10)
ax_spiral.set_aspect("equal")
ax_spiral.set_xlim(-4.2, 4.2)
ax_spiral.set_ylim(-4.2, 4.2)
ax_spiral.set_xlabel("x")
ax_spiral.set_ylabel("y")
ax_spiral.set_title(f"r = exp({b} theta): tangents (red) all meet the rays at {math.degrees(math.atan(1 / b)):.1f} deg")
ax_spiral.grid(True, alpha=0.3)

theta_positive = np.linspace(0.3, 4 * np.pi, 400)
ax_log.plot(theta_positive, b * theta_positive, color="tab:blue", linewidth=2, label="log spiral r = exp(0.2 theta): straight")
ax_log.plot(theta_positive, np.log(0.25 * theta_positive), color="tab:orange", linewidth=2, label="Archimedean r = 0.25 theta: curved")
ax_log.set_xlabel("theta (radians, linear scale)")
ax_log.set_ylabel("ln r (natural log of the radius)")
ax_log.set_title("Log form: ln r = ln a + b theta")
ax_log.grid(True, alpha=0.3)
ax_log.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

### 5.2 Self-similarity: zooming in reproduces the spiral

The same spiral drawn in four windows, each $W = e^{2\pi b}$ times smaller than the last. Because scaling by $W$ is a full turn, every panel shows **the same picture**. No circle, parabola or Archimedean spiral has this property.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

b = 0.2
growth_per_turn = math.exp(2 * math.pi * b)
theta = np.linspace(-12 * np.pi, 2 * np.pi, 20000)
r = np.exp(b * theta)
x = r * np.cos(theta)
y = r * np.sin(theta)

fig, axes = plt.subplots(1, 4, figsize=(14, 3.8))
for level, ax in enumerate(axes):
    half_width = 4.0 / growth_per_turn**level
    ax.plot(x, y, color="tab:blue", linewidth=1.5)
    ax.set_xlim(-half_width, half_width)
    ax.set_ylim(-half_width, half_width)
    ax.set_aspect("equal")
    ax.set_xticks([-half_width, 0, half_width])
    ax.set_yticks([-half_width, 0, half_width])
    ax.tick_params(labelsize=8)
    ax.set_title(f"zoom x{growth_per_turn**level:.1f}", fontsize=10)
    ax.grid(True, alpha=0.3)
plt.suptitle(f"r = exp(0.2 theta), each window {growth_per_turn:.2f} times smaller: identical pictures")
plt.tight_layout()
plt.show()
```

### 5.3 Golden spiral, Fibonacci arcs and a nautilus-like spiral

The Fibonacci quarter-circle construction (black arcs in their squares) hugs the golden spiral (gold). A spiral with the nautilus-like growth of about $3$ per turn (teal), started at the same point, winds noticeably more tightly: the golden spiral reaches the outer squares in fewer turns.

```python
import math

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Arc, Rectangle

golden_ratio = (1 + math.sqrt(5)) / 2
fibonacci = [1, 1, 2, 3, 5, 8, 13, 21]


def rotate_90(vector):
    return np.array([-vector[1], vector[0]])


fig, ax = plt.subplots(figsize=(9, 6.5))

# Each Fibonacci arc is a counter-clockwise quarter circle of radius F_n; its centre lies to the left of the heading.
point = np.array([0.0, 0.0])
heading = np.array([1.0, 0.0])
end_points = []
all_corners = []
for radius in fibonacci:
    centre = point + radius * rotate_90(heading)
    start_angle = math.degrees(math.atan2(*(point - centre)[::-1]))
    ax.add_patch(Arc(centre, 2 * radius, 2 * radius, theta1=start_angle, theta2=start_angle + 90, color="black", linewidth=1.5))
    end = centre + radius * heading
    corner = point + end - centre
    square = np.array([centre, point, corner, end, centre])
    ax.plot(square[:, 0], square[:, 1], color="gray", linewidth=0.8)
    all_corners.extend(square)
    end_points.append(end)
    point = end
    heading = rotate_90(heading)

# Pole of the golden spiral through the last two arc end points: E_n - P = phi * rot90(E_(n-1) - P)
E_last, E_before = end_points[-1], end_points[-2]
rotation = np.array([[0.0, -1.0], [1.0, 0.0]])
pole = np.linalg.solve(np.eye(2) - golden_ratio * rotation, E_last - golden_ratio * rotation @ E_before)

relative = E_last - pole
r_end = np.hypot(*relative)
angle_end = math.atan2(relative[1], relative[0])
for name, growth_per_turn, colour in [("golden spiral (x6.85 per turn)", golden_ratio**4, "goldenrod"), ("nautilus-like (x3 per turn)", 3.0, "teal")]:
    b = math.log(growth_per_turn) / (2 * math.pi)
    theta = np.linspace(angle_end - 8 * np.pi, angle_end, 4000)
    r = r_end * np.exp(b * (theta - angle_end))
    ax.plot(pole[0] + r * np.cos(theta), pole[1] + r * np.sin(theta), color=colour, linewidth=2.2, label=name)

ax.plot([], [], color="black", label="Fibonacci quarter circles")
corners = np.array(all_corners)
ax.set_aspect("equal")
ax.set_xlim(corners[:, 0].min() - 1, corners[:, 0].max() + 1)
ax.set_ylim(corners[:, 1].min() - 1, corners[:, 1].max() + 1)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Golden spiral vs Fibonacci arcs vs a nautilus-like spiral")
ax.legend(loc="upper left", bbox_to_anchor=(1.02, 1.0), fontsize=9)
plt.tight_layout()
plt.show()
```

### 5.4 The falcon's dive

Constant-angle approach paths to a target at the origin for several angles $\alpha$, all starting $100$ m away. $\alpha = 0$ is the straight dive; larger $\alpha$ curves more and is longer, $r_0 / \cos\alpha$. With $\alpha = 40^\circ$ (red, the angle discussed for peregrines) the bird sweeps about $221^\circ$ around the prey before it is within $1$ m.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

r0 = 100.0
fig, ax = plt.subplots(figsize=(8, 7))
for alpha_degrees, colour in [(0, "black"), (20, "tab:blue"), (40, "crimson"), (60, "tab:green"), (75, "tab:purple")]:
    alpha = math.radians(alpha_degrees)
    if alpha_degrees == 0:
        ax.plot([r0, 0], [0, 0], color=colour, linewidth=2, label="alpha = 0 deg (straight), 100 m")
        continue
    theta = np.linspace(0, 12 * np.pi, 20000)
    r = r0 * np.exp(-theta / math.tan(alpha))
    length = r0 / math.cos(alpha)
    ax.plot(r * np.cos(theta), r * np.sin(theta), color=colour, linewidth=2, label=f"alpha = {alpha_degrees} deg, length {length:.0f} m")

ax.plot(0, 0, "k*", markersize=14)
ax.text(3, -8, "prey", fontsize=11)
ax.plot(r0, 0, "ko")
ax.text(r0 - 6, -8, "start", fontsize=11)
ax.set_aspect("equal")
ax.set_xlim(-60, 110)
ax.set_ylim(-60, 90)
ax.set_xlabel("x (m)")
ax.set_ylabel("y (m)")
ax.set_title("Constant-angle pursuit paths are logarithmic spirals")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left", fontsize=9)
plt.show()
```

## 7. Solutions

1. $\psi = \operatorname{arccot} 0.25 = \arctan 4 \approx 75.96^\circ$; pitch $\approx 14.04^\circ$; $W = e^{0.5\pi} \approx 4.810$.
2. $\ln r = \ln 3 - 0.1\theta$: slope $-0.1$, intercept $\ln 3 \approx 1.0986$. It shrinks as $\theta$ increases (counter-clockwise it winds inward).
3. Scaling by $\lambda$ equals rotating by $-\ln\lambda / b = -\ln(0.5)/0.2 = \ln 2 / 0.2 \approx 3.466$ rad $\approx 198.6^\circ$, counter-clockwise (the opposite sense to Example 2).
4. Estimate: $3^6 = 729$ and $3^7 = 2187$, so a bit over $6$ turns. Exactly $\log_3 1000 = \ln 1000 / \ln 3 \approx 6.29$ turns.
5. $b = \cot 80^\circ \approx 0.1763$, $W = e^{2\pi \times 0.1763} \approx 3.03$ — essentially the nautilus figure. A nautilus is close to an $80^\circ$ equiangular spiral; the golden spiral is a $73^\circ$ one.
6. $s = \sqrt{1 + b^2}/b$: $b = 0.1$: $10.05$; golden: $3.414$; $b = 1$: $\sqrt2 \approx 1.414$. As $b \to 0$ the spiral approaches a circle, the turns shrink more and more slowly, and the geometric series of turn lengths converges more and more slowly ($s \approx 1/b$).
7. $1/z = a^{-1}e^{-(b + i)\theta}$. Substituting $\theta' = -\theta$: $1/z = a^{-1}e^{(b + i)\theta'}$ — **the same** $b$, with $a \to 1/a$ and the direction of travel reversed. In log form, $w \mapsto -w$ reflects the line through the origin; a line of slope $1/b$ stays a line of slope $1/b$. (The point reflection $w \mapsto -w$ preserves slopes.)
8. Path length $= r_0 / \cos 30^\circ = 200 / 0.8660 \approx 230.9$ m. Turning angle: $\Delta\theta = \ln(200/2) / \cot 30^\circ = 4.605 / 1.732 \approx 2.659$ rad $\approx 152.3^\circ$.
9. In triangle $OPT$ the angle at $O$ is $90^\circ$, and the (acute) angle at $P$ between the line $PO$ and the tangent is $\psi$ (for $b > 0$, $\psi < 90^\circ$). So $|PT| = |OP| / \cos\psi = r/\cos\psi$, which is the arc length from 2.4. Checked numerically below.

## 8. Conclusion

- A logarithmic spiral $r = ae^{b\theta}$ is the curve whose **log form is a straight line**: $\ln r = \ln a + b\theta$. Each turn multiplies the radius by $W = e^{2\pi b}$.
- It is **equiangular**: the tangent meets every radius at the angle $\psi$ with $\cot\psi = b$, and it is the only curve with that property.
- It is **self-similar**: scaling by $\lambda$ is the same as rotating by $\ln\lambda / b$ — Bernoulli's *eadem mutata resurgo*.
- It winds around its centre infinitely often but has **finite length** $r\sqrt{1 + b^2}/b = r/\cos\psi$ from the pole (Torricelli).
- In complex form it is $e^{(b + i)\theta}$, the exponential of a straight line — the picture that connects it to the complex logarithm and to $e^{At}$ for a rotation-plus-decay matrix.
- The **golden spiral** is the special case $b = \ln\varphi / (\pi/2)$, growing by $\varphi$ per quarter turn. Nautilus shells are close to logarithmic spirals but grow by roughly $3$ per turn, **not** $\varphi^4 \approx 6.85$: the nautilus is not a golden spiral. Galaxy arms are fitted roughly by logarithmic spirals, and a constant-angle approach — proposed by Tucker for diving peregrine falcons — traces one exactly.
