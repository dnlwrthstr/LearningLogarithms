# Change of Base; Common, Natural and Binary Logarithms

There are infinitely many logarithms — one for every base $b > 0$, $b \ne 1$ — but a calculator has only two buttons, `ln` and `log`. This notebook shows why two buttons (in fact, one) are enough: every logarithm is a **constant multiple** of every other. Then it explains why three bases won: $10$ for decimal arithmetic, $e$ for calculus, and $2$ for computers.

This is notebook 04 of module 00 (**Foundations**) in the **Learning Logarithms** series.

## 1. Introduction

Two everyday questions, and a calculator with only `ln` and `log` (base 10):

1. **Addresses.** A file is 5,000,000 bytes long. How many bits does a byte offset into it need? We need the smallest $n$ with $2^n \ge 5{,}000{,}000$, i.e. $\log_2 5{,}000{,}000$.
2. **Doubling time.** An investment returns $7\%$ per year. How many years until it doubles? We need $1.07^n = 2$, i.e. $n = \log_{1.07} 2$.

Neither base is on the calculator. Notebook 03 already contained the trick: take $\log_{10}$ of both sides of $1.07^n = 2$ and use the power law,

$$
n \log_{10} 1.07 = \log_{10} 2 \quad\Longrightarrow\quad n = \frac{\log_{10} 2}{\log_{10} 1.07} \approx \frac{0.30103}{0.029384} \approx 10.24 \text{ years}.
$$

In general, **any** logarithm is a ratio of two logarithms in a base you do have:

$$
\log_b x = \frac{\log_a x}{\log_a b} = \frac{\ln x}{\ln b} = \frac{\log_{10} x}{\log_{10} b}.
$$

For the file: $\log_2 5{,}000{,}000 = \dfrac{\log_{10} 5{,}000{,}000}{\log_{10} 2} \approx \dfrac{6.69897}{0.30103} \approx 22.25$, so $2^{22} < 5{,}000{,}000 < 2^{23}$ and $23$ bits are needed.

## 2. Theory

### 2.1 The change-of-base formula

Let $b, a > 0$, both $\ne 1$, and $x > 0$. Put $y = \log_b x$, so $b^y = x$. Apply $\log_a$ to both sides and use the power law:

$$
\log_a(b^y) = \log_a x \quad\Longrightarrow\quad y \log_a b = \log_a x \quad\Longrightarrow\quad \boxed{\;\log_b x = \frac{\log_a x}{\log_a b}\;}
$$

($\log_a b \ne 0$ because $b \ne 1$.) In exponent form the same fact reads $x = b^y = \left(a^{\log_a b}\right)^y = a^{y \log_a b}$: **any power of $b$ is a power of $a$** with the exponent rescaled by the fixed factor $\log_a b$.

### 2.2 All logarithms are the same function, rescaled

Fix the "calculator base" $a$. Then

$$
\log_b x = \underbrace{\frac{1}{\log_a b}}_{\text{a constant}} \cdot \log_a x .
$$

So the graph of $\log_b$ is the graph of $\log_a$ **stretched vertically** by a constant factor. Changing the base of a logarithm is like changing units: metres to feet, Celsius-degree *differences* to Fahrenheit ones. The useful conversion factors:

| conversion | factor |
|---|---|
| $\log_2 x = \ln x \,/\, \ln 2$ | $1/\ln 2 \approx 1.4427$ |
| $\log_{10} x = \ln x \,/\, \ln 10$ | $1/\ln 10 \approx 0.43429$ |
| $\ln x = \log_{10} x \,/\, \log_{10} e$ | $\ln 10 \approx 2.302585$ |
| $\log_2 x = \log_{10} x \,/\, \log_{10} 2$ | $1/\log_{10} 2 \approx 3.3219$ |

This is also why the *ratio* $\dfrac{\log_a x}{\log_a y}$ from notebook 03's table of false laws is not $\log_a \dfrac{x}{y}$: it is $\log_y x$.

### 2.3 Swapping base and argument

Take $x = a$ in the change-of-base formula: $\log_b a = \dfrac{\log_a a}{\log_a b} = \dfrac{1}{\log_a b}$, i.e.

$$
\log_b a \cdot \log_a b = 1 .
$$

Example: $\log_8 2 = \tfrac13$ and $\log_2 8 = 3$. More generally the factors chain like unit conversions:

$$
\log_a b \cdot \log_b c = \frac{\ln b}{\ln a}\cdot\frac{\ln c}{\ln b} = \frac{\ln c}{\ln a} = \log_a c .
$$

### 2.4 Base 10: common logarithms and decimal shifting

Our numbers are written in base 10, so multiplying by $10$ just moves the decimal point — and adds exactly $1$ to the common logarithm:

$$
\log_{10}(10^k \cdot m) = k + \log_{10} m .
$$

Every positive number can be written as $m \times 10^k$ with $1 \le m < 10$ (scientific notation). Then $\log_{10} x = k + \log_{10} m$, where

- the **characteristic** $k$ is an integer: where the decimal point is,
- the **mantissa** $\log_{10} m \in [0, 1)$: which digits appear.

So a table of $\log_{10} m$ for $1 \le m < 10$ — **one decade** — covers every positive number. That was the case for Briggs's base-10 tables, and it is why base 10 dominated hand computation for 350 years.

| $x$ | scientific notation | $\log_{10} x$ | characteristic | mantissa |
|---|---|---|---|---|
| $3720$ | $3.72 \times 10^{3}$ | $3.5705$ | $3$ | $0.5705$ |
| $3.72$ | $3.72 \times 10^{0}$ | $0.5705$ | $0$ | $0.5705$ |
| $0.0372$ | $3.72 \times 10^{-2}$ | $-1.4295$ | $-2$ | $0.5705$ |

For numbers below $1$ the log is negative, but table users kept the mantissa positive and wrote the negative characteristic with a bar: $\log_{10} 0.0372 = \bar{2}.5705$, meaning $-2 + 0.5705$. The digit count of notebook 01, $\lfloor \log_{10} N \rfloor + 1$, is "characteristic plus one".

### 2.5 Base $e$: the natural logarithm (preview)

The number $e = 2.71828\ldots$ looks like a strange choice for a base, but calculus makes it the only natural one. Every log function is $c \cdot \ln x$ for some constant $c$, and only the choice $c = 1$, i.e. base $e$, makes the slope at $x = 1$ equal to $1$:

$$
\frac{d}{dx} \log_b x = \frac{1}{x \ln b}, \qquad \frac{d}{dx}\ln x = \frac{1}{x}.
$$

Equivalently, $\ln x$ is the area under the hyperbola $1/t$ from $t = 1$ to $t = x$, and $e^x$ is its own derivative. Every other base drags a factor $\ln b$ through every derivative and integral. Module 01 builds this properly: $e$ from compound interest (notebook 01/01), $\ln$ as an area (01/02), and the derivatives (01/03).

Two consequences you will meet constantly: continuous growth at rate $r$ doubles in $\ln 2 / r \approx 0.693/r$ time units (the "rule of 70", and its cousin the rule of 72 for annual compounding), and Python's `math.log(x)` with one argument is $\ln x$.

### 2.6 Base 2: bits and doublings

$\log_2 x$ counts **doublings**: how many times you double $1$ to reach $x$, or halve $x$ to reach $1$.

- A positive integer $n$ needs $\lfloor \log_2 n \rfloor + 1$ binary digits (bits) — the base-2 version of digit counting. Python stores this as `n.bit_length()`.
- $n$ bits distinguish $2^n$ things, so choosing one of $N$ equally likely options takes $\log_2 N$ yes/no questions (module 05).
- Binary search halves the candidates each step, so it needs about $\lceil \log_2 N \rceil$ comparisons: $30$ for a billion items, since $\log_2 10^9 \approx 29.9$.

### 2.7 Units and notation

Because logs to different bases differ by constant factors, the base acts like a **unit**: information in **bits** ($\log_2$), **nats** ($\ln$) or decimal digits/**hartleys** ($\log_{10}$); $1$ nat $= 1/\ln 2 \approx 1.443$ bits, and $1$ decimal digit $= \log_2 10 \approx 3.32$ bits.

The bare symbol "log" means different things in different places — always write the base:

| where | "log" means |
|---|---|
| pure mathematics, Python `math.log`, `numpy.log` | $\ln$ |
| calculators, engineering, chemistry (pH) | $\log_{10}$ |
| computer science, information theory | $\log_2$ |

ISO 80000-2 recommends $\operatorname{lb}$ for $\log_2$, $\ln$ for $\log_e$ and $\operatorname{lg}$ for $\log_{10}$, though "lg" is also used for $\log_2$ in some computer-science texts.

## 3. Worked Examples

### Example 1 — $\log_5 200$ two ways

$$
\log_5 200 = \frac{\ln 200}{\ln 5} = \frac{5.29832}{1.60944} \approx 3.2920, \qquad
\log_5 200 = \frac{\log_{10} 200}{\log_{10} 5} = \frac{2.30103}{0.69897} \approx 3.2920 .
$$

Sanity check: $5^3 = 125 < 200 < 625 = 5^4$ ✓.

### Example 2 — swapping base and argument

$\log_9 27 \cdot \log_{27} 9 = 1$: indeed $\log_9 27 = \tfrac32$ (since $9^{3/2} = 27$) and $\log_{27} 9 = \tfrac23$.

Chain: $\log_2 3 \cdot \log_3 4 \cdot \log_4 5 \cdots \log_{63} 64 = \log_2 64 = 6$ — the middle bases cancel like units.

### Example 3 — characteristic and mantissa

Given $\log_{10} 2.5 = 0.39794$:

- $\log_{10} 250{,}000 = \log_{10}(2.5 \times 10^5) = 5.39794$, so $250{,}000$ has $5 + 1 = 6$ digits.
- $\log_{10} 0.0025 = \log_{10}(2.5 \times 10^{-3}) = -3 + 0.39794 = -2.60206$ (table notation $\bar{3}.39794$).

### Example 4 — doubling times at $7\%$, three ways

- Exact, annual compounding: $n = \dfrac{\ln 2}{\ln 1.07} = \dfrac{0.693147}{0.067659} \approx 10.24$ years.
- Rule of 72: $72 / 7 \approx 10.29$ years.
- Continuous growth at $7\%$: $\ln 2 / 0.07 \approx 9.90$ years.

### Example 5 — bits

How many bits does the number $1{,}000{,}000$ need? $\log_2 10^6 = 6 \log_2 10 = 6 \times 3.3219 = 19.93$, so $\lfloor 19.93 \rfloor + 1 = 20$ bits. Check: $2^{19} = 524{,}288 \le 10^6 < 1{,}048{,}576 = 2^{20}$ ✓.

## 4. Python Examples

### 4.1 Change of base, implemented

A two-line `log_base` reproduces `math.log(x, b)` — which is in fact how CPython computes the two-argument form: as $\ln x / \ln b$.

```python
import math


def log_base(x, base):
    # change of base: everything from the natural log
    return math.log(x) / math.log(base)


cases = [(5_000_000, 2), (2, 1.07), (200, 5), (0.001, 10), (8, 0.5)]
for x, base in cases:
    ours = log_base(x, base)
    via_log10 = math.log10(x) / math.log10(base)
    print(f"log_{base}({x}) = {ours:.10f}   via log10: {via_log10:.10f}   math.log(x, b): {math.log(x, base):.10f}")

print()
print("bits needed for offsets into 5,000,000 bytes:", math.ceil(math.log2(5_000_000)))
print("2**22 =", 2**22, "  2**23 =", 2**23)
```

### 4.2 `math.log(1000, 10)` is not $3$

Because the two-argument form divides two rounded natural logs, the result can miss an exact power by one unit in the last place. `math.log10` and `math.log2` are separate routines that, in practice, return exact results for exact powers of their base. This matters when you take a floor.

```python
import math

print("math.log(1000, 10) =", math.log(1000, 10))
print("math.log10(1000)   =", math.log10(1000))
print("math.log(125, 5)   =", math.log(125, 5))
print()

print(f"{'n':>8} {'floor(math.log(n,10))+1':>24} {'floor(log10(n))+1':>18} {'len(str(n))':>12}")
for n in [999, 1000, 10**5, 10**6]:
    digits_two_arg = math.floor(math.log(n, 10)) + 1
    digits_log10 = math.floor(math.log10(n)) + 1
    print(f"{n:>8} {digits_two_arg:>24} {digits_log10:>18} {len(str(n)):>12}")

print()
print("compare with a tolerance:", math.isclose(math.log(1000, 10), 3))
print("exact integer tools: (10**6).bit_length() =", (10**6).bit_length(), "  len(str(10**6)) =", len(str(10**6)))
```

The two-argument formula claims $1000$ and $10^6$ have $3$ and $6$ digits. For integers, prefer exact tools (`len(str(n))`, `n.bit_length()`); for floats, prefer `math.log10`/`math.log2` and compare with a tolerance.

### 4.3 Characteristic and mantissa

`math.floor` gives the characteristic, and the remainder is the mantissa — always in $[0, 1)$, even for numbers below $1$.

```python
import math

for x in [3720, 3.72, 0.0372, 250_000, 0.0025]:
    log_value = math.log10(x)
    characteristic = math.floor(log_value)
    mantissa = log_value - characteristic
    leading_digits = 10**mantissa
    print(f"x = {x:>9}: log10 x = {log_value:8.5f} = {characteristic:>2} + {mantissa:.5f}   (10**mantissa = {leading_digits:.4f})")
```

### 4.4 Reciprocal and chain identities

Numerically with floats, and exactly with SymPy.

```python
import math

import sympy as sp

pairs = [(2, "2", 8, "8"), (9, "9", 27, "27"), (10, "10", math.e, "e"), (1.07, "1.07", 2, "2")]
for a, a_name, b, b_name in pairs:
    product = math.log(a, b) * math.log(b, a)
    print(f"log_{b_name}({a_name}) * log_{a_name}({b_name}) = {product:.15f}   isclose to 1: {math.isclose(product, 1)}")

# Chain log_2 3 * log_3 4 * ... * log_63 64 = log_2 64 = 6
chain_product = 1.0
for k in range(2, 64):
    chain_product *= math.log(k + 1, k)
print("chain product:", chain_product, " isclose to 6:", math.isclose(chain_product, 6))

a, b, c = sp.symbols("a b c", positive=True)
print("sympy, reciprocal:", sp.simplify(sp.log(a, b) * sp.log(b, a)))
print("sympy, chain:     ", sp.simplify(sp.log(b, a) * sp.log(c, b)), "= log_a c")
```

### 4.5 Doubling times: exact versus the rules of 70 and 72

The exact doubling time for annual growth rate $r$ is $\ln 2 / \ln(1 + r)$; `math.log1p(r)` computes $\ln(1 + r)$ accurately for small $r$ (notebook 03/04).

```python
import math

print(f"{'rate':>6} {'exact years':>12} {'rule of 72':>11} {'rule of 70':>11} {'continuous ln2/r':>17}")
for rate in [0.01, 0.02, 0.05, 0.07, 0.10, 0.20]:
    exact_years = math.log(2) / math.log1p(rate)
    print(f"{rate:>6.0%} {exact_years:>12.2f} {72 / (100 * rate):>11.2f} {70 / (100 * rate):>11.2f} {math.log(2) / rate:>17.2f}")
```

## 5. Visualizations

### 5.1 One curve, three vertical scales

$\log_2 x$, $\ln x$ and $\log_{10} x$ all pass through $(1, 0)$ and differ only by a vertical stretch. The right panel divides each by $\ln x$: the ratios are flat lines at $1/\ln 2$, $1$ and $1/\ln 10$ — the conversion factors of 2.2.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

x_values = np.linspace(0.05, 20, 500)
curves = [("$\\log_2 x$", np.log2, "tab:blue"), ("$\\ln x$", np.log, "tab:green"), ("$\\log_{10} x$", np.log10, "tab:orange")]

fig, (ax_curves, ax_ratio) = plt.subplots(1, 2, figsize=(13, 5))
for label, function, colour in curves:
    ax_curves.plot(x_values, function(x_values), color=colour, label=label)
ax_curves.axhline(0, color="black", linewidth=0.6)
ax_curves.set_xlabel("x (linear scale)")
ax_curves.set_ylabel("value")
ax_curves.set_ylim(-4, 5)
ax_curves.grid(True, alpha=0.3)
ax_curves.legend(loc="lower right")
ax_curves.set_title("Three bases, one shape")

ratio_x = x_values[np.abs(x_values - 1) > 0.05]   # avoid 0/0 at x = 1
for label, function, colour in curves:
    ax_ratio.plot(ratio_x, function(ratio_x) / np.log(ratio_x), color=colour, label=label + r" / $\ln x$")
ax_ratio.text(10, 1 / math.log(2) + 0.05, f"1/ln 2 = {1 / math.log(2):.4f}", color="tab:blue")
ax_ratio.text(10, 1 + 0.05, "1", color="tab:green")
ax_ratio.text(10, 1 / math.log(10) + 0.05, f"1/ln 10 = {1 / math.log(10):.4f}", color="tab:orange")
ax_ratio.set_ylim(0, 1.8)
ax_ratio.set_xlabel("x (linear scale)")
ax_ratio.set_ylabel(r"ratio to $\ln x$")
ax_ratio.grid(True, alpha=0.3)
ax_ratio.set_title("The ratios are constants: change of base")
plt.tight_layout()
plt.show()
```

### 5.2 One logarithmic axis, three rulers

Place numbers at position $\ln x$. The same axis can be labelled in bits ($\log_2 x$), nats ($\ln x$) or decimal digits ($\log_{10} x$) — like a thermometer with Celsius and Fahrenheit scales, except that all three scales share the same zero (at $x = 1$).

```python
import math

import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(11, 3.6))
x_max = 1100
ax.hlines([2, 1, 0], 0, math.log(x_max), color="black", linewidth=1)

for power in range(0, 11):
    position = power * math.log(2)
    ax.vlines(position, 2, 2.15, color="tab:blue")
    ax.text(position, 2.22, f"{2**power}", ha="center", fontsize=9, color="tab:blue")
for nats in range(0, 7):
    ax.vlines(nats, 1, 1.15, color="tab:green")
    ax.text(nats, 1.22, f"e^{nats}", ha="center", fontsize=9, color="tab:green")
for power in range(0, 4):
    position = power * math.log(10)
    ax.vlines(position, 0, 0.15, color="tab:orange")
    ax.text(position, 0.22, f"{10**power}", ha="center", fontsize=9, color="tab:orange")

ax.text(-0.3, 2, "bits (log2)", ha="right", va="center", color="tab:blue")
ax.text(-0.3, 1, "nats (ln)", ha="right", va="center", color="tab:green")
ax.text(-0.3, 0, "digits (log10)", ha="right", va="center", color="tab:orange")
ax.set_xlim(-2.2, math.log(x_max) + 0.3)
ax.set_ylim(-0.3, 2.6)
ax.axis("off")
ax.set_title("Same positions, different units: 2^10 = 1024 sits just past 10^3 = 1000")
plt.show()
```

Ten bits ($1024$) almost exactly match three decimal digits ($1000$): $10 \log_{10} 2 \approx 3.0103$. That coincidence is why $2^{10}$ is called a "kilo" in computing.

### 5.3 Doubling time versus growth rate

The exact doubling time $\ln 2 / \ln(1 + r)$ on log–log axes, with the rule of 72. On these axes $72/r$ is a straight line of slope $-1$; the exact curve stays within about $5\%$ of it for rates from $1\%$ to $20\%$ and drifts away for larger rates.

```python
import matplotlib.pyplot as plt
import numpy as np

rates = np.logspace(np.log10(0.005), np.log10(1.0), 300)
exact_years = np.log(2) / np.log1p(rates)
rule_of_72 = 72 / (100 * rates)

fig, ax = plt.subplots(figsize=(8, 5.5))
ax.plot(rates * 100, exact_years, color="tab:blue", label=r"exact: $\ln 2 / \ln(1 + r)$")
ax.plot(rates * 100, rule_of_72, "--", color="crimson", label="rule of 72: 72 / (100 r)")
ax.plot(7, np.log(2) / np.log(1.07), "o", color="tab:blue")
ax.annotate("7%: 10.24 years", (7, np.log(2) / np.log(1.07)), textcoords="offset points", xytext=(10, 10))
ax.set_xscale("log")
ax.set_yscale("log")
ax.grid(True, which="both", alpha=0.3)
ax.set_xlabel("annual growth rate in percent (log scale)")
ax.set_ylabel("years to double (log scale)")
ax.legend()
ax.set_title("Doubling time = log base (1 + r) of 2")
plt.show()
```

## 7. Solutions

1. $\log_3 50 = \ln 50 / \ln 3 \approx 3.5609$; $\log_{1.5} 10 = \ln 10 / \ln 1.5 \approx 5.6789$; $\log_{0.2} 7 = \ln 7 / \ln 0.2 \approx -1.2091$ (negative: the base is below $1$ and $7 > 1$).
2. $\log_4 8 \cdot \log_8 4 = 1$ (reciprocal identity). $\log_3 5 \cdot \log_5 9 = \log_3 9 = 2$ (chain). $\dfrac{\log_{10} 64}{\log_{10} 4} = \log_4 64 = 3$.
3. $\log_{10}(6.02 \times 10^{23}) = 23.77960$: characteristic $23$, mantissa $0.77960$. $\log_{10} 0.000602 = \log_{10}(6.02 \times 10^{-4}) = -4 + 0.77960 = -3.22040$: characteristic $-4$, mantissa $0.77960$ (table notation $\bar{4}.77960$).
4. $10^{12} = (10^3)^4 \approx (2^{10})^4 = 2^{40}$, so $\log_2 10^{12} \approx 40$ and binary search needs about $40$ comparisons. Exactly: $12 \log_2 10 \approx 39.86$, so $\lceil 39.86 \rceil = 40$ comparisons.
5. Doubling: $\ln 2 / \ln 1.15 \approx 4.96$ hours. Factor 1000: $\ln 1000 / \ln 1.15 \approx 49.43$ hours. Their ratio is $\dfrac{\ln 1000}{\ln 2} = \log_2 1000 \approx 9.97$ — "growing 1000-fold" is "doubling about ten times".
6. Take $\log_b$ of both sides: $\log_b\!\big(a^{\log_b c}\big) = \log_b c \cdot \log_b a$ and $\log_b\!\big(c^{\log_b a}\big) = \log_b a \cdot \log_b c$. These are equal, and $\log_b$ is injective, so the numbers are equal. For $a = 4$, $b = 2$, $c = 9$: $4^{\log_2 9} = 9^{\log_2 4} = 9^2 = 81$.
7. $\log_{b^k} x = \dfrac{\log_b x}{\log_b b^k} = \dfrac{1}{k}\log_b x$. So $\log_{100} 10 = \tfrac12 \log_{10} 10 = \tfrac12$; $\log_{\sqrt2} 8 = 2 \log_2 8 = 6$; $\log_{1/9} 27 = \log_{3^{-2}} 27 = -\tfrac12 \log_3 27 = -\tfrac32$.
8. A quantity $I$ in nats is $\ln N$, in bits $\log_2 N = \ln N / \ln 2$, so $1$ nat $= 1/\ln 2 \approx 1.4427$ bits; likewise $\log_2 N = \log_{10} N \cdot \log_2 10$, so $1$ digit $= \log_2 10 \approx 3.3219$ bits. $8000$ bits $= 8000 \ln 2 \approx 5545.2$ nats $= 8000 \log_{10} 2 \approx 2408.2$ decimal digits.
9. With the two-argument `math.log` exactly two values fail: $n = 1000$ and $n = 10^6$. Both are exact powers of ten, where $\ln n / \ln 10$ is rounded to just **below** the integer, so the floor drops by one. (The other powers of ten happen to round correctly; every other $n$ is far enough from an integer log that rounding cannot matter.) With `math.log10` there are no failures.

Numerical checks:

```python
import math

print("Ex 1:", round(math.log(50) / math.log(3), 4), round(math.log(10) / math.log(1.5), 4), round(math.log(7) / math.log(0.2), 4))
print("Ex 2:", round(math.log(8, 4) * math.log(4, 8), 12), round(math.log(5, 3) * math.log(9, 5), 12), math.log10(64) / math.log10(4))
print("Ex 3:", round(math.log10(6.02e23), 5), round(math.log10(0.000602), 5))
print("Ex 4:", round(12 * math.log2(10), 4), "->", math.ceil(12 * math.log2(10)), "comparisons")
doubling_hours = math.log(2) / math.log(1.15)
thousandfold_hours = math.log(1000) / math.log(1.15)
print("Ex 5:", round(doubling_hours, 2), round(thousandfold_hours, 2), "ratio", round(thousandfold_hours / doubling_hours, 4))
print("Ex 6:", 4 ** math.log2(9), 9 ** math.log2(4))
print("Ex 7:", math.log(10, 100), math.log(8, math.sqrt(2)), math.log(27, 1 / 9))
print("Ex 8:", round(8000 * math.log(2), 1), "nats,", round(8000 * math.log10(2), 1), "digits")

failures_two_arg = [n for n in range(1, 10**6 + 1) if math.floor(math.log(n, 10)) + 1 != len(str(n))]
failures_log10 = [n for n in range(1, 10**6 + 1) if math.floor(math.log10(n)) + 1 != len(str(n))]
print("Ex 9: failures with math.log(n, 10):", failures_two_arg)
print("      failures with math.log10(n):  ", failures_log10)
```

## 8. Conclusion

- **Change of base:** $\log_b x = \dfrac{\log_a x}{\log_a b}$, proved by taking $\log_a$ of $b^y = x$. Every logarithm is a constant multiple of $\ln$; changing the base is changing units.
- **Reciprocal and chain:** $\log_b a \cdot \log_a b = 1$ and $\log_a b \cdot \log_b c = \log_a c$.
- **Base 10** made hand computation work: shifting the decimal point changes only the integer **characteristic**, so one decade of **mantissas** covered every number.
- **Base $e$** is natural for calculus: $\dfrac{d}{dx}\ln x = \dfrac1x$ with no stray constant (module 01).
- **Base 2** counts doublings and bits: $\lfloor \log_2 n \rfloor + 1$ bits, about $\log_2 N$ steps of binary search, doubling time $\log_{1+r} 2$.
- In Python, `math.log(x, b)` is $\ln x / \ln b$ and inherits rounding (`math.log(1000, 10)` $= 2.9999999999999996$). Use `math.log10`, `math.log2`, `int.bit_length` or `len(str(n))` where exactness matters, and compare floats with a tolerance.
