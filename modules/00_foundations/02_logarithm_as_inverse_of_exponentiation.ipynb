# The Logarithm as the Inverse of Exponentiation

Notebook 01 showed what logarithms are *for*: a table of powers turns multiplication into addition. This notebook makes the idea precise. We define $b^y$ for **every** real exponent $y$ — not just whole numbers — so that the table has no gaps, and then define the logarithm as the function that undoes it:

$$
b^y = x \quad\Longleftrightarrow\quad y = \log_b x .
$$

This is notebook 02 of module 00 (**Foundations**) in the **Learning Logarithms** series.

## 1. Introduction

Two equations that look almost the same:

$$
2^x = 8 \qquad\text{and}\qquad 2^x = 10 .
$$

The first is easy: $2 \cdot 2 \cdot 2 = 8$, so $x = 3$. The second has no whole-number answer. Since $2^3 = 8 < 10 < 16 = 2^4$, the answer lies somewhere between $3$ and $4$ — but what does "$2$ to the power $3.3$" even *mean*? You cannot multiply $2$ by itself $3.3$ times.

So the question "what is that $x$?" really contains two questions:

1. **What do non-integer powers mean?** We need $2^{3.3}$, $2^{1/2}$ and $2^{\sqrt{2}}$ to be honest numbers, with the familiar exponent laws still true.
2. **Is there exactly one answer?** If $y \mapsto 2^y$ is a strictly increasing function that takes every positive value, then $2^x = 10$ has exactly one solution. That solution gets a name: $x = \log_2 10 \approx 3.3219$.

Once both questions are answered, the logarithm is simply the **inverse function** of the exponential — in the same way $\sqrt{\phantom{x}}$ undoes squaring on $x \ge 0$.

## 2. Theory

### 2.1 Filling the gaps in the table

Start with what everyone agrees on: for a positive integer $n$, $b^n = b \cdot b \cdots b$ ($n$ factors). From this,

$$
b^m \, b^n = b^{m+n}, \qquad (b^m)^n = b^{mn}.
$$

Every extension below is chosen so that **these two laws keep working**. Nothing is a free choice.

| Exponent | Definition | Forced because… |
|---|---|---|
| $0$ | $b^0 = 1$ | $b^0 \cdot b^n = b^{0+n} = b^n$, so $b^0$ must be $1$ |
| $-n$ | $b^{-n} = \dfrac{1}{b^n}$ | $b^{-n} \cdot b^n = b^0 = 1$ |
| $\tfrac1q$ | $b^{1/q} = \sqrt[q]{b}$ | $(b^{1/q})^q = b^{1} = b$ |
| $\tfrac pq$ | $b^{p/q} = \left(\sqrt[q]{b}\right)^p$ | $b^{p/q} = (b^{1/q})^p$ |

For $b > 0$ the root $\sqrt[q]{b}$ is the unique **positive** number whose $q$-th power is $b$, so all of these are well-defined positive numbers. (One checks that $b^{2/4} = b^{1/2}$, i.e. the answer does not depend on how the fraction is written — this is exactly where $b > 0$ is needed, see 2.5.)

**Real exponents.** An irrational number such as $\sqrt2 = 1.41421356\ldots$ is squeezed between rationals: $1.4 < 1.41 < 1.414 < \cdots < \sqrt 2 < \cdots < 1.415 < 1.42 < 1.5$. For $b > 1$ the map $y \mapsto b^y$ is increasing on rationals, so the numbers

$$
2^{1.4},\; 2^{1.41},\; 2^{1.414},\; 2^{1.4142},\; \dots
$$

increase and are bounded above by $2^{1.5}$. They converge to a limit, and that limit **defines** $2^{\sqrt2} \approx 2.6651$. The resulting function $y \mapsto b^y$ on all of $\mathbb{R}$ is continuous and still obeys both exponent laws. Section 4.2 watches this convergence happen.

### 2.2 The exponential function

For a fixed base $b > 0$, $b \neq 1$, the **exponential function** $f(y) = b^y$ has:

- **domain** all real numbers $\mathbb{R}$,
- **range** the positive reals $(0, \infty)$ — it is never zero or negative, and it takes every positive value (by continuity, since $b^y \to 0$ at one end and $b^y \to \infty$ at the other),
- it is **strictly increasing** if $b > 1$ and **strictly decreasing** if $0 < b < 1$.

Strictly monotone means **injective**: different exponents give different values. Injective plus "takes every positive value" means that for each $x > 0$ there is **exactly one** $y$ with $b^y = x$. That is the answer to both questions in the introduction.

### 2.3 Definition of the logarithm

For $b > 0$, $b \neq 1$ and $x > 0$, the **logarithm of $x$ to base $b$** is the unique real number $y$ with $b^y = x$:

$$
\boxed{\; b^y = x \quad\Longleftrightarrow\quad y = \log_b x \;}
$$

| | exponential $y \mapsto b^y$ | logarithm $x \mapsto \log_b x$ |
|---|---|---|
| domain | $\mathbb{R}$ | $(0, \infty)$ |
| range | $(0, \infty)$ | $\mathbb{R}$ |
| key point | $(0, 1)$ | $(1, 0)$ |
| asymptote | $y$-values $\to 0$ (horizontal) | $x \to 0^+$ (vertical) |

The domain and range swap — as for every pair of inverse functions. In words: **$\log_b x$ is the exponent you must put on $b$ to get $x$.**

### 2.4 Special values and the inverse identities

Each line is an exponent fact read in log form.

| exponent form | log form |
|---|---|
| $b^0 = 1$ | $\log_b 1 = 0$ |
| $b^1 = b$ | $\log_b b = 1$ |
| $b^{-1} = 1/b$ | $\log_b \tfrac1b = -1$ |
| $b^{1/2} = \sqrt b$ | $\log_b \sqrt b = \tfrac12$ |

Because $\log_b$ and $b^{(\cdot)}$ are inverse functions, applying one after the other does nothing:

$$
b^{\log_b x} = x \quad (x > 0), \qquad \log_b\!\left(b^y\right) = y \quad (y \in \mathbb{R}).
$$

The first says "raise $b$ to *the exponent that gives $x$*" — of course you get $x$. The second says "which exponent gives $b^y$?" — of course it is $y$. Note the different domains: $\log_b(b^y)$ works for **every** $y$, but $b^{\log_b x}$ only makes sense for $x > 0$.

### 2.5 Why $b > 0$ and $b \neq 1$

- **$b = 1$:** $1^y = 1$ for every $y$. The function is constant, not injective; $1^y = 5$ has no solution and $1^y = 1$ has infinitely many. No inverse.
- **$b = 0$:** $0^y = 0$ for $y > 0$ and is undefined for $y \le 0$. Same problem.
- **$b < 0$:** take $b = -8$. Then $(-8)^{1/3} = -2$ (since $(-2)^3 = -8$), but $\tfrac13 = \tfrac26$ and $(-8)^{2/6} = \sqrt[6]{(-8)^2} = \sqrt[6]{64} = 2$. The same exponent gives two different answers. Worse, $(-8)^{1/2}$ is not a real number at all. Negative bases jump between positive, negative and undefined on every interval, so there is no continuous function to invert.

With $b > 0$ the argument is restricted too: $b^y > 0$ always, so **$\log_b x$ exists only for $x > 0$**. $\log_b 0$ and $\log_b(-3)$ are undefined in the real numbers. (Module 07 shows what happens with complex numbers.)

### 2.6 Negative and fractional logarithms

The sign of $\log_b x$ (for $b > 1$) tells you where $x$ sits relative to $1$:

$$
0 < x < 1 \;\Rightarrow\; \log_b x < 0, \qquad x = 1 \;\Rightarrow\; \log_b x = 0, \qquad x > 1 \;\Rightarrow\; \log_b x > 0 .
$$

Negative logs are perfectly normal: $\log_{10} 0.001 = -3$ because $10^{-3} = 0.001$. **The logarithm can be negative; its argument cannot.**

Fractional logs come from roots: $\log_4 2 = \tfrac12$ because $4^{1/2} = 2$, and $\log_{27} 9 = \tfrac23$ because $27^{2/3} = (\sqrt[3]{27})^2 = 9$. A trick that always works when both numbers are powers of a common base: write them as powers of that base and compare exponents.

For a base $0 < b < 1$ the signs flip: $\log_{1/2} 8 = -3$ since $(1/2)^{-3} = 8$. Bases below $1$ are legal but rare.

### 2.7 The picture: mirror image in $y = x$

The point $(y, x)$ lies on the graph of $x = b^y$ exactly when $(x, y)$ lies on the graph of $y = \log_b x$. Swapping coordinates is reflection in the line $y = x$. So **the graph of $\log_b x$ is the graph of $b^x$ reflected in the diagonal**: $(3, 8)$ on $y = 2^x$ becomes $(8, 3)$ on $y = \log_2 x$; the horizontal asymptote $y = 0$ becomes the vertical asymptote $x = 0$.

And the core picture from notebook 01 still holds: on the $x$-axis, $1, 2, 4, 8, 16$ are spaced further and further apart, but $\log_2$ sends them to $0, 1, 2, 3, 4$ — equal ratios to equal steps.

## 3. Worked Examples

### Example 1 — trapping $\log_2 10$ by halving the interval

We know $2^3 = 8 < 10 < 16 = 2^4$, so $3 < \log_2 10 < 4$.

- Midpoint $3.5$: $2^{3.5} = 8\sqrt2 \approx 11.31 > 10$, so $3 < \log_2 10 < 3.5$.
- Midpoint $3.25$: $2^{3.25} = 8 \cdot 2^{1/4} \approx 8 \times 1.1892 = 9.51 < 10$, so $3.25 < \log_2 10 < 3.5$.
- Midpoint $3.375$: $2^{3.375} = 8 \cdot 2^{3/8} \approx 8 \times 1.2968 = 10.37 > 10$, so $3.25 < \log_2 10 < 3.375$.

Every step halves the interval and uses only square roots — the way the earliest tables were computed (notebook 03/01). Continued, it converges to $\log_2 10 = 3.32193\ldots$ (section 4.1).

### Example 2 — exact fractional and negative logs

| question | rewrite in exponent form | answer |
|---|---|---|
| $\log_{27} 9$ | $(3^3)^y = 3^2 \Rightarrow 3y = 2$ | $\tfrac23$ |
| $\log_2 \tfrac18$ | $2^y = 2^{-3}$ | $-3$ |
| $\log_{1/3} 81$ | $(3^{-1})^y = 3^4 \Rightarrow -y = 4$ | $-4$ |
| $\log_{10} \sqrt[5]{10}$ | $10^y = 10^{1/5}$ | $\tfrac15$ |
| $\log_8 \tfrac14$ | $(2^3)^y = 2^{-2} \Rightarrow 3y = -2$ | $-\tfrac23$ |

### Example 3 — the inverse identities

- $5^{\log_5 7} = 7$.
- $\log_3(3^{-2.5}) = -2.5$.
- $10^{2\log_{10} 3}$: write $2 \log_{10} 3$ as an exponent of an exponent: $10^{2 \log_{10} 3} = \left(10^{\log_{10} 3}\right)^2 = 3^2 = 9$.

### Example 4 — solving for the base

$\log_b 32 = \tfrac52$ means $b^{5/2} = 32$. Raise both sides to the power $\tfrac25$: $b = 32^{2/5} = (\sqrt[5]{32})^2 = 2^2 = 4$. Check: $4^{5/2} = (\sqrt4)^5 = 2^5 = 32$ ✓. Also $4 > 0$ and $4 \ne 1$, so it is a legal base.

### Example 5 — domain of a log expression

Where is $f(x) = \log_2(x - 3)$ defined? We need the argument positive: $x - 3 > 0$, so the domain is $x > 3$. Its graph is $\log_2 x$ shifted right by $3$, with vertical asymptote $x = 3$, and $f(4) = \log_2 1 = 0$, $f(11) = \log_2 8 = 3$.

## 4. Python Examples

### 4.1 Solving $2^x = 10$ by bisection

The bisection of Example 1, continued until the interval is tiny. It uses only the fact that $2^y$ is increasing.

```python
import math


def log_by_bisection(x, base, low, high, steps=40):
    # Requires base > 1 and base**low <= x <= base**high
    for _ in range(steps):
        middle = (low + high) / 2
        if base**middle < x:
            low = middle
        else:
            high = middle
    return (low + high) / 2


for step_count in [1, 2, 3, 5, 10, 20, 40]:
    estimate = log_by_bisection(10, 2, 3, 4, steps=step_count)
    print(f"{step_count:>2} halvings: log2(10) ~ {estimate:.10f}")

print("math.log2(10)  =", math.log2(10))
print("2 ** log2(10)  =", 2 ** math.log2(10))
```

After $n$ halvings the interval has width $2^{-n}$, so $40$ halvings pin the answer to about $10^{-12}$.

### 4.2 Real exponents as limits of rational ones

$2^{\sqrt2}$ is defined as the limit of $2^{r}$ over rationals $r \to \sqrt2$. Truncating the decimal expansion of $\sqrt2$ gives rational exponents $14/10$, $141/100$, … and each power $2^{p/q}$ can be computed using only an integer power and a $q$-th root.

```python
import math
from fractions import Fraction

sqrt_two = math.sqrt(2)

for digits in range(1, 9):
    exponent = Fraction(math.floor(sqrt_two * 10**digits), 10**digits)
    # b^(p/q) = (q-th root of b)^p
    root = 2 ** (1 / exponent.denominator)
    rational_power = root ** exponent.numerator
    print(f"2^({float(exponent):<10}) = {rational_power:.10f}")

print(f"2^sqrt(2)          = {2 ** sqrt_two:.10f}")
```

The powers increase towards $2^{\sqrt2} \approx 2.6651441$, gaining roughly one correct digit per digit of the exponent: the table from notebook 01 now has no gaps.

### 4.3 The inverse identities, numerically and symbolically

Check $b^{\log_b x} = x$ and $\log_b(b^y) = y$ for a few bases, with a tolerance — floating point never gives exact equality.

```python
import math

import sympy as sp

for base in [2, 10, 0.5, 1.07]:
    x = 37.5
    y = -1.25
    round_trip_x = base ** math.log(x, base)
    round_trip_y = math.log(base**y, base)
    print(f"base {base:>4}: b^(log_b x) = {round_trip_x:.15f}  isclose: {math.isclose(round_trip_x, x)}"
          f" | log_b(b^y) = {round_trip_y:.15f}  isclose: {math.isclose(round_trip_y, y)}")

# Symbolically the identities are exact
b = sp.symbols("b", positive=True)
x = sp.symbols("x", positive=True)
y = sp.symbols("y", real=True)
print()
print("simplify(b**log(x, b)) =", sp.simplify(b ** sp.log(x, b)))
print("simplify(log(b**y, b)) =", sp.simplify(sp.expand_log(sp.log(b**y, b), force=True)))
print("log(1, b) =", sp.log(1, b), "   log(b, b) =", sp.log(b, b))
exact_logs = [sp.log(9, 27), sp.log(sp.Rational(1, 8), 2), sp.log(81, sp.Rational(1, 3))]
print("exact logs (Example 2):", [sp.expand_log(value, force=True) for value in exact_logs])
```

### 4.4 What goes wrong outside the domain

Python refuses logs of non-positive numbers, and a negative base raised to a fractional power leaves the real numbers: Python 3 returns a **complex** number for `(-8) ** (1/3)` (the principal cube root, not $-2$), while `math.pow` raises an error.

```python
import math

for argument in [0, -3]:
    try:
        math.log(argument)
    except ValueError as error:
        print(f"math.log({argument}) -> ValueError: {error}")

print("(-8) ** (1/3)      =", (-8) ** (1 / 3))
try:
    math.pow(-8, 1 / 3)
except ValueError as error:
    print("math.pow(-8, 1/3)  -> ValueError:", error)

# The real cube root has to be taken by hand
print("real cube root of -8 =", -(8 ** (1 / 3)))
print("1 ** y for several y:", [1**y for y in [-2, 0.5, 3, 10]], "-> base 1 cannot be inverted")
```

### 4.5 SymPy solves $2^x = 10$ exactly

SymPy writes the answer with natural logarithms, as $1 + \ln 5 / \ln 2$. That equals $\ln 10 / \ln 2$ (because $\ln 10 = \ln 2 + \ln 5$, notebook 03) — a preview of the change-of-base formula in notebook 04.

```python
import sympy as sp

x = sp.symbols("x", real=True)
solution = sp.solve(sp.Eq(2**x, 10), x)
print("solution:", solution)
print("numerically:", [sp.N(s, 12) for s in solution])
print("same as log(10, 2)?", sp.simplify(solution[0] - sp.log(10, 2)) == 0)
```

## 5. Visualizations

### 5.1 $2^x$ and $\log_2 x$: mirror images in $y = x$

Each marked point $(a, 2^a)$ on the exponential has a partner $(2^a, a)$ on the logarithm; the dotted segments joining them are perpendicular to the diagonal.

```python
import matplotlib.pyplot as plt
import numpy as np

y_values = np.linspace(-4, 3.2, 400)
x_values = np.linspace(0.02, 9.2, 400)

fig, ax = plt.subplots(figsize=(7, 7))
ax.plot(y_values, 2.0**y_values, color="tab:blue", label=r"$y = 2^x$")
ax.plot(x_values, np.log2(x_values), color="tab:orange", label=r"$y = \log_2 x$")
ax.plot([-4, 9.5], [-4, 9.5], "k--", linewidth=1, label=r"$y = x$")

for exponent in [-2, -1, 0, 1, 2, 3]:
    value = 2.0**exponent
    ax.plot(exponent, value, "o", color="tab:blue")
    ax.plot(value, exponent, "o", color="tab:orange")
    ax.plot([exponent, value], [value, exponent], ":", color="gray")
ax.annotate("(3, 8)", (3, 8), textcoords="offset points", xytext=(8, -4))
ax.annotate("(8, 3)", (8, 3), textcoords="offset points", xytext=(-10, -18))

ax.axhline(0, color="black", linewidth=0.6)
ax.axvline(0, color="black", linewidth=0.6)
ax.set_xlim(-4, 9.5)
ax.set_ylim(-4, 9.5)
ax.set_aspect("equal")
ax.set_xlabel("x (linear scale)")
ax.set_ylabel("y (linear scale)")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left")
ax.set_title("The logarithm is the exponential reflected in y = x")
plt.show()
```

The exponential hugs the negative $x$-axis (horizontal asymptote $y = 0$); after reflection the logarithm hugs the negative $y$-axis (vertical asymptote $x = 0$). Both pass through the mirror pair $(0, 1)$ and $(1, 0)$.

### 5.2 Filling the gaps: integer, half and sixteenth exponents

Integer exponents give a handful of isolated dots on $y = 2^x$ (Stifel's table). Halves fill in between, sixteenths almost trace the curve — and the real exponents of 2.1 fill it completely.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(1, 3, figsize=(13, 4), sharey=True)
smooth = np.linspace(-2, 4, 300)
for ax, denominator in zip(axes, [1, 2, 16]):
    rational_exponents = np.arange(-2 * denominator, 4 * denominator + 1) / denominator
    ax.plot(smooth, 2.0**smooth, color="lightgray", linewidth=3, zorder=1)
    ax.plot(rational_exponents, 2.0**rational_exponents, "o", markersize=4, color="tab:blue", zorder=2)
    ax.axhline(10, color="crimson", linewidth=1, linestyle="--")
    ax.set_title(f"exponents p/{denominator}")
    ax.set_xlabel("exponent x")
    ax.grid(True, alpha=0.3)
axes[0].set_ylabel(r"$2^x$ (linear scale)")
axes[0].text(-1.9, 10.5, "target 10", color="crimson")
plt.suptitle("Rational exponents fill the gaps; the value 10 is hit at x = log2(10) = 3.32...")
plt.tight_layout()
plt.show()
```

### 5.3 Logarithms to several bases

All curves pass through $(1, 0)$ because $\log_b 1 = 0$, and each passes through $(b, 1)$ because $\log_b b = 1$. Bases above $1$ give increasing curves; base $\tfrac12$ gives the decreasing one, the reflection of $\log_2 x$ in the $x$-axis ($\log_{1/2} x = -\log_2 x$).

```python
import math

import matplotlib.pyplot as plt
import numpy as np

x_values = np.linspace(0.05, 12, 500)
bases = [(2, "tab:blue"), (math.e, "tab:green"), (10, "tab:orange"), (0.5, "tab:purple")]

fig, ax = plt.subplots(figsize=(9, 5.5))
for base, colour in bases:
    label = "e" if math.isclose(base, math.e) else f"{base:g}"
    ax.plot(x_values, np.log(x_values) / np.log(base), color=colour, label=f"base {label}")
    if base > 1:
        ax.plot(base, 1, "o", color=colour)
ax.plot(1, 0, "ko")
ax.annotate("(1, 0): log_b 1 = 0", (1, 0), textcoords="offset points", xytext=(10, -22))
ax.axhline(0, color="black", linewidth=0.6)
ax.axhline(1, color="gray", linewidth=0.8, linestyle=":")
ax.text(5, 1.08, "y = 1", color="gray")
ax.set_xlim(0, 12)
ax.set_ylim(-4, 4)
ax.set_xlabel("x (linear scale)")
ax.set_ylabel(r"$\log_b x$")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower right", bbox_to_anchor=(1, 0.25))
ax.set_title("Each log curve crosses y = 1 at x = b")
plt.show()
```

9. See the code below: first step $n$ down while $b^n > x$ and up while $b^{n+1} \le x$, then bisect on $[n, n+1]$. $\log_2 0.01 \approx -6.6439$ and $\log_{10} 12345 \approx 4.0915$.

Numerical checks:

```python
import math

import sympy as sp

def exact(value):
    return sp.expand_log(value, force=True)


print("Ex 1:", exact(sp.log(243, 3)), exact(sp.log(sp.Rational(1, 10000), 10)), exact(sp.log(8, 16)))
print("Ex 2:", [exact(sp.log(1, 5)), exact(sp.log(9, 9)), exact(sp.log(3, 9)), exact(sp.log(10, 100)), exact(sp.log(sp.Rational(1, 27), 3))])
print("Ex 3:", 6 ** math.log(11, 6), math.log(4**-0.3, 4), 2 ** (3 * math.log2(5)))
print("Ex 4: 3^4.5 =", 3**4.5, "  log3(100) =", math.log(100, 3))
print("Ex 6:", 125 ** (1 / 3), 8 ** (-2 / 3), "check (1/4)^(-3/2) =", 0.25 ** (-1.5))
print("Ex 7: log_(1/2)(10) =", math.log(10, 0.5), " -log2(10) =", -math.log2(10))


def log_by_bisection(x, base, steps=60):
    if x <= 0 or base <= 1:
        raise ValueError("need x > 0 and base > 1")
    n = 0
    while base**n > x:
        n -= 1
    while base ** (n + 1) <= x:
        n += 1
    low, high = n, n + 1
    for _ in range(steps):
        middle = (low + high) / 2
        if base**middle < x:
            low = middle
        else:
            high = middle
    return (low + high) / 2


for x, base in [(0.01, 2), (12345, 10)]:
    estimate = log_by_bisection(x, base)
    print(f"Ex 9: log_{base}({x}) = {estimate:.6f}  math.log: {math.log(x, base):.6f}  close: {math.isclose(estimate, math.log(x, base))}")
```

## 8. Conclusion

- Powers $b^y$ are extended from whole numbers to $0$, negative integers, fractions and finally all reals, each step **forced** by keeping $b^m b^n = b^{m+n}$ and $(b^m)^n = b^{mn}$: $b^0 = 1$, $b^{-n} = 1/b^n$, $b^{p/q} = (\sqrt[q]{b})^p$, and irrational exponents as limits.
- For $b > 0$, $b \neq 1$, the function $y \mapsto b^y$ is strictly monotone from $\mathbb{R}$ onto $(0, \infty)$, so it has an inverse: $\;b^y = x \iff y = \log_b x$, defined for $x > 0$ with values in all of $\mathbb{R}$.
- Special values and identities: $\log_b 1 = 0$, $\log_b b = 1$, $b^{\log_b x} = x$, $\log_b(b^y) = y$.
- Negative bases break fractional powers, base $1$ is constant, and the argument of a log must be positive. The log itself may be negative (arguments below $1$) or fractional (roots).
- Geometrically $\log_b x$ is $b^x$ reflected in $y = x$, and it turns doublings $1, 2, 4, 8$ into equal steps $0, 1, 2, 3$.
