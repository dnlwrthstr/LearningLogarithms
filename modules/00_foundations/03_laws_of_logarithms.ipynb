# The Laws of Logarithms

Napier's table worked because of one rule: the logarithm of a product is the sum of the logarithms. This notebook proves that rule and its two companions — the quotient and power laws — directly from the exponent laws, shows how they tame numbers far too large to write down, and collects the tempting "laws" that are **false**.

This is notebook 03 of module 00 (**Foundations**) in the **Learning Logarithms** series.

## 1. Introduction

Three questions that look hopeless by hand:

1. **Which is bigger, $2^{1000}$ or $3^{630}$?** Each has about 300 digits. Computing either is out of the question.
2. **How many digits does $\binom{100}{50} = \dfrac{100!}{50!\,50!}$ have?** The numerator alone has 158 digits.
3. **A savings account grows $5\%$ per year. After how many years has it doubled?** We need $n$ with $1.05^n = 2$ — the unknown sits in the exponent.

All three fall to the same three rules. For $x, y > 0$ and any real $k$:

$$
\log_b(xy) = \log_b x + \log_b y, \qquad \log_b \frac{x}{y} = \log_b x - \log_b y, \qquad \log_b (x^k) = k \log_b x .
$$

- (1) The power law turns $2^{1000}$ vs $3^{630}$ into $1000 \log_{10} 2 \approx 301.03$ vs $630 \log_{10} 3 \approx 300.58$. So $2^{1000}$ is bigger — about $10^{0.45} \approx 2.8$ times bigger.
- (2) The product and quotient laws turn $\log_{10}\binom{100}{50}$ into a **sum** of 100 small logarithms (section 4.3).
- (3) The power law brings the exponent down: $n \log_{10} 1.05 = \log_{10} 2$, so $n \approx 14.2$ years.

These laws are not new facts. They are the exponent laws $b^m b^n = b^{m+n}$ and $(b^m)^k = b^{mk}$, **read in log form**.

**Prerequisites**

- Tables of powers and multiplying by adding exponents — [notebook 01](./01_why_logarithms_exist.ipynb).
- The definition $b^y = x \iff y = \log_b x$, real exponents, $b^{\log_b x} = x$ — [notebook 02](./02_logarithm_as_inverse_of_exponentiation.ipynb).

After completing this notebook you will be able to:

- Prove the product, quotient and power laws from the exponent laws.
- Expand a log of a product/quotient/power into simple pieces, and condense a sum of logs back into one log — by hand and with SymPy's `expand_log` and `logcombine`.
- Spot the common false "laws", such as $\log_b(x + y) = \log_b x + \log_b y$, and give a counterexample.
- Track the domain: why $\log_b(x^2) = 2\log_b|x|$ and not $2\log_b x$.
- Compare and estimate astronomically large numbers using logarithms.

## 2. Theory

### 2.1 The exponent laws we start from

For a base $b > 0$ and real exponents $m, n$ (notebook 02):

$$
b^m \, b^n = b^{m+n}, \qquad \frac{b^m}{b^n} = b^{m-n}, \qquad (b^m)^k = b^{mk}.
$$

And the log is defined by $b^y = x \iff y = \log_b x$, with $b > 0$, $b \ne 1$, $x > 0$. Every proof below has the same three steps: **name the logs, use an exponent law, read the result in log form**.

### 2.2 Product law

Let $x, y > 0$ and put $m = \log_b x$, $n = \log_b y$, so $x = b^m$ and $y = b^n$. Then

$$
xy = b^m b^n = b^{m+n} \quad\Longrightarrow\quad \log_b(xy) = m + n = \log_b x + \log_b y .
$$

| exponent form | log form |
|---|---|
| $b^m \cdot b^n = b^{m+n}$ | $\log_b(xy) = \log_b x + \log_b y$ |

### 2.3 Quotient law

With the same $m, n$: $\dfrac{x}{y} = \dfrac{b^m}{b^n} = b^{m-n}$, so

$$
\log_b \frac{x}{y} = \log_b x - \log_b y .
$$

The special case $x = 1$ gives the **reciprocal rule** $\log_b \dfrac1y = -\log_b y$: taking a reciprocal flips the sign of the log.

### 2.4 Power law

Let $x > 0$, $k$ any real number, and $m = \log_b x$. Then $x^k = (b^m)^k = b^{mk}$, so

$$
\log_b (x^k) = k\, m = k \log_b x .
$$

Roots are powers with $k = 1/q$: $\log_b \sqrt[q]{x} = \dfrac{1}{q}\log_b x$. This is the law that **brings an unknown exponent down** to where ordinary algebra can reach it.

Together, the three laws say the logarithm climbs down the hierarchy of operations by one level:

| numbers $x, y$ | logarithms |
|---|---|
| $x \cdot y$ | $\log_b x + \log_b y$ |
| $x / y$ | $\log_b x - \log_b y$ |
| $x^k$ | $k \cdot \log_b x$ |
| $\sqrt[k]{x}$ | $\log_b x \,/\, k$ |

### 2.5 What is **not** a law

The logarithm does **not** simplify sums, and it does not commute with multiplication of logs. Each "law" below is false; the counterexample uses $\log_{10}$.

| tempting but false | counterexample | correct statement |
|---|---|---|
| $\log_b(x + y) = \log_b x + \log_b y$ | $\log_{10}(10 + 10) = 1.301$, but $1 + 1 = 2$ | $\log_b x + \log_b y = \log_b(xy)$ |
| $\log_b(xy) = \log_b x \cdot \log_b y$ | $\log_{10}(10 \cdot 10) = 2$, but $1 \cdot 1 = 1$ | sum, not product |
| $\dfrac{\log_b x}{\log_b y} = \log_b \dfrac{x}{y}$ | $\dfrac{\log_{10} 100}{\log_{10} 10} = 2$, but $\log_{10} 10 = 1$ | difference, not quotient (the ratio is a change of base — notebook 04) |
| $(\log_b x)^k = k \log_b x$ | $(\log_{10} 1000)^2 = 9$, but $2 \log_{10} 1000 = 6$ | $\log_b(x^k) = k\log_b x$ |
| $\log_b(x - y) = \log_b x - \log_b y$ | $\log_{10}(1000 - 10) \approx 2.996$, but $3 - 1 = 2$ | $\log_b x - \log_b y = \log_b(x/y)$ |

The one honest thing you can do with a sum is **factor** it: $\log_b(x + y) = \log_b x + \log_b\!\left(1 + \dfrac{y}{x}\right)$. That identity is the seed of `log1p` and `logaddexp` (notebook 03/04).

Note also that $(\log_b x)^2$ and $\log_b(x^2)$ are different things. Write $(\log_b x)^2$ with brackets; $\log_b x^2$ conventionally means $\log_b(x^2)$.

### 2.6 Mind the domain

Each law needs **both sides** to make sense. The product law requires $x > 0$ and $y > 0$:

- $\log_{10}\big((-2)(-5)\big) = \log_{10} 10 = 1$ is fine, but $\log_{10}(-2) + \log_{10}(-5)$ is undefined.
- $\log_b(x^2)$ is defined for all $x \neq 0$, while $2\log_b x$ only for $x > 0$. The correct identity for all $x \ne 0$ is $\log_b(x^2) = 2\log_b |x|$.

When you **expand**, you may shrink the domain; when you **condense**, you may enlarge it. Both can create or lose solutions of an equation (the "extraneous root" trap of notebook 02/02). This is also why SymPy refuses to expand $\ln(xy)$ unless it knows $x, y > 0$ or is told `force=True`.

### 2.7 The picture: lengths add

On a logarithmic ruler (notebook 01, section 5.2), the number $x$ sits at distance $\log_{10} x$ from $1$. The product law says: **to multiply, lay two lengths end to end**. The segment from $1$ to $2$ plus the segment from $1$ to $3$ reaches $6$. The power law says: $x^3$ is three copies of the segment for $x$. This is precisely how a slide rule works (notebook 03/02).

### 2.8 Why only logarithms do this (remark)

Any function $f$ on $(0,\infty)$ with $f(xy) = f(x) + f(y)$ satisfies $f(1) = 0$, $f(1/x) = -f(x)$ and $f(x^n) = n f(x)$ — exactly like a logarithm. If $f$ is also continuous (and not identically zero), it must be $f(x) = \log_b x$ for some base $b$. So the product law is not just *a* property of logarithms; it essentially **characterizes** them. The choice of base only rescales $f$ — that is notebook 04.

## 3. Worked Examples

### Example 1 — expanding

$$
\log_2 \frac{8x^3}{\sqrt y} = \log_2 8 + \log_2 x^3 - \log_2 y^{1/2} = 3 + 3\log_2 x - \tfrac12 \log_2 y \qquad (x, y > 0).
$$

### Example 2 — condensing

$$
2\ln a + \tfrac13 \ln b - \ln c = \ln a^2 + \ln b^{1/3} - \ln c = \ln \frac{a^2\, \sqrt[3]{b}}{c} \qquad (a, b, c > 0).
$$

### Example 3 — exact values without a calculator

- $\log_6 4 + \log_6 9 = \log_6 36 = 2$.
- $\log_3 54 - \log_3 2 = \log_3 27 = 3$.
- $\log_{10} 5 + \log_{10} 20 = \log_{10} 100 = 2$.
- $\tfrac12 \log_{10} 10^{6} = \log_{10} \sqrt{10^6} = \log_{10} 10^3 = 3$.

### Example 4 — $2^{1000}$ versus $3^{630}$

By the power law, $\log_{10} 2^{1000} = 1000 \times 0.30103 = 301.03$ and $\log_{10} 3^{630} = 630 \times 0.47712 = 300.59$. The logarithm is increasing, so the bigger log belongs to the bigger number: $2^{1000} > 3^{630}$. The ratio is $10^{301.03 - 300.59} = 10^{0.44} \approx 2.8$.

### Example 5 — doubling at $5\%$ per year

Solve $1.05^n = 2$. Take $\log_{10}$ of both sides (both are positive) and use the power law:

$$
n \log_{10} 1.05 = \log_{10} 2 \quad\Longrightarrow\quad n = \frac{\log_{10} 2}{\log_{10} 1.05} = \frac{0.30103}{0.021189} \approx 14.21 .
$$

So the money doubles during the 15th year. (The "rule of 72" estimate $72 / 5 = 14.4$ is close — module 04 explains why.)

### Example 6 — how many digits has $\binom{100}{50}$?

$$
\log_{10} \binom{100}{50} = \sum_{k=51}^{100} \log_{10} k \;-\; \sum_{k=1}^{50} \log_{10} k \approx 29.0039 ,
$$

using the product and quotient laws to replace one enormous division by $100$ small logs. A positive integer with $\log_{10} N \approx 29.0$ has $\lfloor 29.0039 \rfloor + 1 = 30$ digits, and its leading digits are $10^{0.0039} \approx 1.009$, so $\binom{100}{50} \approx 1.009 \times 10^{29}$.

## 4. Python Examples

### 4.1 Checking the three laws numerically

Random positive inputs, every law checked with a tolerance.

```python
import math
import random

random.seed(1)
all_close = True
for _ in range(10_000):
    x = random.uniform(1e-6, 1e6)
    y = random.uniform(1e-6, 1e6)
    k = random.uniform(-10, 10)
    base = random.choice([2, math.e, 10, 1.07, 0.5])
    product_ok = math.isclose(math.log(x * y, base), math.log(x, base) + math.log(y, base), abs_tol=1e-9)
    quotient_ok = math.isclose(math.log(x / y, base), math.log(x, base) - math.log(y, base), abs_tol=1e-9)
    power_ok = math.isclose(math.log(x**k, base), k * math.log(x, base), rel_tol=1e-9, abs_tol=1e-9)
    all_close = all_close and product_ok and quotient_ok and power_ok

print("product, quotient and power laws hold on 10,000 random cases:", all_close)
```

### 4.2 The false laws, numerically

Each row compares the left side with the tempting right side. None agree.

```python
import math

x, y = 1000.0, 10.0
log = math.log10
false_laws = [
    ("log(x + y)  vs log x + log y", log(x + y), log(x) + log(y)),
    ("log(x * y)  vs log x * log y", log(x * y), log(x) * log(y)),
    ("log(x / y)  vs log x / log y", log(x / y), log(x) / log(y)),
    ("log(x)**2   vs 2 log x      ", log(x) ** 2, 2 * log(x)),
    ("log(x - y)  vs log x - log y", log(x - y), log(x) - log(y)),
]
print(f"x = {x}, y = {y}, base 10")
for description, left, right in false_laws:
    print(f"{description}: {left:8.4f} vs {right:8.4f}  equal? {math.isclose(left, right)}")

# The only honest rewrite of a sum: factor out the bigger term
print()
print("log(x + y)              =", log(x + y))
print("log x + log(1 + y / x)  =", log(x) + log(1 + y / x))
```

### 4.3 The three introduction problems

`math.comb` gives $\binom{100}{50}$ exactly, so we can check the log-sum against the true digit count.

```python
import math

# (1) 2^1000 versus 3^630
log_two_power = 1000 * math.log10(2)
log_three_power = 630 * math.log10(3)
print(f"(1) log10(2^1000) = {log_two_power:.4f}, log10(3^630) = {log_three_power:.4f}")
print(f"    ratio 2^1000 / 3^630 ~ {10 ** (log_two_power - log_three_power):.4f}  (exact check: {2**1000 / 3**630:.4f})")

# (2) digits of C(100, 50) from a sum of logs
log_binomial = sum(math.log10(k) for k in range(51, 101)) - sum(math.log10(k) for k in range(1, 51))
print(f"(2) log10 C(100,50) = {log_binomial:.4f} -> {math.floor(log_binomial) + 1} digits, leading part {10 ** (log_binomial % 1):.4f}")
exact = math.comb(100, 50)
print(f"    exact: {exact}  ({len(str(exact))} digits)")

# (3) doubling time at 5 percent
years = math.log10(2) / math.log10(1.05)
print(f"(3) 1.05^n = 2  ->  n = {years:.4f} years;  check 1.05**n = {1.05**years:.12f}")
```

### 4.4 Expanding and condensing with SymPy

`sp.expand_log` applies the laws left to right (one log → many), `sp.logcombine` right to left (many → one). Both act only when the result is valid: with plain symbols SymPy does not know $x > 0$, so it leaves the expression alone unless the symbols are declared positive or `force=True` is passed.

```python
import sympy as sp

x_any, y_any = sp.symbols("x y")                    # no assumptions
x, y = sp.symbols("x y", positive=True)
a, b, c = sp.symbols("a b c", positive=True)

print("no assumptions:      ", sp.expand_log(sp.log(x_any * y_any)))
print("force=True:          ", sp.expand_log(sp.log(x_any * y_any), force=True))
print("positive symbols:    ", sp.expand_log(sp.log(x * y)))
print()

# Example 1 (base 2): sympy writes log_2 as log(.)/log(2)
expression = sp.log(8 * x**3 / sp.sqrt(y), 2)
expanded = sp.expand_log(expression)
print("Example 1 expanded:  ", sp.expand(expanded))
print("  equals 3 + 3 log2 x - log2(y)/2 ?",
      sp.simplify(expanded - (3 + 3 * sp.log(x, 2) - sp.log(y, 2) / 2)) == 0)

# Example 2: condense
condensed = sp.logcombine(2 * sp.log(a) + sp.log(b) / 3 - sp.log(c))
print("Example 2 condensed: ", condensed)

# The domain subtlety: log(x^2) for a real (possibly negative) x
x_real = sp.symbols("x", real=True)
print()
print("log(x**2), x real:   ", sp.expand_log(sp.log(x_real**2)))
print("with force=True:     ", sp.expand_log(sp.log(x_real**2), force=True), " <- wrong for x < 0")
```

Note the last two lines: `force=True` makes SymPy apply the law blindly and produce $2\ln x$, which is undefined for negative $x$ even though $\ln(x^2)$ is fine there. Use `force=True` only when you know the arguments are positive.

## 5. Visualizations

### 5.1 The product law on a log ruler: lengths add

The blue segment from $1$ to $2$ has length $\log_{10} 2$, the orange one from $1$ to $3$ has length $\log_{10} 3$. Laid end to end they reach exactly $6$.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(11, 3))
ax.hlines(0, 0, 1, color="black")
for value in range(1, 11):
    position = np.log10(value)
    ax.vlines(position, 0, 0.12, color="black")
    ax.text(position, 0.17, str(value), ha="center", fontsize=10)

length_two = np.log10(2)
length_three = np.log10(3)
ax.annotate("", xy=(length_two, -0.12), xytext=(0, -0.12), arrowprops=dict(arrowstyle="<->", color="tab:blue", lw=2))
ax.text(length_two / 2, -0.25, "log10 2", ha="center", color="tab:blue")
ax.annotate("", xy=(length_two + length_three, -0.12), xytext=(length_two, -0.12),
            arrowprops=dict(arrowstyle="<->", color="tab:orange", lw=2))
ax.text(length_two + length_three / 2, -0.25, "log10 3", ha="center", color="tab:orange")
ax.vlines(np.log10(6), -0.4, 0.12, color="crimson", linestyle="--")
ax.text(np.log10(6), -0.48, "lands on 6 = 2 x 3", ha="center", color="crimson")

ax.set_xlim(-0.05, 1.05)
ax.set_ylim(-0.6, 0.35)
ax.axis("off")
ax.set_title("Product law: log10 2 + log10 3 = log10 6 (position on the ruler = log10 x)")
plt.show()
```

### 5.2 Product law true, sum "law" false

Fix $y = 4$ and vary $x$. The curves $\ln(4x)$ and $\ln x + \ln 4$ coincide; the curve $\ln(x + 4)$ is a different function entirely. The gap $\ln x + \ln 4 - \ln(x + 4)$ is not even of constant sign: it vanishes only at $x = 4/3$, where $4x = x + 4$.

```python
import matplotlib.pyplot as plt
import numpy as np

x_values = np.linspace(0.05, 10, 400)
y_fixed = 4.0

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(x_values, np.log(x_values * y_fixed), color="tab:blue", linewidth=4, alpha=0.5, label=r"$\ln(4x)$")
ax.plot(x_values, np.log(x_values) + np.log(y_fixed), "k--", label=r"$\ln x + \ln 4$")
ax.plot(x_values, np.log(x_values + y_fixed), color="crimson", label=r"$\ln(x + 4)$  (not equal!)")
ax.plot(4 / 3, np.log(16 / 3), "o", color="crimson")
ax.annotate("only crossing: x = 4/3", (4 / 3, np.log(16 / 3)), textcoords="offset points", xytext=(15, -25))
ax.set_xlabel("x (linear scale)")
ax.set_ylabel("value")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower right")
ax.set_title("The log of a product splits; the log of a sum does not")
plt.show()
```

### 5.3 The power law on a log–log plot

On log–log axes, the power function $x^k$ becomes a straight line of slope $k$, because $\log_{10}(x^k) = k \log_{10} x$. The power law turns "raise to the $k$" into "multiply the height by $k$" — the basis of every log–log plot (notebook 02/03).

```python
import matplotlib.pyplot as plt
import numpy as np

x_values = np.logspace(-1, 2, 200)
fig, ax = plt.subplots(figsize=(8, 6))
for exponent, colour in [(0.5, "tab:green"), (1, "tab:blue"), (2, "tab:orange"), (3, "tab:red"), (-1, "tab:purple")]:
    ax.plot(x_values, x_values**exponent, color=colour, label=f"$x^{{{exponent:g}}}$  (slope {exponent:g})")
ax.set_xscale("log")
ax.set_yscale("log")
ax.grid(True, which="both", alpha=0.3)
ax.set_xlabel("x (log scale)")
ax.set_ylabel(r"$x^k$ (log scale)")
ax.legend()
ax.set_title(r"Power law: $\log_{10}(x^k) = k \log_{10} x$, so powers are straight lines")
plt.show()
```

## 8. Conclusion

- The three laws are the exponent laws in log form, each proved in one line by naming $m = \log_b x$, $n = \log_b y$:
  $\log_b(xy) = \log_b x + \log_b y$, $\;\log_b(x/y) = \log_b x - \log_b y$, $\;\log_b(x^k) = k\log_b x$.
- Consequences: $\log_b(1/x) = -\log_b x$ and $\log_b \sqrt[q]{x} = \tfrac1q \log_b x$; together the laws drop every operation one level down the hierarchy.
- There is **no** law for $\log_b(x \pm y)$. The product of logs, the quotient of logs and $(\log_b x)^k$ are not what they look like.
- Domains matter: the laws need positive arguments, $\log_b(x^2) = 2\log_b|x|$, and condensing can introduce extraneous solutions. SymPy's `expand_log`/`logcombine` only act when positivity is known, or when forced.
- Geometrically, multiplying is laying lengths end to end on a log ruler, and powers become straight lines of slope $k$ on log–log axes. Practically, the laws let us compare $2^{1000}$ with $3^{630}$, count the digits of $\binom{100}{50}$, and bring an unknown exponent down to solve $1.05^n = 2$.
