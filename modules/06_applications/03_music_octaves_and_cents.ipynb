# Music: Octaves, Equal Temperament and Cents

This notebook shows that musical pitch is a logarithmic scale. The ear hears **ratios** of frequency as intervals, so the logarithm (base $2$, because the octave is a doubling) is the natural ruler for music. With it we can explain the twelve keys of the piano octave, the spacing of guitar frets, the **cent** as a unit of pitch, and why tuning is always a compromise.

This is notebook 03 of module 06 (**Applications**) in the **Learning Logarithms** series.

## 1. Introduction

Two puzzles that every musician has seen:

1. **Why does a piano have 12 keys per octave** (7 white, 5 black)? Why not 10, which would be tidier, or 7?
2. **Why are the frets on a guitar closer together further up the neck?** A ruler has equally spaced marks; a fretboard does not.

Both have the same answer. The note A that orchestras tune to vibrates at $440$ Hz. The A an octave higher is $880$ Hz, the next $1760$ Hz; the A an octave lower is $220$ Hz. To the ear these are equally spaced steps — each "one octave" — but in hertz the steps are $220$, $440$, $880$: each step **doubles** the frequency.

So musical pitch is a multiplicative scale, heard additively — exactly what a logarithm converts. Once we measure intervals with $\log_2$ of the frequency ratio:

- dividing the octave into equal steps means equal **ratios**, $2^{1/12}$ each, which forces the frets into a geometric pattern;
- the number $12$ turns out to come from the best rational approximations of $\log_2 (3/2)$, a fact about logarithms and continued fractions.

## 2. Theory

### 2.1 Intervals are ratios; size is $\log_2$

Two notes with frequencies $f_1 < f_2$ form an **interval**. What the ear hears is (to a good approximation, over the middle of the hearing range) the **ratio** $f_2/f_1$, not the difference. The octave is the ratio $2$. A natural measure of interval size is the number of octaves:

$$
\text{size in octaves} = \log_2\frac{f_2}{f_1} \quad\Longleftrightarrow\quad \frac{f_2}{f_1} = 2^{\text{size}}.
$$

Stacking intervals multiplies their ratios and **adds** their sizes: a fifth ($3/2$) plus a fourth ($4/3$) is $\tfrac32 \cdot \tfrac43 = 2$, one octave, and indeed $\log_2\tfrac32 + \log_2\tfrac43 = \log_2 2 = 1$.

### 2.2 Equal temperament

**Twelve-tone equal temperament** (12-TET) divides the octave into $12$ equal steps (semitones). Equal steps on a log scale are equal **ratios**, so each semitone multiplies the frequency by

$$
s = 2^{1/12} \approx 1.059463.
$$

Twelve of them make $s^{12} = 2$. Numbering the notes as in the MIDI standard, with A4 $= 440$ Hz as note $69$,

$$
f(n) = 440 \cdot 2^{(n - 69)/12} \quad\Longleftrightarrow\quad n = 69 + 12 \log_2\frac{f}{440}.
$$

The note number is a logarithm of the frequency. Middle C is note $60$: $f = 440 \cdot 2^{-9/12} \approx 261.63$ Hz. A standard 88-key piano runs from A0 ($27.5$ Hz, note $21$) to C8 ($\approx 4186$ Hz, note $108$).

The value of $2^{1/12}$ was computed in the 1580s, independently by Zhu Zaiyu in China and Simon Stevin in the Netherlands, although equal temperament took a long time to become the standard tuning.

### 2.3 Cents

To compare small differences of pitch, the octave is divided into $1200$ **cents** (Alexander Ellis, 1880s), $100$ per equal-tempered semitone:

$$
c = 1200 \log_2\frac{f_2}{f_1} \quad\Longleftrightarrow\quad \frac{f_2}{f_1} = 2^{c/1200}.
$$

Cents add when intervals stack. One cent is the ratio $2^{1/1200} \approx 1.000578$, a change of about $0.06\%$.

### 2.4 Just intervals and the cost of equal temperament

A string or air column vibrates at a fundamental $f$ and at its **harmonics** $2f, 3f, 4f, 5f, \dots$. Two notes sound smooth together when their harmonics line up, which happens when their ratio is a simple fraction. These **just** intervals are:

| interval | just ratio | cents (just) | 12-TET semitones | cents (12-TET) | error |
|---|---|---|---|---|---|
| octave | $2/1$ | $1200.00$ | $12$ | $1200$ | $0$ |
| perfect fifth | $3/2$ | $701.96$ | $7$ | $700$ | $-1.96$ |
| perfect fourth | $4/3$ | $498.04$ | $5$ | $500$ | $+1.96$ |
| major third | $5/4$ | $386.31$ | $4$ | $400$ | $+13.69$ |
| minor third | $6/5$ | $315.64$ | $3$ | $300$ | $-15.64$ |
| major sixth | $5/3$ | $884.36$ | $9$ | $900$ | $+15.64$ |

The fifths and fourths of 12-TET are almost perfect; the thirds and sixths are noticeably off. Equal temperament accepts these errors in exchange for being able to play in every key equally well.

### 2.5 Frets: equal ratios mean geometric spacing

For a string at fixed tension the frequency is inversely proportional to the vibrating length: $f \propto 1/L$. To raise the pitch by $n$ semitones the vibrating length must shrink to $L \cdot 2^{-n/12}$. So fret $n$ sits at distance

$$
d_n = L\left(1 - 2^{-n/12}\right)
$$

from the nut. The 12th fret is at $L/2$ (the octave is at the middle of the string). The gap between consecutive frets is

$$
d_n - d_{n-1} = L\,2^{-n/12}\left(2^{1/12} - 1\right),
$$

which shrinks by the same factor $2^{-1/12} \approx 0.944$ from each fret to the next: a **geometric** sequence. On a log scale of string length the frets are equally spaced. Luthiers' traditional "rule of 18" divides the remaining length by about $18$ to place the next fret; the exact divisor is $1/(1 - 2^{-1/12}) \approx 17.817$.

### 2.6 Why 12?

Start on a note and keep going up by just fifths ($\times 3/2$). If after $k$ fifths we landed exactly $m$ octaves higher, we would have a closed cycle of $k$ notes:

$$
\left(\tfrac32\right)^{k} = 2^{m} \quad\Longleftrightarrow\quad k \log_2\tfrac32 = m \quad\Longleftrightarrow\quad \log_2\tfrac32 = \frac{m}{k}.
$$

That is impossible: $3^k = 2^{m+k}$ would make an odd number equal an even one. So $\log_2\frac32 \approx 0.5849625$ is **irrational**, and no cycle of pure fifths ever closes. The best we can do is approximate it by fractions $m/k$, and the best approximations come from its **continued fraction**:

$$
\log_2\tfrac32 = [0; 1, 1, 2, 2, 3, 1, 5, \dots], \qquad
\text{convergents } \frac11, \frac12, \frac35, \frac{7}{12}, \frac{24}{41}, \frac{31}{53}, \dots
$$

The convergent $3/5$ is the five-note (pentatonic) scale. The convergent $\mathbf{7/12}$ says that $12$ fifths are almost exactly $7$ octaves: $(3/2)^{12} \approx 129.75$ versus $2^7 = 128$. The mismatch, $(3/2)^{12}/2^7 \approx 1.01364$ or about $23.46$ cents, is the **Pythagorean comma**. Equal temperament spreads it evenly over the $12$ fifths, shrinking each by $23.46/12 \approx 1.96$ cents — the error in the table above. The next good divisions are $41$ and $53$ notes per octave, far too many keys for a practical keyboard. Twelve is the sweet spot, and as a bonus its thirds are tolerable too.

## 3. Worked Examples

### Example 1 — three semitones above A4

C5 is $3$ semitones above A4: $f = 440 \cdot 2^{3/12} \approx 440 \times 1.18921 \approx 523.25$ Hz.

### Example 2 — how far apart are 440 Hz and 450 Hz?

$c = 1200 \log_2(450/440) = 1200 \log_2 1.02273 \approx 38.9$ cents — well under half a semitone, but clearly audible as out of tune.

### Example 3 — the equal-tempered fifth

The fifth above A4 in 12-TET is E5, $440 \cdot 2^{7/12} \approx 659.26$ Hz. The just fifth would be $440 \times 3/2 = 660$ Hz. The gap is $1200 \log_2(660/659.26) \approx 1.96$ cents.

### Example 4 — fret positions on a 650 mm scale

- Fret 1: $650\,(1 - 2^{-1/12}) \approx 650 \times 0.05613 \approx 36.48$ mm from the nut.
- Fret 7: $650\,(1 - 2^{-7/12}) \approx 650 \times 0.33258 \approx 216.18$ mm.
- Fret 12: $650\,(1 - 2^{-1}) = 325$ mm, exactly half way.

### Example 5 — the Pythagorean comma

$(3/2)^{12} = 531441/4096 \approx 129.746$, while $2^7 = 128$. Ratio $531441/524288 \approx 1.013643$, which is $1200 \log_2 1.013643 \approx 23.46$ cents.

### Example 6 — note number from frequency

A tone of $300$ Hz: $n = 69 + 12 \log_2(300/440) \approx 69 - 6.63 = 62.37$. It lies about $37$ cents above D4 (note $62$, $293.66$ Hz).

## 4. Python Examples

### 4.1 The equal-tempered octave

```python
import math

NOTE_NAMES = ["C", "C#", "D", "D#", "E", "F", "F#", "G", "G#", "A", "A#", "B"]


def note_frequency(note_number, a4=440.0):
    return a4 * 2 ** ((note_number - 69) / 12)


def note_name(note_number):
    octave = note_number // 12 - 1
    return f"{NOTE_NAMES[note_number % 12]}{octave}"


semitone = 2 ** (1 / 12)
print("semitone ratio 2**(1/12) =", round(semitone, 6), "   twelve of them:", round(semitone**12, 12))
print()
for note_number in range(60, 73):
    print(f"{note_name(note_number):>4}  note {note_number}  {note_frequency(note_number):8.2f} Hz")
print()
print("piano range:", note_name(21), round(note_frequency(21), 2), "Hz to", note_name(108), round(note_frequency(108), 2), "Hz")
```

### 4.2 Cents, and note numbers from frequencies

```python
import math


def cents(frequency_low, frequency_high):
    return 1200 * math.log2(frequency_high / frequency_low)


def note_number_from_frequency(frequency, a4=440.0):
    return 69 + 12 * math.log2(frequency / a4)


print("440 -> 450 Hz:", round(cents(440, 450), 2), "cents")
print("A4 -> C5     :", round(440 * 2 ** (3 / 12), 2), "Hz")
print("300 Hz is note", round(note_number_from_frequency(300), 3))
print("one cent is the ratio", round(2 ** (1 / 1200), 6))
print("fifth + fourth =", round(cents(1, 3 / 2) + cents(1, 4 / 3), 6), "cents")
```

### 4.3 Just intervals versus 12-TET

```python
import math
from fractions import Fraction

just_intervals = [
    ("octave", Fraction(2, 1), 12),
    ("perfect fifth", Fraction(3, 2), 7),
    ("perfect fourth", Fraction(4, 3), 5),
    ("major third", Fraction(5, 4), 4),
    ("minor third", Fraction(6, 5), 3),
    ("major sixth", Fraction(5, 3), 9),
]

print(f"{'interval':>15} {'ratio':>6} {'just cents':>11} {'12-TET':>7} {'error':>8}")
for name, ratio, semitones in just_intervals:
    just_cents = 1200 * math.log2(ratio)
    tempered_cents = 100 * semitones
    print(f"{name:>15} {str(ratio):>6} {just_cents:>11.2f} {tempered_cents:>7} {tempered_cents - just_cents:>+8.2f}")

print()
print("A4 fifth: 12-TET", round(440 * 2 ** (7 / 12), 2), "Hz, just", 440 * 3 / 2, "Hz")
```

### 4.4 Fret positions

```python
scale_length = 650.0   # mm, typical classical guitar

previous_position = 0.0
print(f"{'fret':>4} {'from nut (mm)':>14} {'gap (mm)':>9} {'gap ratio':>10}")
previous_gap = None
for fret in range(1, 13):
    position = scale_length * (1 - 2 ** (-fret / 12))
    gap = position - previous_position
    ratio_text = "" if previous_gap is None else f"{gap / previous_gap:.4f}"
    print(f"{fret:>4} {position:>14.2f} {gap:>9.2f} {ratio_text:>10}")
    previous_position = position
    previous_gap = gap

print()
print("rule of 18, exact divisor:", round(1 / (1 - 2 ** (-1 / 12)), 3))
```

Every gap is $0.9439 = 2^{-1/12}$ times the one before: the frets form a geometric sequence, and the 12th fret lands at exactly $325$ mm.

### 4.5 Continued fraction of $\log_2(3/2)$ and the Pythagorean comma

```python
import math
from fractions import Fraction


def continued_fraction(value, terms):
    coefficients = []
    for _ in range(terms):
        whole = math.floor(value)
        coefficients.append(whole)
        fractional = value - whole
        if fractional < 1e-12:
            break
        value = 1 / fractional
    return coefficients


def convergents(coefficients):
    results = []
    numerator_prev, numerator = 1, coefficients[0]
    denominator_prev, denominator = 0, 1
    results.append(Fraction(numerator, denominator))
    for a in coefficients[1:]:
        numerator_prev, numerator = numerator, a * numerator + numerator_prev
        denominator_prev, denominator = denominator, a * denominator + denominator_prev
        results.append(Fraction(numerator, denominator))
    return results


fifth_in_octaves = math.log2(3 / 2)
coefficients = continued_fraction(fifth_in_octaves, 8)
print("log2(3/2) =", fifth_in_octaves)
print("continued fraction:", coefficients)
for fraction in convergents(coefficients):
    steps, notes_per_octave = fraction.numerator, fraction.denominator
    fifth_error = 1200 * steps / notes_per_octave - 1200 * fifth_in_octaves
    print(f"  {steps}/{notes_per_octave}: fifth = {steps} of {notes_per_octave} steps, error {fifth_error:+.3f} cents")

comma = Fraction(3, 2) ** 12 / 2**7
print()
print("Pythagorean comma:", comma, "=", round(float(comma), 6), "=", round(1200 * math.log2(comma), 3), "cents")
print("comma / 12 =", round(1200 * math.log2(comma) / 12, 3), "cents")
```

## 5. Visualizations

### 5.1 The piano keyboard on a linear and a log axis

The 88 key frequencies. On a linear axis the low notes are crushed together and the top octave takes up half the width. On a log axis (base 2) every key is the same distance from its neighbour — which is how the keyboard is actually built and how it sounds.

```python
import matplotlib.pyplot as plt
import numpy as np

note_numbers = np.arange(21, 109)
frequencies = 440 * 2.0 ** ((note_numbers - 69) / 12)
is_a = (note_numbers - 69) % 12 == 0

fig, (ax_linear, ax_log) = plt.subplots(2, 1, figsize=(12, 5))
for ax in (ax_linear, ax_log):
    ax.vlines(frequencies, 0, 1, color="tab:blue", linewidth=0.8)
    ax.vlines(frequencies[is_a], 0, 1, color="crimson", linewidth=2)
    ax.set_yticks([])

ax_linear.set_title("88 piano keys; red lines are the A's (27.5, 55, 110, ... 3520 Hz)")
ax_linear.set_xlabel("frequency in Hz (linear scale)")
ax_linear.grid(True, axis="x", alpha=0.3)

ax_log.set_xscale("log", base=2)
ax_log.set_xticks(frequencies[is_a])
ax_log.set_xticklabels([f"{f:g}" for f in frequencies[is_a]])
ax_log.set_xlabel("frequency in Hz (log scale, base 2): every octave has the same width")
ax_log.grid(True, which="both", axis="x", alpha=0.3)

plt.tight_layout()
plt.show()
```

### 5.2 The fretboard

Frets drawn to scale on a 650 mm string. The positions shrink geometrically; plotted against $\log_2$ of the remaining string length they become equally spaced.

```python
import matplotlib.pyplot as plt
import numpy as np

scale_length = 650.0
frets = np.arange(0, 20)
positions = scale_length * (1 - 2.0 ** (-frets / 12))
remaining_length = scale_length - positions

fig, (ax_board, ax_log) = plt.subplots(2, 1, figsize=(12, 5))

ax_board.add_patch(plt.Rectangle((0, 0), scale_length, 1, color="burlywood", alpha=0.6))
ax_board.vlines(positions, 0, 1, color="dimgray", linewidth=2)
for fret, position in zip(frets[1:], positions[1:]):
    if fret in (1, 3, 5, 7, 9, 12, 15, 17, 19):
        ax_board.text(position, 1.08, str(fret), ha="center", fontsize=9)
ax_board.set_xlim(-10, scale_length + 10)
ax_board.set_ylim(0, 1.3)
ax_board.set_yticks([])
ax_board.set_xlabel("distance from the nut (mm, linear scale)")
ax_board.set_title("Guitar frets on a 650 mm scale: fret 12 is exactly half way")

ax_log.plot(-np.log2(remaining_length / scale_length), np.zeros_like(frets), "|", markersize=30, color="dimgray")
ax_log.set_yticks([])
ax_log.set_xlabel("-log2(remaining length / L)  (in octaves): the frets are equally spaced, 1/12 apart")
ax_log.grid(True, axis="x", alpha=0.3)

plt.tight_layout()
plt.show()
```

### 5.3 How far is 12-TET from the just intervals?

```python
import math

import matplotlib.pyplot as plt

intervals = [("fifth 3/2", 3 / 2, 7), ("fourth 4/3", 4 / 3, 5), ("major third 5/4", 5 / 4, 4),
             ("minor third 6/5", 6 / 5, 3), ("major sixth 5/3", 5 / 3, 9), ("minor sixth 8/5", 8 / 5, 8)]
names = [name for name, _, _ in intervals]
errors = [100 * semitones - 1200 * math.log2(ratio) for _, ratio, semitones in intervals]

fig, ax = plt.subplots(figsize=(10, 5))
colours = ["tab:blue" if abs(error) < 5 else "tab:orange" for error in errors]
bars = ax.bar(names, errors, color=colours)
for bar, error in zip(bars, errors):
    offset = 0.6 if error >= 0 else -1.6
    ax.text(bar.get_x() + bar.get_width() / 2, error + offset, f"{error:+.2f}", ha="center", fontsize=10)
ax.axhline(0, color="black", linewidth=0.8)
ax.set_ylim(-20, 20)
ax.set_ylabel("12-TET minus just interval (cents)")
ax.set_title("Equal temperament: fifths and fourths nearly pure, thirds and sixths off by ~14-16 cents")
ax.grid(True, axis="y", alpha=0.3)
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

### 5.4 Which divisions of the octave give a good fifth?

For each number of notes per octave $N$, take the best available fifth (the nearest multiple of $1200/N$ cents to $701.96$) and plot its error. The deep dips are at $N = 12, 41, 53$ — denominators of convergents of $\log_2(3/2)$ — and at multiples of them ($24, 36, \dots$ simply repeat 12-TET). The convergent $N = 5$ (pentatonic) is marked too: its fifth is still $18$ cents off. $12$ is the first $N$ with an error below $2$ cents; the next improvement needs $N = 29$ ($1.49$ cents) or $N = 41$ ($0.48$ cents).

```python
import math

import matplotlib.pyplot as plt
import numpy as np

just_fifth_cents = 1200 * math.log2(3 / 2)
divisions = np.arange(1, 61)
best_errors = []
for notes_per_octave in divisions:
    step = 1200 / notes_per_octave
    best_steps = round(just_fifth_cents / step)
    best_errors.append(abs(best_steps * step - just_fifth_cents))
best_errors = np.array(best_errors)

fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(divisions, best_errors, "o-", markersize=4, color="tab:blue")
for highlight in [5, 12, 29, 41, 53]:
    ax.plot(highlight, best_errors[highlight - 1], "o", color="crimson", markersize=8)
    ax.annotate(f"N={highlight}\n{best_errors[highlight - 1]:.2f} c", xy=(highlight, best_errors[highlight - 1]),
                xytext=(0, -32), textcoords="offset points", ha="center", fontsize=9,
                bbox=dict(boxstyle="round", facecolor="white", edgecolor="none", alpha=0.85))
ax.set_yscale("log")
ax.set_ylim(0.01, 800)
ax.set_xlabel("notes per octave N")
ax.set_ylabel("error of best fifth (cents, log scale)")
ax.set_title("Equal divisions of the octave: how well they approximate the fifth 3/2")
ax.grid(True, which="both", alpha=0.3)
plt.show()
```

## 7. Solutions

1. $440 \cdot 2^{7/12} \approx 659.26$ Hz, the E5 a fifth above A4.
2. $1200 \log_2(4/3) \approx 498.04$ cents: a just perfect fourth.
3. $650\,(1 - 2^{-5/12}) \approx 650 \times 0.24915 \approx 163.05$ mm.
4. $27.5 \to 55 \to 110 \to 220 \to 440 \to 880 \to 1760 \to 3520$ is $7$ doublings, and $4186$ is a bit more, so "a little over $7$ octaves". Exactly: $\log_2(4186.01/27.5) \approx 7.25$, which equals $87/12 = 7.25$ — the piano spans $87$ semitones.
5. Just E: $261.63 \times 5/4 \approx 327.03$ Hz. The tempered E is $329.63 - 327.03 \approx 2.6$ Hz higher, which is $1200 \log_2(329.63/327.03) \approx 13.69$ cents.
6. The gap $g_n = d_n - d_{n-1} = L\,2^{-n/12}(2^{1/12} - 1)$, so $g_{n+1}/g_n = 2^{-1/12} \approx 0.9439$. Also the remaining length after fret $n-1$ is $L\,2^{-(n-1)/12}$, and $g_n$ is that length times $1 - 2^{-1/12}$, i.e. the remaining length divided by $1/(1 - 2^{-1/12}) \approx 17.817$. Dividing by $18$ is a convenient approximation.
7. Suppose $\log_2 3 = p/q$ with positive integers $p, q$. Then $3 = 2^{p/q}$, so $3^q = 2^p$. The left side is odd and the right side is even — contradiction. If $k$ just fifths equalled $m$ octaves, then $\log_2(3/2) = \log_2 3 - 1 = m/k$ would be rational, and so would $\log_2 3$.
8. One step of 53-TET is $1200/53 \approx 22.64$ cents. Fifth: $31$ steps $\approx 701.89$ cents, error $\approx -0.07$ cents (12-TET: $-1.96$). Major third: $17$ steps $\approx 384.91$ cents, error $\approx -1.41$ cents (12-TET: $+13.69$). 53-TET is far more accurate — at the price of 53 keys per octave.
9. $\tfrac32 \cdot \tfrac43 = 2$, so their cents add to $1200$; their 12-TET versions ($700 + 500$) also add to $1200$. Hence $\text{err}(4/3) = 500 - (1200 - c_{3/2}) = -(700 - c_{3/2}) = -\text{err}(3/2)$. Likewise $\tfrac65 \cdot \tfrac53 = 2$ and $300 + 900 = 1200$, so their errors are opposite.

## 8. Conclusion

- The ear hears pitch intervals as frequency **ratios**, so interval size is a logarithm: $\log_2(f_2/f_1)$ octaves, or $1200 \log_2(f_2/f_1)$ **cents**. Stacking intervals multiplies ratios and adds cents.
- **Equal temperament** makes every semitone the ratio $2^{1/12} \approx 1.0595$; note numbers are logarithms of frequency: $n = 69 + 12 \log_2(f/440)$, $f = 440 \cdot 2^{(n-69)/12}$.
- Just intervals ($3/2$, $4/3$, $5/4$, $6/5$) come from the harmonic series. 12-TET misses the fifth by only $1.96$ cents but the thirds and sixths by $14$–$16$ cents.
- **Frets** sit at $L(1 - 2^{-n/12})$: equal ratios of string length, hence geometric spacing on a ruler and equal spacing on a log scale.
- **Why 12:** no stack of pure fifths closes an octave cycle because $\log_2 3$ is irrational; the continued fraction of $\log_2(3/2)$ has convergents $3/5, 7/12, 24/41, 31/53$, and $7/12$ is the first that is both accurate (Pythagorean comma $23.46$ cents, spread as $1.96$ cents per fifth) and small enough for a keyboard.
