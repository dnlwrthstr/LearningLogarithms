# Decibels, Richter and pH: Logarithmic Perception

This notebook explains why so many everyday scales — loudness in **decibels**, earthquake **magnitudes**, acidity as **pH**, the brightness of stars in **magnitudes** — are logarithms. The quantities they measure span many powers of ten, and our senses respond to **ratios**, not differences. A logarithm turns both facts into a convenient number between roughly $0$ and $15$.

This is notebook 01 of module 06 (**Applications**) in the **Learning Logarithms** series.

## 1. Introduction

The quietest sound a healthy young ear can detect carries an intensity of about

$$
I_0 = 10^{-12}\ \text{W/m}^2,
$$

and sound becomes painful at roughly $1\ \text{W/m}^2$. Between the two lies a factor of $10^{12}$ — a million million. Try drawing that on a ruler: if the threshold of hearing is $1$ mm from zero, the threshold of pain is $10^{9}$ m away, more than twice the distance to the Moon. A whisper, a conversation and a jet engine cannot be shown on the same linear axis at all.

And yet the ear handles the whole range comfortably, and it does something curious: going from $1$ to $2$ violins feels like a big change, going from $50$ to $51$ violins feels like none. Both add one violin; only the first doubles the sound. The ear notices **relative** change.

Both observations point to the logarithm:

- it compresses $12$ powers of ten into the numbers $0$ to $12$ (or $0$ to $120$ decibels), and
- it turns equal **ratios** into equal **steps**, which matches how we perceive.

The same reasoning produced the Richter scale for earthquakes, the pH scale for acidity and the magnitude scale for stars.

## 2. Theory

### 2.1 The Weber–Fechner law

In the 1830s Ernst Heinrich Weber measured the smallest change in a weight that people could notice. He found that the **just noticeable difference** $\Delta I$ is roughly proportional to the stimulus $I$ itself:

$$
\frac{\Delta I}{I} \approx \text{constant}.
$$

Adding $1$ g to $100$ g is about as noticeable as adding $10$ g to $1000$ g. Gustav Fechner (1860) turned this into a law about perceived magnitude $S$: if every just-noticeable step of **relative** size adds the same amount to the sensation, then

$$
dS = k\,\frac{dI}{I} \quad\Longrightarrow\quad S = k \ln\frac{I}{I_0},
$$

where $I_0$ is the threshold at which sensation starts. Sensation grows with the **logarithm** of the stimulus.

The law is an approximation — it holds well in the middle of the range and less well near the thresholds, and later work (Stevens' power law) fits some senses better. But it explains why engineers chose logarithmic units: one unit of the scale is one fixed **ratio** of the physical quantity, and that is roughly one fixed step of perception.

### 2.2 Decibels

The **bel** (named after Alexander Graham Bell) is the base-10 logarithm of a power or intensity ratio; the **decibel** is a tenth of a bel. The sound intensity level is

$$
L = 10 \log_{10}\frac{I}{I_0}\ \text{dB}, \qquad I_0 = 10^{-12}\ \text{W/m}^2.
$$

Exponent form and log form, side by side:

$$
\frac{I}{I_0} = 10^{L/10} \quad\Longleftrightarrow\quad L = 10 \log_{10}\frac{I}{I_0}.
$$

So $0$ dB is the threshold of hearing (not silence), every $+10$ dB is a factor $10$ in intensity, and every $+3$ dB is almost exactly a factor $2$, because $10 \log_{10} 2 \approx 3.01$.

### 2.3 Amplitudes: why 20 instead of 10

Microphones measure **sound pressure** $p$, an amplitude, not intensity. Intensity is proportional to the square of the amplitude, $I \propto p^2$. The power law of logarithms then gives

$$
10 \log_{10}\frac{I}{I_0} = 10 \log_{10}\left(\frac{p}{p_0}\right)^2 = 20 \log_{10}\frac{p}{p_0},
$$

with reference pressure $p_0 = 20\ \mu\text{Pa}$. The same rule holds for voltages and any other amplitude: **10 for powers, 20 for amplitudes**, so that a decibel means the same thing either way. Doubling an amplitude is $20 \log_{10} 2 \approx 6.02$ dB.

### 2.4 Adding sound sources

Decibels do **not** add. Two independent (incoherent) sources add their **intensities**, so to combine levels we must leave log space, add, and come back:

$$
L_\text{total} = 10 \log_{10}\left(10^{L_1/10} + 10^{L_2/10}\right).
$$

For two equal sources $L_1 = L_2 = L$, this is $L + 10 \log_{10} 2 \approx L + 3.01$ dB. When one source is much louder, the quieter one barely matters: the increase is

$$
\Delta L = 10 \log_{10}\left(1 + 10^{-(L_1 - L_2)/10}\right),
$$

which is at most $3.01$ dB and falls quickly as the gap $L_1 - L_2$ grows. (This "leave log space, add, come back" pattern is exactly the `logaddexp` function of notebook 03/04.)

### 2.5 Earthquake magnitude

Charles Richter's 1935 **local magnitude** was defined from the largest swing $A$ on a standard seismograph, corrected for distance:

$$
M = \log_{10} A - \log_{10} A_0(\text{distance}).
$$

So **one unit of magnitude means a factor of $10$ in recorded amplitude**. Energy is a different matter. The empirical Gutenberg–Richter energy relation is approximately

$$
\log_{10} E \approx 1.5\,M + 4.8 \qquad (E \text{ in joules}),
$$

so one unit of magnitude corresponds to roughly a factor $10^{1.5} \approx 31.6$ in **energy**, and two units to roughly $10^{3} = 1000$. The constants are fitted to data, so treat these as order-of-magnitude statements, not exact laws. Modern seismology reports the **moment magnitude**, $M_w = \tfrac23\left(\log_{10} M_0 - 9.1\right)$ with the seismic moment $M_0$ in N·m, which was designed to agree with Richter's scale where both apply; its factor $\tfrac23$ builds in the same $10^{1.5}$ per unit.

### 2.6 pH

The acidity of a water solution is measured by its hydrogen-ion concentration $[\mathrm{H}^+]$ in mol/L, which ranges over about $14$ powers of ten. Søren Sørensen (1909) introduced

$$
\mathrm{pH} = -\log_{10}[\mathrm{H}^+] \quad\Longleftrightarrow\quad [\mathrm{H}^+] = 10^{-\mathrm{pH}}.
$$

The minus sign makes typical values positive. Pure water at $25^\circ$C has $[\mathrm{H}^+] = 10^{-7}$, so pH $7$. Lower pH means **more** acid, and one pH unit is a factor $10$ in concentration. (Strictly, chemists use the *activity* of the ions rather than the concentration; for dilute solutions the two agree closely.)

### 2.7 Stellar magnitudes

Ancient Greek star catalogues sorted visible stars into six brightness classes, from first (brightest) to sixth (faintest). In 1856 Norman Pogson made this precise: five magnitudes are a factor of exactly $100$ in flux $F$, so

$$
m_1 - m_2 = -2.5 \log_{10}\frac{F_1}{F_2} \quad\Longleftrightarrow\quad \frac{F_1}{F_2} = 10^{-0.4\,(m_1 - m_2)}.
$$

One magnitude is a factor $100^{1/5} \approx 2.512$. The scale runs backwards (brighter means smaller), a fossil of the ancient classes — and it, too, reflects the eye's roughly logarithmic response.

### 2.8 The common pattern

| Scale | Quantity | Formula | One unit means |
|---|---|---|---|
| sound level | intensity $I$ | $10 \log_{10}(I/I_0)$ | factor $10^{0.1} \approx 1.26$ |
| sound level | pressure $p$ | $20 \log_{10}(p/p_0)$ | factor $10^{0.05} \approx 1.12$ |
| Richter magnitude | amplitude $A$ | $\log_{10}(A/A_0)$ | factor $10$ (energy roughly $\times 31.6$) |
| pH | $[\mathrm{H}^+]$ | $-\log_{10}[\mathrm{H}^+]$ | factor $10$ (downwards) |
| stellar magnitude | flux $F$ | $-2.5 \log_{10}(F/F_0)$ | factor $2.512$ (downwards) |

Every one of them is "a fixed ratio is a fixed step".

## 3. Worked Examples

### Example 1 — intensity to decibels

A busy office has intensity $I = 10^{-5}\ \text{W/m}^2$. Then

$$
L = 10 \log_{10}\frac{10^{-5}}{10^{-12}} = 10 \log_{10} 10^{7} = 70\ \text{dB}.
$$

### Example 2 — two equal sources

Two machines each produce $70$ dB. Their intensities are each $10^{7} I_0$, so together $2 \times 10^7 I_0$:

$$
L = 10 \log_{10}(2 \times 10^7) = 70 + 10 \log_{10} 2 \approx 73.01\ \text{dB}.
$$

Doubling the number of sources adds $3$ dB, not $70$.

### Example 3 — a loud and a quieter source

$80$ dB and $70$ dB together: $10^{8} I_0 + 10^{7} I_0 = 1.1 \times 10^{8} I_0$, so

$$
L = 80 + 10 \log_{10} 1.1 \approx 80.41\ \text{dB}.
$$

The $70$ dB source adds less than half a decibel.

### Example 4 — doubling the pressure amplitude

$20 \log_{10} 2 \approx 6.02$ dB. Because intensity goes as the square of pressure, doubling the pressure quadruples the intensity, and $10 \log_{10} 4 = 6.02$ dB as well. The two formulas agree, as they must.

### Example 5 — magnitude 7 versus magnitude 5

Two units of magnitude: the seismogram amplitude is $10^2 = 100$ times larger, and the energy is roughly $10^{1.5 \times 2} = 10^3 = 1000$ times larger.

### Example 6 — pH of a solution

A solution has $[\mathrm{H}^+] = 3.2 \times 10^{-4}$ mol/L. Then

$$
\mathrm{pH} = -\log_{10}(3.2 \times 10^{-4}) = -(\log_{10} 3.2 - 4) = 4 - 0.505 \approx 3.49.
$$

The whole part ($3$) comes from the power of ten, the decimals from the mantissa $3.2$ — exactly Briggs' trick from notebook 00/01.

### Example 7 — star brightness

A first-magnitude star compared with a sixth-magnitude star: $m_1 - m_2 = -5$, so $F_1/F_2 = 10^{-0.4 \times (-5)} = 10^{2} = 100$.

## 4. Python Examples

### 4.1 Decibel conversion functions

Two small functions, one for each direction. Note that the inverse of $L = 10 \log_{10}(I/I_0)$ is $I = I_0 \cdot 10^{L/10}$.

```python
import math

REFERENCE_INTENSITY = 1e-12   # W/m^2, threshold of hearing


def intensity_to_db(intensity):
    return 10 * math.log10(intensity / REFERENCE_INTENSITY)


def db_to_intensity(level_db):
    return REFERENCE_INTENSITY * 10 ** (level_db / 10)


for intensity in [1e-12, 1e-10, 1e-5, 1e-2, 1.0]:
    print(f"I = {intensity:8.0e} W/m^2  ->  {intensity_to_db(intensity):6.1f} dB")

print()
print("70 dB is", db_to_intensity(70), "W/m^2")
print("+3 dB multiplies intensity by", round(10 ** 0.3, 4))
print("+10 dB multiplies intensity by", 10 ** 1.0)
```

### 4.2 Adding sources: leave log space, add, come back

`np.logaddexp` does the same job for natural logarithms; here we write the base-10 version explicitly.

```python
import math


def combine_levels(levels_db):
    total_relative_intensity = sum(10 ** (level / 10) for level in levels_db)
    return 10 * math.log10(total_relative_intensity)


print("70 + 70 dB         ->", round(combine_levels([70, 70]), 2), "dB")
print("80 + 70 dB         ->", round(combine_levels([80, 70]), 2), "dB")
print("ten sources of 60  ->", round(combine_levels([60] * 10), 2), "dB")
print("80 + 60 dB         ->", round(combine_levels([80, 60]), 3), "dB")
```

### 4.3 Amplitude versus intensity

A doubled pressure amplitude gives the same decibel change whether we use $20 \log_{10}$ on pressure or $10 \log_{10}$ on intensity.

```python
import math

REFERENCE_PRESSURE = 20e-6   # Pa

pressure = 0.02              # Pa, about a normal conversation
level_from_pressure = 20 * math.log10(pressure / REFERENCE_PRESSURE)
print(f"p = {pressure} Pa -> {level_from_pressure:.2f} dB SPL")

ratio = 2
print(f"pressure x{ratio}: 20*log10({ratio}) = {20 * math.log10(ratio):.3f} dB")
print(f"intensity x{ratio**2}: 10*log10({ratio**2}) = {10 * math.log10(ratio**2):.3f} dB")
print("agree:", math.isclose(20 * math.log10(ratio), 10 * math.log10(ratio**2)))

# inverse-square law: intensity ~ 1 / distance^2, so doubling distance loses 20*log10(2) dB
print(f"doubling distance from a point source: {-20 * math.log10(2):.2f} dB")
```

### 4.4 Earthquake magnitudes: amplitude and energy

Using the approximate energy relation $\log_{10} E \approx 1.5 M + 4.8$.

```python
import math


def amplitude_ratio(magnitude_1, magnitude_2):
    return 10 ** (magnitude_1 - magnitude_2)


def energy_joules(magnitude):
    return 10 ** (1.5 * magnitude + 4.8)


print(f"one unit of magnitude: energy factor ~ {10 ** 1.5:.1f}")
for big, small in [(6, 5), (7, 5), (9, 6)]:
    print(f"M{big} vs M{small}: amplitude x{amplitude_ratio(big, small):,.0f}, "
          f"energy x{energy_joules(big) / energy_joules(small):,.0f}")

print()
for magnitude in [4, 6, 8]:
    print(f"M{magnitude}: E ~ {energy_joules(magnitude):.1e} J")
```

### 4.5 pH both ways

```python
import math


def ph_from_concentration(hydrogen_concentration):
    return -math.log10(hydrogen_concentration)


def concentration_from_ph(ph):
    return 10 ** (-ph)


print("[H+] = 3.2e-4  -> pH", round(ph_from_concentration(3.2e-4), 2))
print("[H+] = 1e-7    -> pH", round(ph_from_concentration(1e-7), 2))
print("pH 2.5         -> [H+] =", f"{concentration_from_ph(2.5):.3e}")
print("pH 3 vs pH 6: acid stronger by a factor", round(concentration_from_ph(3) / concentration_from_ph(6)))
```

### 4.6 Stellar magnitudes

```python
import math


def flux_ratio(magnitude_1, magnitude_2):
    return 10 ** (-0.4 * (magnitude_1 - magnitude_2))


print("one magnitude      : factor", round(flux_ratio(0, 1), 4), "  (100**(1/5) =", round(100 ** 0.2, 4), ")")
print("magnitude 1 vs 6   : factor", round(flux_ratio(1, 6), 6))
print("magnitude -1 vs 6  : factor", round(flux_ratio(-1, 6), 1))
```

## 5. Visualizations

### 5.1 Twelve powers of ten on one axis

On a logarithmic axis, everyday sounds spread out evenly. The levels below are typical, rounded values — real sounds vary a lot. The top axis shows the same positions in decibels: the dB scale **is** the log axis, relabelled.

```python
import matplotlib.pyplot as plt
import numpy as np

sounds = [
    ("threshold of hearing", 0),
    ("rustling leaves", 20),
    ("quiet library", 40),
    ("conversation", 60),
    ("busy traffic", 80),
    ("loud concert", 110),
    ("threshold of pain", 120),
]
reference_intensity = 1e-12

fig, ax = plt.subplots(figsize=(11, 4))
for index, (name, level) in enumerate(sounds):
    intensity = reference_intensity * 10 ** (level / 10)
    ax.plot(intensity, 0, "o", color="tab:blue", markersize=8)
    height = 0.35 if index % 2 == 0 else -0.35
    ax.annotate(f"{name}\n{level} dB", xy=(intensity, 0), xytext=(intensity, height),
                ha="center", va="center", fontsize=9,
                arrowprops=dict(arrowstyle="-", color="gray"))

ax.set_xscale("log")
ax.set_xlim(1e-13, 1e1)
ax.set_ylim(-0.6, 0.6)
ax.set_yticks([])
ax.set_xlabel("sound intensity I in W/m^2 (log scale)")
ax.grid(True, which="both", axis="x", alpha=0.3)

top_axis = ax.twiny()
top_axis.set_xscale("log")
top_axis.set_xlim(ax.get_xlim())
tick_intensities = [reference_intensity * 10.0 ** (level / 10) for level in range(0, 121, 20)]
top_axis.set_xticks(tick_intensities)
top_axis.set_xticklabels([str(level) for level in range(0, 121, 20)])
top_axis.minorticks_off()
top_axis.set_xlabel("sound level in dB (the same axis, relabelled)")
ax.set_title("Sound intensities span a factor of 10^12", pad=40)
plt.tight_layout()
plt.show()
```

### 5.2 Weber–Fechner: equal ratios feel like equal steps

Left: intensities that double each time, $I_0, 2I_0, 4I_0, \dots$, plotted on a linear axis — the steps grow without bound. Right: the same intensities on a log axis, with Fechner's sensation $S = \log_2(I/I_0)$ — every doubling is the same step, which is roughly how they sound.

```python
import matplotlib.pyplot as plt
import numpy as np

doublings = np.arange(0, 11)
intensities = 2.0 ** doublings          # in units of I_0
sensation = np.log2(intensities)

fig, (ax_linear, ax_log) = plt.subplots(1, 2, figsize=(12, 4.5))

ax_linear.plot(intensities, sensation, "o-", color="tab:orange")
ax_linear.set_xlabel("intensity I / I_0 (linear scale)")
ax_linear.set_ylabel("sensation S (doublings)")
ax_linear.set_title("Linear axis: equal sensation steps look unequal")
ax_linear.grid(True, alpha=0.3)

ax_log.plot(intensities, sensation, "o-", color="tab:blue")
ax_log.set_xscale("log", base=2)
ax_log.set_xlabel("intensity I / I_0 (log scale, base 2)")
ax_log.set_ylabel("sensation S (doublings)")
ax_log.set_title("Log axis: S = log2(I / I_0) is a straight line")
ax_log.grid(True, which="both", alpha=0.3)

plt.tight_layout()
plt.show()
```

### 5.3 How much does a second source add?

The increase in total level when a second, quieter source joins, as a function of how much quieter it is. Equal sources add $3.01$ dB; a source $10$ dB quieter adds only about $0.4$ dB.

```python
import matplotlib.pyplot as plt
import numpy as np

level_gap = np.linspace(0, 25, 300)
increase = 10 * np.log10(1 + 10 ** (-level_gap / 10))

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(level_gap, increase, color="tab:blue", label="increase in total level")
for gap in [0, 10, 20]:
    value = 10 * np.log10(1 + 10 ** (-gap / 10))
    ax.plot(gap, value, "o", color="crimson")
    ax.annotate(f"{value:.2f} dB", xy=(gap, value), xytext=(gap + 0.8, value + 0.15), fontsize=10)

ax.set_ylim(-0.1, 3.4)

ax.set_xlabel("how much quieter the second source is, L1 - L2 (dB)")
ax.set_ylabel("increase over the louder source alone (dB)")
ax.set_title("Adding sources: decibels do not add")
ax.grid(True, alpha=0.3)
ax.legend()
plt.show()
```

### 5.4 pH and magnitude: two more log axes

Left: approximate pH of familiar liquids, with the hydrogen-ion concentration on a log axis. Right: approximate earthquake energy versus magnitude — a straight line on a log axis with slope $1.5$ decades per magnitude unit.

```python
import matplotlib.pyplot as plt
import numpy as np

liquids = [("stomach acid", 1.5), ("lemon juice", 2.0), ("coffee", 5.0), ("pure water", 7.0),
           ("seawater", 8.1), ("household ammonia", 11.5)]

fig, (ax_ph, ax_quake) = plt.subplots(1, 2, figsize=(13, 5))

for index, (name, ph) in enumerate(liquids):
    concentration = 10 ** (-ph)
    ax_ph.barh(index, concentration, color="tab:green", alpha=0.7)
    ax_ph.text(concentration * 1.5, index, f"pH {ph}", va="center", fontsize=9)
ax_ph.set_yticks(range(len(liquids)))
ax_ph.set_yticklabels([name for name, _ in liquids])
ax_ph.set_xscale("log")
ax_ph.set_xlim(1e-13, 1)
ax_ph.set_xlabel("[H+] in mol/L (log scale)")
ax_ph.set_title("pH = -log10[H+]")
ax_ph.grid(True, which="both", axis="x", alpha=0.3)

magnitudes = np.linspace(2, 9.5, 100)
energies = 10 ** (1.5 * magnitudes + 4.8)
ax_quake.plot(magnitudes, energies, color="tab:red")
ax_quake.set_yscale("log")
ax_quake.set_xlabel("magnitude M (linear)")
ax_quake.set_ylabel("approximate energy E in joules (log scale)")
ax_quake.set_title("log10 E ~ 1.5 M + 4.8")
ax_quake.grid(True, which="both", alpha=0.3)

plt.tight_layout()
plt.show()
```

In both panels the logarithmic axis turns a huge range into an even one. The pH bars all look comparable in length only because the axis is logarithmic: the stomach-acid bar is $10^{10}$ times longer than the ammonia bar.

## 7. Solutions

1. $I = 10^{-12} \cdot 10^{90/10} = 10^{-3}$ W/m². The difference is $30$ dB, i.e. a factor $10^{3} = 1000$.
2. pH $= -\log_{10} 10^{-9} = 9$ (basic). $[\mathrm{H}^+] = 10^{-2.5} \approx 3.16 \times 10^{-3}$ mol/L.
3. Ten equal intensities: $60 + 10 \log_{10} 10 = 70$ dB.
4. Three units: $32^3 \approx 32{,}768$, so "about thirty thousand". Exactly with the relation: $10^{4.5} \approx 31{,}623$.
5. Going from $1$ m to $8$ m is three doublings of distance, each costing $20 \log_{10} 2 \approx 6.02$ dB: $100 - 3 \times 6.02 \approx 81.94$ dB. Equivalently $100 - 20 \log_{10} 8$.
6. No. The concentrations are $10^{-2}$ and $10^{-4}$; their average is $(10^{-2} + 10^{-4})/2 = 5.05 \times 10^{-3}$, so pH $= -\log_{10}(5.05 \times 10^{-3}) \approx 2.30$. The strong acid dominates — averaging logarithms is not the same as averaging the quantities.
7. $L_\text{total} = 10 \log_{10}(10^{L_1/10} + 10^{(L_1 - d)/10}) = L_1 + 10 \log_{10}(1 + 10^{-d/10})$. Since $d \ge 0$, $10^{-d/10} \le 1$, so $\Delta L \le 10 \log_{10} 2 \approx 3.01$. Setting $\Delta L = 0.5$: $10^{-d/10} = 10^{0.05} - 1 \approx 0.122$, so $d = -10 \log_{10} 0.122 \approx 9.14$ dB.
8. Integrating $dS/dI = k/I$ gives $S = k \ln I + C$; $S(I_0) = 0$ forces $C = -k \ln I_0$, so $S = k \ln(I/I_0)$. A $1$ dB change is an intensity factor $10^{0.1} \approx 1.259$, an increase of about $26\%$.

## 8. Conclusion

- Many physical quantities span a dozen or more powers of ten, and our senses respond roughly to **ratios** (Weber's law). Fechner's integration of $dS = k\,dI/I$ gives $S = k \ln(I/I_0)$: sensation grows like a logarithm.
- **Decibels**: $L = 10 \log_{10}(I/I_0)$ for powers and intensities, $20 \log_{10}(p/p_0)$ for amplitudes, because $I \propto p^2$. $+3$ dB is $\times 2$ in intensity, $+10$ dB is $\times 10$.
- Levels of independent sources combine as $10 \log_{10}(10^{L_1/10} + 10^{L_2/10})$: leave log space, add, come back. Two equal sources give $+3.01$ dB.
- **Earthquakes**: one magnitude unit is a factor $10$ in seismogram amplitude and, by an empirical relation, roughly $10^{1.5} \approx 32$ in energy.
- **pH** $= -\log_{10}[\mathrm{H}^+]$: one unit is a factor $10$ in acidity, downwards. **Star magnitudes**: five magnitudes are a factor $100$, so $m_1 - m_2 = -2.5 \log_{10}(F_1/F_2)$.
- Geometrically, each of these scales is just a log axis with friendly labels: equal ratios become equal steps.
