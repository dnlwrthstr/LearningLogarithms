# Log Returns and Continuous Compounding

This notebook shows why finance measures growth with **logarithms**. Prices evolve by multiplication — each day's price is yesterday's price times a factor — and the logarithm turns that chain of multiplications into a sum. The result is the **log return**: it adds up over time, it treats gains and losses symmetrically, and it is exactly the rate in continuous compounding $A = Pe^{rt}$.

This is notebook 02 of module 06 (**Applications**) in the **Learning Logarithms** series.

## 1. Introduction

You invest EUR 100. In year one the investment gains $50\%$; in year two it loses $50\%$. The average return is

$$
\frac{+50\% + (-50\%)}{2} = 0\%,
$$

so you should be back where you started. You are not:

$$
100 \times 1.5 = 150, \qquad 150 \times 0.5 = 75.
$$

You have lost a quarter of your money. The trouble is that percentage returns **multiply**, but we averaged them by **adding**. A $50\%$ loss is applied to a bigger base than the $50\%$ gain was, and to undo a $50\%$ loss you would need a $100\%$ gain.

The logarithm repairs this. Measure each year by $\ln(\text{end}/\text{start})$ instead:

$$
\ln 1.5 + \ln 0.5 = 0.405 - 0.693 = -0.288 = \ln 0.75.
$$

The log returns **add up** to the log of the total growth factor — the multiplication law $\ln(xy) = \ln x + \ln y$ at work, exactly as in Napier's tables. This notebook builds that idea into the standard toolkit of quantitative finance.

## 2. Theory

### 2.1 Simple returns multiply

A price that moves from $P_0$ to $P_1$ has **simple return**

$$
R = \frac{P_1 - P_0}{P_0} = \frac{P_1}{P_0} - 1, \qquad P_1 = P_0 (1 + R).
$$

Over several periods the growth factors multiply:

$$
\frac{P_T}{P_0} = (1 + R_1)(1 + R_2)\cdots(1 + R_T).
$$

So the total simple return is **not** $R_1 + R_2 + \cdots + R_T$.

### 2.2 Log returns add

The **log return** (or continuously compounded return) is

$$
r = \ln\frac{P_1}{P_0} = \ln(1 + R) \quad\Longleftrightarrow\quad \frac{P_1}{P_0} = e^{r}, \quad R = e^r - 1.
$$

Taking $\ln$ of the product in 2.1 turns it into a sum:

$$
\ln\frac{P_T}{P_0} = r_1 + r_2 + \cdots + r_T \quad\Longleftrightarrow\quad P_T = P_0\, e^{\,r_1 + r_2 + \cdots + r_T}.
$$

This is the whole point: **log returns are additive over time.** Geometrically, plotting $\ln P_t$ instead of $P_t$ turns every percentage move into a fixed-length step, so a log-scaled price chart shows a constant growth rate as a straight line.

Log returns are also **symmetric**: a move of $+r$ followed by $-r$ returns to the start, because $e^{r} e^{-r} = 1$. Simple returns are not: $+50\%$ needs $-33.3\%$ to undo it, and $-50\%$ needs $+100\%$.

### 2.3 Small returns: $\ln(1+R) \approx R$

The series of notebook 01/04 gives

$$
\ln(1 + R) = R - \frac{R^2}{2} + \frac{R^3}{3} - \cdots
$$

For daily returns of about $1\%$ the difference is about $R^2/2 = 0.00005$, negligible. For yearly returns of $50\%$ it is not. The log return is **always smaller** than the simple return (for $R \neq 0$), because $\ln$ is concave and lies below its tangent line $y = R$ at $R = 0$.

### 2.4 Continuous compounding

An annual rate $r$ compounded $m$ times a year grows $P$ to $P(1 + r/m)^{mt}$ after $t$ years. As $m \to \infty$,

$$
\left(1 + \frac{r}{m}\right)^{mt} \longrightarrow e^{rt}, \qquad A = P e^{rt} \quad\Longleftrightarrow\quad r = \frac{1}{t}\ln\frac{A}{P}.
$$

(Taking logs shows why: $mt \ln(1 + r/m) \approx mt \cdot r/m = rt$.) So a **continuously compounded rate is exactly a log return per unit time**. Conversions:

| from | to | formula |
|---|---|---|
| continuous rate $r$ | effective annual rate $R$ | $R = e^{r} - 1$ |
| effective annual rate $R$ | continuous rate $r$ | $r = \ln(1 + R)$ |
| nominal rate $j$ compounded $m$ times | continuous rate $r$ | $r = m \ln(1 + j/m)$ |

The doubling time at continuous rate $r$ solves $e^{rT} = 2$: $T = \ln 2 / r \approx 0.693 / r$ — the "rule of 70" (the familiar rule of 72 is tuned for annual compounding at typical rates).

### 2.5 Averages: arithmetic versus geometric

The arithmetic mean of simple returns $\bar R = \frac1T \sum R_t$ **overstates** the growth that actually happened. The true per-period growth rate is the **geometric mean**

$$
\left(\prod_t (1 + R_t)\right)^{1/T} - 1 = \exp\left(\frac1T \sum_t r_t\right) - 1,
$$

which is just the arithmetic mean of the **log** returns, mapped back. For returns with mean $\mu$ and volatility $\sigma$, a second-order expansion gives the useful rule

$$
\text{mean log return} \approx \mu - \frac{\sigma^2}{2},
$$

the **volatility drag**: the more a price swings, the more its long-run growth falls short of its average return.

### 2.6 A caveat: across assets, simple returns add

Log returns add over **time**. Across **assets** in a portfolio it is the other way round: a portfolio with weights $w_i$ has simple return $R_p = \sum_i w_i R_i$ exactly, but $\ln(1 + R_p) \neq \sum_i w_i r_i$ in general. Use log returns to chain periods, simple returns to combine holdings.

## 3. Worked Examples

### Example 1 — the $+50\%$, $-50\%$ trap

Simple returns: $R_1 = 0.5$, $R_2 = -0.5$, sum $0$. Growth factor $1.5 \times 0.5 = 0.75$, total return $-25\%$.

Log returns: $r_1 = \ln 1.5 \approx 0.4055$, $r_2 = \ln 0.5 \approx -0.6931$, sum $\approx -0.2877 = \ln 0.75$. The log returns get it right.

### Example 2 — three months

Monthly returns $+10\%$, $-5\%$, $+3\%$. Multiply: $1.10 \times 0.95 \times 1.03 = 1.07635$, a total of $7.635\%$.

Log returns: $\ln 1.10 + \ln 0.95 + \ln 1.03 \approx 0.09531 - 0.05129 + 0.02956 = 0.07358$, and $e^{0.07358} \approx 1.07635$ ✓. (Adding the simple returns would give $8\%$.)

### Example 3 — compounding frequency

EUR 1000 at a nominal $5\%$ for $10$ years:

| compounding | formula | value (EUR) |
|---|---|---|
| annually | $1000 \cdot 1.05^{10}$ | $1628.89$ |
| monthly | $1000 \cdot (1 + 0.05/12)^{120}$ | $1647.01$ |
| continuously | $1000 \cdot e^{0.5}$ | $1648.72$ |

More frequent compounding helps, but with rapidly diminishing returns: the limit is $e^{rt}$.

### Example 4 — converting rates

An effective annual rate of $8\%$ corresponds to a continuous rate $r = \ln 1.08 \approx 7.696\%$. Conversely, a continuous rate of $8\%$ gives an effective annual rate $e^{0.08} - 1 \approx 8.329\%$.

### Example 5 — doubling time

At a continuous rate of $7\%$: $T = \ln 2 / 0.07 \approx 9.90$ years. The rule of 70 gives $70 / 7 = 10$.

### Example 6 — the portfolio caveat

Half the money in asset A ($+20\%$), half in B ($-20\%$). The portfolio's simple return is $0.5 \times 0.2 + 0.5 \times (-0.2) = 0$, and indeed EUR 50 becomes 60 and EUR 50 becomes 40, total 100. The weighted log returns are $0.5 \ln 1.2 + 0.5 \ln 0.8 \approx -0.0204$, which is **not** $\ln(1 + 0) = 0$.

## 4. Python Examples

### 4.1 Simple and log returns from a price list

`np.diff(np.log(prices))` is the standard one-step recipe for log returns.

```python
import numpy as np

prices = np.array([100.0, 150.0, 75.0])

simple_returns = prices[1:] / prices[:-1] - 1
log_returns = np.diff(np.log(prices))

print("simple returns :", simple_returns)
print("log returns    :", np.round(log_returns, 4))
print()
print("sum of simple returns :", simple_returns.sum())
print("product of factors - 1:", np.prod(1 + simple_returns) - 1)
print("sum of log returns    :", round(log_returns.sum(), 4))
print("ln(P_T / P_0)         :", round(np.log(prices[-1] / prices[0]), 4))
print("exp(sum of logs) - 1  :", round(np.exp(log_returns.sum()) - 1, 4))
```

### 4.2 Example 2 in code, and the conversion $r = \ln(1+R)$

`np.log1p(R)` computes $\ln(1 + R)$ accurately even for tiny $R$ (notebook 03/04), and `np.expm1(r)` computes $e^r - 1$.

```python
import numpy as np

monthly_simple = np.array([0.10, -0.05, 0.03])
monthly_log = np.log1p(monthly_simple)

print("log returns        :", np.round(monthly_log, 5))
print("sum of log returns :", round(monthly_log.sum(), 5))
print("total via logs     :", round(np.expm1(monthly_log.sum()), 5))
print("total via product  :", round(np.prod(1 + monthly_simple) - 1, 5))
print("naive sum          :", round(monthly_simple.sum(), 5))
```

### 4.3 How good is $\ln(1+R) \approx R$?

```python
import numpy as np

print(f"{'R':>8} {'ln(1+R)':>10} {'R - ln(1+R)':>12} {'R^2/2':>10}")
for simple_return in [0.001, 0.01, 0.05, 0.10, 0.50, -0.10, -0.50]:
    log_return = np.log1p(simple_return)
    print(f"{simple_return:>8.3f} {log_return:>10.5f} {simple_return - log_return:>12.6f} {simple_return**2 / 2:>10.6f}")
```

For $|R| \le 1\%$ the two agree to about five decimal places; the error is close to $R^2/2$, the first term of the series that we dropped. The error column is always positive: the log return is always the smaller of the two.

### 4.4 Compounding frequency and the limit $e^{rt}$

```python
import math

principal = 1000
rate = 0.05
years = 10

print(f"{'compounding':>14} {'periods/yr':>10} {'value':>10}")
for name, periods in [("annual", 1), ("quarterly", 4), ("monthly", 12), ("daily", 365), ("hourly", 8760)]:
    value = principal * (1 + rate / periods) ** (periods * years)
    print(f"{name:>14} {periods:>10} {value:>10.2f}")
print(f"{'continuous':>14} {'inf':>10} {principal * math.exp(rate * years):>10.2f}")

print()
print("effective 8% -> continuous", round(math.log(1.08), 5))
print("continuous 8% -> effective", round(math.expm1(0.08), 5))
print("doubling time at 7% continuous:", round(math.log(2) / 0.07, 2), "years")
```

### 4.5 A simulated price path

We simulate ten years of daily prices by multiplying by $e^{r_t}$ each day, with random log returns $r_t$ (normal, mean $0.0003$, standard deviation $0.02$ — roughly a volatile stock). Then we compare three ways of summarising the path.

```python
import numpy as np

rng = np.random.default_rng(seed=4)
trading_days = 2520                     # about 10 years
daily_log_returns = rng.normal(loc=0.0003, scale=0.02, size=trading_days)

start_price = 100.0
prices = start_price * np.exp(np.concatenate([[0.0], np.cumsum(daily_log_returns)]))
daily_simple_returns = prices[1:] / prices[:-1] - 1

total_log_growth = np.log(prices[-1] / prices[0])
print(f"final price                    : {prices[-1]:.2f}")
print(f"ln(P_T / P_0)                  : {total_log_growth:.6f}")
print(f"sum of log returns             : {daily_log_returns.sum():.6f}")
print(f"  agree:", np.isclose(total_log_growth, daily_log_returns.sum()))
print(f"sum of simple returns          : {daily_simple_returns.sum():.6f}   <- not ln(P_T/P_0), not the total return")
print(f"actual total simple return     : {prices[-1] / prices[0] - 1:.6f}")
print()
mean_simple = daily_simple_returns.mean()
variance_simple = daily_simple_returns.var()
print(f"arithmetic mean simple return  : {mean_simple:.6f}")
print(f"mean log return                : {daily_log_returns.mean():.6f}")
print(f"mean simple - variance / 2     : {mean_simple - variance_simple / 2:.6f}   <- volatility drag")
```

The sum of the log returns reproduces $\ln(P_T/P_0)$ to machine precision — it has to, because it is a telescoping sum of $\ln P_t - \ln P_{t-1}$. The sum of simple returns overshoots by roughly $T\sigma^2/2 = 2520 \times 0.0002 \approx 0.5$, and the mean log return sits close to $\mu - \sigma^2/2$.

## 5. Visualizations

### 5.1 $\ln(1+R)$ lies below its tangent line

The log return equals the simple return to first order, and is always below it.

```python
import matplotlib.pyplot as plt
import numpy as np

simple_return = np.linspace(-0.8, 1.5, 400)
log_return = np.log1p(simple_return)

fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(simple_return, log_return, color="tab:blue", label="log return r = ln(1 + R)")
ax.plot(simple_return, simple_return, "--", color="gray", label="tangent line r = R")
ax.fill_between(simple_return, log_return, simple_return, color="tab:orange", alpha=0.2, label="gap R - ln(1 + R)")
for point in [0.5, -0.5]:
    ax.plot(point, np.log1p(point), "o", color="crimson")
    ax.annotate(f"R = {point:+.0%}\nr = {np.log1p(point):+.3f}", xy=(point, np.log1p(point)),
                xytext=(point + 0.1, np.log1p(point) - 0.45), fontsize=10)

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlabel("simple return R")
ax.set_ylabel("return")
ax.set_title("Log return versus simple return")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left")
plt.show()
```

### 5.2 Compounding more often converges to $e^{r}$

One year at $r = 100\%$ (to make the effect visible) compounded $m$ times: $(1 + 1/m)^m$ creeps up to $e \approx 2.718$. The horizontal axis is logarithmic, because $m$ ranges over many powers of ten.

```python
import matplotlib.pyplot as plt
import numpy as np

periods_per_year = np.unique(np.logspace(0, 6, 60).astype(int))
growth_factor = (1 + 1 / periods_per_year) ** periods_per_year

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(periods_per_year, growth_factor, "o-", markersize=3, color="tab:blue", label="(1 + 1/m)^m")
ax.axhline(np.e, color="crimson", linestyle="--", label="e = 2.71828...")
for m, name, offset in [(1, "annual", (25, 10)), (12, "monthly", (25, -30)), (365, "daily", (25, -30))]:
    value = (1 + 1 / m) ** m
    ax.annotate(f"{name}: {value:.4f}", xy=(m, value), xytext=offset, textcoords="offset points",
                fontsize=9, arrowprops=dict(arrowstyle="->", color="gray"))
ax.set_xscale("log")
ax.set_xlabel("compounding periods per year m (log scale)")
ax.set_ylabel("growth factor after one year")
ax.set_title("Continuous compounding is the limit m -> infinity")
ax.grid(True, which="both", alpha=0.3)
ax.legend(loc="lower right")
plt.show()
```

### 5.3 The price path: linear axis, log axis, and the running sums

Top: the simulated price on a linear and on a logarithmic axis. Bottom: the running sum of log returns tracks $\ln(P_t/P_0)$ exactly (the two curves coincide), while the running sum of simple returns drifts upward, away from the truth.

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(seed=4)
trading_days = 2520
daily_log_returns = rng.normal(loc=0.0003, scale=0.02, size=trading_days)
prices = 100.0 * np.exp(np.concatenate([[0.0], np.cumsum(daily_log_returns)]))
years = np.arange(trading_days + 1) / 252

daily_simple_returns = prices[1:] / prices[:-1] - 1
running_log_sum = np.concatenate([[0.0], np.cumsum(daily_log_returns)])
running_simple_sum = np.concatenate([[0.0], np.cumsum(daily_simple_returns)])
true_log_growth = np.log(prices / prices[0])

fig, axes = plt.subplots(2, 2, figsize=(13, 8))
axes[0, 0].plot(years, prices, color="tab:blue")
axes[0, 0].set_title("price, linear axis")
axes[0, 0].set_ylabel("price (linear scale)")

axes[0, 1].plot(years, prices, color="tab:blue")
axes[0, 1].set_yscale("log")
axes[0, 1].set_title("price, log axis: equal % moves have equal height")
axes[0, 1].set_ylabel("price (log scale)")

axes[1, 0].plot(years, true_log_growth, color="black", linewidth=3, alpha=0.4, label="ln(P_t / P_0)")
axes[1, 0].plot(years, running_log_sum, color="tab:blue", linewidth=1, label="running sum of log returns")
axes[1, 0].plot(years, running_simple_sum, color="tab:orange", linewidth=1, label="running sum of simple returns")
axes[1, 0].set_title("only the log returns add up correctly")
axes[1, 0].set_ylabel("cumulative value")
axes[1, 0].legend(loc="upper left", fontsize=9)

axes[1, 1].plot(years, running_simple_sum - true_log_growth, color="tab:orange", linewidth=4, alpha=0.5,
                label="simple sum - log growth")
axes[1, 1].plot(years, 0.5 * np.cumsum(np.concatenate([[0.0], daily_simple_returns**2])), "--", color="gray",
                label="half the running sum of R_t^2")
axes[1, 1].set_title("the gap grows like sum of R_t^2 / 2")
axes[1, 1].set_ylabel("simple sum - log growth")
axes[1, 1].legend(loc="upper left", fontsize=9)

for ax in axes.flat:
    ax.set_xlabel("years")
    ax.grid(True, which="both", alpha=0.3)
plt.tight_layout()
plt.show()
```

### 5.4 The $+50\% / -50\%$ coin game

Each round a fair coin decides whether your wealth is multiplied by $1.5$ or by $0.5$. The expected simple return per round is exactly $0$, so the **average** wealth over many players stays at $1$. But the expected **log** return per round is $\tfrac12(\ln 1.5 + \ln 0.5) \approx -0.144$, and that is what a typical player experiences. On a log axis each path is a random walk with negative drift.

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(seed=1)
players = 20000
rounds = 50
factors = np.where(rng.random((players, rounds)) < 0.5, 1.5, 0.5)
wealth = np.cumprod(factors, axis=1)
wealth = np.hstack([np.ones((players, 1)), wealth])
round_index = np.arange(rounds + 1)

expected_log_return = 0.5 * (np.log(1.5) + np.log(0.5))

fig, ax = plt.subplots(figsize=(10, 5.5))
for path in wealth[:30]:
    ax.plot(round_index, path, color="gray", alpha=0.3, linewidth=0.8)
ax.plot(round_index, np.median(wealth, axis=0), color="tab:blue", linewidth=2, label="median wealth (simulated)")
ax.plot(round_index, np.exp(expected_log_return * round_index), "--", color="crimson",
        label="exp(n * mean log return)")
ax.axhline(1.0, color="black", linewidth=1, label="expected wealth = 1")
ax.set_yscale("log")
ax.set_xlabel("round n")
ax.set_ylabel("wealth (log scale)")
ax.set_title("A fair game in simple returns, a losing game in log returns")
ax.grid(True, which="both", alpha=0.3)
ax.legend(loc="lower left")
plt.show()

print("mean log return per round:", round(expected_log_return, 4))
print("simulated mean wealth after 50 rounds  :", round(wealth[:, -1].mean(), 3))
print("simulated median wealth after 50 rounds:", f"{np.median(wealth[:, -1]):.2e}")
```

The simulated mean wealth is noisy (it is carried by a handful of very lucky players), while the median zigzags tightly around $e^{-0.144\,n}$ (it alternates between even and odd rounds). Long-run growth is governed by the **mean log return**, not the mean simple return.

## 7. Solutions

1. Simple returns: $100/80 - 1 = 25\%$ and $80/100 - 1 = -20\%$ (sum $+5\%$). Log returns: $\ln 1.25 \approx 0.2231$ and $\ln 0.8 \approx -0.2231$ (sum $0$). Only the log returns are symmetric.
2. $500 e^{0.04 \times 15} = 500 e^{0.6} \approx$ EUR 911.06.
3. $e^{0.06} - 1 \approx 6.184\%$ effective; $\ln 1.06 \approx 5.827\%$ continuous.
4. Rule of 70: $70/3 \approx 23.3$ years. Exactly: $\ln 2 / 0.03 \approx 23.10$ years.
5. Log returns add: $252 \times 0.0004 = 0.1008$. Simple: $e^{0.1008} - 1 \approx 10.61\%$.
6. Each two-year cycle multiplies by $1.1 \times 0.9 = 0.99$, a log return of $\ln 0.99 \approx -0.01005$. Ten cycles: $0.99^{10} \approx 0.9044$, so about $90.4\%$ is left.
7. Let $f(R) = R - \ln(1 + R)$ on $R > -1$. Then $f(0) = 0$ and $f'(R) = 1 - \frac{1}{1 + R} = \frac{R}{1 + R}$, which is negative for $-1 < R < 0$ and positive for $R > 0$. So $f$ has its minimum $0$ at $R = 0$ and $f(R) > 0$ elsewhere. Summing $\ln(1 + R_t) \le R_t$ over $t$ gives $\sum r_t \le \sum R_t$.
8. Each round multiplies the expected wealth by $\tfrac12(1.5 + 0.5) = 1$, so $E[W_n] = 1$. The log wealth is a sum of $n$ independent terms with mean $-0.1438$, so its median is near $-0.1438\,n$ and the median wealth near $e^{-14.38} \approx 5.7 \times 10^{-7}$ for $n = 100$. The mean is held up by a vanishingly small fraction of players with enormous wealth (for instance $1.5^{100} \approx 4 \times 10^{17}$ with probability $2^{-100}$); almost everyone loses.
9. $\ln(1 + r/m)^m = m \ln(1 + r/m) = m\left(\frac{r}{m} - \frac{r^2}{2m^2} + \cdots\right) = r - \frac{r^2}{2m} + \cdots \to r$. Since $\exp$ is continuous, $(1 + r/m)^m \to e^r$.

Numerical checks:

## 8. Conclusion

- Prices grow by **multiplication**, so simple returns compound: $P_T/P_0 = \prod (1 + R_t)$. Adding or averaging simple returns is wrong, which is why $+50\%$ then $-50\%$ leaves $75\%$.
- The **log return** $r = \ln(P_1/P_0) = \ln(1 + R)$ turns the product into a sum: $\ln(P_T/P_0) = \sum r_t$, and $P_T = P_0 e^{\sum r_t}$. Log returns are additive over time and symmetric ($+r$ then $-r$ is break-even).
- For small returns $\ln(1 + R) \approx R$ with error about $R^2/2$; the log return is always the smaller one.
- **Continuous compounding** $A = Pe^{rt}$ is the limit of $(1 + r/m)^{mt}$; a continuous rate is a log return per unit time. Convert with $r = \ln(1 + R)$ and $R = e^r - 1$; doubling time is $\ln 2 / r$.
- Long-run growth is governed by the mean **log** return, roughly $\mu - \sigma^2/2$: volatility drags growth down. Across assets in a portfolio, however, it is the simple returns that combine linearly.
- Geometrically: on a log price axis, equal percentage moves are equal steps, and constant growth is a straight line.
