# How Napier and Briggs Built the Tables: Repeated Square Roots

A table of logarithms is only useful if someone has filled it in. This notebook reconstructs how that was done **without calculus, without series and without $e$**: Henry Briggs computed $\log_{10} 2$ to about fourteen decimal places using nothing but square roots, one multiplication per entry and a great deal of patience.

This is notebook 01 of module 03 (**Computing logarithms**) in the **Learning Logarithms** series.

## 1. Introduction

Suppose it is 1617 and you have just been converted to base-10 logarithms. You know exactly two values:

$$
\log_{10} 1 = 0, \qquad \log_{10} 10 = 1.
$$

Now find $\log_{10} 2$ — the number $y$ with $10^{y} = 2$ — to fourteen decimal places. You have no calculus (Newton is not yet born), no power series, no calculator; only pen, paper and the four operations, plus the slow but reliable school method for **square roots**.

Here is the difficulty. Square roots give you the logarithms of a few special numbers for free:

$$
\log_{10} \sqrt{10} = \tfrac12, \qquad \log_{10} \sqrt{\sqrt{10}} = \tfrac14, \qquad \log_{10} 10^{1/8} = \tfrac18, \;\dots
$$

but $2$ is not among them, and nothing in this list seems to bring you closer to it. Briggs found a way: keep taking square roots until the numbers are **so close to 1 that the logarithm becomes a straight line**. Near $1$, $\log_{10}(1 + \varepsilon)$ is simply proportional to $\varepsilon$, and one ratio unlocks every number. His *Arithmetica Logarithmica* (1624) describes taking 54 successive square roots of $10$, each carried to more than thirty decimal places, and its tables gave $\log_{10}$ of the integers $1$–$20{,}000$ and $90{,}000$–$100{,}000$ to fourteen decimals.

## 2. Theory

### 2.1 Square roots halve logarithms

Taking a square root halves the exponent:

$$
\left(10^{y}\right)^{1/2} = 10^{y/2} \quad\Longleftrightarrow\quad \log_{10}\sqrt{x} = \tfrac12 \log_{10} x .
$$

Starting from $10 = 10^1$ and taking $n$ square roots gives

$$
10^{1/2^n}, \qquad \log_{10} 10^{1/2^n} = \frac{1}{2^n}.
$$

The same is true for any starting number $x > 0$: after $n$ square roots we have $x^{1/2^n}$ with logarithm $\log_{10} x / 2^n$. As $n$ grows, the logarithm shrinks to $0$, so the number itself creeps towards $10^0 = 1$. Write

$$
x^{1/2^n} = 1 + \varepsilon_n , \qquad \varepsilon_n \to 0 .
$$

Each square root roughly **halves** $\varepsilon$: $\sqrt{1 + \varepsilon} \approx 1 + \varepsilon/2$ when $\varepsilon$ is small (square both sides: $(1 + \varepsilon/2)^2 = 1 + \varepsilon + \varepsilon^2/4$).

### 2.2 Near 1 the logarithm is a straight line

If $\varepsilon \approx \varepsilon/2$ after one more root, and the logarithm is also exactly halved, then for small $\varepsilon$ the logarithm and $\varepsilon$ shrink **in step**. Their ratio settles down to a constant:

$$
\log_{10}(1 + \varepsilon) \approx k\,\varepsilon \qquad (\varepsilon \text{ small}).
$$

Briggs measured $k$ from the square roots of $10$, whose logarithms he knew exactly. After $n = 54$ roots,

$$
10^{1/2^{54}} = 1 + \varepsilon_{10}, \qquad \varepsilon_{10} \approx 1.2781914932 \times 10^{-16},
$$

and since $\log_{10}$ of the left side is exactly $1/2^{54}$,

$$
k \approx \frac{1/2^{54}}{\varepsilon_{10}} = \frac{1}{2^{54}\,\varepsilon_{10}} \approx 0.4342944819 .
$$

In modern language $k = \log_{10} e = 1/\ln 10$: the slope of $\log_{10}$ at $x = 1$. Briggs found $e$'s fingerprint a century before Euler named it. (Notebook 01/04 shows the full series $\ln(1+\varepsilon) = \varepsilon - \varepsilon^2/2 + \dots$; Briggs only needed its first term.)

### 2.3 The method for $\log_{10} 2$

Do the same with $2$. After $n$ square roots, $2^{1/2^n} = 1 + \varepsilon_2$ with

$$
\log_{10} 2^{1/2^n} = \frac{\log_{10} 2}{2^n} \approx k\,\varepsilon_2 \quad\Longrightarrow\quad \log_{10} 2 \approx 2^n k\,\varepsilon_2 = \frac{\varepsilon_2}{\varepsilon_{10}} .
$$

So $\log_{10} 2$ is the **ratio of how far the two numbers still are from 1** after the same number of square roots. In exponent form and log form side by side:

| Exponent form | Log form |
|---|---|
| $10^{1/2^n} = 1 + \varepsilon_{10}$ | $\log_{10}(1 + \varepsilon_{10}) = 2^{-n}$ |
| $2^{1/2^n} = 1 + \varepsilon_{2}$ | $\log_{10}(1 + \varepsilon_{2}) = 2^{-n}\log_{10} 2$ |
| divide the two lines | $\log_{10} 2 \approx \varepsilon_2 / \varepsilon_{10}$ |

The same ratio, with $\ln$ in place of $\log_{10}$, is the limit

$$
\ln x = \lim_{N \to \infty} N\left(x^{1/N} - 1\right),
$$

and Briggs's formula is this limit for $N = 2^{54}$, divided by the same limit for $x = 10$.

### 2.4 How accurate is it?

The neglected term is the next one in $\log_{10}(1+\varepsilon) = k(\varepsilon - \varepsilon^2/2 + \dots)$, so the **relative** error of the linear rule is about $\varepsilon/2$. Every further square root halves $\varepsilon$ and therefore halves the error: one more correct binary digit per root, roughly three more decimal digits per ten roots. After 54 roots $\varepsilon_2 \approx 4 \times 10^{-17}$, good for about sixteen digits — but **only if every root is carried to about $16 + 16 = 32$ significant figures**, because the information lives in the digits after the long run of zeros in $1.0000000000000000384\dots$. That is why Briggs worked to thirty-odd places, and why ordinary 16-digit floating-point numbers fail at this game (section 5.3).

### 2.5 From a few primes to a whole table

The laws of logarithms mean only **primes** need the expensive treatment:

$$
\log_{10} 4 = 2\log_{10} 2,\quad \log_{10} 5 = \log_{10} 10 - \log_{10} 2,\quad \log_{10} 6 = \log_{10} 2 + \log_{10} 3,\quad \log_{10} 8 = 3\log_{10} 2,\quad \log_{10} 9 = 2\log_{10} 3 .
$$

From $\log_{10} 2$, $\log_{10} 3$ and $\log_{10} 7$ the whole table $\log_{10} 1, \dots, \log_{10} 10$ follows by addition. Briggs also used shortcuts such as $2^{10} = 1024 = 1000 \times 1.024$, which gives $\log_{10} 2 = (3 + \log_{10} 1.024)/10$: a number close to $1$ needs far fewer square roots (Exercise 6).

### 2.6 Napier's kinematic definition (briefly)

Napier's 1614 definition, three years earlier, came from motion rather than square roots. Two points start together with the same speed $10^7$:

- $P$ moves along a segment of length $10^7$ towards its end $B$, with speed always equal to its **remaining distance** $x$ to $B$, so it slows down and never quite arrives;
- $Q$ moves along an unbounded line with **constant** speed $10^7$.

When $P$'s remaining distance is $x$, Napier defined the logarithm of $x$ to be the distance $y$ that $Q$ has covered. In equal times $P$'s distance shrinks by equal **ratios** while $Q$ advances by equal **steps** — a geometric sequence paired with an arithmetic one, as in Stifel's table. In modern notation, $x = 10^7 e^{-t}$ and $y = 10^7 t$, so

$$
\operatorname{NapLog}(x) = 10^7 \ln\frac{10^7}{x} .
$$

It decreases as $x$ grows, $\operatorname{NapLog}(10^7) = 0$, and its product rule carries an awkward scale factor: $\operatorname{NapLog}(xy/10^7) = \operatorname{NapLog}(x) + \operatorname{NapLog}(y)$. Napier's tables listed these logarithms for the sines used in astronomy. When Briggs visited him in Edinburgh, they agreed that a table with $\log_{10} 1 = 0$ and $\log_{10} 10 = 1$ would be far more convenient — and Briggs set out to build it with the method above.

## 3. Worked Examples

### Example 1 — the first few square roots of 10, by hand

$\sqrt{10} \approx 3.1623$, $\sqrt{3.1623} \approx 1.7783$, $\sqrt{1.7783} \approx 1.3335$, $\sqrt{1.3335} \approx 1.1548$. Their logarithms are exactly $\tfrac12, \tfrac14, \tfrac18, \tfrac1{16}$. Check the last: $1.1548^{16} \approx 10.0$.

The distance from 1 goes $2.16,\ 0.778,\ 0.333,\ 0.155$: roughly halving, as predicted, and the halving gets more exact as we approach $1$.

### Example 2 — a rough $k$ from only four roots

After four roots, $\varepsilon_{10} \approx 0.1548$ and $\log_{10}(1.1548) = 1/16 = 0.0625$, so $k \approx 0.0625 / 0.1548 \approx 0.404$. The true value is $0.4343$; the error ($7\%$) is roughly $\varepsilon/2 \approx 8\%$, as section 2.4 predicts. More roots fix that.

### Example 3 — $\log_{10} 2$ from ten roots

Ten square roots of $2$ give $2^{1/1024} \approx 1.000677130$, ten roots of $10$ give $10^{1/1024} \approx 1.002251$. Then

$$
\log_{10} 2 \approx \frac{0.000677130}{0.002251} \approx 0.3008 ,
$$

already correct to three digits (true value $0.30103$). With 54 roots Briggs's ratio is correct to about sixteen digits.

### Example 4 — the rest of the table

With $\log_{10} 2 = 0.30103$ and $\log_{10} 3 = 0.47712$:

- $\log_{10} 5 = 1 - 0.30103 = 0.69897$,
- $\log_{10} 6 = 0.30103 + 0.47712 = 0.77815$,
- $\log_{10} 1.5 = \log_{10} 3 - \log_{10} 2 = 0.17609$,
- $\log_{10} 7.2 = \log_{10}(8 \cdot 9 / 10) = 3(0.30103) + 2(0.47712) - 1 = 0.85733$.

### Example 5 — Napier's logarithm of half the radius

For $x = 10^7/2$ (the sine of $30^\circ$ on a circle of radius $10^7$), $\operatorname{NapLog}(x) = 10^7 \ln 2 \approx 6{,}931{,}472$. Halving $x$ again adds the same amount: $\operatorname{NapLog}(10^7/4) \approx 13{,}862{,}944$ — equal ratios, equal steps.

## 4. Python Examples

### 4.1 Repeated square roots with the `decimal` module

Floats carry about 16 significant digits — not enough (section 2.4). Python's standard-library `decimal` module does arithmetic with as many digits as we ask for, and `Decimal.sqrt()` is correctly rounded at that precision. We use 60 digits, about twice what Briggs used.

```python
from decimal import Decimal, getcontext

getcontext().prec = 60


def repeated_square_roots(start, count):
    value = Decimal(start)
    for _ in range(count):
        value = value.sqrt()
    return value


for count in [1, 2, 3, 4, 10, 20, 54]:
    root = repeated_square_roots(10, count)
    print(f"{count:>2} roots of 10: {str(root)[:42]}")
```

After 54 roots the number is $1.000\,000\,000\,000\,000\,127\,8\dots$: fifteen zeros, then the information. Briggs carried this to more than thirty places by hand.

### 4.2 Briggs's constant $k$

```python
from decimal import Decimal, getcontext

getcontext().prec = 60


def repeated_square_roots(start, count):
    value = Decimal(start)
    for _ in range(count):
        value = value.sqrt()
    return value


roots_taken = 54
epsilon_ten = repeated_square_roots(10, roots_taken) - 1
k_briggs = 1 / (Decimal(2) ** roots_taken * epsilon_ten)
k_exact = Decimal(1) / Decimal(10).ln()     # log10(e) = 1 / ln(10), the modern value

print("epsilon_10       =", f"{epsilon_ten:.10e}")
print("k from 54 roots  =", str(k_briggs)[:22])
print("1 / ln(10)       =", str(k_exact)[:22])
print("relative error   =", f"{abs(k_briggs / k_exact - 1):.2e}", "  (compare epsilon/2 =", f"{epsilon_ten / 2:.2e})")
```

The relative error equals $\varepsilon/2$ almost exactly — the size of the first neglected term.

### 4.3 $\log_{10} 2$ the Briggs way

The function below takes any $x > 0$, square-roots it and $10$ the same number of times, and returns the ratio $\varepsilon_x / \varepsilon_{10}$.

```python
import math
from decimal import Decimal, getcontext

getcontext().prec = 60


def repeated_square_roots(start, count):
    value = Decimal(start)
    for _ in range(count):
        value = value.sqrt()
    return value


def briggs_log10(x, roots_taken=54):
    epsilon_x = repeated_square_roots(x, roots_taken) - 1
    epsilon_ten = repeated_square_roots(10, roots_taken) - 1
    return epsilon_x / epsilon_ten


log10_two_briggs = briggs_log10(2)
log10_two_exact = Decimal(2).log10()

print("Briggs, 54 roots :", str(log10_two_briggs)[:22])
print("decimal log10(2) :", str(log10_two_exact)[:22])
print("math.log10(2)    :", repr(math.log10(2)))
print("absolute error   :", f"{abs(log10_two_briggs - log10_two_exact):.1e}")
```

Sixteen correct digits — more than Briggs's fourteen-place tables needed, and as many as a Python float can hold.

### 4.4 A table of $\log_{10} 1, \dots, \log_{10} 10$ from three primes

Only $2$, $3$ and $7$ go through the square-root mill. Everything else is assembled with the laws of logarithms, exactly as in section 2.5.

```python
import math
from decimal import Decimal, getcontext

getcontext().prec = 60


def repeated_square_roots(start, count):
    value = Decimal(start)
    for _ in range(count):
        value = value.sqrt()
    return value


def briggs_log10(x, roots_taken=54):
    return (repeated_square_roots(x, roots_taken) - 1) / (repeated_square_roots(10, roots_taken) - 1)


log2, log3, log7 = briggs_log10(2), briggs_log10(3), briggs_log10(7)
table = {
    1: Decimal(0),
    2: log2,
    3: log3,
    4: 2 * log2,
    5: 1 - log2,
    6: log2 + log3,
    7: log7,
    8: 3 * log2,
    9: 2 * log3,
    10: Decimal(1),
}

print(" x   Briggs-style log10(x)   math.log10(x)        agree?")
for x, value in table.items():
    print(f"{x:>2}   {float(value):.15f}     {math.log10(x):.15f}   {math.isclose(float(value), math.log10(x), abs_tol=1e-15)}")
```

### 4.5 Napier's logarithm

$\operatorname{NapLog}(x) = 10^7 \ln(10^7 / x)$. Check the scaled product rule and compare with Briggs's version.

```python
import math

RADIUS = 10**7


def napier_log(x):
    return RADIUS * math.log(RADIUS / x)


print("NapLog(10^7)     =", napier_log(RADIUS))
print("NapLog(10^7 / 2) =", round(napier_log(RADIUS / 2), 1))

x, y = 6_000_000, 3_500_000
left = napier_log(x * y / RADIUS)
right = napier_log(x) + napier_log(y)
print(f"NapLog(xy/10^7) = {left:.4f},  NapLog(x) + NapLog(y) = {right:.4f},  equal: {math.isclose(left, right)}")
print("NapLog decreases:", napier_log(2_000_000) > napier_log(8_000_000))
```

## 5. Visualizations

### 5.1 Square roots march towards 1 — and the logarithm keeps pace

Left: the distance $\varepsilon_n = 10^{1/2^n} - 1$ on a log scale. After the first few roots it falls on a straight line with slope "halve per step". Right: the ratio $\log_{10}(1+\varepsilon_n)/\varepsilon_n = 2^{-n}/\varepsilon_n$, which settles on Briggs's constant $k = 0.4343$.

```python
import matplotlib.pyplot as plt
import numpy as np
from decimal import Decimal, getcontext

getcontext().prec = 60

root_counts = np.arange(0, 41)
epsilons = []
value = Decimal(10)
for count in root_counts:
    if count > 0:
        value = value.sqrt()
    epsilons.append(float(value - 1))
epsilons = np.array(epsilons)
ratio_estimates = 2.0 ** (-root_counts) / epsilons

fig, (ax_eps, ax_k) = plt.subplots(1, 2, figsize=(12, 4.5))
ax_eps.semilogy(root_counts, epsilons, "o-", markersize=3)
ax_eps.set_xlabel("number of square roots n")
ax_eps.set_ylabel(r"$\varepsilon_n = 10^{1/2^n} - 1$ (log scale)")
ax_eps.set_title("Each square root halves the distance to 1")
ax_eps.grid(True, which="both", alpha=0.3)

ax_k.plot(root_counts, ratio_estimates, "o-", markersize=3, color="tab:orange")
ax_k.axhline(1 / np.log(10), color="black", linestyle="--", label=r"$k = 1/\ln 10 \approx 0.4343$")
ax_k.set_xlabel("number of square roots n")
ax_k.set_ylabel(r"$2^{-n} / \varepsilon_n$")
ax_k.set_title("The ratio log / distance settles on k")
ax_k.legend()
ax_k.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 5.2 Why the linear rule works: zooming in on $\log_{10}$ near 1

Zoom in on the graph of $\log_{10} x$ around $x = 1$ and it becomes indistinguishable from its tangent line $k(x - 1)$. Each square root zooms in by a factor of two.

```python
import matplotlib.pyplot as plt
import numpy as np

k = 1 / np.log(10)
fig, axes = plt.subplots(1, 3, figsize=(13, 4))
for ax, half_width in zip(axes, [2.0, 0.3, 0.02]):
    x = np.linspace(max(1 - half_width, 0.05), 1 + half_width, 400)
    ax.plot(x, np.log10(x), label=r"$\log_{10} x$", linewidth=2)
    ax.plot(x, k * (x - 1), "--", label=r"$k(x-1)$")
    ax.set_title(f"window 1 ± {half_width}")
    ax.set_xlabel("x")
    ax.grid(True, alpha=0.3)
axes[0].set_ylabel("value")
axes[0].legend()
plt.tight_layout()
plt.show()
```

### 5.3 Error versus number of roots: 60 digits against floats

The Briggs estimate of $\log_{10} 2$ after $n$ roots, computed two ways. With 60-digit decimals the error halves with every root (a straight line on the log axis). With ordinary floats the error halves only until about $n \approx 27$; after that, $1 + \varepsilon$ cannot store enough digits of $\varepsilon$ and the error **grows** again — the computer version of a careless Briggs who rounded to sixteen places.

```python
import math

import matplotlib.pyplot as plt
import numpy as np
from decimal import Decimal, getcontext

getcontext().prec = 60
true_value = Decimal(2).log10()

root_counts = np.arange(1, 61)
decimal_errors = []
float_errors = []
decimal_two, decimal_ten = Decimal(2), Decimal(10)
float_two, float_ten = 2.0, 10.0
for count in root_counts:
    decimal_two, decimal_ten = decimal_two.sqrt(), decimal_ten.sqrt()
    float_two, float_ten = math.sqrt(float_two), math.sqrt(float_ten)
    decimal_errors.append(float(abs((decimal_two - 1) / (decimal_ten - 1) - true_value)))
    if float_ten > 1.0:
        float_errors.append(abs((float_two - 1) / (float_ten - 1) - float(true_value)))
    else:
        float_errors.append(np.nan)   # 10^(1/2^n) has rounded to exactly 1.0

fig, ax = plt.subplots(figsize=(9, 5))
ax.semilogy(root_counts, decimal_errors, "o-", markersize=3, label="decimal, 60 digits")
ax.semilogy(root_counts, float_errors, "s-", markersize=3, label="float, ~16 digits")
ax.axvline(54, color="gray", linestyle=":", label="Briggs: 54 roots")
ax.set_xlabel("number of square roots n")
ax.set_ylabel(r"error in $\log_{10} 2$ (log scale)")
ax.set_title("More roots help only if the digits are kept")
ax.grid(True, which="both", alpha=0.3)
ax.legend()
plt.show()

best_float = int(root_counts[np.nanargmin(float_errors)])
print("best float result after", best_float, "roots, error", f"{np.nanmin(float_errors):.1e}")
print("decimal error after 54 roots:", f"{decimal_errors[53]:.1e}")
```

### 5.4 Napier's two moving points

$P$ (geometric: equal ratios in equal times) and $Q$ (arithmetic: equal steps in equal times), sampled at equal time intervals. Each vertical link pairs a number $x$ with its Napier logarithm.

```python
import matplotlib.pyplot as plt
import numpy as np

RADIUS = 1.0                       # work in units of 10^7
times = np.arange(0, 8) * 0.25
remaining_distance = RADIUS * np.exp(-times)     # P: distance still to go to B
q_position = RADIUS * times                       # Q: distance covered = NapLog

fig, ax = plt.subplots(figsize=(11, 3.8))
ax.hlines(1, 0, RADIUS, color="tab:blue")
ax.hlines(0, 0, 2.0, color="tab:orange")
for x, y in zip(remaining_distance, q_position):
    p_position = RADIUS - x                       # P measured from the start A
    ax.plot([p_position, y], [1, 0], color="gray", alpha=0.5)
    ax.plot(p_position, 1, "o", color="tab:blue")
    ax.plot(y, 0, "o", color="tab:orange")
    ax.text(y, -0.2, f"{y:.2f}", ha="center", fontsize=9)
ax.text(0, 1.18, "P on AB: remaining distance x shrinks by the same ratio each step", color="tab:blue")
ax.text(0, -0.45, "Q: distance covered y = NapLog(x) grows by the same step (units of 10^7)", color="tab:orange")
ax.text(RADIUS, 1.05, "B", fontsize=11)
ax.set_ylim(-0.6, 1.35)
ax.set_xlim(-0.05, 2.1)
ax.axis("off")
ax.set_title("Napier's kinematic logarithm")
plt.show()
```

2. Need $2.3/2^n < 10^{-6}$, i.e. $2^n > 2.3 \times 10^6$. Since $2^{21} \approx 2.10 \times 10^6$ and $2^{22} \approx 4.19 \times 10^6$, the answer is $n = 22$.
3. $\operatorname{NapLog}(10^7) = 0$: at the start $P$ has its whole distance $10^7$ left and $Q$ has not moved. A smaller remaining distance means more time has passed, so $Q$ is further along: $\operatorname{NapLog}(10^6) > \operatorname{NapLog}(9 \times 10^6)$. Napier's logarithm **decreases**.
4. $2^{54} = 2^4 \cdot (2^{10})^5 \approx 16 \times 10^{15} = 1.6 \times 10^{16}$, so $\varepsilon_{10} \approx 2.3/1.6 \times 10^{-16} \approx 1.4 \times 10^{-16}$ (exact: $1.278 \times 10^{-16}$). Then $10^{1/2^{54}} = 1.000\,000\,000\,000\,000\,13\dots$, with 15 zeros after the decimal point.
5. Both agree with `math.log10` to about 16 significant digits (the error is about $10^{-17}$), the limit of a float.
6. Linear: $k \cdot 0.024 = 0.0104231$, so $\log_{10} 2 \approx 0.3010423$ — error $1.2 \times 10^{-5}$. Two-term: $k(0.024 - 0.000288) = 0.0102980$, so $\log_{10} 2 \approx 0.3010298$ — error $2 \times 10^{-7}$. A number already close to 1 needs few or no square roots.
7. Need $\varepsilon/2 \approx \ln 10/2^{n+1} < 10^{-30}$, i.e. $2^{n+1} > 2.3 \times 10^{30}$; $2^{101} \approx 2.5 \times 10^{30}$, so $n = 100$ roots. Then $\varepsilon \approx 2 \times 10^{-30}$, and keeping 30 significant digits **of $\varepsilon$** needs about $30 + 30 = 60$ digits of working precision.
8. SymPy simplifies the difference to $0$ (check below). Since $\ln(10^7/x) = \ln 10^7 - \ln x$, we get $\ln x = 7\ln 10 - \operatorname{NapLog}(x)/10^7$ and so $\log_{10} x = 7 - \operatorname{NapLog}(x)/(10^7 \ln 10)$.
9. The two-term rule leaves a relative error of order $\varepsilon^2$ instead of $\varepsilon$. At $n = 20$ the one-term error is about $2 \times 10^{-7}$ and the two-term error about $4 \times 10^{-13}$; the two-term rule reaches the $10^{-17}$ level after about $n = 28$ roots instead of $54$. The code prints the exact figures.

Numerical checks:

```python
import math
from decimal import Decimal, getcontext

import sympy as sp

getcontext().prec = 60


def repeated_square_roots(start, count):
    value = Decimal(start)
    for _ in range(count):
        value = value.sqrt()
    return value


def briggs_log10(x, roots_taken=54):
    return (repeated_square_roots(x, roots_taken) - 1) / (repeated_square_roots(10, roots_taken) - 1)


print("Ex 1:", round(2 * 0.30103 + 0.47712, 5), math.log10(12), math.log10(0.5), round(1 - 2 * 0.30103, 5), math.log10(2.5))

n = 0
while repeated_square_roots(10, n) - 1 >= Decimal("1e-6"):
    n += 1
print("Ex 2: roots needed =", n)

print("Ex 4: epsilon_10 after 54 roots =", f"{repeated_square_roots(10, 54) - 1:.4e}", "->", str(repeated_square_roots(10, 54))[:22])

for x in [7, 11]:
    estimate = briggs_log10(x)
    print(f"Ex 5: log10({x}) Briggs {float(estimate):.16f}  math {math.log10(x):.16f}  error {abs(estimate - Decimal(x).log10()):.1e}")

k = 1 / math.log(10)
for label, log_1024 in [("linear  ", k * 0.024), ("two-term", k * (0.024 - 0.024**2 / 2))]:
    estimate = (3 + log_1024) / 10
    print(f"Ex 6: {label} log10(2) = {estimate:.7f}, error {abs(estimate - math.log10(2)):.1e}")

roots = 0
while Decimal(10).ln() / Decimal(2) ** (roots + 1) >= Decimal("1e-30"):
    roots += 1
print("Ex 7: roots for 30 digits =", roots, " error then", f"{abs(briggs_log10(2, roots) / Decimal(2).log10() - 1):.1e}")

x, y = sp.symbols("x y", positive=True)
radius = sp.Integer(10) ** 7
nap_log = lambda value: radius * sp.log(radius / value)
print("Ex 8: product rule difference =", sp.simplify(sp.expand_log(nap_log(x * y / radius) - nap_log(x) - nap_log(y), force=True)))


def two_term_log10_two(roots_taken):
    eps_two = repeated_square_roots(2, roots_taken) - 1
    eps_ten = repeated_square_roots(10, roots_taken) - 1
    return (eps_two - eps_two**2 / 2) / (eps_ten - eps_ten**2 / 2)


true_log2 = Decimal(2).log10()
print("Ex 9: n=20 one-term error", f"{abs(briggs_log10(2, 20) - true_log2):.1e}", " two-term error", f"{abs(two_term_log10_two(20) - true_log2):.1e}")
target = abs(briggs_log10(2, 54) - true_log2)
roots = 1
while abs(two_term_log10_two(roots) - true_log2) > target:
    roots += 1
print("      two-term rule matches the 54-root one-term error after", roots, "roots")
```

## 8. Conclusion

- Square roots **halve** logarithms: $\log_{10} x^{1/2^n} = \log_{10} x / 2^n$. Taking enough of them drives any positive number to within $\varepsilon$ of $1$.
- Near $1$ the logarithm is a straight line: $\log_{10}(1 + \varepsilon) \approx k\varepsilon$ with $k = \log_{10} e = 1/\ln 10 \approx 0.4343$. Briggs measured $k$ from 54 square roots of $10$.
- **Briggs's method** in one formula: after the same number of square roots, $\log_{10} x \approx \varepsilon_x / \varepsilon_{10}$ — equivalently $\ln x = \lim N(x^{1/N} - 1)$. Every extra root buys one more binary digit, provided every root is carried to about twice the target precision; 16-digit floats break down after about 27 roots.
- Only primes need the expensive treatment; the laws of logarithms assemble the rest of the table from $\log_{10} 2$, $\log_{10} 3$, $\log_{10} 7, \dots$
- Napier's original logarithm came from two moving points — one geometric, one arithmetic — and equals $10^7 \ln(10^7/x)$ in modern terms: decreasing, with $\operatorname{NapLog}(10^7) = 0$. Base 10 with $\log_{10} 1 = 0$ was Briggs's improvement.
