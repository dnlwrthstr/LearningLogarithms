# The Slide Rule: a Logarithm You Can Hold

A **slide rule** is two rulers with logarithmic scales that slide past each other. Sliding adds lengths, the scales turn lengths into logarithms, and so the instrument multiplies. For about 350 years it was the engineer's calculator: bridges, aircraft and power stations were designed on slide rules until cheap electronic calculators replaced them in the 1970s.

This is notebook 02 of module 03 (**Computing logarithms**) in the **Learning Logarithms** series.

## 1. Introduction

It is 1960. You are an engineer, and before lunch you need a few hundred products like

$$
3.7 \times 4.6, \qquad \frac{8.31 \times 293}{101.3}, \qquad \sqrt{50},
$$

each to about three significant figures. Long multiplication takes a minute and invites mistakes. A table of logarithms (notebook 01) is reliable but slow: two lookups, an addition, a reverse lookup. You want something faster — a device that does the "look up, add, look up" in one motion.

The key observation is that **adding is easy for a ruler**. Put two ordinary rulers side by side, slide the top one so its $0$ sits over $2$ on the bottom one, and the bottom ruler under the top's $3$ reads $5$: the rulers have added $2 + 3$. Now replace the uniform markings by **logarithmic** ones — print each number $x$ at distance $\log_{10} x$ from the left end — and the same motion adds logarithms:

$$
\log_{10} 2 + \log_{10} 3 = \log_{10} 6 .
$$

The rulers have multiplied. Edmund Gunter drew a single logarithmic scale around 1620 and measured distances on it with a pair of dividers; William Oughtred put two logarithmic scales against each other in the 1620s, and that is the slide rule. It stayed essentially the same idea, with more and more specialised scales, until pocket electronic calculators (the HP-35 of 1972 was the first scientific one that fit in a pocket) made it obsolete within a few years.

## 2. Theory

### 2.1 A logarithmic scale

On a scale of length $L$, the number $x \in [1, 10]$ is printed at distance

$$
d(x) = L \log_{10} x
$$

from the left end (the **index**, marked $1$). So $1$ is at $0$, $10$ is at $L$, $2$ is at $0.301L$, $3$ at $0.477L$, and $\sqrt{10} \approx 3.16$ exactly in the middle. Equal ratios get equal lengths: $1 \to 2$, $2 \to 4$ and $4 \to 8$ all have length $0.301L$. A typical rule had $L = 25\text{ cm}$.

### 2.2 Multiplication: adding two lengths

A slide rule has a fixed **D** scale and an identical **C** scale on the sliding middle strip. To compute $a \times b$:

1. slide C so that its left index $1$ is above $a$ on D (the slide is now shifted by $L\log_{10} a$);
2. move the cursor to $b$ on C;
3. read the product on D under the cursor.

The cursor's distance from the left end of D is

$$
L \log_{10} a + L \log_{10} b = L \log_{10}(ab),
$$

so D shows $ab$ there. Exponent form and log form side by side:

| Exponent form | Log form (lengths on the rule) |
|---|---|
| $a = 10^{\alpha},\ b = 10^{\beta}$ | $d(a) = L\alpha,\ d(b) = L\beta$ |
| $ab = 10^{\alpha + \beta}$ | $d(ab) = d(a) + d(b)$ |

### 2.3 Running off the end, and the decimal point

If $ab > 10$ the cursor would land beyond the right end of D. The fix: use the **right** index $10$ of C instead. Sliding C's $10$ over $a$ shifts the slide by $L\log_{10} a - L$, so the cursor lands at $L\log_{10}(ab) - L = L\log_{10}(ab/10)$, and D shows $ab/10$.

More generally, a slide rule only ever handles the **digits** of a number — its mantissa in $[1, 10)$. The user writes $x = m \times 10^{e}$ and tracks the power of ten in their head:

$$
\log_{10} x = \log_{10} m + e, \qquad 0 \le \log_{10} m < 1 .
$$

$37 \times 460$ is set exactly like $3.7 \times 4.6$; the rough estimate $40 \times 500 = 20{,}000$ says the answer $1.702$ means $17{,}020$. Estimation was a core slide-rule skill.

### 2.4 Division: subtracting lengths

To compute $a / b$: move the cursor to $a$ on D, slide C until $b$ on C is under the cursor, and read D under C's index. The index sits at distance $L\log_{10} a - L\log_{10} b = L\log_{10}(a/b)$.

### 2.5 Squares and square roots: the A and B scales

The **A** scale (above D) and **B** scale (on the slide, above C) pack **two** decades, $1$ to $100$, into the same length $L$:

$$
d_A(x) = \frac{L}{2}\log_{10} x, \qquad x \in [1, 100] .
$$

A number $y$ on D at distance $L\log_{10} y$ sits under the point of A at the same distance, which is $\frac{L}{2}\log_{10} x$ with $x = y^2$. So the cursor on $y$ (on D) reads $y^2$ on A, and the cursor on $x$ (on A) reads $\sqrt{x}$ on D:

$$
\log_{10} y^2 = 2\log_{10} y .
$$

For a square root you must pick the correct **half** of A: numbers with an odd number of digits before the decimal point (like $5$ or $500$) go on the left half, an even number (like $50$) on the right half. Other scales used the same trick: the **K** scale packs three decades for cubes, and the inverted **CI** scale runs right to left to give $1/x$.

### 2.6 Accuracy: a constant relative error

A human reads a scale to within a fixed **distance** $\delta$, say $0.1$ mm, wherever on the rule they look. A position error $\delta$ changes $\log_{10} x$ by $\delta / L$, so

$$
\frac{\Delta x}{x} \approx \ln 10 \cdot \frac{\delta}{L} .
$$

With $\delta = 0.1$ mm and $L = 250$ mm, each reading has a relative error of about $0.09\%$, the **same** at $x = 1.2$ as at $x = 9.8$ — a logarithmic scale spreads its precision evenly over ratios. A multiplication involves about three settings and readings, so the result is good to about $0.15\%$: **three significant figures**. That is exactly the precision that engineering data (material strengths, loads, measurements) usually had anyway.

## 3. Worked Examples

### Example 1 — $2 \times 3$ on a 25 cm rule

$d(2) = 25 \times 0.30103 = 7.526$ cm and $d(3) = 25 \times 0.47712 = 11.928$ cm. Slide C's $1$ to $7.526$ cm, move the cursor to C's $3$ at $7.526 + 11.928 = 19.454$ cm, and read D there: $10^{19.454/25} = 10^{0.7782} = 6.00$.

### Example 2 — $4 \times 5$: off the end

$d(4) + d(5) = 15.05 + 17.47 = 32.52 > 25$ cm. Use the right index: C's $10$ over D's $4$, cursor to C's $5$ at $32.52 - 25 = 7.53$ cm (the position of $2$); D reads $2.00$. Estimate: $4 \times 5 \approx 20$, so the answer is $20.0$.

### Example 3 — $8 / 2$

Cursor on D's $8$ at $22.58$ cm, C's $2$ under the cursor; C's left index is then at $22.58 - 7.53 = 15.05$ cm, where D reads $4.00$.

### Example 4 — $\sqrt{50}$ with the A scale

$50$ has two digits, so use the right half of A. On A it sits at $\frac{25}{2}\log_{10} 50 = 12.5 \times 1.69897 = 21.24$ cm. The cursor carries this down to D, which reads $10^{21.24/25} = 10^{0.8495} = 7.07$. Check: $7.07^2 = 49.98$.

### Example 5 — how good is the answer?

With $\delta = 0.1$ mm per setting or reading and three of them, the typical relative error is $\sqrt{3} \times \ln 10 \times 0.1/250 \approx 0.0016$. For $3.7 \times 4.6 = 17.02$ that is $\pm 0.03$: a slide-rule user would report $17.0$.

## 4. Python Examples

### 4.1 A slide-rule simulator

The simulator works only with **positions** on the rule (in cm). It splits each number into mantissa and power of ten, adds or subtracts positions, folds the result back onto the rule with the right index when needed, and converts the final position back into a number.

```python
import math

RULE_LENGTH_CM = 25.0


def split_scientific(x):
    # x = mantissa * 10**exponent with 1 <= mantissa < 10
    exponent = math.floor(math.log10(x))
    return x / 10**exponent, exponent


def position_on_d(mantissa):
    return RULE_LENGTH_CM * math.log10(mantissa)


def number_at_position(position_cm):
    return 10 ** (position_cm / RULE_LENGTH_CM)


def slide_rule_multiply(a, b):
    mantissa_a, exponent_a = split_scientific(a)
    mantissa_b, exponent_b = split_scientific(b)
    cursor = position_on_d(mantissa_a) + position_on_d(mantissa_b)
    exponent = exponent_a + exponent_b
    if cursor >= RULE_LENGTH_CM:           # off the end: use the right index of C
        cursor -= RULE_LENGTH_CM
        exponent += 1
    return number_at_position(cursor) * 10**exponent, cursor


def slide_rule_divide(a, b):
    mantissa_a, exponent_a = split_scientific(a)
    mantissa_b, exponent_b = split_scientific(b)
    index_position = position_on_d(mantissa_a) - position_on_d(mantissa_b)
    exponent = exponent_a - exponent_b
    if index_position < 0:                 # left index falls off: read under the right index
        index_position += RULE_LENGTH_CM
        exponent -= 1
    return number_at_position(index_position) * 10**exponent, index_position


for a, b in [(2, 3), (4, 5), (3.7, 4.6), (37, 460)]:
    result, cursor = slide_rule_multiply(a, b)
    print(f"{a} x {b}: cursor at {cursor:5.2f} cm on D -> {result:.6g}   (exact {a * b:g})")
for a, b in [(8, 2), (3, 7)]:
    result, index_position = slide_rule_divide(a, b)
    print(f"{a} / {b}: index at {index_position:5.2f} cm on D -> {result:.6g}   (exact {a / b:.6g})")
```

A perfect simulator returns exact answers: the logarithms do all the work. The slide rule's real limitation is the human eye, which we add in section 4.3.

### 4.2 Squares and roots on the A scale

```python
import math

RULE_LENGTH_CM = 25.0


def position_on_a(x):
    # A scale: two decades (1..100) in the same length
    return RULE_LENGTH_CM / 2 * math.log10(x)


def number_on_d(position_cm):
    return 10 ** (position_cm / RULE_LENGTH_CM)


def number_on_a(position_cm):
    return 10 ** (2 * position_cm / RULE_LENGTH_CM)


for x in [5, 50, 2, 20]:
    cursor = position_on_a(x)
    half = "left" if cursor < RULE_LENGTH_CM / 2 else "right"
    print(f"sqrt({x:>2}): A position {cursor:5.2f} cm ({half} half) -> D reads {number_on_d(cursor):.4f}   (exact {math.sqrt(x):.4f})")

for y in [1.5, 7.07]:
    cursor = RULE_LENGTH_CM * math.log10(y)
    print(f"{y}^2: D position {cursor:5.2f} cm -> A reads {number_on_a(cursor):.4f}   (exact {y**2:.4f})")
```

### 4.3 Adding the human: reading errors

Model each setting or reading as an independent position error with standard deviation $\delta = 0.1$ mm $= 0.01$ cm. A multiplication has three: setting the index over $a$, setting the cursor on $b$, reading the result. We also round the reading to what a user would write down (three significant figures).

```python
import math

import numpy as np

RULE_LENGTH_CM = 25.0
READING_ERROR_CM = 0.01
rng = np.random.default_rng(seed=1)


def noisy_multiply(a, b, trials):
    errors = rng.normal(0, READING_ERROR_CM, size=(trials, 3)).sum(axis=1)
    cursor = RULE_LENGTH_CM * (math.log10(a) + math.log10(b)) + errors
    return 10 ** (cursor / RULE_LENGTH_CM)


trials = 100_000
for a, b in [(2, 3), (3.7, 4.6), (1.2, 1.3), (9.1, 8.7)]:
    results = noisy_multiply(a, b, trials)
    relative_errors = results / (a * b) - 1
    one_reading = float(f"{results[0]:.3g}")
    print(f"{a} x {b} = {a * b:7.3f}: one user reads {one_reading:<6}  relative error std {relative_errors.std():.5f}")

predicted = math.sqrt(3) * math.log(10) * READING_ERROR_CM / RULE_LENGTH_CM
print(f"predicted from the formula: {predicted:.5f}")
```

The relative error is the same, about $0.16\%$, whether the product is $1.56$ or $79$. That is three significant figures, everywhere on the rule.

### 4.4 One setting, a whole table

Once C is set against D, **every** pair of aligned numbers has the same ratio. Set C's $1$ over D's $2.54$ and the rule is a complete inch-to-centimetre conversion table: read any number of inches on C and its centimetres on D, without moving the slide again. This is why slide rules were so fast for repeated work.

```python
import math

RULE_LENGTH_CM = 25.0
factor = 2.54
shift = RULE_LENGTH_CM * math.log10(factor)        # C's index over 2.54 on D

print(" inches on C   position on D   D reads (cm)")
for inches in [1, 1.5, 2, 3, 3.5]:
    position = shift + RULE_LENGTH_CM * math.log10(inches)
    print(f"{inches:>10}     {position:8.2f} cm     {10 ** (position / RULE_LENGTH_CM):8.3f}")
for inches in [5, 8]:                               # these run off the right end: use C's right index
    position = shift + RULE_LENGTH_CM * math.log10(inches) - RULE_LENGTH_CM
    print(f"{inches:>10}     {position:8.2f} cm     {10 ** (position / RULE_LENGTH_CM):8.3f}  x 10")
```

## 5. Visualizations

### 5.1 Drawing the C and D scales: $2 \times 3$

The D scale is fixed; the C scale is shifted right by $d(2) = L\log_{10} 2$. The red cursor sits over $3$ on C — and over $6$ on D.

```python
import matplotlib.pyplot as plt
import numpy as np


def scale_ticks(decades=1):
    # (value, tick height, label) for a log scale from 1 to 10**decades
    ticks = []
    for decade in range(decades):
        factor = 10**decade
        for whole in range(1, 10):
            step = 0.1 if whole < 5 else 0.5
            for fraction in np.arange(0, 1, step):
                value = (whole + fraction) * factor
                is_whole = np.isclose(fraction, 0)
                height = 0.45 if is_whole else (0.3 if np.isclose(fraction, 0.5) else 0.18)
                label = f"{int(round(value))}" if is_whole else ""
                ticks.append((value, height, label))
    ticks.append((10.0**decades, 0.45, f"{10**decades}"))
    return ticks


def draw_scale(ax, name, y, offset=0.0, length=1.0, decades=1, ticks_up=True, colour="black"):
    # one strip of a slide rule: the value v sits at offset + length * log10(v) / decades
    ax.add_patch(plt.Rectangle((offset, y - 0.5), length, 1.0, facecolor="ivory", edgecolor="gray"))
    sign = 1 if ticks_up else -1
    base = y - 0.5 if ticks_up else y + 0.5
    for value, height, label in scale_ticks(decades):
        x = offset + length * np.log10(value) / decades
        ax.plot([x, x], [base, base + sign * height], color=colour, linewidth=0.8)
        if label:
            ax.text(x, base + sign * (height + 0.08), label, ha="center",
                    va="bottom" if ticks_up else "top", fontsize=8, color=colour)
    ax.text(offset - 0.02, y, name, ha="right", va="center", fontsize=11, fontweight="bold")


shift = np.log10(2)
fig, ax = plt.subplots(figsize=(13, 3.2))
draw_scale(ax, "C", y=1.0, offset=shift, ticks_up=False, colour="tab:blue")
draw_scale(ax, "D", y=0.0, offset=0.0, ticks_up=True)
cursor_x = shift + np.log10(3)
ax.axvline(cursor_x, color="crimson", linewidth=1.5)
ax.annotate("", xy=(shift, -0.75), xytext=(0, -0.75), arrowprops=dict(arrowstyle="<->", color="tab:green"))
ax.text(shift / 2, -0.95, "log10 2", ha="center", color="tab:green", fontsize=9)
ax.annotate("", xy=(cursor_x, -0.75), xytext=(shift, -0.75), arrowprops=dict(arrowstyle="<->", color="tab:purple"))
ax.text((shift + cursor_x) / 2, -0.95, "log10 3", ha="center", color="tab:purple", fontsize=9)
ax.text(cursor_x + 0.01, 1.65, "cursor: 3 on C over 6 on D", color="crimson", fontsize=10)
ax.set_xlim(-0.08, 1.35)
ax.set_ylim(-1.15, 1.9)
ax.axis("off")
ax.set_title("2 x 3 = 6: the lengths log10 2 and log10 3 add up to log10 6")
plt.show()
```

Notice that with this setting **every** pair on the two scales is in the ratio $2$: C's $1.5$ is over D's $3$, C's $2$ over D's $4$, C's $4$ over D's $8$. A slide rule set for one multiplication is a complete multiplication table for that factor.

```python
import matplotlib.pyplot as plt
import numpy as np


def scale_ticks(decades=1):
    # (value, tick height, label) for a log scale from 1 to 10**decades
    ticks = []
    for decade in range(decades):
        factor = 10**decade
        for whole in range(1, 10):
            step = 0.1 if whole < 5 else 0.5
            for fraction in np.arange(0, 1, step):
                value = (whole + fraction) * factor
                is_whole = np.isclose(fraction, 0)
                height = 0.45 if is_whole else (0.3 if np.isclose(fraction, 0.5) else 0.18)
                label = f"{int(round(value))}" if is_whole else ""
                ticks.append((value, height, label))
    ticks.append((10.0**decades, 0.45, f"{10**decades}"))
    return ticks


def draw_scale(ax, name, y, offset=0.0, length=1.0, decades=1, ticks_up=True, colour="black"):
    # one strip of a slide rule: the value v sits at offset + length * log10(v) / decades
    ax.add_patch(plt.Rectangle((offset, y - 0.5), length, 1.0, facecolor="ivory", edgecolor="gray"))
    sign = 1 if ticks_up else -1
    base = y - 0.5 if ticks_up else y + 0.5
    for value, height, label in scale_ticks(decades):
        x = offset + length * np.log10(value) / decades
        ax.plot([x, x], [base, base + sign * height], color=colour, linewidth=0.8)
        if label:
            ax.text(x, base + sign * (height + 0.08), label, ha="center",
                    va="bottom" if ticks_up else "top", fontsize=8, color=colour)
    ax.text(offset - 0.02, y, name, ha="right", va="center", fontsize=11, fontweight="bold")


fig, (ax_divide, ax_wrap) = plt.subplots(2, 1, figsize=(13, 6))

shift = np.log10(8) - np.log10(2)          # slide moved so that C's 2 is under D's 8
draw_scale(ax_divide, "C", y=1.0, offset=shift, ticks_up=False, colour="tab:blue")
draw_scale(ax_divide, "D", y=0.0, ticks_up=True)
ax_divide.axvline(np.log10(8), color="crimson", linewidth=1.5)
ax_divide.axvline(shift, color="tab:green", linewidth=1.5, linestyle="--")
ax_divide.text(np.log10(8) + 0.01, 1.65, "cursor: 2 on C under 8 on D", color="crimson", fontsize=10)
ax_divide.text(shift + 0.01, -0.85, "C's index over 4 on D:  8 / 2 = 4", color="tab:green", fontsize=10)
ax_divide.set_title("Division: subtract lengths")

shift = np.log10(4) - 1                    # C's right index 10 over D's 4
draw_scale(ax_wrap, "C", y=1.0, offset=shift, ticks_up=False, colour="tab:blue")
draw_scale(ax_wrap, "D", y=0.0, ticks_up=True)
cursor_x = shift + np.log10(5)
ax_wrap.axvline(cursor_x, color="crimson", linewidth=1.5)
ax_wrap.text(cursor_x + 0.01, 1.65, "cursor: 5 on C over 2 on D  ->  4 x 5 = 20", color="crimson", fontsize=10)
ax_wrap.set_title("Product larger than 10: use C's right index")

for ax in (ax_divide, ax_wrap):
    ax.set_xlim(-0.65, 1.65)
    ax.set_ylim(-1.1, 1.9)
    ax.axis("off")
plt.tight_layout()
plt.show()
```

### 5.4 Why a logarithmic scale: relative error on a log rule and on a linear rule

Same reading error ($0.1$ mm, three readings) on two 25 cm rules: our logarithmic one for numbers $1$–$10$, and a hypothetical **linear** one for numbers $1$–$10$ ($2.78$ cm per unit). The linear rule's error is a constant **absolute** error, which is a large relative error for small numbers; the logarithmic rule's relative error is flat. Only near the top of its range does the linear rule do better. Both axes of the right panel are logarithmic.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

RULE_LENGTH_CM = 25.0
READING_ERROR_CM = 0.01
rng = np.random.default_rng(seed=2)
trials = 20_000

true_values = np.geomspace(1.05, 9.5, 25)
log_rule_spread = []
linear_rule_spread = []
for value in true_values:
    position_errors = rng.normal(0, READING_ERROR_CM, size=trials) * math.sqrt(3)
    log_reading = 10 ** (math.log10(value) + position_errors / RULE_LENGTH_CM)
    linear_reading = value + position_errors * 9 / RULE_LENGTH_CM     # 9 units over 25 cm
    log_rule_spread.append(np.std(log_reading / value - 1))
    linear_rule_spread.append(np.std(linear_reading / value - 1))

fig, (ax_hist, ax_compare) = plt.subplots(1, 2, figsize=(13, 4.5))
samples = 10 ** (math.log10(3.7 * 4.6) + rng.normal(0, READING_ERROR_CM, size=(trials, 3)).sum(axis=1) / RULE_LENGTH_CM)
ax_hist.hist(samples, bins=60, color="tab:blue", alpha=0.75)
ax_hist.axvline(3.7 * 4.6, color="crimson", label="exact 17.02")
ax_hist.set_xlabel("slide-rule result for 3.7 x 4.6")
ax_hist.set_ylabel("count")
ax_hist.set_title("20,000 simulated users")
ax_hist.legend()

ax_compare.loglog(true_values, log_rule_spread, "o-", label="logarithmic rule")
ax_compare.loglog(true_values, linear_rule_spread, "s-", label="linear rule, same length")
ax_compare.set_xlabel("value being read (log scale)")
ax_compare.set_ylabel("relative error, std (log scale)")
ax_compare.set_title("Log scale: equal relative precision everywhere")
ax_compare.grid(True, which="both", alpha=0.3)
ax_compare.legend()
plt.tight_layout()
plt.show()

print(f"spread of the 3.7 x 4.6 results: {samples.std():.3f}  (relative {samples.std() / 17.02:.4f})")
```

### 5.5 A circular slide rule never runs off the end

Wrap the D scale around a circle: $x$ sits at angle $2\pi \log_{10} x$, and $1$ and $10$ coincide. Adding angles is multiplication, and going past the top simply continues into the next decade — the "use the right index" rule is built in. The arcs show $\log_{10} 4$ (green) followed by $\log_{10} 5$ (purple), ending at the digits $2$, i.e. $20$.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(6, 6))
ax.add_patch(plt.Circle((0, 0), 1.0, fill=False, color="black"))
for value in np.concatenate([np.arange(1, 2, 0.1), np.arange(2, 5, 0.2), np.arange(5, 10, 0.5)]):
    angle = np.pi / 2 - 2 * np.pi * np.log10(value)        # clockwise from the top
    is_whole = np.isclose(value, round(value))
    inner = 0.88 if is_whole else 0.94
    ax.plot([inner * np.cos(angle), np.cos(angle)], [inner * np.sin(angle), np.sin(angle)], color="black", linewidth=0.8)
    if is_whole:
        ax.text(1.1 * np.cos(angle), 1.1 * np.sin(angle), f"{int(round(value))}", ha="center", va="center", fontsize=11)

for start, length, colour, radius in [(0, np.log10(4), "tab:green", 0.75), (np.log10(4), np.log10(5), "tab:purple", 0.62)]:
    fractions = np.linspace(start, start + length, 100)
    angles = np.pi / 2 - 2 * np.pi * fractions
    ax.plot(radius * np.cos(angles), radius * np.sin(angles), color=colour, linewidth=3)
end_angle = np.pi / 2 - 2 * np.pi * (np.log10(4) + np.log10(5))
ax.plot([0, 0.85 * np.cos(end_angle)], [0, 0.85 * np.sin(end_angle)], color="crimson", linewidth=1.5)
ax.text(0, 0.25, "4 x 5 -> digits 2 (= 20)", ha="center", color="crimson", fontsize=10)
ax.set_xlim(-1.25, 1.25)
ax.set_ylim(-1.25, 1.25)
ax.set_aspect("equal")
ax.axis("off")
ax.set_title("Circular scale: angle = 2 pi log10 x (mod one turn)")
plt.show()
```

## 7. Solutions

1. $d(1.5) = 25 \times 0.17609 = 4.40$ cm, $d(5) = 25 \times 0.69897 = 17.47$ cm, $d(7.5) = 25 \times 0.87506 = 21.88$ cm, and $4.40 + 17.47 = 21.88$ ✓ since $1.5 \times 5 = 7.5$.

```python
import math

import numpy as np

L = 25.0
d = lambda x: L * math.log10(x)

print("Ex 1:", round(d(1.5), 2), round(d(5), 2), round(d(7.5), 2), " sum", round(d(1.5) + d(5), 2))
position = d(6) + d(7) - L
print("Ex 2: cursor at", round(position, 2), "cm -> D reads", round(10 ** (position / L), 4), "-> 42")
position = d(9) - d(4)
print("Ex 3: index at", round(position, 2), "cm -> D reads", round(10 ** (position / L), 4))
relative = math.sqrt(3) * math.log(10) * 0.01 / L
print(f"Ex 4: 0.37 * 46 = {0.37 * 46:.4f}, relative error {relative:.4f} -> +/- {0.37 * 46 * relative:.3f}")
for x in [5, 500]:
    position = L / 2 * math.log10(x) % (L / 2) if x >= 100 else L / 2 * math.log10(x)
    print(f"Ex 5: sqrt({x}) -> A position {position:.2f} cm, D digits {10 ** (position / L):.4f}, exact {math.sqrt(x):.4f}")
relative = math.log(10) * 0.02 / L
print(f"Ex 6: relative {relative:.5f}; at 2: {2 * relative:.4f}, at 9: {9 * relative:.4f}")
circle_position = (math.log10(4) + math.log10(5)) % 1
print("Ex 7: circular position", round(circle_position, 5), "-> digits", round(10**circle_position, 6))
position = L / 3 * math.log10(200)
print("Ex 8: K position", round(position, 2), "cm -> D reads", round(10 ** (position / L), 4), " exact", round(200 ** (1 / 3), 4))

rng = np.random.default_rng(seed=3)
exact = 2.5 * 3.1 * 4.2 / 1.7
log_result = math.log10(2.5) + math.log10(3.1) + math.log10(4.2) - math.log10(1.7)
samples = 10 ** (log_result + rng.normal(0, 0.01, size=(200_000, 7)).sum(axis=1) / L)
print(f"Ex 9: exact {exact:.3f}; Monte Carlo relative spread {np.std(samples / exact - 1):.5f}; formula {math.sqrt(7) * math.log(10) * 0.01 / L:.5f}")
```

## 8. Conclusion

- A slide rule prints $x$ at distance $L\log_{10} x$. Sliding adds distances, and $\log_{10} a + \log_{10} b = \log_{10}(ab)$ turns the sum back into a product: **multiplication by adding lengths**, $10^{\alpha}\cdot 10^{\beta} = 10^{\alpha+\beta}$ in metal and wood.
- Division subtracts lengths; products beyond $10$ use the right index (the rule works modulo one decade); the user tracks the power of ten by estimation.
- The A/B scales pack two decades into the same length, so they show squares of D: $\log_{10} y^2 = 2\log_{10} y$. The K scale (three decades) does cubes; CI gives reciprocals.
- A reading error of fixed **length** is a fixed **relative** error, $\Delta x/x \approx \ln 10\,\delta/L$ — about $0.1\%$ per reading on a 25 cm rule, three significant figures per calculation, evenly over the whole scale.
- The same "fixed relative precision" idea is how computers store numbers: floating point is a logarithmic number line, the subject of notebook 04.
