# Bits, Digits and Counting

This notebook shows that the logarithm is the natural way to **count choices**: how many yes/no questions pin down one object among $N$, how many digits a number needs in base $b$, how many bits a computer needs to store it, and how strong a password really is. All four questions have the same answer — a logarithm — and the reason is that logarithms turn the *multiplication* of independent choices into the *addition* of their sizes.

This is notebook 01 of module 05 (**Information and probability**) in the **Learning Logarithms** series.

## 1. Introduction

A friend thinks of one person among the roughly **one million** inhabitants of a city. You may ask only yes/no questions. How many questions do you need, in the worst case, to be sure of the answer?

Asking "Is it Alice? Is it Bob? …" could take a million questions. A cleverer strategy splits the remaining candidates in half each time ("Is the surname in the first half of the alphabetical list?"). After one question $500{,}000$ candidates remain, after two $250{,}000$, after $k$ questions about $10^6 / 2^k$. We are done when only one remains:

$$
\frac{10^6}{2^k} \le 1 \quad\Longleftrightarrow\quad 2^k \ge 10^6 \quad\Longleftrightarrow\quad k \ge \log_2 10^6 \approx 19.93.
$$

So **20 questions** are enough for a million things — the parlour game "Twenty Questions" is a logarithm in disguise. Without the logarithm the question "for which $k$ is $2^k \ge 10^6$?" is a trial-and-error search over powers of two; with it, it is one line: $k = \lceil \log_2 N \rceil$.

The same pattern appears everywhere we count possibilities:

| Question | Exponent form | Log form |
|---|---|---|
| yes/no questions to find one of $N$ | $2^k \ge N$ | $k = \lceil \log_2 N \rceil$ |
| digits of $N$ in base $b$ | $b^{d-1} \le N < b^d$ | $d = \lfloor \log_b N \rfloor + 1$ |
| bits to store $0, \dots, N-1$ | $2^k \ge N$ | $k = \lceil \log_2 N \rceil$ |
| strength of a password with $N$ equally likely values | $N = 2^H$ | $H = \log_2 N$ bits |

- What a logarithm is and why it turns products into sums — [Why logarithms exist](../00_foundations/01_why_logarithms_exist.ipynb).
- The laws of logarithms, especially $\log_b(xy) = \log_b x + \log_b y$ and $\log_b x^k = k \log_b x$ — [The laws of logarithms](../00_foundations/03_laws_of_logarithms.ipynb).
- Change of base, $\log_b x = \ln x / \ln b$ — [Change of base](../00_foundations/04_change_of_base_and_common_bases.ipynb).
- Floor and ceiling: $\lfloor x \rfloor$ is the largest integer $\le x$, $\lceil x \rceil$ the smallest integer $\ge x$.

## 2. Theory

### 2.1 Halving: the question-counting argument

A yes/no question splits the remaining candidates into two groups; whatever the answer, one group survives. In the **worst case** the larger group survives, so the best a question can do is split as evenly as possible. With $k$ questions there are at most $2^k$ distinct answer sequences (yes-yes-no-…), and each sequence can point to at most one object. Hence

$$
\text{$k$ questions can distinguish at most } 2^k \text{ objects} \quad\Longrightarrow\quad k \ge \log_2 N.
$$

Since $k$ is a whole number, the minimum is

$$
k_{\min}(N) = \lceil \log_2 N \rceil,
$$

and halving achieves it. This is a **lower bound and a strategy** at once — the same argument proves that binary search (notebook 06/04) is optimal among comparison-based searches.

### 2.2 Digits in base $b$

A positive integer $N$ has $d$ digits in base $b$ exactly when it is at least the smallest $d$-digit number $b^{d-1}$ and less than the smallest $(d+1)$-digit number $b^d$:

$$
b^{d-1} \le N < b^{d}
\quad\Longleftrightarrow\quad
d - 1 \le \log_b N < d
\quad\Longleftrightarrow\quad
d = \lfloor \log_b N \rfloor + 1.
$$

The middle step takes $\log_b$ of all three parts, which preserves the inequalities because $\log_b$ is increasing for $b > 1$. The whole-number part of $\log_b N$ counts digits; the fractional part tells you where $N$ sits *inside* its block of $d$-digit numbers (it reappears in notebook 05 as the mantissa behind Benford's law).

Changing base just rescales: $\log_b N = \log_{10} N / \log_{10} b$. A number has about $\log_2 10 \approx 3.32$ times as many binary digits as decimal digits.

### 2.3 Bits to store a number

$k$ bits have $2^k$ patterns, so they can store the integers $0, 1, \dots, 2^k - 1$. To store $N$ different values (an index into a list of length $N$, say) you need

$$
k = \lceil \log_2 N \rceil \text{ bits}.
$$

To store **the number $N$ itself** (values $0, \dots, N$) you need its number of binary digits, $\lfloor \log_2 N \rfloor + 1$ — the two formulas differ only when $N$ is a power of two. Example: $256$ values ($0$–$255$) fit in $8$ bits, but the number $256$ itself needs $9$.

### 2.4 Search spaces measured in bits

A password drawn uniformly at random from an alphabet of $A$ symbols, $L$ symbols long, is one of $N = A^L$ equally likely strings. Its **strength** is the number of yes/no questions (or, in the worst case, roughly the $\log_2$ of the number of guesses) an attacker needs:

$$
H = \log_2 A^L = L \log_2 A \text{ bits}.
$$

The power law of logarithms turns an unwieldy number like $95^{12} \approx 5.4 \times 10^{23}$ into a sum: $12 \times 6.57 \approx 78.8$ bits. Each extra bit **doubles** the attacker's work. This applies only if the password really is chosen at random — a human-chosen word has far fewer bits than its length suggests.

### 2.5 Why the logarithm is the right measure of "amount of choice"

Suppose we want a number $I(N)$ that says "how much choice" there is among $N$ equally likely options. Two requirements are natural:

1. **Monotone:** more options means more choice, $I(N) < I(N+1)$.
2. **Additive for independent choices:** choosing a card ($52$ options) *and* rolling a die ($6$ options) is one choice among $52 \times 6 = 312$ outcomes, and it should count as the two amounts added: $I(52 \cdot 6) = I(52) + I(6)$. In general $I(MN) = I(M) + I(N)$.

The logarithm satisfies both, since $\log_b(MN) = \log_b M + \log_b N$. Conversely, from additivity, $I(2^k) = k\, I(2)$ and $I(N^k) = k\, I(N)$; one can show (Hartley 1928 used this measure; the uniqueness argument is in Shannon 1948) that every monotone additive $I$ is a **multiple of a logarithm**:

$$
I(N) = c \log_2 N .
$$

The constant $c$ just chooses the unit. With $c = 1$ (base 2) the unit is the **bit**; with base $10$ it is the **decimal digit** (also called a hartley or ban); with base $e$ it is the **nat**. One decimal digit is worth $\log_2 10 \approx 3.32$ bits.

A product of possibilities, a sum of bits — this is the same multiplication-to-addition trick that Napier used, now applied to counting.

## 3. Worked Examples

### Example 1 — Twenty Questions for the whole world

There are about $8.1 \times 10^9$ people. $\log_2(8.1 \times 10^9) = \log_2 8.1 + 9 \log_2 10 \approx 3.02 + 29.90 = 32.92$, so $\lceil 32.92 \rceil = 33$ questions identify anyone on Earth. Going from a city of a million to the whole planet only costs $13$ extra questions.

### Example 2 — digits of $3^{50}$

$\log_{10} 3^{50} = 50 \log_{10} 3 \approx 50 \times 0.47712 = 23.856$. So $3^{50}$ has $\lfloor 23.856 \rfloor + 1 = 24$ decimal digits. In binary: $\log_2 3^{50} = 50 \times 1.58496 = 79.25$, so $80$ bits.

### Example 3 — how many bits for a colour?

A screen colour has red, green and blue channels, each with $256$ levels. The number of colours is $256^3 = 16{,}777{,}216$, and

$$
\log_2 256^3 = 3 \log_2 256 = 3 \times 8 = 24 \text{ bits}.
$$

Additivity in action: three independent channels of $8$ bits each give $24$ bits.

### Example 4 — two passwords compared

- $8$ random lowercase letters: $8 \log_2 26 \approx 8 \times 4.700 = 37.6$ bits.
- $4$ random words from a list of $7776$ words: $4 \log_2 7776 \approx 4 \times 12.925 = 51.7$ bits.

The four-word passphrase has about $14$ more bits, i.e. it is $2^{14} \approx 16{,}000$ times harder to brute-force — and easier to remember.

### Example 5 — when does a decimal number need 64 bits?

A 64-bit unsigned integer holds values up to $2^{64} - 1$. Its number of decimal digits is $\lfloor 64 \log_{10} 2 \rfloor + 1 = \lfloor 19.27 \rfloor + 1 = 20$. So any 20-digit decimal number *might* overflow, and every 19-digit number fits ($10^{19} - 1 < 2^{64}$ since $\log_2 10^{19} \approx 63.12$).

## 4. Python Examples

### 4.1 Playing Twenty Questions by halving

The function below finds a secret integer in $\{0, \dots, N-1\}$ by asking "is it less than $m$?" and counts the questions. The worst case over all secrets should equal $\lceil \log_2 N \rceil$.

```python
import math


def questions_to_find(secret, n_candidates):
    low, high = 0, n_candidates          # the secret is in [low, high)
    questions = 0
    while high - low > 1:
        middle = (low + high) // 2
        questions += 1                   # "is the secret less than middle?"
        if secret < middle:
            high = middle
        else:
            low = middle
    return questions


for n_candidates in [2, 10, 1000, 1_000_000]:
    worst_case = max(questions_to_find(s, n_candidates) for s in range(0, n_candidates, max(1, n_candidates // 5000)))
    print(f"N = {n_candidates:>9,}: worst case seen {worst_case:>2} questions, ceil(log2 N) = {math.ceil(math.log2(n_candidates))}")
```

For $N = 10^6$ we sampled every 200th secret; the worst case is already $20$, as predicted.

### 4.2 Digits in any base

`math.log(N, b)` can be off by a rounding error exactly at powers of the base (`math.log(1000, 10)` is `2.9999999999999996`), so a robust digit counter checks the result with exact integer arithmetic.

```python
import math


def digits_by_log(n, base):
    return math.floor(math.log(n, base)) + 1


def digits_by_division(n, base):
    count = 0
    while n > 0:
        n //= base
        count += 1
    return count


for n, base in [(999, 10), (1000, 10), (255, 2), (256, 2), (3**50, 10), (3**50, 2), (10**15, 16)]:
    print(f"{n:>24} in base {base:>2}: log formula {digits_by_log(n, base):>2}, exact {digits_by_division(n, base):>2}")
```

Notice $N = 1000$ in base 10: the floating-point log is just below $3$, so the log formula gives $3$ digits instead of $4$. At an exact power of the base the formula is right *mathematically* but fragile *numerically*. Python's integers have a built-in exact answer for base 2: `n.bit_length()`.

```python
import math

for n in [1, 255, 256, 1_000_000, 2**64 - 1, 2**64]:
    print(f"{n:>22}: bit_length = {n.bit_length():>2},  floor(log2 n) + 1 = {math.floor(math.log2(n)) + 1:>2}")
```

Look at $2^{64} - 1$: the float version of the formula says $65$ bits, but the number fits in $64$. Converting $2^{64} - 1$ to a float rounds it **up** to exactly $2^{64}$ (a double has only $53$ bits of precision), and $\log_2 2^{64} = 64$. The logarithm is the right *idea*, but near powers of the base the exact integer method is the right *tool*.

### 4.3 Bits per decimal digit

How many bits does each extra decimal digit cost? By change of base, a $d$-digit number needs about $d \log_2 10 \approx 3.32\,d$ bits. The ratio settles towards $\log_2 10$ as $d$ grows.

```python
import math

for decimal_digits in [1, 3, 10, 30, 100, 1000]:
    largest = 10**decimal_digits - 1
    bits = largest.bit_length()
    print(f"{decimal_digits:>5} decimal digits -> {bits:>5} bits   ratio {bits / decimal_digits:.4f}")
print(f"log2(10) = {math.log2(10):.4f}")
```

### 4.4 Password strength in bits

A small table of common password recipes, each assumed to be chosen uniformly at random.

```python
import math

recipes = [
    ("4-digit PIN", 10, 4),
    ("8 lowercase letters", 26, 8),
    ("8 printable ASCII chars", 95, 8),
    ("12 printable ASCII chars", 95, 12),
    ("4 words from 7776", 7776, 4),
    ("6 words from 7776", 7776, 6),
]

guesses_per_second = 1e10
seconds_per_year = 365.25 * 24 * 3600
for name, alphabet_size, length in recipes:
    bits = length * math.log2(alphabet_size)
    years_to_search_all = 2**bits / guesses_per_second / seconds_per_year
    print(f"{name:<26} {bits:6.1f} bits   exhaustive search at 1e10/s: {years_to_search_all:9.2e} years")
```

Every $10$ bits multiplies the search time by $2^{10} \approx 1000$. On the bit scale the numbers are easy to compare; as raw counts ($10^4$ versus $95^{12} \approx 5 \times 10^{23}$) they are not.

### 4.5 Additivity, checked

```python
import math

cards, die_faces, coin_sides = 52, 6, 2
joint_outcomes = cards * die_faces * coin_sides

bits_separately = math.log2(cards) + math.log2(die_faces) + math.log2(coin_sides)
bits_jointly = math.log2(joint_outcomes)
print(f"log2(52) + log2(6) + log2(2) = {bits_separately:.6f}")
print(f"log2(52 * 6 * 2 = {joint_outcomes})   = {bits_jointly:.6f}")
print("equal:", math.isclose(bits_separately, bits_jointly))

print()
print("units of the same amount of choice (N = 1,000,000):")
n = 1_000_000
print(f"  {math.log2(n):.3f} bits,  {math.log10(n):.3f} decimal digits,  {math.log(n):.3f} nats")
```

## 5. Visualizations

### 5.1 Halving a million candidates

Each question halves the candidate count. On a linear axis the curve collapses to the floor after a few steps; on a log axis it is a **straight line** going down by one factor of 2 per question, reaching $1$ at question $20$.

```python
import matplotlib.pyplot as plt
import numpy as np

n_candidates = 1_000_000
questions = np.arange(0, 21)
remaining = np.ceil(n_candidates / 2.0**questions)

fig, (ax_linear, ax_log) = plt.subplots(1, 2, figsize=(12, 4.5))
ax_linear.plot(questions, remaining, "o-", color="tab:blue")
ax_linear.set_xlabel("questions asked")
ax_linear.set_ylabel("candidates remaining (linear scale)")
ax_linear.set_title("Linear axis")
ax_linear.grid(True, alpha=0.3)

ax_log.plot(questions, remaining, "o-", color="tab:blue")
ax_log.set_yscale("log")
ax_log.axhline(1, color="crimson", linestyle="--", label="one candidate left")
ax_log.set_xlabel("questions asked")
ax_log.set_ylabel("candidates remaining (log scale)")
ax_log.set_title("Log axis: a straight line, slope = -log10(2) per question")
ax_log.grid(True, which="both", alpha=0.3)
ax_log.legend()
plt.tight_layout()
plt.show()
```

### 5.2 Digits as a staircase over a log axis

The number of decimal digits of $N$ is constant on each block $[10^{d-1}, 10^d)$ and jumps by one at every power of ten. On a log-scaled $N$ axis the steps are **equally wide**, and the staircase hugs the line $\log_{10} N + 1$ from below. Binary digits make a staircase that is $3.32$ times steeper.

```python
import matplotlib.pyplot as plt
import numpy as np

n_values = np.unique(np.round(np.logspace(0, 6, 3000)).astype(np.int64))
decimal_digits = np.array([len(str(n)) for n in n_values])
binary_digits = np.array([int(n).bit_length() for n in n_values])

fig, ax = plt.subplots(figsize=(10, 5))
ax.step(n_values, binary_digits, where="post", color="tab:orange", label="binary digits (bits)")
ax.step(n_values, decimal_digits, where="post", color="tab:blue", label="decimal digits")
ax.plot(n_values, np.log2(n_values) + 1, color="tab:orange", linestyle=":", label="log2(N) + 1")
ax.plot(n_values, np.log10(n_values) + 1, color="tab:blue", linestyle=":", label="log10(N) + 1")
ax.set_xscale("log")
ax.set_xlabel("N (log scale)")
ax.set_ylabel("number of digits")
ax.set_title("Digits of N: floor(log_b N) + 1")
ax.grid(True, which="both", alpha=0.3)
ax.legend()
plt.show()
```

### 5.3 The bit ladder of passwords

Plotting each recipe at its bit count puts all of them on one additive scale. The vertical lines are a rough guide to what is feasible to search exhaustively at $10^{10}$ guesses per second.

```python
import math

import matplotlib.pyplot as plt

recipes = [
    ("4-digit PIN", 10, 4),
    ("8 lowercase", 26, 8),
    ("8 printable ASCII", 95, 8),
    ("4 words (7776)", 7776, 4),
    ("12 printable ASCII", 95, 12),
    ("6 words (7776)", 7776, 6),
]
names = [name for name, _, _ in recipes]
bits = [length * math.log2(alphabet) for _, alphabet, length in recipes]

guesses_per_second = 1e10
one_day_bits = math.log2(guesses_per_second * 86400)
one_millennium_bits = math.log2(guesses_per_second * 86400 * 365.25 * 1000)

fig, ax = plt.subplots(figsize=(10, 4.5))
ax.barh(names, bits, color="tab:green", alpha=0.75)
for y, value in enumerate(bits):
    ax.text(value + 1, y, f"{value:.1f}", va="center")
ax.axvline(one_day_bits, color="crimson", linestyle="--", label=f"one day of guessing ({one_day_bits:.0f} bits)")
ax.axvline(one_millennium_bits, color="purple", linestyle="--", label=f"1000 years of guessing ({one_millennium_bits:.0f} bits)")
ax.set_xlabel("strength in bits = L * log2(A)   (each bit doubles the search)")
ax.set_xlim(0, 90)
ax.set_title("Password search spaces on the bit scale")
ax.legend(loc="lower right")
ax.grid(True, axis="x", alpha=0.3)
plt.tight_layout()
plt.show()
```

The distance between the "one day" and "1000 years" lines is only $\log_2(365{,}250) \approx 18.5$ bits: on the logarithmic scale a factor of a third of a million is less than three extra random lowercase letters.

## 7. Solutions

1. Card: $\lceil \log_2 52 \rceil = \lceil 5.70 \rceil = 6$. Chessboard: $\log_2 64 = 6$ exactly, so $6$ (row in $3$ questions, column in $3$ — additivity again).
2. $\log_{10} 2^{200} = 200 \times 0.30103 = 60.206$, so $61$ digits. $\log_{16} 10^{12} = 12 / \log_{10} 16 = 12 / 1.20412 = 9.966$, so $10$ hexadecimal digits.
3. IDs: $\lceil \log_2 5000 \rceil = \lceil 12.29 \rceil = 13$ bits. The number $5000$: $\lfloor 12.29 \rfloor + 1 = 13$ bits as well (they differ only at powers of two).
4. Estimate: $30 \times 3.32 \approx 99.6$, so about $100$ bits. Exactly: $\log_2(10^{30} - 1) \approx 99.66$, so `bit_length` is $100$.
5. $10 \log_2 62 \approx 10 \times 5.954 = 59.5$ bits. Lowercase: $L \times 4.700 \ge 59.54$ gives $L \ge 12.67$, so $13$ letters.
6. With $k - 1$ questions there are only $2^{k-1}$ answer sequences. If $k = \lceil \log_2 N \rceil$ then $2^{k-1} < N$, so by the pigeonhole principle two different secrets get the same answer sequence and cannot be told apart. Hence some secret needs the $k$-th question.
7. Each three-way question at best leaves a third of the candidates (the "equal" branch catches one, the other two split the rest), so after $k$ questions about $N / 3^k$ remain and $k \approx \log_3 N$. The ratio is $\log_2 N / \log_3 N = \ln 3 / \ln 2 = \log_2 3 \approx 1.585$: three-way questions are worth $1.585$ bits each.
8. Induction: $I(N^{k}) = I(N^{k-1} \cdot N) = I(N^{k-1}) + I(N) = \dots = k I(N)$. And $I(1) = I(1 \cdot 1) = 2 I(1)$ forces $I(1) = 0$. For $I = \ln$: $\ln N / \ln 2 = \log_2 N$ by the change-of-base formula — checked below.
9. $\frac{d}{db}\frac{b}{\ln b} = \frac{\ln b - 1}{(\ln b)^2} = 0$ at $b = e \approx 2.718$. Base $3$ has $3 / \ln 3 \approx 2.731$, base $2$ has $2 / \ln 2 \approx 2.885$: base $3$ is the most economical integer base, about $5.4\%$ better than binary.

Numerical checks:

```python
import math

import sympy as sp

print("Ex 1:", math.ceil(math.log2(52)), round(math.log2(64), 12))
print("Ex 2:", len(str(2**200)), len(format(10**12, "x")))
print("Ex 3:", math.ceil(math.log2(5000)), (5000).bit_length())
print("Ex 4: estimate", 30 * 3.32, " exact bit_length", (10**30 - 1).bit_length())
bits_62 = 10 * math.log2(62)
print("Ex 5:", round(bits_62, 2), "bits; lowercase letters needed:", math.ceil(bits_62 / math.log2(26)))
print("Ex 7: log2(3) =", round(math.log2(3), 4))
for n in [3, 5, 7]:
    print(f"Ex 8: ln({n})/ln(2) = {math.log(n) / math.log(2):.6f}, log2({n}) = {math.log2(n):.6f}")

b = sp.symbols("b", positive=True)
economy = b / sp.log(b)
critical_points = sp.solve(sp.diff(economy, b), b)
print("Ex 9: minimiser", critical_points, "=", [float(c) for c in critical_points])
print(f"      b/ln b: base 2 -> {2 / math.log(2):.4f}, base e -> {math.e:.4f}, base 3 -> {3 / math.log(3):.4f}")
print(f"      base 3 saves {100 * (1 - (3 / math.log(3)) / (2 / math.log(2))):.1f}% over base 2")
```

## 8. Conclusion

- Counting choices is a logarithm: $k$ yes/no questions distinguish $2^k$ objects, so identifying one of $N$ takes $\lceil \log_2 N \rceil$ questions. $\;2^k \ge N \iff k \ge \log_2 N$.
- A positive integer has $\lfloor \log_b N \rfloor + 1$ digits in base $b$; storing $N$ distinct values takes $\lceil \log_2 N \rceil$ bits. In code, prefer exact integer tools (`bit_length`, `len(str(n))`) at exact powers of the base, where floating-point logs can fall just short.
- A random password of length $L$ from $A$ symbols carries $L \log_2 A$ bits; every bit doubles the attacker's work, and the log scale makes astronomically different search spaces comparable.
- The logarithm is the **only** monotone measure of choice that is additive for independent choices — the bit, the decimal digit and the nat are the same idea in different bases ($1$ decimal digit $= \log_2 10 \approx 3.32$ bits).
- The picture: halving a set of candidates is a straight line on a log axis, and digit counts are an even staircase on a log-scaled number line.
