# Entropy and Information

This notebook measures **information**. Claude Shannon's 1948 paper *A Mathematical Theory of Communication* answered a question that sounds philosophical but turned out to be an engineering problem: how much does a message tell you? The answer is a logarithm — the **surprise** of an event with probability $p$ is $-\log_2 p$ bits — and its average over all messages, the **entropy**, is the hard limit on how far any data can be compressed.

This is notebook 02 of module 05 (**Information and probability**) in the **Learning Logarithms** series.

## 1. Introduction

A weather station in a desert sends one message a day: `sunny` on $99\%$ of days, `rain` on $1\%$. A station in a temperate city sends `sunny` or `rain` with probability $50\%$ each. Both send one word per day. Which station sends more **information**?

Intuitively the desert station is boring: you could predict `sunny` without listening and be right $99$ times out of $100$. But on the day it says `rain`, that single message is big news. We want a number that captures this:

- a **certain** message (probability $1$) carries **no** information;
- a **rare** message carries **a lot**;
- two **independent** messages together carry the **sum** of their information.

The last requirement is the key. Independent probabilities **multiply** ($p \cdot q$) but information should **add**. Turning products into sums is exactly what a logarithm does — the same trick that Napier used for astronomical arithmetic and that notebook 01 used for counting equally likely choices. The result is

$$
\text{surprise}(p) = \log_2 \frac{1}{p} = -\log_2 p \text{ bits}, \qquad \text{entropy} = \text{average surprise}.
$$

The desert station averages only $0.08$ bits per day; the city station a full $1$ bit. The practical meaning: a year of desert reports can be compressed into about $30$ bits, a year of city reports cannot be squeezed below $365$.

Telegraph engineers had already discovered the idea by hand: Morse code gives the most common English letter, E, the shortest symbol (a single dot), so that frequent letters cost little.

## 2. Theory

### 2.1 Surprise: what the axioms force

Let $I(p)$ be the information carried by an event of probability $p \in (0, 1]$. Ask for:

1. $I(1) = 0$ — a certain event tells nothing.
2. $I$ is **decreasing** and **continuous** — rarer means more surprising, and a tiny change in $p$ makes a tiny change in $I$.
3. $I(pq) = I(p) + I(q)$ — two independent events (joint probability $pq$) carry the sum of their information.

Write $p = 2^{-x}$, i.e. $x = -\log_2 p$, and set $f(x) = I(2^{-x})$. Axiom 3 becomes $f(x + y) = f(x) + f(y)$, Cauchy's functional equation, whose only continuous solutions are $f(x) = c\,x$. Therefore

$$
I(p) = c \cdot (-\log_2 p), \qquad c > 0.
$$

Choosing $c = 1$ measures information in **bits**:

| Exponent form | Log form |
|---|---|
| $p = 2^{-I}$ | $I = -\log_2 p = \log_2 \dfrac{1}{p}$ |

A fair coin flip ($p = 1/2$) carries $1$ bit; one specific card from a deck ($p = 1/52$) carries $\log_2 52 \approx 5.70$ bits — matching notebook 01, where $N$ equally likely options were worth $\log_2 N$ bits. Using $\ln$ instead gives the unit **nat**: $1 \text{ nat} = \log_2 e \approx 1.443$ bits.

### 2.2 Entropy: expected surprise

A random source emits symbol $i$ with probability $p_i$. The **entropy** is the expected value of the surprise:

$$
H(X) = E\!\left[-\log_2 p(X)\right] = \sum_{i} p_i \log_2 \frac{1}{p_i} = -\sum_i p_i \log_2 p_i \quad \text{bits per symbol},
$$

with the convention $0 \log_2 0 = 0$ (justified by $\lim_{p \to 0^+} p \log_2 p = 0$: impossible symbols contribute nothing).

For two outcomes with probabilities $p$ and $1 - p$ this is the **binary entropy function**

$$
h(p) = -p \log_2 p - (1 - p) \log_2 (1 - p),
$$

which is $0$ at $p = 0$ and $p = 1$ (no uncertainty) and has its maximum $h(1/2) = 1$ bit.

### 2.3 The uniform distribution maximizes entropy

For $n$ outcomes, $0 \le H(X) \le \log_2 n$, with equality on the right exactly for the uniform distribution. Proof with the inequality $\ln x \le x - 1$ (equality only at $x = 1$):

$$
H(X) - \log_2 n = \sum_i p_i \log_2 \frac{1}{n p_i}
= \frac{1}{\ln 2} \sum_i p_i \ln \frac{1}{n p_i}
\le \frac{1}{\ln 2} \sum_i p_i \left(\frac{1}{n p_i} - 1\right)
= \frac{1}{\ln 2}\left(\sum_i \frac{1}{n} - \sum_i p_i\right) = 0.
$$

(The sums run over the $i$ with $p_i > 0$; if some $p_i = 0$ the first bracket sum is $< 1$ and the inequality is strict.) Equality needs $n p_i = 1$ for all $i$. So **entropy measures how spread out a distribution is**: all mass on one outcome gives $0$, a perfectly even spread gives $\log_2 n$, the "amount of choice" of notebook 01.

### 2.4 Additivity and the chain rule

For independent $X$ and $Y$, $p(x, y) = p(x) p(y)$, and the log turns the product into a sum inside the expectation:

$$
H(X, Y) = E[-\log_2 p(X) - \log_2 p(Y)] = H(X) + H(Y).
$$

Shannon also required a **grouping** property — a choice can be split into stages and the entropies add, weighted by how often each stage is reached:

$$
H\!\left(\tfrac12, \tfrac13, \tfrac16\right) = H\!\left(\tfrac12, \tfrac12\right) + \tfrac12\, H\!\left(\tfrac23, \tfrac13\right).
$$

Together with continuity and "uniform choices among more options carry more information", this grouping property determines $H$ up to the choice of unit (Shannon 1948, Theorem 2). The logarithm is not a convenient choice; it is the only one.

### 2.5 The compression bound

Encode each symbol with a binary codeword that can be decoded unambiguously (a **prefix code**: no codeword is the start of another). The **Kraft inequality** says codeword lengths $\ell_i$ are possible exactly when

$$
\sum_i 2^{-\ell_i} \le 1 .
$$

Shannon's **source coding theorem** states: the average length $L = \sum p_i \ell_i$ of any such code satisfies

$$
H(X) \le L, \qquad\text{and lengths } \ell_i = \left\lceil -\log_2 p_i \right\rceil \text{ achieve } L < H(X) + 1.
$$

So the ideal codeword length for a symbol is **its surprise**: frequent symbols get short codewords, rare ones long — Morse's rule, made exact. Coding blocks of $k$ symbols at once pushes the overhead down to $1/k$ bit per symbol, so $H$ is the true limit of lossless compression. (The lengths $\lceil -\log_2 p_i \rceil$ satisfy Kraft because $2^{-\lceil -\log_2 p_i \rceil} \le 2^{\log_2 p_i} = p_i$, and the $p_i$ sum to $1$.)

## 3. Worked Examples

### Example 1 — the two weather stations

City: $h(1/2) = \tfrac12 \cdot 1 + \tfrac12 \cdot 1 = 1$ bit per day.

Desert: `sunny` has surprise $-\log_2 0.99 \approx 0.0145$ bits, `rain` has $-\log_2 0.01 \approx 6.644$ bits. Weighted by how often each is sent:

$$
h(0.01) = 0.99 \times 0.0145 + 0.01 \times 6.644 \approx 0.0144 + 0.0664 = 0.0808 \text{ bits per day}.
$$

Over a year: $365 \times 0.0808 \approx 29.5$ bits, against $365$ bits for the city.

### Example 2 — a fair die

Six outcomes, each with surprise $\log_2 6$. The average of a constant is the constant: $H = \log_2 6 \approx 2.585$ bits, the maximum possible for six outcomes.

### Example 3 — a lopsided four-letter alphabet

Symbols A, B, C, D with probabilities $\tfrac12, \tfrac14, \tfrac18, \tfrac18$. Surprises: $1, 2, 3, 3$ bits.

$$
H = \tfrac12 \cdot 1 + \tfrac14 \cdot 2 + \tfrac18 \cdot 3 + \tfrac18 \cdot 3 = 0.5 + 0.5 + 0.375 + 0.375 = 1.75 \text{ bits}.
$$

The code A = `0`, B = `10`, C = `110`, D = `111` uses codeword lengths equal to the surprises and averages exactly $1.75$ bits — better than the $2$ bits of the obvious fixed-length code, and as good as possible, because $L \ge H$.

### Example 4 — entropy in nats

The same distribution in nats: multiply by $\ln 2$: $1.75 \times 0.6931 \approx 1.213$ nats. Changing the base of the logarithm only changes the unit, $H_{\text{nats}} = H_{\text{bits}} \ln 2$.

### Example 5 — independent sources add

A fair coin ($1$ bit) and a fair die ($2.585$ bits) together: $12$ equally likely pairs, $\log_2 12 = \log_2 2 + \log_2 6 \approx 3.585$ bits.

## 4. Python Examples

### 4.1 Surprise and entropy functions

`np.log2(0)` is $-\infty$, and $0 \times (-\infty)$ is `nan`, so the entropy function drops zero probabilities explicitly — the $0 \log_2 0 = 0$ convention in code.

```python
import numpy as np


def surprise_bits(p):
    return -np.log2(p)


def entropy_bits(probabilities):
    p = np.asarray(probabilities, dtype=float)
    assert np.isclose(p.sum(), 1.0), "probabilities must sum to 1"
    p = p[p > 0]                      # convention: 0 * log2(0) = 0
    return float(np.sum(p * -np.log2(p)))


print("surprise of sunny (0.99):", round(float(surprise_bits(0.99)), 4), "bits")
print("surprise of rain  (0.01):", round(float(surprise_bits(0.01)), 4), "bits")
print()
print("H(desert)          =", round(entropy_bits([0.99, 0.01]), 4))
print("H(city)            =", round(entropy_bits([0.5, 0.5]), 4))
print("H(fair die)        =", round(entropy_bits([1 / 6] * 6), 4), " log2(6) =", round(np.log2(6), 4))
print("H(1/2,1/4,1/8,1/8) =", round(entropy_bits([0.5, 0.25, 0.125, 0.125]), 4))
print("H(certain)         =", entropy_bits([1.0, 0.0, 0.0]))
print("in nats:", round(entropy_bits([0.5, 0.25, 0.125, 0.125]) * np.log(2), 4))
```

### 4.2 Additivity and grouping, checked

```python
import numpy as np


def entropy_bits(probabilities):
    p = np.asarray(probabilities, dtype=float)
    p = p[p > 0]
    return float(np.sum(p * -np.log2(p)))


coin = np.array([0.5, 0.5])
die = np.full(6, 1 / 6)
joint = np.outer(coin, die).ravel()          # independent: p(x, y) = p(x) p(y)
print(f"H(coin) + H(die) = {entropy_bits(coin) + entropy_bits(die):.6f}")
print(f"H(coin, die)     = {entropy_bits(joint):.6f}")

left = entropy_bits([1 / 2, 1 / 3, 1 / 6])
right = entropy_bits([1 / 2, 1 / 2]) + 0.5 * entropy_bits([2 / 3, 1 / 3])
print(f"grouping: {left:.6f} vs {right:.6f}  equal: {np.isclose(left, right)}")
```

### 4.3 Entropy of English letters from a sample

We estimate letter probabilities from a short passage (written for this notebook), keeping the $26$ letters plus the space and ignoring punctuation and case. This is a **first-order** estimate: it treats each character as independent of the previous ones.

```python
SAMPLE_TEXT = """
The lighthouse keeper woke before dawn, as he had done every morning for thirty years.
He climbed the narrow stairs, trimmed the wick, and watched the grey sea turn slowly to silver.
Most days nothing happened at all. A fishing boat would pass, a gull would complain about the wind,
and the tide would come in and go out again exactly when the almanac said it would.
He wrote the same three words in his logbook almost every evening: calm, clear, quiet.
Once in a long while the logbook said something else. There was the night of the great storm,
when the waves broke over the rocks and the lamp was the only thing that stood between a ship and the reef.
There was the morning a whale surfaced so close to the shore that he could hear it breathe.
Those were the pages his grandchildren wanted to read. Nobody ever asked about the calm days,
because a calm day told them nothing they did not already expect. The rare entries were the ones
that carried the news, and the keeper understood, without ever putting it into numbers,
that the less likely a message is, the more it tells you when it finally arrives.
"""

import collections
import math

characters = [c for c in SAMPLE_TEXT.lower().replace("\n", " ") if c.isalpha() or c == " "]
cleaned = " ".join("".join(characters).split())      # collapse repeated spaces
counts = collections.Counter(cleaned)
total = sum(counts.values())

entropy_estimate = 0.0
for symbol, count in counts.items():
    p = count / total
    entropy_estimate -= p * math.log2(p)

print("characters used:", total, " distinct symbols:", len(counts))
print("most common:", [(s if s != " " else "space", round(c / total, 3)) for s, c in counts.most_common(6)])
print(f"estimated entropy: {entropy_estimate:.3f} bits per character")
print(f"uniform over 27 symbols: {math.log2(27):.3f} bits per character")
print(f"8-bit ASCII spends: 8 bits per character")
```

Even this crude estimate is well below $\log_2 27 \approx 4.75$ bits: English letters are far from equally likely. Shannon's 1951 paper *Prediction and Entropy of Printed English* estimated, from large letter-frequency tables, a first-order value of about $4$ bits per character, and showed that taking context into account (an `h` after `t`, a `u` after `q`) brings the entropy down to roughly one bit per character. That redundancy is what makes text so compressible.

### 4.4 Shannon code lengths for the sample

Assign each symbol a codeword of length $\lceil -\log_2 p \rceil$. Check Kraft's inequality and compare the average length with the entropy: it must land in $[H, H + 1)$.

```python
SAMPLE_TEXT = """
The lighthouse keeper woke before dawn, as he had done every morning for thirty years.
He climbed the narrow stairs, trimmed the wick, and watched the grey sea turn slowly to silver.
Most days nothing happened at all. A fishing boat would pass, a gull would complain about the wind,
and the tide would come in and go out again exactly when the almanac said it would.
He wrote the same three words in his logbook almost every evening: calm, clear, quiet.
Once in a long while the logbook said something else. There was the night of the great storm,
when the waves broke over the rocks and the lamp was the only thing that stood between a ship and the reef.
There was the morning a whale surfaced so close to the shore that he could hear it breathe.
Those were the pages his grandchildren wanted to read. Nobody ever asked about the calm days,
because a calm day told them nothing they did not already expect. The rare entries were the ones
that carried the news, and the keeper understood, without ever putting it into numbers,
that the less likely a message is, the more it tells you when it finally arrives.
"""

import collections
import math

characters = [c for c in SAMPLE_TEXT.lower().replace("\n", " ") if c.isalpha() or c == " "]
cleaned = " ".join("".join(characters).split())
counts = collections.Counter(cleaned)
total = sum(counts.values())

entropy = 0.0
average_length = 0.0
kraft_sum = 0.0
for symbol, count in counts.most_common():
    p = count / total
    length = math.ceil(-math.log2(p))
    entropy -= p * math.log2(p)
    average_length += p * length
    kraft_sum += 2.0**-length

for symbol in [" ", "e", "t", "z" if "z" in counts else "q"]:
    if symbol in counts:
        p = counts[symbol] / total
        print(f"symbol {symbol!r}: p = {p:.4f}, surprise = {-math.log2(p):.2f} bits, codeword length = {math.ceil(-math.log2(p))}")
print(f"Kraft sum = {kraft_sum:.4f} (must be <= 1)")
print(f"H = {entropy:.3f} <= average length {average_length:.3f} < H + 1 = {entropy + 1:.3f}")
```

### 4.5 A real compressor against the bound

Generate a long sequence of biased coin flips ($P(1) = 0.1$), pack them $8$ to a byte, and compress with `zlib` (the DEFLATE algorithm behind ZIP and PNG). No lossless compressor can beat $h(0.1)$ bits per flip on average.

```python
import zlib

import numpy as np

rng = np.random.default_rng(0)
n_flips = 400_000
for p_one in [0.5, 0.1, 0.01]:
    flips = (rng.random(n_flips) < p_one).astype(np.uint8)
    packed = np.packbits(flips).tobytes()
    compressed = zlib.compress(packed, level=9)
    bits_per_flip = 8 * len(compressed) / n_flips
    binary_entropy = -(p_one * np.log2(p_one) + (1 - p_one) * np.log2(1 - p_one))
    print(f"P(1) = {p_one:<5}: entropy bound {binary_entropy:.3f} bits/flip, zlib achieves {bits_per_flip:.3f} bits/flip")
```

`zlib` never beats the bound (it cannot). On the biased sources it stays roughly $20$ to $40\%$ above it: a general-purpose compressor looks for repeated byte patterns and does not know that the data are independent biased coin flips, whereas a code built from the probabilities themselves could approach $h(p)$. For the fair coin no compressor can help: $1$ bit per flip is already optimal, and `zlib` even adds a little overhead.

## 5. Visualizations

### 5.1 Surprise is a straight line on a log axis

$-\log_2 p$ grows without bound as $p \to 0$. With a log-scaled $p$ axis the curve becomes a straight line: each halving of $p$ adds exactly one bit.

```python
import matplotlib.pyplot as plt
import numpy as np

p = np.linspace(0.001, 1, 1000)
p_log = np.logspace(-3, 0, 400)

fig, (ax_linear, ax_log) = plt.subplots(1, 2, figsize=(12, 4.5))
ax_linear.plot(p, -np.log2(p), color="tab:purple")
ax_linear.set_xlabel("probability p (linear scale)")
ax_linear.set_ylabel("surprise -log2(p) in bits")
ax_linear.set_title("Rare events are very surprising")
ax_linear.grid(True, alpha=0.3)

ax_log.plot(p_log, -np.log2(p_log), color="tab:purple")
for k in range(0, 10):
    ax_log.plot(2.0**-k, k, "o", color="tab:orange")
ax_log.set_xscale("log")
ax_log.set_xlabel("probability p (log scale)")
ax_log.set_ylabel("surprise -log2(p) in bits")
ax_log.set_title("Each halving of p adds one bit")
ax_log.grid(True, which="both", alpha=0.3)
plt.tight_layout()
plt.show()
```

### 5.2 The binary entropy curve

$h(p)$ is symmetric about $p = 1/2$, where it reaches $1$ bit. The dashed lines show the two parts of the average: the contribution $-p \log_2 p$ of the outcome with probability $p$ and the contribution of its complement.

```python
import matplotlib.pyplot as plt
import numpy as np

p = np.linspace(1e-6, 1 - 1e-6, 1000)
part_one = -p * np.log2(p)
part_two = -(1 - p) * np.log2(1 - p)
h = part_one + part_two

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(p, h, color="black", linewidth=2, label="h(p)")
ax.plot(p, part_one, "--", color="tab:blue", label="-p log2 p")
ax.plot(p, part_two, "--", color="tab:orange", label="-(1-p) log2(1-p)")
for p_marked, label in [(0.01, "desert"), (0.5, "city")]:
    h_marked = -(p_marked * np.log2(p_marked) + (1 - p_marked) * np.log2(1 - p_marked))
    ax.plot(p_marked, h_marked, "o", color="crimson")
    ax.annotate(f"{label}: {h_marked:.3f} bits", (p_marked, h_marked), textcoords="offset points", xytext=(10, 8))
ax.set_xlabel("p = P(rain)")
ax.set_ylabel("entropy in bits")
ax.set_title("Binary entropy h(p): maximal uncertainty at p = 1/2")
ax.set_ylim(0, 1.12)
ax.grid(True, alpha=0.3)
ax.legend(loc="upper right")
plt.show()
```

### 5.3 No distribution beats the uniform one

Draw $20{,}000$ random probability distributions on $4$ outcomes and compute their entropies. Every one of them is below $\log_2 4 = 2$ bits; only the uniform distribution touches the line.

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(1)
n_outcomes = 4
distributions = rng.dirichlet(np.ones(n_outcomes), size=20_000)
entropies = -np.sum(distributions * np.log2(distributions), axis=1)

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.hist(entropies, bins=80, color="tab:green", alpha=0.75)
ax.axvline(np.log2(n_outcomes), color="crimson", linestyle="--", label="log2(4) = 2 bits (uniform)")
ax.set_xlabel("entropy in bits")
ax.set_ylabel("number of random distributions")
ax.set_title(f"Entropies of random 4-outcome distributions (max seen: {entropies.max():.4f})")
ax.legend(loc="upper left")
ax.grid(True, alpha=0.3)
plt.show()
```

### 5.4 Letter surprises: frequent letters get short codes

For each symbol of the sample text, the bar shows its surprise $-\log_2 p$ — the ideal codeword length. Space and `e` cost three bits or less; letters that appear only once or twice in the sample cost nine or ten bits.

```python
SAMPLE_TEXT = """
The lighthouse keeper woke before dawn, as he had done every morning for thirty years.
He climbed the narrow stairs, trimmed the wick, and watched the grey sea turn slowly to silver.
Most days nothing happened at all. A fishing boat would pass, a gull would complain about the wind,
and the tide would come in and go out again exactly when the almanac said it would.
He wrote the same three words in his logbook almost every evening: calm, clear, quiet.
Once in a long while the logbook said something else. There was the night of the great storm,
when the waves broke over the rocks and the lamp was the only thing that stood between a ship and the reef.
There was the morning a whale surfaced so close to the shore that he could hear it breathe.
Those were the pages his grandchildren wanted to read. Nobody ever asked about the calm days,
because a calm day told them nothing they did not already expect. The rare entries were the ones
that carried the news, and the keeper understood, without ever putting it into numbers,
that the less likely a message is, the more it tells you when it finally arrives.
"""

import collections

import matplotlib.pyplot as plt
import numpy as np

characters = [c for c in SAMPLE_TEXT.lower().replace("\n", " ") if c.isalpha() or c == " "]
cleaned = " ".join("".join(characters).split())
counts = collections.Counter(cleaned)
total = sum(counts.values())

symbols = [s for s, _ in counts.most_common()]
probabilities = np.array([counts[s] / total for s in symbols])
surprises = -np.log2(probabilities)
entropy = float(np.sum(probabilities * surprises))
labels = ["spc" if s == " " else s for s in symbols]

fig, ax = plt.subplots(figsize=(11, 4.5))
ax.bar(labels, surprises, color="tab:purple", alpha=0.75)
ax.axhline(entropy, color="black", linestyle="--", label=f"entropy (average surprise) = {entropy:.2f} bits")
ax.set_xlabel("symbol, most frequent first")
ax.set_ylabel("surprise -log2(p) in bits")
ax.set_title("Surprise of each character in the sample text")
ax.legend(loc="upper left")
ax.grid(True, axis="y", alpha=0.3)
plt.show()
```

5. The pair has $36$ equally likely values, so $H = \log_2 36 \approx 5.170$ bits. The sum takes $11$ values with probabilities $\tfrac{1}{36}, \tfrac{2}{36}, \dots, \tfrac{6}{36}, \dots, \tfrac{1}{36}$; its entropy is about $3.274$ bits. Summing **loses information**: $(1, 6)$, $(2, 5)$, … all give $7$ and can no longer be told apart.
6. $h'(p) = -\log_2 p - \tfrac{1}{\ln 2} + \log_2(1 - p) + \tfrac{1}{\ln 2} = \log_2 \tfrac{1-p}{p}$, which is zero only when $\tfrac{1-p}{p} = 1$, i.e. $p = \tfrac12$. SymPy check below.
7. Each term $p_i \log_2 (1/p_i)$ is $\ge 0$ because $0 < p_i \le 1$ gives $1/p_i \ge 1$ and so $\log_2(1/p_i) \ge 0$. (The inequality $\ln x \le x - 1$ with $x = p_i$ gives the same: $-\ln p_i \ge 1 - p_i \ge 0$.) The sum is $0$ only if every term is $0$, i.e. every $p_i$ is $0$ or $1$; since they sum to $1$, exactly one is $1$.
8. Left side $\approx 1.8464$ bits. Right side: $h(0.4) \approx 0.9710$ plus $0.6 \times H(\tfrac12, \tfrac13, \tfrac16) \approx 0.6 \times 1.4591 = 0.8755$, total $1.8464$ ✓.
9. Lengths $\lceil 1.32 \rceil, \lceil 1.74 \rceil, \lceil 2.32 \rceil, \lceil 3.32 \rceil = 2, 2, 3, 4$. Kraft: $\tfrac14 + \tfrac14 + \tfrac18 + \tfrac1{16} = 0.6875 \le 1$. Average length $L = 0.8 + 0.6 + 0.6 + 0.4 = 2.4$ bits, and $1.846 \le 2.4 < 2.846$ ✓. (A better code, `0`, `10`, `110`, `111`, achieves $1.9$ bits; the Shannon code only guarantees the bound.)

Numerical checks:

```python
import math

import numpy as np
import sympy as sp


def entropy_bits(probabilities):
    p = np.asarray(probabilities, dtype=float)
    p = p[p > 0]
    return float(np.sum(p * -np.log2(p)))


print("Ex 1:", round(math.log2(6), 3), round(math.log2(52), 3), round(math.log2(6 * 52), 3))
print("Ex 2:", entropy_bits([0.5, 0.25, 0.25]))
print("Ex 3:", entropy_bits([1 / 8] * 8))
print("Ex 4: estimate", round(0.01 * math.log2(100), 4), " exact", round(entropy_bits([0.99, 0.01]), 4))

pair_counts = np.ones((6, 6))
sum_counts = np.array([sum(1 for a in range(1, 7) for b in range(1, 7) if a + b == s) for s in range(2, 13)])
print("Ex 5: H(pair) =", round(entropy_bits(pair_counts.ravel() / 36), 4), " H(sum) =", round(entropy_bits(sum_counts / 36), 4))

p = sp.symbols("p", positive=True)
h = -p * sp.log(p, 2) - (1 - p) * sp.log(1 - p, 2)
derivative = sp.simplify(sp.diff(h, p))
print("Ex 6: h'(p) =", derivative, "  critical points:", sp.solve(derivative, p))

distribution = [0.4, 0.3, 0.2, 0.1]
left = entropy_bits(distribution)
right = entropy_bits([0.4, 0.6]) + 0.6 * entropy_bits([0.3 / 0.6, 0.2 / 0.6, 0.1 / 0.6])
print(f"Ex 8: {left:.4f} vs {right:.4f}")

lengths = [math.ceil(-math.log2(q)) for q in distribution]
kraft = sum(2.0**-length for length in lengths)
average_length = sum(q * length for q, length in zip(distribution, lengths))
print("Ex 9: lengths", lengths, " Kraft sum", kraft, " L =", round(average_length, 4), " H =", round(left, 4))
```

## 8. Conclusion

- The information in an event of probability $p$ is its **surprise** $-\log_2 p$ bits. $\;p = 2^{-I} \iff I = -\log_2 p$. Requiring that independent events add their information **forces** the logarithm.
- **Entropy** is expected surprise, $H(X) = -\sum p_i \log_2 p_i$. It is $0$ for a certain outcome and at most $\log_2 n$, reached only by the uniform distribution. Changing the log base changes only the unit (bits, nats, decimal digits).
- The binary entropy $h(p)$ peaks at $1$ bit for a fair coin; a predictable source like the desert station carries very little information per message.
- English letters have a first-order entropy of about $4$ bits per character, well below the $\log_2 27 \approx 4.75$ of a uniform alphabet — the redundancy that compressors exploit.
- **Compression bound**: no lossless code can average fewer than $H$ bits per symbol, and codeword lengths $\lceil -\log_2 p_i \rceil$ achieve less than $H + 1$. The ideal length of a codeword is the surprise of its symbol.
- The picture: on a log-scaled probability axis, surprise is a straight line — each halving of $p$ costs one more bit.
