# Multiplicative Processes: the Log-Normal Distribution and Benford's Law

This notebook follows randomness that **multiplies** instead of adds. A quantity that grows by random percentages — a population, a company's revenue, the size of a river basin, a price — is a product of many random factors. Its logarithm is a *sum*, so the central limit theorem applies to the logarithm, and the quantity itself becomes **log-normal**. The same logarithmic view explains a famous oddity: in many real-world tables, about $30\%$ of the numbers start with the digit $1$ and fewer than $5\%$ start with $9$.

This is notebook 05 of module 05 (**Information and probability**) in the **Learning Logarithms** series.

## 1. Introduction

Take any large table of naturally occurring numbers — populations of towns, areas of lakes, account balances, physical constants — and look only at the **first digit** of each number. A natural guess is that each of $1, 2, \dots, 9$ appears about $1/9 \approx 11\%$ of the time. It does not. Typically

| first digit | 1 | 2 | 3 | 4 | 5 | 6 | 7 | 8 | 9 |
|---|---|---|---|---|---|---|---|---|---|
| share | 30.1% | 17.6% | 12.5% | 9.7% | 7.9% | 6.7% | 5.8% | 5.1% | 4.6% |

The astronomer Simon Newcomb noticed in 1881 that the early pages of books of logarithm tables were far more worn than the later ones: people looked up numbers starting with $1$ more often than numbers starting with $9$. The physicist Frank Benford rediscovered the pattern in 1938 and checked it on many data sets, and it now carries his name. It is used, for example, as a screening test for fabricated accounting figures, because people who invent numbers tend to spread the first digits evenly.

Why should nature prefer the digit $1$? Without logarithms it is a mystery. With them it is almost obvious: the first digit of $x$ is decided by the **fractional part of $\log_{10} x$**, and on a logarithmic ruler the stretch from $1$ to $2$ is much longer than the stretch from $9$ to $10$:

$$
\log_{10} 2 - \log_{10} 1 \approx 0.301, \qquad \log_{10} 10 - \log_{10} 9 \approx 0.046 .
$$

Quantities produced by **multiplicative** processes spread out evenly on that ruler — and that is where both the log-normal distribution and Benford's law come from.

## 2. Theory

### 2.1 Products of random factors

Let a quantity start at $X_0 > 0$ and in each period be multiplied by an independent positive random factor $G_i$ (for example $G_i = 1 + R_i$ with a random growth rate $R_i > -1$):

$$
X_n = X_0 \, G_1 G_2 \cdots G_n .
$$

The logarithm turns the product into a sum:

| Exponent form | Log form |
|---|---|
| $X_n = X_0 \prod_{i=1}^n G_i$ | $\ln X_n = \ln X_0 + \sum_{i=1}^n \ln G_i$ |

If the $\ln G_i$ are independent with mean $m$ and variance $s^2$, the **central limit theorem** says that the sum is approximately normal for large $n$:

$$
\ln X_n \approx \mathcal N\!\left(\ln X_0 + n m,\; n s^2\right).
$$

So $X_n$ itself is approximately **log-normal**. Additive noise piles up into a bell curve; multiplicative noise piles up into a bell curve **on a log axis**.

### 2.2 The log-normal distribution

$X$ is log-normal with parameters $\mu, \sigma$ when $\ln X \sim \mathcal N(\mu, \sigma^2)$, i.e. $X = e^{Y}$ with $Y$ normal. Its density follows from the change of variables $y = \ln x$, $dy = dx / x$:

$$
f_X(x) = \frac{1}{x \,\sigma \sqrt{2\pi}} \exp\!\left(-\frac{(\ln x - \mu)^2}{2\sigma^2}\right), \qquad x > 0 .
$$

It is skewed to the right: a long tail of very large values, no negative values. Its three centres differ:

- **median** $e^{\mu}$ (half of $Y$ lies below $\mu$, and $e^{(\cdot)}$ is increasing),
- **mean** $E[X] = e^{\mu + \sigma^2/2}$ (the long right tail pulls it up),
- **mode** $e^{\mu - \sigma^2}$ (the peak of the density).

On a log-scaled $x$ axis the histogram of $X$ is a symmetric bell — the geometric picture of the whole notebook.

### 2.3 The first digit lives in the mantissa

Write any $x > 0$ as $x = 10^{\log_{10} x}$ and split the exponent into its whole part and its fractional part, the **mantissa** $\{\log_{10} x\} = \log_{10} x - \lfloor \log_{10} x \rfloor \in [0, 1)$:

$$
x = 10^{\lfloor \log_{10} x \rfloor} \cdot 10^{\{\log_{10} x\}} .
$$

The first factor only moves the decimal point (notebook 01: the whole part counts digits). The second, $10^{\{\log_{10} x\}} \in [1, 10)$, is $x$ written in scientific notation, and its integer part is the first digit. Hence

$$
\text{first digit of } x = d \quad\Longleftrightarrow\quad \log_{10} d \le \{\log_{10} x\} < \log_{10}(d + 1).
$$

### 2.4 Benford's law from a uniform mantissa

If the mantissa $\{\log_{10} X\}$ is **uniformly distributed** on $[0, 1)$, the probability of the first digit $d$ is the length of its interval:

$$
P(d) = \log_{10}(d + 1) - \log_{10} d = \log_{10}\!\left(1 + \frac1d\right), \qquad d = 1, \dots, 9 .
$$

The nine probabilities telescope to $\log_{10} 10 - \log_{10} 1 = 1$. In exponent form: the first digit is $d$ when $X$ lands in $[d, d + 1) \times 10^k$, and on a log ruler these intervals have lengths $0.301, 0.176, \dots, 0.046$ per decade.

### 2.5 When is the mantissa uniform?

- **Wide multiplicative spread.** If $\log_{10} X$ has a smooth distribution spread over several units (several orders of magnitude of $X$), wrapping it around the unit interval gives an almost flat mantissa. The log-normal from 2.1 has $\operatorname{sd}(\log_{10} X) = \sigma / \ln 10$, which keeps growing with $n$ — so multiplicative processes drift towards Benford's law the longer they run.
- **Scale invariance.** If a law of first digits is to hold whatever the units (dollars or euros, metres or feet), multiplying all data by a constant $c$ must not change it. Multiplying by $c$ **shifts** $\log_{10} X$ by $\log_{10} c$, and the only distribution of the mantissa that is unchanged by every shift on the circle $[0, 1)$ is the uniform one. Benford's law is the only scale-invariant first-digit law.
- **Powers of a number.** For $2^n$, $\log_{10} 2^n = n \log_{10} 2$. Because $\log_{10} 2$ is irrational, the equidistribution theorem says the fractional parts $\{n \log_{10} 2\}$ fill $[0, 1)$ evenly as $n$ runs through $1, 2, 3, \dots$ So the first digits of the powers of two follow Benford's law exactly in the limit.

And when it fails: data confined to less than one order of magnitude (adult heights in centimetres all start with $1$ or $2$), assigned numbers (telephone numbers, postcodes), or uniformly distributed quantities do not follow Benford's law.

## 3. Worked Examples

### Example 1 — the Benford probabilities

$P(1) = \log_{10} 2 \approx 0.3010$, $P(2) = \log_{10} 1.5 \approx 0.1761$, $P(9) = \log_{10}(10/9) \approx 0.0458$. A number starting with $1$ is $0.3010 / 0.0458 \approx 6.6$ times as common as one starting with $9$.

### Example 2 — the first digit of $2^{100}$ without computing it

$\log_{10} 2^{100} = 100 \times 0.30103 = 30.103$. The whole part $30$ says $2^{100}$ has $31$ digits; the mantissa $0.103$ lies in $[\log_{10} 1, \log_{10} 2) = [0, 0.301)$, so the first digit is $1$. More precisely $10^{0.103} \approx 1.27$, so $2^{100} \approx 1.27 \times 10^{30}$ (exactly $1{,}267{,}650{,}600{,}228{,}229{,}401{,}496{,}703{,}205{,}376$).

### Example 3 — first digit at most 3

$P(d \le 3) = \log_{10} 2 + \log_{10}\tfrac32 + \log_{10}\tfrac43 = \log_{10}\left(2 \cdot \tfrac32 \cdot \tfrac43\right) = \log_{10} 4 \approx 0.602$. The product telescopes — a sum of logs is the log of a product.

### Example 4 — centres of a log-normal

$\mu = 0$, $\sigma = 1$: median $e^0 = 1$, mean $e^{0.5} \approx 1.649$, mode $e^{-1} \approx 0.368$. Mode $<$ median $<$ mean, as for every right-skewed distribution.

### Example 5 — a multiplicative growth process

Each year a town's population is multiplied by a random factor whose logarithm has mean $m = 0.01$ and standard deviation $s = 0.1$. After $n = 100$ years, $\ln(X_{100}/X_0)$ is approximately normal with mean $100 \times 0.01 = 1$ and standard deviation $\sqrt{100} \times 0.1 = 1$. So the typical (median) town has grown by a factor $e^1 \approx 2.7$, but the **average** growth factor is $e^{1 + 1/2} \approx 4.5$, pulled up by a few towns that boomed.

## 4. Python Examples

### 4.1 Simulating a multiplicative process

$20{,}000$ independent "companies" start at size $1$ and are multiplied each year for $100$ years by a factor drawn uniformly from $[0.75, 1.30]$. That factor is **not** log-normal, yet the logarithm of the final size is close to normal, as the central limit theorem predicts.

```python
import numpy as np

rng = np.random.default_rng(2024)
n_companies, n_years = 20_000, 100
factors = rng.uniform(0.75, 1.30, size=(n_companies, n_years))
final_sizes = np.prod(factors, axis=1)
log_sizes = np.log(final_sizes)

# CLT prediction from the moments of ln(G) for G ~ Uniform(0.75, 1.30)
grid = np.linspace(0.75, 1.30, 200_001)
mean_log_factor = np.mean(np.log(grid))
var_log_factor = np.mean(np.log(grid) ** 2) - mean_log_factor**2
print(f"predicted ln X: mean {n_years * mean_log_factor:.3f}, sd {np.sqrt(n_years * var_log_factor):.3f}")
print(f"simulated ln X: mean {log_sizes.mean():.3f}, sd {log_sizes.std():.3f}")

skewness_of_x = np.mean((final_sizes - final_sizes.mean()) ** 3) / final_sizes.std() ** 3
skewness_of_log_x = np.mean((log_sizes - log_sizes.mean()) ** 3) / log_sizes.std() ** 3
print(f"skewness of X: {skewness_of_x:.2f}   skewness of ln X: {skewness_of_log_x:.3f}")
print(f"median of X: {np.median(final_sizes):.3f}, mean of X: {final_sizes.mean():.3f}, largest: {final_sizes.max():.1f}")
```

$X$ is strongly skewed, $\ln X$ is nearly symmetric. The median company has grown by a factor of about $3.5$, but the mean is more than three times larger, inflated by a few lucky companies that grew by a factor of a thousand or more — a typical log-normal signature.

### 4.2 Log-normal centres, checked by simulation

```python
import numpy as np

rng = np.random.default_rng(7)
mu, sigma = 0.0, 1.0
samples = np.exp(rng.normal(mu, sigma, size=2_000_000))

# the mode of X is not exp(mode of ln X) = e^mu: estimate it from a histogram of X itself
x_counts, x_edges = np.histogram(samples, bins=np.linspace(0, 3, 301))
x_mode_estimate = 0.5 * (x_edges[np.argmax(x_counts)] + x_edges[np.argmax(x_counts) + 1])

print(f"median: simulated {np.median(samples):.4f}, formula e^mu = {np.exp(mu):.4f}")
print(f"mean  : simulated {samples.mean():.4f}, formula e^(mu + sigma^2/2) = {np.exp(mu + sigma**2 / 2):.4f}")
print(f"mode  : histogram {x_mode_estimate:.3f}, formula e^(mu - sigma^2) = {np.exp(mu - sigma**2):.4f}")
```

### 4.3 First digits from the mantissa

Reading the first digit off $\{\log_{10} x\}$ and comparing with the decimal string. For huge exact integers we take $\log_{10}$ from Python's arbitrary-precision integers via `math.log10`, which accepts them directly.

```python
import math


def first_digit_from_log(x):
    mantissa = math.log10(x) % 1.0
    return int(10**mantissa)


def first_digit_from_string(n):
    return int(str(n)[0])


for exponent in [10, 50, 100, 1000]:
    n = 2**exponent
    mantissa = math.log10(n) % 1.0
    print(f"2^{exponent:<5} log10 = {math.log10(n):10.4f}  mantissa {mantissa:.4f}  10^mantissa = {10**mantissa:.4f}"
          f"  first digit: log {first_digit_from_log(n)}, string {first_digit_from_string(n)}")
```

### 4.4 Benford's law for the powers of two

The first digits of $2^1, 2^2, \dots, 2^{10000}$, counted exactly from the integers, against $\log_{10}(1 + 1/d)$.

```python
import math
from collections import Counter

n_powers = 10_000
counts = Counter()
power = 1
for _ in range(n_powers):
    power *= 2
    counts[int(str(power)[0])] += 1

print(" d   observed   Benford")
for d in range(1, 10):
    print(f" {d}   {counts[d] / n_powers:.4f}     {math.log10(1 + 1 / d):.4f}")
```

### 4.5 A chi-square test against Benford's law

For counts $O_d$ from $N$ numbers and expected counts $E_d = N \log_{10}(1 + 1/d)$, the statistic $\chi^2 = \sum_d (O_d - E_d)^2 / E_d$ has approximately a chi-square distribution with $8$ degrees of freedom if the data follow Benford's law; values above about $15.5$ happen only $5\%$ of the time. We test the multiplicative companies, an additive process (sums of the same number of uniform steps), and uniformly distributed numbers.

```python
import numpy as np

rng = np.random.default_rng(11)
benford = np.log10(1 + 1 / np.arange(1, 10))


def first_digits(values):
    values = np.asarray(values, dtype=float)
    mantissas = np.mod(np.log10(values), 1.0)
    return np.floor(10**mantissas + 1e-12).astype(int)


def chi_square_against_benford(values):
    digits = first_digits(values)
    observed = np.array([np.sum(digits == d) for d in range(1, 10)])
    expected = len(digits) * benford
    return float(np.sum((observed - expected) ** 2 / expected)), observed / len(digits)


data_sets = {
    "multiplicative (product of 100 factors)": np.prod(rng.uniform(0.75, 1.30, size=(20_000, 100)), axis=1),
    "multiplicative, 400 factors": np.prod(rng.uniform(0.75, 1.30, size=(20_000, 400)), axis=1),
    "additive (sum of 100 uniform steps)": np.sum(rng.uniform(0, 2, size=(20_000, 100)), axis=1),
    "uniform on [1, 1000)": rng.uniform(1, 1000, size=20_000),
}
print("critical value at 5%, 8 degrees of freedom: 15.51")
for name, values in data_sets.items():
    chi_square, shares = chi_square_against_benford(values)
    print(f"{name:<42} chi^2 = {chi_square:9.1f}   share of 1s = {shares[0]:.3f}")
```

The additive process sits in a narrow band around its mean ($100$, with a standard deviation of about $6$), so nearly every value starts with $8$, $9$ or $1$ — no Benford there. Uniform numbers have nearly equal first-digit shares. Both multiplicative data sets spread over several orders of magnitude and pass the test comfortably.

### 4.6 Scale invariance

Convert the "multiplicative, 400 factors" data from one currency to another (multiply by $0.92$) or from feet to metres (multiply by $0.3048$): the Benford fit does not care. Do the same to uniform data and the first-digit shares change completely.

```python
import numpy as np

rng = np.random.default_rng(12)


def first_digit_shares(values):
    mantissas = np.mod(np.log10(values), 1.0)
    digits = np.floor(10**mantissas + 1e-12).astype(int)
    return np.array([np.mean(digits == d) for d in range(1, 10)])


multiplicative = np.prod(rng.uniform(0.75, 1.30, size=(20_000, 400)), axis=1)
uniform = rng.uniform(1, 1000, size=20_000)
for scale in [1.0, 0.92, 0.3048, 7.0]:
    print(f"x {scale:<6}  multiplicative: {np.round(first_digit_shares(scale * multiplicative)[:4], 3)}"
          f"   uniform: {np.round(first_digit_shares(scale * uniform)[:4], 3)}")
print(f"Benford   d = 1..4         : {np.round(np.log10(1 + 1 / np.arange(1, 5)), 3)}")
```

## 5. Visualizations

### 5.1 The same data on a linear and a log axis

The final company sizes of section 4.1: on a linear axis a skewed pile near zero with a long right tail; on a log axis a symmetric bell that matches the log-normal density.

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(2024)
final_sizes = np.prod(rng.uniform(0.75, 1.30, size=(20_000, 100)), axis=1)
mu_hat, sigma_hat = np.log(final_sizes).mean(), np.log(final_sizes).std()

fig, (ax_linear, ax_log) = plt.subplots(1, 2, figsize=(12, 4.5))
# normalise by ALL samples (some lie beyond 15), so the bars are comparable with the density
bin_width = 0.1
ax_linear.hist(final_sizes, bins=np.arange(0, 15 + bin_width, bin_width), weights=np.full(final_sizes.size, 1 / (final_sizes.size * bin_width)),
              color="tab:blue", alpha=0.6, label="simulated")
x = np.linspace(0.01, 15, 1000)
lognormal_density = np.exp(-((np.log(x) - mu_hat) ** 2) / (2 * sigma_hat**2)) / (x * sigma_hat * np.sqrt(2 * np.pi))
ax_linear.plot(x, lognormal_density, color="crimson", label="log-normal density")
ax_linear.set_xlabel("final size X (linear scale)")
ax_linear.set_ylabel("density")
ax_linear.set_title("Skewed on a linear axis")
ax_linear.legend(loc="upper right")
ax_linear.grid(True, alpha=0.3)

log_bins = np.logspace(np.log10(final_sizes.min()), np.log10(final_sizes.max()), 80)
ax_log.hist(final_sizes, bins=log_bins, color="tab:blue", alpha=0.6, label="simulated")
ax_log.set_xscale("log")
bin_centres = np.sqrt(log_bins[:-1] * log_bins[1:])
widths_in_ln = np.log(log_bins[1:]) - np.log(log_bins[:-1])
expected_counts = len(final_sizes) * widths_in_ln * np.exp(-((np.log(bin_centres) - mu_hat) ** 2) / (2 * sigma_hat**2)) / (sigma_hat * np.sqrt(2 * np.pi))
ax_log.plot(bin_centres, expected_counts, color="crimson", label="normal curve in ln X")
ax_log.set_xlabel("final size X (log scale)")
ax_log.set_ylabel("count per log-spaced bin")
ax_log.set_title("A symmetric bell on a log axis")
ax_log.legend(loc="upper right")
ax_log.grid(True, which="both", alpha=0.3)
plt.tight_layout()
plt.show()
```

### 5.2 Sample paths on a log axis

Twenty companies over $100$ years. On a log axis each path is a random walk — a sum of independent steps $\ln G_i$ — and the spread grows like $\sqrt{n}$, exactly as for additive noise.

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(5)
factors = rng.uniform(0.75, 1.30, size=(20, 100))
paths = np.cumprod(np.hstack([np.ones((20, 1)), factors]), axis=1)
years = np.arange(101)

grid = np.linspace(0.75, 1.30, 200_001)
mean_log, sd_log = np.mean(np.log(grid)), np.std(np.log(grid))

fig, ax = plt.subplots(figsize=(10, 5))
for path in paths:
    ax.plot(years, path, color="tab:blue", alpha=0.5, linewidth=1)
ax.plot(years, np.exp(years * mean_log), color="black", linewidth=2, label="median exp(n m)")
ax.plot(years, np.exp(years * mean_log + 2 * np.sqrt(years) * sd_log), "--", color="crimson", label="median x exp(+-2 s sqrt(n))")
ax.plot(years, np.exp(years * mean_log - 2 * np.sqrt(years) * sd_log), "--", color="crimson")
ax.set_yscale("log")
ax.set_xlabel("year n")
ax.set_ylabel("size X_n (log scale)")
ax.set_title("Multiplicative growth: a random walk on the log scale")
ax.legend(loc="upper left")
ax.grid(True, which="both", alpha=0.3)
plt.show()
```

### 5.3 Benford's law is a log ruler

Top: one decade of a logarithmic ruler, divided at $1, 2, \dots, 10$. The width of each segment is $\log_{10}(1 + 1/d)$ — the probability of first digit $d$ under a uniform mantissa. Bottom: observed first-digit shares for the powers of two and the multiplicative companies, against Benford's prediction.

```python
import matplotlib.pyplot as plt
import numpy as np

digits = np.arange(1, 10)
benford = np.log10(1 + 1 / digits)

powers_counts = np.zeros(9)
power = 1
for _ in range(10_000):
    power *= 2
    powers_counts[int(str(power)[0]) - 1] += 1

rng = np.random.default_rng(2024)
companies = np.prod(rng.uniform(0.75, 1.30, size=(20_000, 400)), axis=1)
company_digits = np.floor(10 ** np.mod(np.log10(companies), 1.0) + 1e-12).astype(int)
company_shares = np.array([np.mean(company_digits == d) for d in digits])

fig, (ax_ruler, ax_bars) = plt.subplots(2, 1, figsize=(11, 7), gridspec_kw={"height_ratios": [1, 2.5]})
colours = plt.cm.viridis(np.linspace(0.1, 0.9, 9))
for d, colour in zip(digits, colours):
    left, right = np.log10(d), np.log10(d + 1)
    ax_ruler.barh(0, right - left, left=left, color=colour, edgecolor="white")
    if d <= 5:
        ax_ruler.text((left + right) / 2, 0, f"{d}\n{100 * (right - left):.1f}%", ha="center", va="center", color="white", fontsize=9)
    else:
        ax_ruler.text((left + right) / 2, 0, f"{d}", ha="center", va="center", color="white", fontsize=9)
ax_ruler.set_xlim(0, 1)
ax_ruler.set_yticks([])
ax_ruler.set_xlabel("mantissa {log10 x}  (position on one decade of a log ruler)")
ax_ruler.set_title("First digit d occupies [log10 d, log10(d+1)) of the mantissa")

width = 0.27
ax_bars.bar(digits - width, benford, width, color="black", alpha=0.7, label="Benford log10(1 + 1/d)")
ax_bars.bar(digits, powers_counts / powers_counts.sum(), width, color="tab:orange", label="powers 2^1 ... 2^10000")
ax_bars.bar(digits + width, company_shares, width, color="tab:blue", label="multiplicative companies (400 factors)")
ax_bars.set_xticks(digits)
ax_bars.set_xlabel("first digit d")
ax_bars.set_ylabel("share")
ax_bars.legend(loc="upper right")
ax_bars.grid(True, axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

### 5.4 Wider spread, better Benford

For log-normal samples $X = e^{Y}$, $Y \sim \mathcal N(0, \sigma^2)$, the plot shows the largest deviation between observed first-digit shares and Benford's law as $\sigma$ grows. By $\sigma \approx 1.5$ — a standard deviation of about $0.65$ in $\log_{10} X$, so that $X$ routinely ranges over a factor of $20$ — the mantissa is essentially uniform and the deviation drops to the sampling noise of about $10^{-3}$ for $200{,}000$ samples.

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(3)
benford = np.log10(1 + 1 / np.arange(1, 10))
sigmas = np.linspace(0.1, 4.0, 40)
max_deviation = []
for sigma in sigmas:
    samples = np.exp(rng.normal(0, sigma, size=200_000))
    digits = np.floor(10 ** np.mod(np.log10(samples), 1.0) + 1e-12).astype(int)
    shares = np.array([np.mean(digits == d) for d in range(1, 10)])
    max_deviation.append(np.max(np.abs(shares - benford)))

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(sigmas, max_deviation, "o-", color="tab:purple")
ax.set_yscale("log")
ax.set_xlabel("sigma = standard deviation of ln X")
ax.set_ylabel("max |share - Benford| (log scale)")
ax.set_title("Log-normal data approach Benford's law as the spread grows")
ax.grid(True, which="both", alpha=0.3)
plt.show()
```

## 7. Solutions

1. $P(9) = \log_{10}(10/9) \approx 0.0458$; $P(1)/P(9) \approx 0.3010 / 0.0458 \approx 6.58$.
2. $50 \times 0.30103 = 15.0515$, mantissa $0.0515 < 0.301$, so the first digit is $1$ ($10^{0.0515} \approx 1.126$). Indeed $2^{50} = 1{,}125{,}899{,}906{,}842{,}624$.
3. Median $e^2 \approx 7.389$, mean $e^{2 + 0.125} = e^{2.125} \approx 8.373$, mode $e^{2 - 0.25} = e^{1.75} \approx 5.755$.
4. $100 \times 0.477 = 47.7$: $48$ digits, mantissa about $0.7$, and $10^{0.7} \approx 5$, so the first digit is about $5$. Exactly: $100 \log_{10} 3 \approx 47.7121$, $10^{0.7121} \approx 5.15$, first digit $5$, $48$ digits.
5. $\sum_{d=1}^{k} \log_{10}\frac{d+1}{d} = \log_{10}\left(\frac21 \cdot \frac32 \cdots \frac{k+1}{k}\right) = \log_{10}(k+1)$. $P(d \ge 5) = 1 - \log_{10} 5 = \log_{10} 2 \approx 0.301$.
6. Doubling adds $\log_{10} 2 \approx 0.301$ to the mantissa (mod $1$). Mantissas in $[\log_{10} 5, 1)$ — first digits $5$ to $9$ — move to $[\log_{10} 10 - 1, \log_{10} 2) = [0, 0.301)$, which is first digit $1$. Their total probability is $1 - \log_{10} 5 = \log_{10} 2 = P(1)$, so the share of $1$s is preserved — scale invariance in action.
7. $f_X(x) = \frac{1}{\sigma\sqrt{2\pi}} e^{-(\ln x - \mu)^2 / (2\sigma^2)} \cdot \frac{1}{x}$, the formula of 2.2. SymPy confirms the integral over $(0, \infty)$ is $1$ (below, for $\mu = 0.3$, $\sigma = 0.7$, after the substitution $x = e^y$, which turns it back into the integral of a normal density).
8. $E[e^Y] = \int e^{y} \frac{1}{\sigma\sqrt{2\pi}} e^{-(y - \mu)^2/(2\sigma^2)} dy$. The exponent $y - \frac{(y - \mu)^2}{2\sigma^2} = -\frac{(y - \mu - \sigma^2)^2}{2\sigma^2} + \mu + \frac{\sigma^2}{2}$, so the integral is $e^{\mu + \sigma^2/2}$ times the integral of a normal density, which is $1$.
9. $P(\text{second} = 0) \approx 0.1197$, $P(\text{second} = 9) \approx 0.0850$: second digits are much closer to uniform ($0.1$) than first digits. The powers of two agree to about two decimals (below).

## 8. Conclusion

- A product of many independent positive factors has a logarithm that is a sum, so by the central limit theorem it is approximately **log-normal**: $\;X = X_0 \prod G_i \iff \ln X = \ln X_0 + \sum \ln G_i$.
- If $\ln X \sim \mathcal N(\mu, \sigma^2)$: median $e^{\mu}$, mean $e^{\mu + \sigma^2/2}$, mode $e^{\mu - \sigma^2}$. Skewed on a linear axis, a symmetric bell on a log axis.
- The first digit of $x$ is read from the **mantissa** $\{\log_{10} x\}$: it is $d$ exactly when the mantissa lies in $[\log_{10} d, \log_{10}(d+1))$.
- A uniform mantissa gives **Benford's law** $P(d) = \log_{10}(1 + 1/d)$: $30.1\%$ ones down to $4.6\%$ nines. Data spread over many orders of magnitude (multiplicative processes, powers of $2$) approach it; it is the only first-digit law that survives a change of units.
- Benford's law fails for narrow, additive, assigned or uniform data — the chi-square test makes the difference visible.
- The picture: on a log ruler, the interval of first digit $1$ is six and a half times as long as the interval of first digit $9$.
