# Exponential Growth and Decay

This notebook solves the differential equation $dy/dt = ky$ — "the rate of change is proportional to the amount" — and shows that every practical question about it (how long until it halves? until it doubles? how old is this sample? what is the rate, given measurements?) is answered by a logarithm.

This is notebook 03 of module 04 (**Logarithms in calculus**) in the **Learning Logarithms** series.

## 1. Introduction

A piece of charcoal from an ancient hearth contains only **30%** of the carbon-14 that a living tree of today contains. How old is it?

Here is what we know. Living things keep exchanging carbon with the atmosphere, so while alive they hold carbon-14 at roughly the atmospheric level. After death the exchange stops and the carbon-14 decays, with a **half-life of about 5,730 years**: after 5,730 years half is left, after 11,460 years a quarter, after 17,190 years an eighth.

30% lies between a half and a quarter, so the charcoal is between one and two half-lives old — between 5,730 and 11,460 years. But *where* in between? Interpolating linearly would be wrong, because decay is not linear: each half-life removes half of what is **left**, not a fixed amount. We need to solve

$$
\left(\tfrac12\right)^{t / 5730} = 0.30 \qquad\text{for } t,
$$

and the unknown is in the **exponent**. Undoing an exponent is exactly what a logarithm does:

$$
\frac{t}{5730} = \log_{1/2} 0.30 = \frac{\ln 0.30}{\ln 0.5} \approx 1.737
\qquad\Longrightarrow\qquad t \approx 9{,}950 \text{ years}.
$$

This is the idea behind radiocarbon dating, developed by Willard Libby and colleagues in the late 1940s (Libby received the 1960 Nobel Prize in Chemistry for it). Real laboratory dates need more care — the atmospheric carbon-14 level has varied over time, so raw ages are corrected with calibration curves — but the logarithm is the core of the calculation. The same calculation, with different constants, gives doubling times of investments and bacteria, the cooling of a cup of coffee, and the rate constant hidden in a column of measurements.

## 2. Theory

### 2.1 Solving $dy/dt = ky$

The equation says the **relative** rate of change is constant: $\dfrac{y'}{y} = k$ — the logarithmic derivative of notebook 01. Since $\dfrac{y'}{y} = \dfrac{d}{dt}\ln|y|$, integrating both sides gives

$$
\ln|y| = kt + C .
$$

The logarithm appears here naturally, from $\int dy/y$ (notebook 02). Now undo it by exponentiating (log form → exponent form):

$$
|y| = e^{C} e^{kt} \qquad\Longrightarrow\qquad y(t) = y_0\, e^{kt}, \qquad y_0 = y(0).
$$

$k > 0$ is **growth**, $k < 0$ is **decay**. Every solution is the same curve $e^{kt}$, stretched vertically by $y_0$.

### 2.2 Half-life and doubling time

For decay ($k < 0$), the **half-life** $T_{1/2}$ is the time with $y(T_{1/2}) = \tfrac12 y_0$:

$$
e^{k T_{1/2}} = \tfrac12 \iff k T_{1/2} = \ln\tfrac12 = -\ln 2 \iff T_{1/2} = \frac{\ln 2}{|k|} .
$$

For growth ($k > 0$), the **doubling time** is, by the same calculation, $T_2 = \dfrac{\ln 2}{k}$.

The half-life does not depend on $y_0$ or on when you start the clock: **any** half-life interval halves the amount. That is what "constant relative rate" means. It also means we can write the solution in base $\tfrac12$ or base $2$ instead of base $e$:

$$
y(t) = y_0 \left(\tfrac12\right)^{t / T_{1/2}} \quad\text{(decay)}, \qquad y(t) = y_0\, 2^{t / T_2} \quad\text{(growth)}.
$$

To recover the elapsed time from the remaining fraction $y/y_0$, take a logarithm:

$$
t = T_{1/2} \cdot \frac{\ln(y_0 / y)}{\ln 2} = T_{1/2} \cdot \log_2\frac{y_0}{y} \qquad \text{("how many halvings?")}.
$$

### 2.3 The rule of 72 (and why the honest number is 69.3)

Money growing at a **continuous** rate $r$ doubles in $T_2 = \ln 2 / r$. With $r$ written in percent, $R = 100 r$:

$$
T_2 = \frac{100 \ln 2}{R} \approx \frac{69.3}{R}.
$$

But most interest is compounded **yearly**: $y = y_0 (1 + r)^t$. Then

$$
T_2 = \frac{\ln 2}{\ln(1 + r)} ,
$$

which is a little longer, because $\ln(1 + r) < r$. Using $\ln(1+r) \approx r - r^2/2$,

$$
T_2 \approx \frac{\ln 2}{r(1 - r/2)} \approx \frac{\ln 2}{r}\left(1 + \frac{r}{2}\right) = \frac{69.3}{R} + 0.35 .
$$

At typical rates of $6\%$–$10\%$ the correction is worth about $69.3 \to 72$ in the numerator: at $R = 8$, $\frac{69.3}{8} + 0.35 \approx 9.01$ and $\frac{72}{8} = 9$, while the exact value is $\frac{\ln 2}{\ln 1.08} \approx 9.006$. And $72$ has many divisors ($2, 3, 4, 6, 8, 9, 12$), which makes the mental arithmetic easy. A rule of this kind already appears in Luca Pacioli's 1494 *Summa de arithmetica*, stated without derivation.

Summary: **69.3** (or 70) for continuous growth, **72** for yearly compounding near 8%.

### 2.4 Newton's law of cooling

A hot object in a room at temperature $T_{\text{env}}$ cools at a rate proportional to the **temperature difference**:

$$
\frac{dT}{dt} = -k\,(T - T_{\text{env}}), \qquad k > 0 .
$$

The difference $D = T - T_{\text{env}}$ satisfies $D' = -kD$ — the decay equation — so $D(t) = D_0 e^{-kt}$ and

$$
T(t) = T_{\text{env}} + (T_0 - T_{\text{env}})\, e^{-kt} .
$$

The temperature does not decay to zero but to the room temperature, and the **gap** halves every $\ln 2 / k$. Two measurements determine $k$:

$$
k = \frac{1}{t}\ln\frac{T_0 - T_{\text{env}}}{T(t) - T_{\text{env}}} .
$$

### 2.5 Finding $k$ from data: straighten the curve with a log

Taking $\ln$ of $y = y_0 e^{kt}$ gives

$$
\ln y = \ln y_0 + k t ,
$$

a **straight line** in $t$ with slope $k$ and intercept $\ln y_0$. So: take logs of the measurements, fit a straight line by least squares, read off the slope. On a plot with a logarithmic $y$-axis (a semi-log plot) exponential data look straight, and the slope is the relative rate — the picture from notebook 01. Fitting in log space also treats a 5% error at $y = 100$ and at $y = 1$ alike, which suits measurements whose error is proportional to their size.

## 3. Worked Examples

### Example 1 — dating the charcoal

$\dfrac{y}{y_0} = 0.30$ and $T_{1/2} = 5730$ years:

$$
t = 5730 \cdot \frac{\ln(1/0.30)}{\ln 2} = 5730 \cdot \frac{1.20397}{0.69315} = 5730 \times 1.73697 \approx 9{,}953 \text{ years}.
$$

The rate constant is $k = -\ln 2 / 5730 \approx -1.2097 \times 10^{-4}$ per year.

### Example 2 — doubling time of bacteria

A culture grows from $1{,}000$ to $8{,}000$ cells in $90$ minutes. That is $\log_2 8 = 3$ doublings, so $T_2 = 30$ minutes and $k = \ln 2 / 30 \approx 0.0231$ per minute. After $4$ hours ($8$ doublings) there are $1000 \cdot 2^8 = 256{,}000$ cells.

### Example 3 — rule of 72

At $6\%$ yearly interest, money doubles in about $72/6 = 12$ years. Exactly: $\ln 2 / \ln 1.06 \approx 11.90$ years. At a $6\%$ **continuous** rate: $\ln 2 / 0.06 \approx 11.55$ years $\approx 69.3/6$.

### Example 4 — the coffee cup

Coffee at $90\,^\circ\text{C}$ in a $20\,^\circ\text{C}$ room is at $70\,^\circ\text{C}$ after $10$ minutes. The gap fell from $70$ to $50$:

$$
k = \frac{1}{10}\ln\frac{70}{50} = \frac{\ln 1.4}{10} \approx 0.03365 \text{ per minute}.
$$

When is it at $50\,^\circ\text{C}$ (gap $30$)? $t = \dfrac{1}{k}\ln\dfrac{70}{30} = \dfrac{0.84730}{0.03365} \approx 25.2$ minutes. The gap halves every $\ln 2 / k \approx 20.6$ minutes.

## 4. Python Examples

### 4.1 Radiocarbon ages from the remaining fraction

The formula $t = T_{1/2}\log_2(y_0/y)$ as a function, applied to a range of fractions.

```python
import math

HALF_LIFE_C14 = 5730.0   # years (approximate)


def radiocarbon_age(fraction_remaining, half_life=HALF_LIFE_C14):
    if not 0 < fraction_remaining <= 1:
        raise ValueError("fraction must be in (0, 1]")
    return half_life * math.log2(1 / fraction_remaining)


for fraction in [1.0, 0.9, 0.5, 0.30, 0.25, 0.10, 0.01, 0.002]:
    print(f"fraction left {fraction:6.3f}  ->  age {radiocarbon_age(fraction):9,.0f} years")

print("\nrate constant k =", -math.log(2) / HALF_LIFE_C14, "per year")
```

Notice how slowly the age grows as the fraction shrinks: going from $1\%$ to $0.2\%$ left (a factor of $5$) adds only $\log_2 5 \approx 2.3$ half-lives. Measuring tiny fractions accurately is what limits the method in practice, to a few tens of thousands of years.

### 4.2 SymPy solves $dy/dt = ky$ and the cooling law

```python
import sympy as sp

t, k = sp.symbols("t k", real=True)
temperature_env, T0 = sp.symbols("T_env T_0", real=True)
y = sp.Function("y")
T = sp.Function("T")

growth_solution = sp.dsolve(sp.Eq(y(t).diff(t), k * y(t)), y(t))
print("dy/dt = k y          ->", growth_solution)

cooling_solution = sp.dsolve(sp.Eq(T(t).diff(t), -k * (T(t) - temperature_env)), T(t), ics={T(0): T0})
print("dT/dt = -k(T - T_env) ->", sp.simplify(cooling_solution))

half_life = sp.solve(sp.Eq(sp.exp(-k * t), sp.Rational(1, 2)), t)
print("half-life for decay rate k:", half_life)
```

### 4.3 Rule of 72 versus the exact doubling time

For yearly rates from 1% to 20%: the exact doubling time $\ln 2 / \ln(1 + r)$, the rule of 72, the rule of 69.3, and the refined $69.3/R + 0.35$.

```python
import math

print(f"{'rate %':>6} {'exact':>8} {'72/R':>8} {'69.3/R':>8} {'69.3/R+0.35':>12} {'continuous ln2/r':>17}")
for rate_percent in [1, 2, 4, 6, 8, 10, 12, 15, 20]:
    r = rate_percent / 100
    exact_yearly = math.log(2) / math.log(1 + r)
    print(f"{rate_percent:>6} {exact_yearly:>8.3f} {72 / rate_percent:>8.3f} {69.3 / rate_percent:>8.3f} "
          f"{69.3 / rate_percent + 0.35:>12.3f} {math.log(2) / r:>17.3f}")
```

The rule of 72 is best near $8\%$; at $1\%$ the rule of 70 does better. The refined rule $69.3/R + 0.35$ is good across the whole range, and $69.3/R$ is exact for continuous compounding.

### 4.4 Newton's law of cooling: coffee

```python
import math

room = 20.0
start = 90.0
after_10_min = 70.0

k = math.log((start - room) / (after_10_min - room)) / 10
print(f"k = {k:.5f} per minute,  gap half-life = {math.log(2) / k:.2f} min")


def coffee_temperature(minutes):
    return room + (start - room) * math.exp(-k * minutes)


def minutes_until(target):
    return math.log((start - room) / (target - room)) / k


for minutes in [0, 10, 20, 30, 60]:
    print(f"t = {minutes:>2} min: {coffee_temperature(minutes):6.2f} C")
print(f"reaches 50 C after {minutes_until(50):.2f} min,  reaches 25 C after {minutes_until(25):.1f} min")
```

### 4.5 Fitting $k$ from noisy measurements

Simulate a decaying quantity with true $k = -0.15$ and $5\%$ multiplicative noise, then fit a line to $\ln y$ with `np.polyfit`. The slope estimates $k$, the intercept estimates $\ln y_0$.

```python
import math

import numpy as np

rng = np.random.default_rng(seed=7)
true_k = -0.15
true_y0 = 200.0
times = np.arange(0, 21, 2.0)
measurements = true_y0 * np.exp(true_k * times) * np.exp(rng.normal(0, 0.05, size=times.size))

slope, intercept = np.polyfit(times, np.log(measurements), 1)
print("measurements:", np.round(measurements, 2))
print(f"fitted k  = {slope:.4f}   (true {true_k})")
print(f"fitted y0 = {math.exp(intercept):.2f}  (true {true_y0})")
print(f"fitted half-life = {math.log(2) / -slope:.3f}   (true {math.log(2) / -true_k:.3f})")
```

## 5. Visualizations

### 5.1 Decay: halving at equal intervals

Carbon-14 decay with the half-life intervals marked. On the linear axis (left) each interval removes half of what is left — big drops at first, small ones later. On the logarithmic axis (right) the same curve is a straight line and every half-life is the **same vertical step**: equal ratios, equal distances.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

half_life = 5730
years = np.linspace(0, 6 * half_life, 600)
fraction = 0.5 ** (years / half_life)

fig, (ax_lin, ax_log) = plt.subplots(1, 2, figsize=(12, 4.5))
for ax in (ax_lin, ax_log):
    ax.plot(years, fraction, color="tab:blue")
    for n in range(1, 6):
        ax.plot(n * half_life, 0.5**n, "o", color="crimson")
        ax.axvline(n * half_life, color="gray", linestyle=":", linewidth=0.8)
    ax.plot(9953, 0.30, "s", color="tab:green", label="charcoal: 30% left, ~9,950 years")
    ax.set_xlabel("years since death")
    ax.legend(loc="upper right")

ax_lin.set_ylabel("fraction of C-14 left (linear scale)")
ax_lin.grid(True, alpha=0.3)
ax_lin.set_title("Linear axis: each half-life halves what is left")
ax_log.set_yscale("log")
ax_log.set_ylabel("fraction of C-14 left (log scale)")
ax_log.grid(True, which="both", alpha=0.3)
ax_log.set_title("Log axis: a straight line, equal steps")
plt.tight_layout()
plt.show()
```

### 5.2 Doubling-time rules versus the truth

The exact yearly doubling time, the rule of 72 and the rule of 69.3 for rates between 1% and 25%. The bottom panel shows the errors in years: 72 crosses zero near $8\%$.

```python
import matplotlib.pyplot as plt
import numpy as np

rate_percent = np.linspace(1, 25, 300)
exact = np.log(2) / np.log(1 + rate_percent / 100)
rule_72 = 72 / rate_percent
rule_693 = 69.3 / rate_percent

fig, (ax_top, ax_bottom) = plt.subplots(2, 1, figsize=(9, 7), sharex=True, gridspec_kw={"height_ratios": [2, 1]})
ax_top.plot(rate_percent, exact, color="black", linewidth=2.5, label="exact: ln 2 / ln(1 + r)")
ax_top.plot(rate_percent, rule_72, "--", color="tab:orange", label="rule of 72")
ax_top.plot(rate_percent, rule_693, ":", color="tab:blue", label="rule of 69.3 (exact for continuous)")
ax_top.set_yscale("log")
ax_top.set_yticks([3, 5, 10, 20, 50])
ax_top.set_yticklabels(["3", "5", "10", "20", "50"])
ax_top.set_ylabel("doubling time in years (log scale)")
ax_top.grid(True, which="both", alpha=0.3)
ax_top.legend(loc="upper right")
ax_top.set_title("Doubling time at yearly compounding")

ax_bottom.plot(rate_percent, rule_72 - exact, "--", color="tab:orange", label="72/R - exact")
ax_bottom.plot(rate_percent, rule_693 - exact, ":", color="tab:blue", label="69.3/R - exact")
ax_bottom.axhline(0, color="black", linewidth=0.8)
ax_bottom.axvline(8, color="gray", linestyle="--", linewidth=0.8)
ax_bottom.set_xlabel("yearly rate R (%)")
ax_bottom.set_ylabel("error (years)")
ax_bottom.grid(True, alpha=0.3)
ax_bottom.legend(loc="upper right")
plt.tight_layout()
plt.show()
```

### 5.3 Newton's law of cooling

Left: the coffee temperature approaches room temperature. Right: the **gap** $T - T_{\text{env}}$ on a log axis is a straight line with slope $-k$ — the same "straighten it with a log" picture.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

room, start = 20.0, 90.0
k = math.log(70 / 50) / 10
minutes = np.linspace(0, 120, 400)
temperature = room + (start - room) * np.exp(-k * minutes)

fig, (ax_temp, ax_gap) = plt.subplots(1, 2, figsize=(12, 4.5))
ax_temp.plot(minutes, temperature, color="tab:red", label="coffee temperature")
ax_temp.axhline(room, color="gray", linestyle="--", label="room 20 C")
ax_temp.plot([10, 25.18], [70, 50], "o", color="black")
ax_temp.set_xlabel("minutes")
ax_temp.set_ylabel("temperature (C)")
ax_temp.grid(True, alpha=0.3)
ax_temp.legend(loc="upper right")
ax_temp.set_title("T(t) = 20 + 70 e^(-kt)")

ax_gap.plot(minutes, temperature - room, color="tab:purple")
ax_gap.set_yscale("log")
ax_gap.set_yticks([2, 5, 10, 20, 50])
ax_gap.set_yticklabels(["2", "5", "10", "20", "50"])
ax_gap.set_xlabel("minutes")
ax_gap.set_ylabel("T - T_env (log scale)")
ax_gap.grid(True, which="both", alpha=0.3)
ax_gap.set_title(f"Gap on a log axis: straight line, slope -k = {-k:.4f}")
plt.tight_layout()
plt.show()
```

### 5.4 The fit from section 4.5

The noisy measurements on a semi-log plot, with the fitted line. On a linear axis it would be a curve that is hard to judge by eye; on a log axis it is a straight line whose slope is $k$.

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(seed=7)
times = np.arange(0, 21, 2.0)
measurements = 200.0 * np.exp(-0.15 * times) * np.exp(rng.normal(0, 0.05, size=times.size))
slope, intercept = np.polyfit(times, np.log(measurements), 1)
fine_times = np.linspace(0, 20, 200)

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(times, measurements, "o", color="tab:blue", label="measurements")
ax.plot(fine_times, np.exp(intercept + slope * fine_times), color="tab:orange",
        label=f"fit: y = {np.exp(intercept):.1f} e^({slope:.4f} t)")
ax.plot(fine_times, 200 * np.exp(-0.15 * fine_times), "--", color="gray", label="truth: y = 200 e^(-0.15 t)")
ax.set_yscale("log")
ax.set_xlabel("t")
ax.set_ylabel("y (log scale)")
ax.grid(True, which="both", alpha=0.3)
ax.legend(loc="upper right")
ax.set_title("Fitting k: a straight line through ln y")
plt.show()
```

## 8. Conclusion

- $\dfrac{dy}{dt} = ky$ means a constant **relative** rate $y'/y = k$. Integrating $dy/y$ produces $\ln|y| = kt + C$, and exponentiating gives $y = y_0 e^{kt}$.
- Every "how long?" question is a logarithm: half-life and doubling time are $T = \ln 2 / |k|$, and the time to reach a fraction $f$ is $T_{1/2}\log_2(1/f)$ — "how many halvings".
- Radiocarbon dating: with $T_{1/2} \approx 5730$ years, 30% remaining means about $9{,}950$ years (before calibration).
- Rule of 72: $\ln 2 \approx 0.693$ gives $69.3/R$ for continuous growth; yearly compounding adds about $0.35$ years, which near $8\%$ is the same as replacing $69.3$ with $72$.
- Newton's law of cooling: the **gap** to room temperature decays exponentially, $T = T_{\text{env}} + (T_0 - T_{\text{env}})e^{-kt}$.
- To find $k$ from data, take logs: $\ln y = \ln y_0 + kt$ is a straight line. On a semi-log plot exponential growth and decay are straight, and the slope is the rate.
