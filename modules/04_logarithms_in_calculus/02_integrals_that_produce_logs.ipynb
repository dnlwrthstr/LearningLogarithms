# Integrals That Produce Logarithms

This notebook is about the integrals whose answer is a logarithm: $\int dx/x$, $\int f'(x)/f(x)\,dx$, $\int \tan x\,dx$, $\int \sec x\,dx$, the partial-fraction integrals of rational functions, and $\int \ln x\,dx$. They all come from one fact: the logarithm fills the single hole in the power rule.

This is notebook 02 of module 04 (**Logarithms in calculus**) in the **Learning Logarithms** series.

## 1. Introduction

The power rule for integrals covers almost every power of $x$:

$$
\int x^n \, dx = \frac{x^{n+1}}{n+1} + C .
$$

It works for $n = 2$, for $n = -2$, for $n = \tfrac12$, for $n = -0.999$. It fails at exactly one exponent: $n = -1$, where it would divide by zero. Yet $\int_1^2 \frac{dx}{x}$ is a perfectly good area under a perfectly good curve — it is about $0.693$. So which function has derivative $1/x$? It cannot be a power of $x$: the derivative of $x^m$ is $m x^{m-1}$, which is never $x^{-1}$ (that would need $m = 0$, and then the derivative is $0$).

Look at what happens **near** the gap. Measure the area from $1$ instead of using an arbitrary constant:

$$
\int_1^x t^{n}\,dt = \frac{x^{n+1} - 1}{n+1} = \frac{x^h - 1}{h}, \qquad h = n + 1 .
$$

As $n \to -1$, i.e. $h \to 0$, the formula has the form $0/0$ — but its values settle down. At $x = 2$:

| $h$ | $1$ | $0.1$ | $0.01$ | $0.001$ |
|---|---|---|---|---|
| $\dfrac{2^h - 1}{h}$ | $1$ | $0.7177$ | $0.6956$ | $0.6934$ |

The limit is $\ln 2 = 0.6931\ldots$ In general

$$
\lim_{h \to 0} \frac{x^h - 1}{h} = \ln x ,
$$

so the natural logarithm is **exactly the function the power rule is missing**. The hole at $n = -1$ is filled not by a power but by a logarithm — and every integral in this notebook inherits its logarithm from that one gap.

## 2. Theory

### 2.1 The gap in the power rule

The antiderivatives of $x^n$, measured from $1$, form a family $F_h(x) = \dfrac{x^h - 1}{h}$ with $h = n + 1 \neq 0$. To see the limit as $h \to 0$, write $x^h = e^{h \ln x}$ (exponent form) and use $e^{u} = 1 + u + \tfrac{u^2}{2} + \cdots$:

$$
\frac{x^h - 1}{h} = \frac{h \ln x + \tfrac12 h^2 (\ln x)^2 + \cdots}{h} = \ln x + \tfrac12 h (\ln x)^2 + \cdots \;\xrightarrow{\;h \to 0\;}\; \ln x .
$$

So $F_0(x) = \ln x$ continues the family smoothly through the gap. Equivalently, $\ln x = \int_1^x \frac{dt}{t}$ is the definition of the natural logarithm as an area (module 01), and the fundamental theorem of calculus gives

$$
\frac{d}{dx} \ln x = \frac1x \qquad (x > 0) .
$$

### 2.2 Why the absolute value: $\int \frac{dx}{x} = \ln|x| + C$

The function $1/x$ is defined for $x < 0$ too, but $\ln x$ is not (logarithms of real numbers need $x > 0$). For $x < 0$, differentiate $\ln(-x)$ with the chain rule:

$$
\frac{d}{dx}\ln(-x) = \frac{1}{-x}\cdot(-1) = \frac1x .
$$

So on the negative side the antiderivative is $\ln(-x)$, and both sides are captured by

$$
\int \frac{dx}{x} = \ln|x| + C .
$$

Geometrically, $1/x$ is an **odd** function: the area under it from $-b$ to $-a$ is the mirror image (negated) of the area from $a$ to $b$, so $\int_{-3}^{-1}\frac{dx}{x} = -\ln 3$.

Two warnings:

- $1/x$ has a vertical asymptote at $0$, so a definite integral **must not cross $0$**. "$\int_{-1}^{1}\frac{dx}{x} = \ln|1| - \ln|-1| = 0$" is wrong: each half diverges, and the integral does not exist.
- Since the domain is two separate intervals, the constant may be different on each side: the most general antiderivative is $\ln|x| + C_1$ for $x > 0$ and $\ln|x| + C_2$ for $x < 0$.

### 2.3 The pattern $\int \frac{f'(x)}{f(x)}\,dx = \ln|f(x)| + C$

Substitute $u = f(x)$, $du = f'(x)\,dx$:

$$
\int \frac{f'(x)}{f(x)}\,dx = \int \frac{du}{u} = \ln|u| + C = \ln|f(x)| + C .
$$

This is notebook 01 read backwards: the derivative of $\ln|f|$ is the logarithmic derivative $f'/f$. Whenever the numerator is (a constant multiple of) the derivative of the denominator, the answer is a logarithm. Examples:

| integrand | $f(x)$ | antiderivative |
|---|---|---|
| $\dfrac{2x}{x^2+1}$ | $x^2+1$ | $\ln(x^2+1)$ (no bars needed: $x^2+1 > 0$) |
| $\dfrac{1}{ax+b}$ | $ax+b$ | $\tfrac1a \ln|ax+b|$ |
| $\tan x = \dfrac{\sin x}{\cos x}$ | $\cos x$ | $-\ln|\cos x| = \ln|\sec x|$ |
| $\cot x = \dfrac{\cos x}{\sin x}$ | $\sin x$ | $\ln|\sin x|$ |
| $\dfrac{1}{x \ln x}$ | $\ln x$ | $\ln|\ln x|$ |

### 2.4 $\int \sec x\,dx$

The secant needs a trick: multiply top and bottom by $\sec x + \tan x$.

$$
\int \sec x\,dx = \int \frac{\sec x(\sec x + \tan x)}{\sec x + \tan x}\,dx = \int \frac{\sec^2 x + \sec x \tan x}{\sec x + \tan x}\,dx .
$$

The numerator is exactly the derivative of the denominator, so by 2.3

$$
\int \sec x\,dx = \ln|\sec x + \tan x| + C .
$$

(This integral has a famous practical history: it gives the spacing of latitude lines on a Mercator map.)

### 2.5 Partial fractions: rational functions become sums of logs

A rational function whose denominator factors into distinct linear factors splits into simple pieces, each of which integrates to a logarithm:

$$
\frac{1}{x^2 - 1} = \frac{1}{(x-1)(x+1)} = \frac{1/2}{x-1} - \frac{1/2}{x+1}
$$

$$
\int \frac{dx}{x^2-1} = \tfrac12\ln|x-1| - \tfrac12\ln|x+1| + C = \tfrac12 \ln\left|\frac{x-1}{x+1}\right| + C .
$$

The last step is the quotient law of logarithms, $\ln a - \ln b = \ln(a/b)$, used to tidy up.

### 2.6 $\int \ln x\,dx$ by parts

Take $u = \ln x$, $dv = dx$, so $du = dx/x$, $v = x$:

$$
\int \ln x\,dx = x\ln x - \int x \cdot \frac1x\,dx = x \ln x - x + C .
$$

Check: $\frac{d}{dx}(x\ln x - x) = \ln x + 1 - 1 = \ln x$. ✓ This antiderivative is the engine of Stirling's approximation in notebook 04.

## 3. Worked Examples

### Example 1 — $\int_1^{e^3} \frac{dx}{x}$

$= \ln e^3 - \ln 1 = 3$. In exponent form: the area under $1/x$ from $1$ to $e^k$ is $k$. Equal **ratios** of the endpoints give equal **areas**: $\int_1^{e}$, $\int_e^{e^2}$, $\int_{e^2}^{e^3}$ are each $1$.

### Example 2 — $\int_{-3}^{-1} \frac{dx}{x}$

$= \ln|-1| - \ln|-3| = 0 - \ln 3 \approx -1.0986$. Negative, as it must be: the curve is below the axis on this interval.

### Example 3 — $\int_0^{\pi/4} \tan x\,dx$

$= \big[-\ln|\cos x|\big]_0^{\pi/4} = -\ln\tfrac{\sqrt2}{2} + \ln 1 = \ln\sqrt2 = \tfrac12 \ln 2 \approx 0.3466$.

### Example 4 — $\int_0^{\pi/4} \sec x\,dx$

$= \big[\ln|\sec x + \tan x|\big]_0^{\pi/4} = \ln(\sqrt2 + 1) - \ln 1 \approx 0.8814$.

### Example 5 — $\int_2^3 \frac{dx}{x^2-1}$

$= \tfrac12\Big[\ln\left|\tfrac{x-1}{x+1}\right|\Big]_2^3 = \tfrac12\left(\ln\tfrac24 - \ln\tfrac13\right) = \tfrac12 \ln\tfrac32 \approx 0.2027$.

### Example 6 — $\int_1^e \ln x\,dx$

$= \big[x \ln x - x\big]_1^e = (e - e) - (0 - 1) = 1$.

### Example 7 — $\int_0^1 \frac{2x}{x^2+1}\,dx$

Numerator $=$ derivative of denominator: $= \big[\ln(x^2+1)\big]_0^1 = \ln 2 \approx 0.6931$.

## 4. Python Examples

### 4.1 Approaching the gap: $\frac{x^h - 1}{h} \to \ln x$

The table from the introduction, extended. The error shrinks roughly in proportion to $h$, as the series in 2.1 predicts ($\text{error} \approx \tfrac12 h (\ln x)^2$).

```python
import math

x = 2.0
print(f"ln 2 = {math.log(x):.10f}\n")
print(f"{'h':>8} {'(2^h - 1)/h':>15} {'error':>12} {'h/2 (ln 2)^2':>14}")
for h in [1, 0.1, 0.01, 0.001, 1e-4, -0.01]:
    approximation = (x**h - 1) / h
    print(f"{h:>8} {approximation:>15.10f} {approximation - math.log(x):>12.2e} {h / 2 * math.log(x)**2:>14.2e}")
```

SymPy computes the limit exactly.

```python
import sympy as sp

x, h = sp.symbols("x h", positive=True)
print("limit of (x^h - 1)/h as h -> 0:", sp.limit((x**h - 1) / h, h, 0))
print("integral of x^n with n symbolic: ", sp.integrate(x**sp.Symbol("n"), x))
```

SymPy itself has to split into two cases — the power rule for $n \neq -1$ and the logarithm for $n = -1$.

### 4.2 SymPy's antiderivatives

SymPy usually omits the absolute value bars (it works over the complex numbers, where $\ln$ is defined for negative arguments). Differentiating each answer confirms it. For $\sec x$ SymPy returns a different-looking but equivalent form, $\tfrac12\ln\frac{1+\sin x}{1-\sin x}$ up to a constant — exercise 8 shows why.

```python
import sympy as sp

x = sp.symbols("x")
integrands = [1 / x, 2 * x / (x**2 + 1), sp.tan(x), sp.cot(x), sp.sec(x), 1 / (x**2 - 1), sp.log(x), 1 / (x * sp.log(x))]
for integrand in integrands:
    antiderivative = sp.integrate(integrand, x)
    check = sp.simplify(sp.diff(antiderivative, x) - integrand)
    print(f"integral of {str(integrand):<16} = {str(antiderivative):<40} derivative check: {check == 0}")
```

### 4.3 The definite integrals of section 3, numerically

A trapezoid rule on a fine grid reproduces every worked example. (The `np.trapezoid` shim keeps the code working on NumPy 1.x and 2.x.)

```python
import math

import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz


def numeric_integral(function, a, b, points=200_001):
    grid = np.linspace(a, b, points)
    return np.trapezoid(function(grid), grid)


cases = [
    ("1/x on [1, e^3]", lambda t: 1 / t, 1, math.e**3, 3.0),
    ("1/x on [-3, -1]", lambda t: 1 / t, -3, -1, -math.log(3)),
    ("tan on [0, pi/4]", np.tan, 0, math.pi / 4, 0.5 * math.log(2)),
    ("sec on [0, pi/4]", lambda t: 1 / np.cos(t), 0, math.pi / 4, math.log(math.sqrt(2) + 1)),
    ("1/(x^2-1) on [2, 3]", lambda t: 1 / (t**2 - 1), 2, 3, 0.5 * math.log(1.5)),
    ("ln x on [1, e]", np.log, 1, math.e, 1.0),
    ("2x/(x^2+1) on [0, 1]", lambda t: 2 * t / (t**2 + 1), 0, 1, math.log(2)),
]
for name, function, a, b, exact in cases:
    numeric = numeric_integral(function, a, b)
    print(f"{name:<22} numeric {numeric:10.6f}   exact {exact:10.6f}   close: {math.isclose(numeric, exact, rel_tol=1e-8)}")
```

### 4.4 Partial fractions with SymPy

`sp.apart` performs the partial-fraction split; integrating the pieces gives a sum of logarithms.

```python
import sympy as sp

x = sp.symbols("x")
rational_functions = [1 / (x**2 - 1), (2 * x + 3) / (x**2 + 3 * x + 2), (x**2 + 1) / (x * (x - 1) * (x + 2))]
for rational in rational_functions:
    pieces = sp.apart(rational, x)
    print("f(x)          =", rational)
    print("partial fracs =", pieces)
    print("integral      =", sp.integrate(pieces, x))
    print()
```

### 4.5 Integrating across the asymptote gives nonsense

Cut a gap $[-\varepsilon, \delta]$ around $0$ out of $[-1, 2]$ and integrate $1/x$ over what is left. The answer is $\ln 2 + \ln(\varepsilon/\delta)$: it depends on **how** the gap closes, so there is no single value — $\int_{-1}^{2}\frac{dx}{x}$ does not exist.

```python
import math

for epsilon, delta in [(1e-3, 1e-3), (1e-6, 1e-6), (1e-6, 2e-6), (1e-6, 1e-9)]:
    left_piece = math.log(epsilon) - math.log(1)      # integral of 1/x from -1 to -epsilon = ln|-eps| - ln|-1|
    right_piece = math.log(2) - math.log(delta)       # integral from delta to 2
    print(f"epsilon = {epsilon:.0e}, delta = {delta:.0e}:  total = {left_piece + right_piece:8.4f}")
```

## 5. Visualizations

### 5.1 The power-rule family closing the gap

The antiderivatives $F_h(x) = \dfrac{x^h - 1}{h}$ for $h = 1, 0.5, 0.2, 0.05, -0.2, -0.5$ (all measured from $1$), and their limit $\ln x$ in black. Positive $h$ lie above $\ln x$, negative $h$ below, and both sides squeeze onto it as $h \to 0$.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0.05, 5, 500)
fig, ax = plt.subplots(figsize=(9, 5.5))
colors = plt.cm.coolwarm(np.linspace(0, 1, 6))
for h, color in zip([1, 0.5, 0.2, 0.05, -0.2, -0.5], colors):
    ax.plot(x, (x**h - 1) / h, color=color, label=f"h = {h}:  (x^h - 1)/h")
ax.plot(x, np.log(x), color="black", linewidth=2.5, label="limit h -> 0:  ln x")
ax.axhline(0, color="gray", linewidth=0.8)
ax.axvline(1, color="gray", linewidth=0.8, linestyle=":")
ax.set_ylim(-4, 4)
ax.set_xlabel("x")
ax.set_ylabel("area under $t^{h-1}$ from 1 to x")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower right", fontsize=9)
ax.set_title("The antiderivatives of $x^{h-1}$ converge to $\\ln x$ as $h \\to 0$")
plt.show()
```

### 5.2 $1/x$ on both sides of zero, and $\ln|x|$

Left: the areas under $1/x$ on $[1, 3]$ and $[-3, -1]$ are mirror images with opposite signs. Right: the antiderivative $\ln|x|$ has two separate branches, one for each side of the asymptote — the reason the constant $C$ can differ on each side and the reason we may never integrate across $0$.

```python
import matplotlib.pyplot as plt
import numpy as np

x_pos = np.linspace(0.1, 4, 300)
x_neg = -x_pos[::-1]

fig, (ax_area, ax_anti) = plt.subplots(1, 2, figsize=(12, 4.5))
ax_area.plot(x_pos, 1 / x_pos, color="tab:blue")
ax_area.plot(x_neg, 1 / x_neg, color="tab:blue", label="y = 1/x")
fill_pos = np.linspace(1, 3, 100)
ax_area.fill_between(fill_pos, 1 / fill_pos, color="tab:green", alpha=0.35, label="area = ln 3")
ax_area.fill_between(-fill_pos, -1 / fill_pos, color="tab:red", alpha=0.35, label="area = -ln 3")
ax_area.axhline(0, color="black", linewidth=0.8)
ax_area.axvline(0, color="black", linewidth=0.8)
ax_area.set_ylim(-4, 4)
ax_area.set_xlabel("x")
ax_area.set_ylabel("1/x")
ax_area.grid(True, alpha=0.3)
ax_area.legend(loc="upper left")
ax_area.set_title("Odd integrand: mirrored areas")

ax_anti.plot(x_pos, np.log(x_pos), color="tab:purple", label="ln|x|, x > 0")
ax_anti.plot(x_neg, np.log(-x_neg), color="tab:orange", label="ln|x|, x < 0")
ax_anti.plot(x_neg, np.log(-x_neg) + 1, color="tab:orange", linestyle="--", alpha=0.7, label="ln|x| + 1 on the left only")
ax_anti.axvline(0, color="black", linewidth=0.8)
ax_anti.axhline(0, color="black", linewidth=0.8)
ax_anti.set_ylim(-2.5, 3.8)
ax_anti.set_xlabel("x")
ax_anti.set_ylabel("antiderivative")
ax_anti.grid(True, alpha=0.3)
ax_anti.legend(loc="upper right", fontsize=9)
ax_anti.set_title("ln|x|: two branches, two constants")
plt.tight_layout()
plt.show()
```

### 5.3 $\tan x$ and its integral $-\ln|\cos x|$

The antiderivative $-\ln|\cos x|$ rises to $+\infty$ wherever $\cos x = 0$, i.e. at the asymptotes of $\tan x$. Between asymptotes it is a smooth bowl whose slope is $\tan x$. The shaded area on $[0, \pi/4]$ equals $\tfrac12\ln 2$, the height of the bowl at $\pi/4$.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-1.5, 4.6, 2000)
cos_x = np.cos(x)
near_asymptote = np.abs(cos_x) < 0.02
tan_x = np.where(near_asymptote, np.nan, np.tan(x))
antiderivative = np.where(near_asymptote, np.nan, -np.log(np.abs(cos_x)))

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x, tan_x, color="tab:blue", label="tan x")
ax.plot(x, antiderivative, color="tab:red", linewidth=2, label="-ln|cos x|")
shade = np.linspace(0, math.pi / 4, 100)
ax.fill_between(shade, np.tan(shade), color="tab:blue", alpha=0.3, label="area on [0, pi/4] = (ln 2)/2")
ax.plot(math.pi / 4, 0.5 * math.log(2), "o", color="tab:red")
for asymptote in [-math.pi / 2, math.pi / 2, 3 * math.pi / 2]:
    ax.axvline(asymptote, color="gray", linestyle=":")
ax.axhline(0, color="black", linewidth=0.8)
ax.set_ylim(-4, 4)
ax.set_xlabel("x")
ax.set_ylabel("value")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower right")
ax.set_title("$\\int \\tan x\\,dx = -\\ln|\\cos x| + C$")
plt.show()
```

## 7. Solutions

1. The numerator is the derivative of $3x + 5$: $\big[\ln|3x+5|\big]_0^1 = \ln 8 - \ln 5 = \ln\tfrac85 \approx 0.4700$.
2. $\big[\ln|\sin x|\big]_{\pi/6}^{\pi/2} = \ln 1 - \ln\tfrac12 = \ln 2 \approx 0.6931$.
3. $\ln|-e| - \ln|-e^2| = 1 - 2 = -1$. The curve $1/x$ is negative for $x < 0$, so the area counts negatively; it is the mirror image of $\int_e^{e^2}\frac{dx}{x} = 1$.
4. $\int_1^{1000}\frac{dx}{x} = \ln 1000 = 3\ln 10 \approx 6.9$. $\int_{1000}^{2000}\frac{dx}{x} = \ln 2 \approx 0.69$. The second interval is longer ($1000$ versus $999$) but has a tenth of the area: area under $1/x$ depends only on the **ratio** of the endpoints ($1000$ versus $2$), not their difference.
5. As $f'/f$ with $f = x^2+3x+2$: $\big[\ln|x^2+3x+2|\big]_0^1 = \ln 6 - \ln 2 = \ln 3$. By partial fractions: $\dfrac{2x+3}{(x+1)(x+2)} = \dfrac{1}{x+1} + \dfrac{1}{x+2}$, so the integral is $\ln 2 + \ln\tfrac32 = \ln 3 \approx 1.0986$.
6. $u = \ln x$, $dv = x\,dx$: $\int x\ln x\,dx = \tfrac{x^2}{2}\ln x - \int \tfrac{x}{2}\,dx = \tfrac{x^2}{2}\ln x - \tfrac{x^2}{4}$. On $[1, e]$: $\left(\tfrac{e^2}{2} - \tfrac{e^2}{4}\right) - \left(0 - \tfrac14\right) = \tfrac{e^2 + 1}{4} \approx 2.0973$.
7. $\dfrac{1}{x(x+1)} = \dfrac1x - \dfrac{1}{x+1}$, so the integral is $\Big[\ln\dfrac{x}{x+1}\Big]_1^N = \ln\dfrac{N}{N+1} - \ln\dfrac12 = \ln\dfrac{2N}{N+1} \to \ln 2$.
8. $\dfrac{d}{dx}\ln|\sec x + \tan x| = \dfrac{\sec x\tan x + \sec^2 x}{\sec x + \tan x} = \dfrac{\sec x(\tan x + \sec x)}{\sec x + \tan x} = \sec x$. For the second form: $\sec x + \tan x = \dfrac{1 + \sin x}{\cos x}$, and squaring, $\dfrac{(1+\sin x)^2}{\cos^2 x} = \dfrac{(1+\sin x)^2}{(1 - \sin x)(1 + \sin x)} = \dfrac{1+\sin x}{1-\sin x}$. Taking $\tfrac12\ln$ of the square gives the claim (for $|x| < \pi/2$ both sides are defined and $\sec x + \tan x > 0$).
9. With $u = \ln t$, $du = dt/t$: $\int \frac{dt}{t\ln t} = \ln|\ln t|$, so $\int_2^x\frac{dt}{t\ln t} = \ln\ln x - \ln\ln 2$. This grows without bound as $x \to \infty$ (just extremely slowly), so the improper integral diverges. It exceeds $3$ when $\ln\ln x > 3 + \ln\ln 2$, i.e. $\ln x > e^3\ln 2 \approx 13.92$, i.e. $x > 2^{e^3} \approx 1.1 \times 10^6$.

## 8. Conclusion

- The power rule $\int x^n dx = \frac{x^{n+1}}{n+1}$ fails only at $n = -1$, and no power of $x$ fills the gap. The limit $\frac{x^h - 1}{h} \to \ln x$ shows that the **natural logarithm is the missing member** of the family of antiderivatives.
- $\int \frac{dx}{x} = \ln|x| + C$. The absolute value covers $x < 0$, where $\frac{d}{dx}\ln(-x) = \frac1x$. The constant may differ on each side, and an integral of $1/x$ must never cross $0$.
- Area under $1/x$ depends only on the **ratio** of the endpoints: $\int_a^b \frac{dx}{x} = \ln\frac{b}{a}$ — the multiplicative-to-additive picture once more.
- $\int \frac{f'}{f}\,dx = \ln|f| + C$ is logarithmic differentiation backwards. It gives $\int\tan x\,dx = -\ln|\cos x|$, $\int\cot x\,dx = \ln|\sin x|$, and (with a trick) $\int\sec x\,dx = \ln|\sec x + \tan x|$.
- Partial fractions turn rational functions into sums of $\frac{A}{x - r}$, which integrate to sums of logarithms; the laws of logarithms then tidy the answer.
- $\int \ln x\,dx = x\ln x - x + C$, by parts.
