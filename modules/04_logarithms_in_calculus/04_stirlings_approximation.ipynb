# Stirling's Approximation

This notebook estimates $n!$ — a product of $n$ numbers — by turning it into a **sum** of logarithms and replacing the sum with an integral. The result, $n! \approx \sqrt{2\pi n}\,(n/e)^n$, is one of the most useful approximations in mathematics: it counts digits of enormous numbers, estimates probabilities in coin-flipping, and underlies the entropy formulas of module 05.

This is notebook 04 of module 04 (**Logarithms in calculus**) in the **Learning Logarithms** series.

## 1. Introduction

Two questions:

1. **How many digits does $100!$ have?**
2. **If you flip a fair coin 100 times, what is the probability of exactly 50 heads?**

The second one is $\dbinom{100}{50} / 2^{100} = \dfrac{100!}{50!\,50!\,2^{100}}$. Both questions are about factorials, and factorials are *huge*: $100!$ is a product of a hundred numbers, most of them two-digit. A pocket calculator overflows at $70!$ (about $10^{100}$), and the double-precision floats used by NumPy give up at $171!$:

$$
170! \approx 7.3 \times 10^{306}, \qquad 171! \approx 1.2 \times 10^{309} > \text{largest float} \approx 1.8 \times 10^{308}.
$$

Is the answer to question 2 close to $50\%$? To $1\%$? Intuition is no help, and the direct calculation divides one astronomically large number by another.

A logarithm turns the product into a sum,

$$
\ln 100! = \ln 1 + \ln 2 + \ln 3 + \cdots + \ln 100 ,
$$

and a sum of values of a smooth function is approximately an **integral**, which we can do exactly: $\int \ln x\,dx = x\ln x - x$ (notebook 02). That one idea gives

$$
\ln n! \approx n \ln n - n + \tfrac12 \ln(2\pi n)
\qquad\Longleftrightarrow\qquad
n! \approx \sqrt{2\pi n}\left(\frac{n}{e}\right)^n ,
$$

and with it, in a few lines: $100!$ has $158$ digits, and the probability of exactly $50$ heads is about $\dfrac{1}{\sqrt{50\pi}} \approx 8\%$.

## 2. Theory

### 2.1 From a product to a sum

In exponent form $n! = 1 \cdot 2 \cdot 3 \cdots n$; in log form

$$
\ln n! = \sum_{k=1}^{n} \ln k .
$$

A product of $n$ numbers is hard to estimate; a sum of $n$ slowly varying numbers is easy, because it is a Riemann sum.

### 2.2 Squeezing the sum between two integrals

The function $\ln x$ is increasing, so on each interval $[k-1, k]$ the value $\ln k$ (the right end) is at least the average of $\ln x$ there, and on $[k, k+1]$ it is at most the average:

$$
\int_{k-1}^{k} \ln x\,dx \;\le\; \ln k \;\le\; \int_{k}^{k+1} \ln x\,dx .
$$

Add these up for $k = 2, \dots, n$ on the left and $k = 1, \dots, n$ on the right (remember $\ln 1 = 0$):

$$
\int_1^n \ln x\,dx \;\le\; \ln n! \;\le\; \int_1^{n+1} \ln x\,dx .
$$

With $\int \ln x\,dx = x\ln x - x$:

$$
n\ln n - n + 1 \;\le\; \ln n! \;\le\; (n+1)\ln(n+1) - n .
$$

Both bounds are $n\ln n - n$ plus terms of size $\ln n$. Since $n \ln n - n$ grows much faster than $\ln n$,

$$
\ln n! = n \ln n - n + O(\ln n), \qquad\text{i.e.}\qquad \ln n! \sim n\ln n .
$$

In exponent form, $n! = (n/e)^n$ times a factor that is "small" compared with the main term. Geometrically: $\ln n!$ is the area of a staircase of unit-width bars of heights $\ln 1, \ln 2, \dots, \ln n$, and the curve $\ln x$ runs through it (section 5.1).

### 2.3 The trapezoid correction: $\tfrac12 \ln n$

The bars overshoot the curve by small triangles. A better approximation of the area under $\ln x$ is the **trapezoid rule**, which uses the average of the two ends of each interval:

$$
\int_1^n \ln x\,dx \approx \tfrac12\ln 1 + \ln 2 + \cdots + \ln(n-1) + \tfrac12\ln n = \ln n! - \tfrac12 \ln n .
$$

So

$$
\ln n! \approx n\ln n - n + 1 + \tfrac12 \ln n .
$$

The error of the trapezoid rule on $[k, k+1]$ is about $\frac{1}{12}|f''| = \frac{1}{12k^2}$, and $\sum 1/k^2$ converges, so the total error tends to a **constant**. The true formula therefore has the form $\ln n! = n\ln n - n + \tfrac12\ln n + c + (\text{something} \to 0)$, and the constant turns out to be not $1$ but

$$
c = \tfrac12 \ln(2\pi) \approx 0.9189 .
$$

Identifying this constant takes more than the trapezoid rule (one route is the Wallis product for $\pi$). Historically, Abraham de Moivre found the form $n! \approx C\,n^{n+1/2}e^{-n}$, and James Stirling showed that $C = \sqrt{2\pi}$; both results date from around 1730.

### 2.4 Stirling's formula, two ways

$$
\boxed{\;\ln n! \approx n\ln n - n + \tfrac12\ln(2\pi n)\;}
\qquad\Longleftrightarrow\qquad
\boxed{\;n! \approx \sqrt{2\pi n}\left(\frac{n}{e}\right)^n\;}
$$

The **ratio** of the two sides tends to $1$ — the **relative** error goes to $0$, roughly like $\frac{1}{12n}$. (The absolute error $n! - \sqrt{2\pi n}(n/e)^n$ grows without bound; only relative accuracy improves.) A sharper version multiplies by $\left(1 + \frac{1}{12n}\right)$.

For digits, divide by $\ln 10$:

$$
\log_{10} n! \approx n\log_{10}\frac{n}{e} + \tfrac12\log_{10}(2\pi n), \qquad \text{digits of } n! = \lfloor\log_{10} n!\rfloor + 1 .
$$

### 2.5 The central binomial coefficient

Apply Stirling to $\dbinom{2n}{n} = \dfrac{(2n)!}{(n!)^2}$:

$$
\binom{2n}{n} \approx \frac{\sqrt{4\pi n}\,(2n/e)^{2n}}{2\pi n\,(n/e)^{2n}} = \frac{4^n}{\sqrt{\pi n}} .
$$

Dividing by $2^{2n} = 4^n$, the probability of exactly $n$ heads in $2n$ fair flips is

$$
P(n \text{ heads in } 2n \text{ flips}) \approx \frac{1}{\sqrt{\pi n}} .
$$

The most likely outcome becomes **less** likely as $n$ grows, but only like $1/\sqrt n$ — the probability spreads over about $\sqrt n$ outcomes near the middle.

### 2.6 `math.lgamma`

The gamma function satisfies $\Gamma(n+1) = n!$, and Python's `math.lgamma(x)` returns $\ln|\Gamma(x)|$ computed directly, without ever forming the huge number. So `math.lgamma(n + 1)` is $\ln n!$ to full double precision, for $n$ up to about $10^{305}$.

## 3. Worked Examples

### Example 1 — digits of $100!$

$\ln 100! \approx 100\ln 100 - 100 + \tfrac12\ln(200\pi) = 460.517 - 100 + 3.222 = 363.739$.

$\log_{10} 100! = 363.739 / \ln 10 = 363.739 / 2.302585 \approx 157.97$, so $100!$ has $\lfloor 157.97 \rfloor + 1 = 158$ digits. (Exact: $100! \approx 9.33 \times 10^{157}$.)

The crude form $n\ln n - n = 360.517$ alone gives $\log_{10} \approx 156.57$, i.e. $157$ digits: it misses by a factor $e^{3.22} \approx 25$, which is the $\sqrt{2\pi n}$ term.

### Example 2 — exactly 50 heads in 100 flips

Here $2n = 100$, $n = 50$: $P \approx \dfrac{1}{\sqrt{50\pi}} = \dfrac{1}{12.533} \approx 0.07979$. The exact value is $\dbinom{100}{50}/2^{100} \approx 0.07959$. Stirling is off by $0.25\%$.

### Example 3 — $10!$

$\sqrt{20\pi}\,(10/e)^{10} = 7.9267 \times 453{,}999.3 \approx 3{,}598{,}696$, versus $10! = 3{,}628{,}800$. The ratio is $0.99170$, i.e. $0.83\%$ too small — close to $\frac{1}{12 \cdot 10} = 0.83\%$. Multiplying by $1 + \frac{1}{120}$ gives $3{,}628{,}685$, off by only $0.003\%$.

### Example 4 — the bounds of 2.2 for $n = 10$

$\ln 10! = 15.104$. Lower bound $10\ln 10 - 10 + 1 = 14.026$; upper bound $11\ln 11 - 10 = 16.377$. The true value sits between them, and Stirling gives $15.096$.

## 4. Python Examples

### 4.1 Floats overflow, logarithms do not

`math.factorial` returns exact Python integers (any size), but converting to a float fails beyond $170!$. Its logarithm, from `math.lgamma`, is a modest number.

```python
import math

for n in [20, 100, 170, 171, 1000, 10**6]:
    log_factorial = math.lgamma(n + 1)
    try:
        as_float = f"{float(math.factorial(n)):.4e}" if n <= 1000 else "(not attempted)"
    except OverflowError:
        as_float = "OverflowError"
    print(f"n = {n:>8}:  ln n! = {log_factorial:>14.3f}   log10 n! = {log_factorial / math.log(10):>12.3f}   float(n!) = {as_float}")
```

### 4.2 The two approximations against the truth

```python
import math


def stirling_crude(n):
    return n * math.log(n) - n


def stirling_full(n):
    return n * math.log(n) - n + 0.5 * math.log(2 * math.pi * n)


print(f"{'n':>8} {'ln n! (lgamma)':>16} {'n ln n - n':>14} {'+ 1/2 ln(2 pi n)':>17} {'difference':>12} {'1/(12n)':>10}")
for n in [1, 2, 5, 10, 50, 100, 1000, 10**4]:
    exact = math.lgamma(n + 1)
    print(f"{n:>8} {exact:>16.6f} {stirling_crude(n):>14.6f} {stirling_full(n):>17.6f} "
          f"{exact - stirling_full(n):>12.2e} {1 / (12 * n):>10.2e}")
```

The difference between $\ln n!$ and the full Stirling formula is almost exactly $\frac{1}{12n}$ — that is the next correction term. In log form an additive error $\varepsilon$ is a relative error $e^{\varepsilon} - 1 \approx \varepsilon$ in $n!$ itself.

### 4.3 Digits of $n!$ — count them, then predict them

```python
import math

for n in [10, 52, 100, 500, 1000]:
    exact_digits = len(str(math.factorial(n)))
    log10_stirling = (n * math.log(n) - n + 0.5 * math.log(2 * math.pi * n)) / math.log(10)
    print(f"{n:>6}! has {exact_digits:>6} digits;  Stirling log10 = {log10_stirling:12.4f} -> {math.floor(log10_stirling) + 1:>6} digits")
```

### 4.4 Exactly $n$ heads in $2n$ flips

Exact probabilities use Python's exact integers via `math.comb`; for very large $n$ we compute the logarithm of the probability with `lgamma` instead, so nothing overflows.

```python
import math


def prob_exactly_half_exact(two_n):
    n = two_n // 2
    return math.comb(two_n, n) / 2**two_n


def prob_exactly_half_lgamma(two_n):
    n = two_n // 2
    log_probability = math.lgamma(two_n + 1) - 2 * math.lgamma(n + 1) - two_n * math.log(2)
    return math.exp(log_probability)


for two_n in [10, 100, 1000, 10_000, 10**6]:
    n = two_n // 2
    exact = prob_exactly_half_exact(two_n) if two_n <= 10_000 else prob_exactly_half_lgamma(two_n)
    stirling = 1 / math.sqrt(math.pi * n)
    print(f"{two_n:>8} flips: P(exactly half heads) = {exact:.6f}   1/sqrt(pi n) = {stirling:.6f}   ratio = {exact / stirling:.5f}")
```

The ratio approaches $1$ like $1 - \frac{1}{8n}$. For $100$ flips the answer to the opening question is just under $8\%$.

### 4.5 SymPy: the limit that defines Stirling's formula

SymPy confirms $\lim_{n\to\infty} \dfrac{n!}{\sqrt{2\pi n}\,(n/e)^n} = 1$, and its asymptotic series for $\ln\Gamma$ reproduces the $\frac{1}{12n}$ term.

```python
import sympy as sp

n = sp.symbols("n", positive=True)
ratio = sp.factorial(n) / (sp.sqrt(2 * sp.pi * n) * (n / sp.E)**n)
print("limit of n! / Stirling:", sp.limit(ratio, n, sp.oo))

x = sp.symbols("x", positive=True)
series = sp.series(sp.loggamma(x + 1) - (x * sp.log(x) - x + sp.log(2 * sp.pi * x) / 2), x, sp.oo, n=4)
print("ln n! - Stirling  =", sp.apart(sp.simplify(sp.expand_log(series.removeO(), force=True)), x), "+ O(1/n^4)")
```

## 5. Visualizations

### 5.1 The staircase and the curve

$\ln n!$ is the total area of the bars of heights $\ln 2, \ln 3, \dots, \ln n$ (each of width $1$, sitting on $[k-1, k]$). The curve $\ln x$ runs below the bar tops; the area under it from $1$ to $n$ is $n\ln n - n + 1$. The small gaps between bars and curve are almost triangles — the trapezoid correction $\tfrac12\ln n$.

```python
import matplotlib.pyplot as plt
import numpy as np

n = 12
k = np.arange(2, n + 1)
x = np.linspace(1, n, 400)

fig, ax = plt.subplots(figsize=(10, 5))
ax.bar(k - 0.5, np.log(k), width=1.0, color="tab:blue", alpha=0.35, edgecolor="tab:blue",
       label=f"bars ln k on [k-1, k]: total = ln {n}! = {np.sum(np.log(k)):.3f}")
ax.plot(x, np.log(x), color="crimson", linewidth=2.5,
        label=f"ln x: area from 1 to {n} = {n * np.log(n) - n + 1:.3f}")
ax.fill_between(x, np.log(x), color="crimson", alpha=0.12)
ax.set_xlim(0.5, n + 0.5)
ax.set_xlabel("x")
ax.set_ylabel("ln x")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left")
ax.set_title("ln n! is a staircase area; the integral of ln x is the area under the curve")
plt.show()
```

### 5.2 Relative error on log-log axes

The relative error of the crude formula $(n/e)^n$ and of the full formula $\sqrt{2\pi n}(n/e)^n$, both computed in log form as $|1 - e^{\text{approx} - \ln n!}|$. On log-log axes the full formula's error is a straight line of slope $-1$ lying on top of $\frac{1}{12n}$: a power law. The crude formula's error tends to $1$ (100%), because it misses the growing factor $\sqrt{2\pi n}$.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

n_values = np.unique(np.logspace(0, 6, 120).astype(int))
exact = np.array([math.lgamma(n + 1) for n in n_values])
crude = n_values * np.log(n_values) - n_values
full = crude + 0.5 * np.log(2 * np.pi * n_values)

relative_error_crude = np.abs(1 - np.exp(crude - exact))
relative_error_full = np.abs(np.expm1(full - exact))

fig, ax = plt.subplots(figsize=(9, 5.5))
ax.plot(n_values, relative_error_crude, color="tab:orange", label="crude: $(n/e)^n$")
ax.plot(n_values, relative_error_full, color="tab:blue", linewidth=2.5, label="Stirling: $\\sqrt{2\\pi n}(n/e)^n$")
ax.plot(n_values, 1 / (12 * n_values), "--", color="black", label="$1/(12n)$")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("n (log scale)")
ax.set_ylabel("relative error (log scale)")
ax.grid(True, which="both", alpha=0.3)
ax.legend(loc="lower left")
ax.set_title("Relative error of Stirling's approximation")
plt.show()
```

### 5.3 How big is $n!$? Digits of $n!$ next to $n^n$ and $2^n$

The number of digits ($\approx\log_{10}$) of $2^n$, $n!$ and $n^n$. On linear axes $\log_{10} 2^n$ is a straight line, $\log_{10} n^n = n\log_{10} n$ curves upward, and $\log_{10} n!$ lies just below it, trailing by about $n\log_{10} e \approx 0.434\,n$ digits — the $(1/e)^n$ factor.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

n_values = np.arange(1, 201)
digits_power_of_two = n_values * math.log10(2)
digits_factorial = np.array([math.lgamma(n + 1) / math.log(10) for n in n_values])
digits_n_to_n = n_values * np.log10(n_values)

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(n_values, digits_n_to_n, color="tab:red", label="$\\log_{10} n^n$")
ax.plot(n_values, digits_factorial, color="tab:blue", linewidth=2.5, label="$\\log_{10} n!$")
ax.plot(n_values, digits_power_of_two, color="tab:green", label="$\\log_{10} 2^n$")
ax.plot(100, digits_factorial[99], "o", color="black")
ax.annotate("100! has 158 digits", xy=(100, digits_factorial[99]), xytext=(110, 70), arrowprops=dict(arrowstyle="->"))
ax.set_xlabel("n")
ax.set_ylabel("log10 of the number (about its digit count)")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left")
ax.set_title("Size of $n!$ compared with $2^n$ and $n^n$")
plt.show()
```

### 5.4 The middle of the binomial distribution

Probability of exactly half heads in $2n$ flips (dots) against $1/\sqrt{\pi n}$ (line), on log-log axes. A straight line of slope $-\tfrac12$: a power law, as section 2.5 predicts.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

half_values = np.unique(np.logspace(0, 5, 40).astype(int))
exact = [math.exp(math.lgamma(2 * n + 1) - 2 * math.lgamma(n + 1) - 2 * n * math.log(2)) for n in half_values]

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(2 * half_values, exact, "o", color="tab:blue", label="exact P(n heads in 2n flips)")
ax.plot(2 * half_values, 1 / np.sqrt(np.pi * half_values), color="tab:orange", label="$1/\\sqrt{\\pi n}$")
probability_100_flips = math.comb(100, 50) / 2**100
ax.plot(100, probability_100_flips, "s", color="crimson", markersize=9,
        label=f"100 flips: {probability_100_flips:.4f}")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("number of flips 2n (log scale)")
ax.set_ylabel("probability (log scale)")
ax.grid(True, which="both", alpha=0.3)
ax.legend(loc="upper right")
ax.set_title("Exactly half heads: probability falls like $1/\\sqrt{n}$")
plt.show()
```

8. $\ln\binom nk = \ln n! - \ln k! - \ln(n-k)! \approx (n\ln n - n) - (k\ln k - k) - ((n-k)\ln(n-k) - (n-k))$. 9. $\log_2 n! = \ln n!/\ln 2 \approx (n\ln n - n)/\ln 2 = n\log_2 n - n\log_2 e = n\log_2 n - 1.443\,n$. For $n = 52$: $\log_2 52! \approx 225.6$ comparisons at least (the crude formula gives $221.4$). For $n = 10^6$: about $1.85 \times 10^7$ comparisons.

```python
import math


def stirling_log(n):
    return n * math.log(n) - n + 0.5 * math.log(2 * math.pi * n)


print("Ex 1:", round(stirling_log(1000) / math.log(10), 3), " digits:", len(str(math.factorial(1000))))
print("Ex 2:", round(math.lgamma(51), 4), round(50 * math.log(50) - 50, 4), round(stirling_log(50), 4),
      " relative error:", round(math.expm1(math.lgamma(51) - stirling_log(50)), 5))
print("Ex 3:", round(20 * math.log(20) - 19, 3), round(math.lgamma(21), 3), round(21 * math.log(21) - 20, 3))
estimate_52 = 52 * (1.716 - 0.434) + 0.5 * math.log10(2 * math.pi * 52)
print("Ex 4: estimate", round(estimate_52, 2), " exact log10:", round(math.lgamma(53) / math.log(10), 3), " digits:", len(str(math.factorial(52))))
print("Ex 5:", round(1 / math.sqrt(500 * math.pi), 6), round(math.comb(1000, 500) / 2**1000, 6))
stirling_10 = math.sqrt(20 * math.pi) * (10 / math.e) ** 10
print("Ex 6:", round(stirling_10, 1), round(stirling_10 * (1 + 1 / 120), 1), " rel. error:", f"{abs(stirling_10 * (1 + 1 / 120) / math.factorial(10) - 1):.2e}")
print("Ex 7:", [round(n / math.exp(math.lgamma(n + 1) / n), 5) for n in [10, 1000, 10**6]])
p = 0.3
print("Ex 8: n H(p) =", round(-1000 * (p * math.log(p) + (1 - p) * math.log(1 - p)), 2), " exact:", round(math.log(math.comb(1000, 300)), 2))
print("Ex 9:", round(math.lgamma(53) / math.log(2), 2), round(52 * math.log2(52) - 52 * math.log2(math.e), 2),
      f"{math.lgamma(10**6 + 1) / math.log(2):.4e}")
```

## 8. Conclusion

- A factorial is a product, so its logarithm is a sum: $\ln n! = \sum_{k=1}^n \ln k$. A sum of a slowly varying function is close to its integral, and $\int \ln x\,dx = x\ln x - x$.
- Squeezing the sum between integrals gives $n\ln n - n + 1 \le \ln n! \le (n+1)\ln(n+1) - n$, so $\ln n! \sim n\ln n - n$.
- The trapezoid correction adds $\tfrac12\ln n$, and the constant is $\tfrac12\ln(2\pi)$ (de Moivre and Stirling, around 1730): $\ln n! \approx n\ln n - n + \tfrac12\ln(2\pi n)$, i.e. $n! \approx \sqrt{2\pi n}\,(n/e)^n$, with relative error about $\frac{1}{12n}$.
- In practice: work with $\ln n!$, not $n!$ — `math.lgamma(n + 1)` never overflows. $100!$ has $158$ digits; $52!$ has $68$.
- Applied to $\binom{2n}{n}$: the chance of exactly half heads in $2n$ flips is about $1/\sqrt{\pi n}$ — about $8\%$ for $100$ flips. Applied to $\binom nk$: $\ln\binom nk \approx nH(k/n)$, the entropy.
