# Logarithmic Differentiation

This notebook shows how taking a logarithm *before* differentiating turns impossible-looking derivatives into routine ones. The logarithm's old trick — products become sums, quotients become differences, powers become multiples — works on derivatives just as it worked on Napier's multiplications.

This is notebook 01 of module 04 (**Logarithms in calculus**) in the **Learning Logarithms** series.

## 1. Introduction

What is the derivative of

$$
f(x) = x^x \qquad (x > 0)?
$$

Both rules you know seem to apply, and both give the wrong answer:

| Rule | Assumes | Gives | Problem |
|---|---|---|---|
| power rule $\dfrac{d}{dx} x^n = n x^{n-1}$ | the **exponent** is a constant | $x \cdot x^{x-1} = x^x$ | the exponent is $x$, not a constant |
| exponential rule $\dfrac{d}{dx} a^x = a^x \ln a$ | the **base** is a constant | $x^x \ln x$ | the base is $x$, not a constant |

At $x = 1$ the first answer is $1$ and the second is $0$. A quick numerical check settles it: $\big(1.001^{1.001} - 0.999^{0.999}\big) / 0.002 \approx 1.000$. Neither rule is right in general, and the true answer turns out to be their **sum**.

The way out is to take a logarithm first. The exponent comes down as a factor,

$$
\ln f(x) = x \ln x,
$$

and $x \ln x$ is an ordinary product that the product rule handles. The same move tames derivatives of long products and quotients such as

$$
g(x) = \frac{x^2 (x+1)^3}{\sqrt{x^2+1}},
$$

where the quotient rule inside a product rule inside a chain rule is a recipe for mistakes. After a logarithm it is a short sum.

## 2. Theory

### 2.1 The method

For a positive function $f$, the chain rule applied to $\ln f(x)$ gives

$$
\frac{d}{dx} \ln f(x) = \frac{f'(x)}{f(x)}
\qquad\Longrightarrow\qquad
f'(x) = f(x) \cdot \frac{d}{dx} \ln f(x).
$$

**Logarithmic differentiation** is this identity used as a procedure:

1. Write $y = f(x)$ and take $\ln$ of both sides.
2. Use the laws of logarithms to break $\ln y$ into a sum: products → sums, quotients → differences, powers → multiples.
3. Differentiate both sides; the left side is $y'/y$ (implicit differentiation).
4. Multiply by $y$.

Step 2 is where the logarithm earns its keep. It is the same law that drove Napier's tables,

$$
\ln(uv) = \ln u + \ln v \qquad\text{(log form)} \qquad\Longleftrightarrow\qquad e^{a} e^{b} = e^{a+b} \qquad\text{(exponent form)},
$$

now applied to functions instead of numbers.

### 2.2 $x^x$, and $u^v$ in general

Take $y = x^x$, $x > 0$. Then $\ln y = x \ln x$, and differentiating,

$$
\frac{y'}{y} = 1 \cdot \ln x + x \cdot \frac1x = \ln x + 1
\qquad\Longrightarrow\qquad
\frac{d}{dx} x^x = x^x (\ln x + 1) = \underbrace{x^x}_{\text{power rule}} + \underbrace{x^x \ln x}_{\text{exponential rule}}.
$$

Both naive answers from the introduction appear — as the two halves of the true one. This is no accident. For $y = u(x)^{v(x)}$ with $u > 0$, $\ln y = v \ln u$ and

$$
\frac{d}{dx} u^v = u^v\left( v' \ln u + v \frac{u'}{u} \right) = \underbrace{v\,u^{v-1} u'}_{\text{treat } v \text{ as constant}} + \underbrace{u^v \ln u \; v'}_{\text{treat } u \text{ as constant}}.
$$

When a quantity depends on $x$ in two places, its derivative is the sum of the contributions from each place, holding the other fixed. The power rule and the exponential rule are each one of those contributions.

Another way to see the same thing: $x^x = e^{x \ln x}$ by definition of real powers, so the chain rule gives $e^{x \ln x}(\ln x + 1)$ directly. Logarithmic differentiation and "rewrite as $e^{\ln}$" are the same calculation.

### 2.3 Products and quotients

For $y = \dfrac{u_1^{a_1} u_2^{a_2} \cdots}{w_1^{b_1} w_2^{b_2} \cdots}$ with all factors positive,

$$
\ln y = a_1 \ln u_1 + a_2 \ln u_2 + \cdots - b_1 \ln w_1 - b_2 \ln w_2 - \cdots
$$

$$
\frac{y'}{y} = a_1 \frac{u_1'}{u_1} + a_2 \frac{u_2'}{u_2} + \cdots - b_1 \frac{w_1'}{w_1} - b_2 \frac{w_2'}{w_2} - \cdots
$$

Every factor contributes one simple term, independently of the others. (If some factors can be negative, apply the method to $|y|$: $\ln|y|$ has the same derivative formula, $\frac{d}{dx}\ln|u| = u'/u$, so the final answer is unchanged. Notebook 02 explains why.)

### 2.4 $f'/f$ is the relative rate of change

The quantity

$$
\frac{f'(x)}{f(x)} = \frac{d}{dx} \ln f(x)
$$

is called the **logarithmic derivative**. It measures change **as a fraction of the current value**: a population with $P'/P = 0.02$ per year is growing by 2% per year, whether it has a hundred members or a billion. Section 2.3 then says:

- **relative rates of a product add**: if revenue $R = p \cdot q$ (price times quantity), then $R'/R = p'/p + q'/q$;
- **relative rates of a quotient subtract**: GDP per person grows at (GDP growth) − (population growth);
- **a power multiplies the rate**: the area of a growing square $A = s^2$ grows at twice the relative rate of its side.

On a plot with a **logarithmic $y$-axis**, the height of the curve is $\ln f$ (up to a constant factor), so the **slope on a log axis is the relative rate**. A constant relative rate is a straight line on such a plot — the exponential functions of notebook 03.

### 2.5 The minimum of $x^x$

On $(0, \infty)$, $\frac{d}{dx} x^x = x^x(\ln x + 1)$. Since $x^x > 0$, the sign is the sign of $\ln x + 1$:

$$
\ln x + 1 = 0 \iff \ln x = -1 \iff x = e^{-1} = \frac1e \approx 0.3679.
$$

For $x < 1/e$ the derivative is negative, for $x > 1/e$ positive, so $x = 1/e$ is the global minimum, with value

$$
\left(\tfrac1e\right)^{1/e} = e^{-1/e} \approx 0.6922.
$$

It is easier to minimise $\ln f = x\ln x$ instead of $f$: because $\ln$ is increasing, $f$ and $\ln f$ have their minima at the same place. This "optimise the log instead" trick returns in maximum likelihood (module 05). Also note $\lim_{x \to 0^+} x^x = 1$, because $x \ln x \to 0$.

## 3. Worked Examples

### Example 1 — $\dfrac{d}{dx} x^x$ at $x = 2$

$y = x^x$, $y' = x^x(\ln x + 1)$. At $x = 2$: $y' = 4(\ln 2 + 1) \approx 4 \times 1.6931 = 6.7726$.

### Example 2 — the messy quotient

$g(x) = \dfrac{x^2 (x+1)^3}{\sqrt{x^2+1}}$, $x > 0$. Take logs:

$$
\ln g = 2\ln x + 3 \ln(x+1) - \tfrac12 \ln(x^2+1).
$$

Differentiate term by term:

$$
\frac{g'}{g} = \frac{2}{x} + \frac{3}{x+1} - \frac{x}{x^2+1}
\qquad\Longrightarrow\qquad
g'(x) = \frac{x^2 (x+1)^3}{\sqrt{x^2+1}} \left( \frac{2}{x} + \frac{3}{x+1} - \frac{x}{x^2+1} \right).
$$

At $x = 1$: $g(1) = \dfrac{1 \cdot 8}{\sqrt 2} = 4\sqrt2 \approx 5.6569$ and $\dfrac{g'}{g} = 2 + \tfrac32 - \tfrac12 = 3$, so $g'(1) = 12\sqrt2 \approx 16.9706$.

### Example 3 — $x^{\sin x}$

$\ln y = \sin x \ln x$, so $\dfrac{y'}{y} = \cos x \ln x + \dfrac{\sin x}{x}$ and

$$
\frac{d}{dx} x^{\sin x} = x^{\sin x}\left(\cos x \ln x + \frac{\sin x}{x}\right).
$$

At $x = \pi/2$: $\cos(\pi/2) = 0$, so $y' = (\pi/2)^1 \cdot \dfrac{1}{\pi/2} = 1$.

### Example 4 — relative rates add

A shop raises its price by 3% per year while the number of items sold grows by 2% per year (both as continuous rates, $p'/p = 0.03$, $q'/q = 0.02$). Revenue $R = pq$ has $R'/R = 0.03 + 0.02 = 0.05$: revenue grows at 5% per year. No need to know $p$ or $q$ themselves.

### Example 5 — the minimum of $x^x$

From section 2.5: the minimum is at $x = 1/e \approx 0.36788$, value $e^{-1/e} \approx 0.69220$. Check by sampling: $0.3^{0.3} \approx 0.6968$, $0.4^{0.4} \approx 0.6931$ — both larger, as they should be.

## 4. Python Examples

### 4.1 SymPy confirms $\frac{d}{dx}x^x$

SymPy differentiates $x^x$ directly; we compare with $x^x(\ln x + 1)$ and with both naive rules.

```python
import sympy as sp

x = sp.symbols("x", positive=True)
f = x**x

derivative = sp.diff(f, x)
our_formula = x**x * (sp.log(x) + 1)
power_rule_guess = x * x**(x - 1)
exponential_rule_guess = x**x * sp.log(x)

print("SymPy:           ", derivative)
print("matches formula: ", sp.simplify(derivative - our_formula) == 0)
print("power + exponential guesses:", sp.simplify(power_rule_guess + exponential_rule_guess - derivative) == 0)
print("value at x = 2:  ", sp.N(derivative.subs(x, 2), 8))
```

### 4.2 A finite-difference check

A central difference $\dfrac{f(x+h) - f(x-h)}{2h}$ approximates $f'(x)$ without any calculus. It agrees with the logarithmic-differentiation formula, and with neither naive rule.

```python
import math


def x_to_the_x(x):
    return x**x


def central_difference(function, x, h=1e-6):
    return (function(x + h) - function(x - h)) / (2 * h)


for x in [0.5, 1.0, 2.0, 3.0]:
    numeric = central_difference(x_to_the_x, x)
    formula = x**x * (math.log(x) + 1)
    power_rule = x**x
    exponential_rule = x**x * math.log(x)
    print(f"x = {x}: numeric {numeric:9.5f}  formula {formula:9.5f}  "
          f"power-rule guess {power_rule:8.4f}  exp-rule guess {exponential_rule:8.4f}  "
          f"agree: {math.isclose(numeric, formula, rel_tol=1e-6)}")
```

### 4.3 The messy quotient, three ways

Direct differentiation with SymPy, logarithmic differentiation done by hand, and a finite difference all agree at $x = 1$.

```python
import math

import sympy as sp

x = sp.symbols("x", positive=True)
g = x**2 * (x + 1)**3 / sp.sqrt(x**2 + 1)

direct = sp.diff(g, x)
log_rate = 2 / x + 3 / (x + 1) - x / (x**2 + 1)
by_logs = g * log_rate

print("difference of the two symbolic answers:", sp.simplify(direct - by_logs))
print("g(1)        =", sp.nsimplify(g.subs(x, 1)), "=", float(g.subs(x, 1)))
print("g'(1)/g(1)  =", log_rate.subs(x, 1))
print("g'(1)       =", sp.simplify(by_logs.subs(x, 1)), "=", float(by_logs.subs(x, 1)))

g_numeric = sp.lambdify(x, g)
h = 1e-6
print("finite diff =", (g_numeric(1 + h) - g_numeric(1 - h)) / (2 * h))
```

### 4.4 A general helper: $\frac{d}{dx} u^v$

The formula $u^v(v' \ln u + v u'/u)$ from section 2.2 as a function, checked against SymPy on a few examples.

```python
import sympy as sp

x = sp.symbols("x", positive=True)


def derivative_of_power(base, exponent):
    base_prime = sp.diff(base, x)
    exponent_prime = sp.diff(exponent, x)
    return base**exponent * (exponent_prime * sp.log(base) + exponent * base_prime / base)


examples = [(x, x), (x, sp.sin(x)), (x**2 + 1, sp.sin(x)), (sp.log(x + 1), x)]
for base, exponent in examples:
    ours = derivative_of_power(base, exponent)
    sympys = sp.diff(base**exponent, x)
    print(f"d/dx ({base})^({exponent}): agrees with SymPy -> {sp.simplify(ours - sympys) == 0}")

print("x^sin(x) at pi/2:", sp.simplify(derivative_of_power(x, sp.sin(x)).subs(x, sp.pi / 2)))
```

### 4.5 The minimum of $x^x$

Solve $\ln x + 1 = 0$ with SymPy, then confirm on a fine grid with NumPy.

```python
import math

import numpy as np
import sympy as sp

x = sp.symbols("x", positive=True)
critical_points = sp.solve(sp.Eq(sp.diff(x * sp.log(x), x), 0), x)
print("critical point of x ln x:", critical_points, "=", float(critical_points[0]))
print("minimum value e^(-1/e):  ", math.exp(-1 / math.e))

grid = np.linspace(0.01, 2, 200_001)
values = grid**grid
index_of_min = np.argmin(values)
print("grid minimum at x =", round(grid[index_of_min], 5), " value", round(values[index_of_min], 5))
print("close to 1/e:", math.isclose(grid[index_of_min], 1 / math.e, abs_tol=1e-4))
```

### 4.6 Relative rates of a product add

Price grows like $p(t) = 10\,e^{0.03t}$ and quantity like $q(t) = 500\,e^{0.02t}$. The relative rate of revenue, computed numerically from $R = pq$, is $0.05$ at every time.

```python
import numpy as np

t = np.linspace(0, 20, 2001)
price = 10 * np.exp(0.03 * t)
quantity = 500 * np.exp(0.02 * t)
revenue = price * quantity

relative_rate_of_revenue = np.gradient(revenue, t) / revenue
print("relative rate of revenue, min and max:", relative_rate_of_revenue[1:-1].min(), relative_rate_of_revenue[1:-1].max())
print("all close to 0.05:", np.allclose(relative_rate_of_revenue[1:-1], 0.05, atol=1e-5))

# The same rate from the slope of ln R
slope_of_log = np.polyfit(t, np.log(revenue), 1)[0]
print("slope of ln R:", round(slope_of_log, 10))
```

## 5. Visualizations

### 5.1 $x^x$, its minimum, and its derivative

The curve dips to its minimum $e^{-1/e} \approx 0.692$ at $x = 1/e$, where the derivative $x^x(\ln x + 1)$ crosses zero. As $x \to 0^+$ the function tends to $1$.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0.005, 2, 800)
f = x**x
f_prime = x**x * (np.log(x) + 1)

fig, (ax_f, ax_df) = plt.subplots(1, 2, figsize=(12, 4.5))
ax_f.plot(x, f, color="tab:blue", label="$x^x$")
x_min = 1 / math.e
ax_f.plot(x_min, x_min**x_min, "o", color="crimson")
ax_f.annotate(f"minimum at x = 1/e\nvalue = {x_min**x_min:.4f}", xy=(x_min, x_min**x_min),
              xytext=(0.6, 2.0), arrowprops=dict(arrowstyle="->"))
ax_f.set_xlabel("x")
ax_f.set_ylabel("$x^x$")
ax_f.set_ylim(0, 4.2)
ax_f.grid(True, alpha=0.3)
ax_f.legend(loc="upper left")
ax_f.set_title("$x^x$ has a minimum at $1/e$")

ax_df.plot(x, f_prime, color="tab:orange", label="$x^x(\\ln x + 1)$")
ax_df.axhline(0, color="black", linewidth=0.8)
ax_df.axvline(x_min, color="crimson", linestyle="--", label="x = 1/e")
ax_df.set_xlabel("x")
ax_df.set_ylabel("derivative")
ax_df.set_ylim(-4, 8)
ax_df.grid(True, alpha=0.3)
ax_df.legend(loc="upper left")
ax_df.set_title("The derivative changes sign at $1/e$")
plt.tight_layout()
plt.show()
```

### 5.2 The derivative is "power rule + exponential rule"

The two naive answers $x^x$ (exponent frozen) and $x^x \ln x$ (base frozen) are each wrong, but their sum is exactly the true derivative.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0.05, 2.2, 600)
power_rule_part = x**x
exponential_rule_part = x**x * np.log(x)
true_derivative = x**x * (np.log(x) + 1)

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(x, power_rule_part, "--", color="tab:blue", label="power-rule part: $x \\cdot x^{x-1} = x^x$")
ax.plot(x, exponential_rule_part, "--", color="tab:green", label="exponential-rule part: $x^x \\ln x$")
ax.plot(x, true_derivative, color="black", linewidth=2.5, label="true derivative: sum of both")
ax.axhline(0, color="gray", linewidth=0.8)
ax.set_xlabel("x")
ax.set_ylabel("value")
ax.set_ylim(-1.5, 12)
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left")
ax.set_title("$\\frac{d}{dx}x^x$ = (exponent held fixed) + (base held fixed)")
plt.show()
```

### 5.3 Slope on a log axis is the relative rate

Two functions with the same **relative** growth rate $0.05$ but very different sizes, $R_1 = 5000\,e^{0.05t}$ and $R_2 = 50\,e^{0.05t}$. On a linear axis (left) $R_1$ climbs much faster in absolute terms. On a logarithmic axis (right) they are **parallel straight lines**: the slope there is $\frac{d}{dt}\ln R = R'/R$, the same for both. A third curve with rate $0.08$ is steeper on the log axis.

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(0, 40, 400)
large = 5000 * np.exp(0.05 * t)
small = 50 * np.exp(0.05 * t)
faster = 50 * np.exp(0.08 * t)

fig, (ax_lin, ax_log) = plt.subplots(1, 2, figsize=(12, 4.5))
for ax in (ax_lin, ax_log):
    ax.plot(t, large, color="tab:blue", label="5000 e^(0.05 t)")
    ax.plot(t, small, color="tab:orange", label="50 e^(0.05 t)")
    ax.plot(t, faster, color="tab:green", label="50 e^(0.08 t)")
    ax.set_xlabel("t")
    ax.legend(loc="upper left")

ax_lin.set_ylabel("value (linear scale)")
ax_lin.grid(True, alpha=0.3)
ax_lin.set_title("Linear axis: absolute change")

ax_log.set_yscale("log")
ax_log.set_ylabel("value (log scale)")
ax_log.grid(True, which="both", alpha=0.3)
ax_log.set_title("Log axis: slope = relative rate f'/f")
plt.tight_layout()
plt.show()
```

## 7. Solutions

1. $\ln y = n \ln x$, so $y'/y = n/x$ and $y' = x^n \cdot n/x = n x^{n-1}$. Nothing assumed $n$ is an integer.
2. $\ln y = 2\ln(x+1) + 3\ln(x+2) + 4\ln(x+3)$, so $\dfrac{y'}{y} = \dfrac{2}{x+1} + \dfrac{3}{x+2} + \dfrac{4}{x+3}$. At $0$: $y(0) = 1 \cdot 8 \cdot 81 = 648$ and $\dfrac{y'}{y} = 2 + \tfrac32 + \tfrac43 = \tfrac{29}{6}$, so $y'(0) = 648 \cdot \tfrac{29}{6} = 3132$.
3. $\ln y = \dfrac{\ln x}{x}$, $\dfrac{y'}{y} = \dfrac{1 - \ln x}{x^2}$, so $y' = x^{1/x}\dfrac{1 - \ln x}{x^2}$. It is positive for $x < e$ and negative for $x > e$: maximum at $x = e$, value $e^{1/e} \approx 1.4447$.
4. Estimate: $2.0\% - 0.5\% = 1.5\%$. With continuous rates this is exact: $(\ln(Y/P))' = Y'/Y - P'/P = 0.015$. With yearly factors: $1.02 / 1.005 = 1.014925$, i.e. $1.49\%$ — the subtraction rule is exact for logarithmic rates and very good for small yearly percentages.
5. $\ln y = (\ln x)^2$, $\dfrac{y'}{y} = \dfrac{2\ln x}{x}$, so $y' = 2 x^{\ln x}\dfrac{\ln x}{x} = 2 x^{\ln x - 1}\ln x$. At $x = e$: $y' = 2e^{0} \cdot 1 = 2$.
6. $y(2) = 4$, $y'(2) = 4(1 + \ln 2) \approx 6.7726$. Tangent: $y = 4 + 6.7726(x - 2)$, i.e. $y \approx 6.7726x - 9.5452$.
7. $\ln y = x^x \ln x$. Differentiate with the product rule, using $(x^x)' = x^x(\ln x + 1)$: $\dfrac{y'}{y} = x^x(\ln x + 1)\ln x + \dfrac{x^x}{x}$. So $y' = x^{x^x} x^x\left(\ln x(\ln x + 1) + \dfrac1x\right)$. At $x = 1$: $y' = 1 \cdot 1 \cdot (0 + 1) = 1$. At $x = 2$: $y = 2^4 = 16$, $x^x = 4$, so $y' = 64(\ln 2 (1 + \ln 2) + \tfrac12) \approx 107.11$.
8. $h(x) = \ln x / x$ has $h'(x) = (1 - \ln x)/x^2$, so its maximum is at $x = e$. Hence $h(\pi) < h(e)$: $\dfrac{\ln \pi}{\pi} < \dfrac1e$, i.e. $e \ln \pi < \pi$, i.e. $\ln(\pi^e) < \ln(e^\pi)$. Because $\ln$ is increasing, $\pi^e < e^\pi$. Numerically $\pi^e \approx 22.459$ and $e^\pi \approx 23.141$.
9. $\ln(uvw) = \ln u + \ln v + \ln w$, so $\dfrac{(uvw)'}{uvw} = \dfrac{u'}{u} + \dfrac{v'}{v} + \dfrac{w'}{w}$; multiply by $uvw$ to get $u'vw + uv'w + uvw'$. The logarithm turns the product into a sum, and the derivative of a sum is the sum of the derivatives.

Numerical checks:

```python
import math

import sympy as sp

x = sp.symbols("x", positive=True)

y2 = (x + 1)**2 * (x + 2)**3 * (x + 3)**4
print("Ex 2: y'(0) =", sp.diff(y2, x).subs(x, 0))

y3 = x**(1 / x)
print("Ex 3: critical points:", sp.solve(sp.diff(sp.log(y3), x), x), " max value e^(1/e) =", math.exp(1 / math.e))

print("Ex 4: continuous 0.020 - 0.005 =", 0.020 - 0.005, "  yearly factors:", 1.02 / 1.005 - 1)

y5 = x**sp.log(x)
print("Ex 5: y'(e) =", sp.simplify(sp.diff(y5, x).subs(x, sp.E)))

slope = 4 * (1 + math.log(2))
print("Ex 6: slope =", round(slope, 4), " intercept =", round(4 - 2 * slope, 4))

y7 = x**(x**x)
d7 = sp.diff(y7, x)
print("Ex 7: y'(1) =", sp.simplify(d7.subs(x, 1)), "  y'(2) =", round(float(d7.subs(x, 2)), 4),
      "  formula at 2:", round(64 * (math.log(2) * (1 + math.log(2)) + 0.5), 4))

print("Ex 8: pi^e =", round(math.pi**math.e, 4), "  e^pi =", round(math.e**math.pi, 4), "  pi^e < e^pi:", math.pi**math.e < math.e**math.pi)

u, v, w = sp.Function("u")(x), sp.Function("v")(x), sp.Function("w")(x)
three_factor = sp.diff(u * v * w, x) - (sp.diff(u, x) * v * w + u * sp.diff(v, x) * w + u * v * sp.diff(w, x))
print("Ex 9: three-factor product rule holds:", sp.simplify(three_factor) == 0)
```

## 8. Conclusion

- **Logarithmic differentiation**: for $f > 0$, $f' = f \cdot (\ln f)'$. Take $\ln$, break the expression into a sum with the laws of logarithms, differentiate, multiply back.
- It is the only practical route to variable powers: $\dfrac{d}{dx}x^x = x^x(\ln x + 1)$, and in general $\dfrac{d}{dx}u^v = v u^{v-1}u' + u^v \ln u\, v'$ — the power rule and the exponential rule **added**, each holding one of the two $x$'s fixed.
- Long products and quotients become sums: every factor contributes one term $\pm a\,u'/u$, independently of the others.
- $f'/f = (\ln f)'$ is the **relative rate of change**. Relative rates of products add, of quotients subtract, and powers multiply them. On a log-scaled axis the slope *is* the relative rate.
- Optimising $\ln f$ instead of $f$ finds the same point (since $\ln$ is increasing) and is often easier: $x^x$ has its minimum $e^{-1/e}$ at $x = 1/e$, and $x^{1/x}$ its maximum $e^{1/e}$ at $x = e$.
